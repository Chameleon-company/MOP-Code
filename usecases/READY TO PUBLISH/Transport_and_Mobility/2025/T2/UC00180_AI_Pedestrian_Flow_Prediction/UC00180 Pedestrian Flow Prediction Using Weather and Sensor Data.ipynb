<div id="section-2" class="usecase-section-heading">
<b>2. Importing Datasets</b>
</div>

This section imports the datasets required for the pedestrian flow prediction model. The data is retrieved from the City of Melbourne Open Data API and loaded into pandas DataFrames for further cleaning, exploration and modelling.

```python
def fetch_data(base_url, dataset, api_key, num_records=99, offset=0):
    all_records = []
    max_offset = 9900  # Maximum number of requests

    while True:
        # Maximum limit check
        if offset > max_offset:
            break

        # Create API request URL
        filters = f'{dataset}/records?limit={num_records}&offset={offset}'
        url = f'{base_url}{filters}&api_key={api_key}'

        # Start request
        try:
            result = requests.get(url, timeout=10)
            result.raise_for_status()
            records = result.json().get('results')
        except requests.exceptions.RequestException as e:
            raise Exception(f"API request failed: {e}")
        if records is None:
            break
        all_records.extend(records)
        if len(records) < num_records:
            break

        # Next cycle offset
        offset += num_records

    # DataFrame all data
    df = pd.DataFrame(all_records)
    return df

#os.environ.get('MELBOURNE_API_KEY')

# Retrieve API key from environment variable
API_KEY = os.environ.get('MELBOURNE_API_KEY')

# Base URL for the City of Melbourne Open Data API
BASE_URL = 'https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/'
```

<div id="section-3-2-5" class="usecase-sub-sub-section-heading">
<b>3.2.5 Visualising Weather Station Features Against Pedestrian Count</b>
</div>

All key features are visually examined against pedestrian count, this is done by merging the datasets into the `combined_df` 

```python
# Create the hourly pedestrian dataset used in the original use case
pedestrian_hourly = (
    pedestrian_counting
    .groupby('date_time')['hourday']
    .sum()
    .reset_index()
)

pedestrian_hourly.columns = ['date_time', 'pedestrian_count']
```

```python
# Make both datetime columns compatible for merging
pedestrian_hourly['date_time'] = (
    pedestrian_hourly['date_time']
    .dt.tz_localize(None)
    .dt.floor('h')
)

weather_stations_df['date_time'] = (
    weather_stations_df['date_time']
    .dt.tz_localize(None)
    .dt.floor('h')
)
```

```python
# Merge the hourly pedestrian and weather datasets
combined_df = pd.merge(
    pedestrian_hourly,
    weather_stations_df,
    on='date_time',
    how='inner'
)

# Plot air temperature against pedestrian count
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=combined_df,
    x='airtemp',
    y='pedestrian_count',
    alpha=0.5,
    color='#059669'
)

plt.title("Air Temperature v. Pedestrian Count")
plt.xlabel("Air Temperature (°C)")
plt.ylabel("Pedestrian Count")
plt.tight_layout()
plt.show()
```

**Air Temperature vs Pedestrian Count Summary**

The scatterplot indicates that most observations are concentrated between **10°C and 25°C**, with pedestrian counts generally clustered at lower values and a smaller number of observations exhibiting much higher pedestrian counts. The highest pedestrian counts occur within this moderate temperature range, while temperatures below **5°C** and above **30°C** are associated with lower pedestrian activity.

<div id="section-3-5" class="usecase-sub-section-heading">
<b>3.5 Calculating Distance from Venues to Pedestrian Sensors</b>
</div>
We have utilised GeoPy, a python library that makes geographical calculations on distance. In this case, the distance is between the Venue location and each pedestrian sensor. This is a key step because the venue dataset does not contain pedestrian sensors or historical pedestrian count data. Linking each venue to its nearest pedestrian sensor allows the closest available pedestrian count data to be used in the analysis.

```python
from geopy.distance import geodesic

# Store calculated venue-to-sensor distances
distance_rows = []

# Loop through each venue
for _, venue in venue_location.iterrows():
    venue_loc = (venue['latitude'], venue['longitude'])

    # Skip venues with missing coordinates
    if pd.isna(venue['latitude']) or pd.isna(venue['longitude']):
        continue

    venue_id = venue['prop_id']
    venue_name = venue['full_name']
    sport = venue['sport']
    wedding = venue['wedding']
    bookable = venue['bookable']
    no_smoking = venue['no_smoking']
    event = venue['event']
    training = venue['training']
    dog_prohib = venue['dog_prohib']

    # Loop through each pedestrian sensor
    for _, sensor in pedestrian_counting.iterrows():
        sensor_loc = (sensor['latitude'], sensor['longitude'])

        # Skip sensors with missing coordinates
        if pd.isna(sensor['latitude']) or pd.isna(sensor['longitude']):
            continue

        sensor_id = sensor['location_id']

        # Calculate the distance between the venue and sensor
        distance = geodesic(venue_loc, sensor_loc).meters

        # Store the result
        distance_rows.append({
            'venue_id': venue_id,
            'venue_name': venue_name,
            'sensor_id': sensor_id,
            'distance_venue': distance,
            'sport': sport,
            'wedding': wedding,
            'bookable': bookable,
            'no_smoking': no_smoking,
            'event': event,
            'training': training,
            'dog_prohib': dog_prohib
        })

# Convert the results to a DataFrame
venue_sensor_distance = pd.DataFrame(distance_rows)

# Preview the calculated distances
venue_sensor_distance.head()
```

<div id="section-3-6" class="usecase-sub-section-heading">
<b>3.6 Nearest Sensor Calculation</b>
</div>
The calculated distances are grouped by venue, and idxmin() is used to identify the pedestrian sensor closest to each venue. Linking each venue to its nearest sensor allows the most geographically relevant pedestrian count data to be used in the subsequent analysis and modelling.

```python
# Select the row with the minimum distance for each venue_id
nearest_sensors = venue_sensor_distance.loc[venue_sensor_distance.groupby('venue_id')['distance_venue'].idxmin()].reset_index(drop=True)

nearest_sensors.head()
```

<div id="section-3-7" class="usecase-sub-section-heading">
<b>3.7 Merging the Datasets</b>
</div>
The final dataset is created by merging pedestrian count, nearest_sensor(venue data) and weather data. 
Merge pedestrian count with nearest sensor left on location id and right on sensor id with inner join. 
Next model_df is merged with weather station data on date_time with inner join. The purpose is to combine all the datasets to extract the key features to be able to apply predicitive models. 

```python
# Check the original pedestrian and weather date ranges
print("Pedestrian range:")
print(
    pedestrian_counting['date_time'].min(),
    "to",
    pedestrian_counting['date_time'].max()
)

print("\nWeather range:")
print(
    weather_stations_df['date_time'].min(),
    "to",
    weather_stations_df['date_time'].max()
)
```

```python
# Convert pedestrian timestamps to hourly datetime values
pedestrian_counting['date_time'] = (
    pd.to_datetime(pedestrian_counting['date_time'])
      .dt.floor('h')
)

# Convert weather timestamps to timezone-naive hourly datetime values
weather_stations_df['date_time'] = (
    pd.to_datetime(weather_stations_df['date_time'])
      .dt.tz_localize(None)
      .dt.floor('h')
)

# Merge pedestrian data with the nearest sensor information
model_df = pd.merge(
    pedestrian_counting,
    nearest_sensors,
    left_on='location_id',
    right_on='sensor_id',
    how='inner'
)

# Merge the combined dataset with the weather data
model_df = pd.merge(
    model_df,
    weather_stations_df,
    on='date_time',
    how='inner'
)

# Preview the merged dataset
model_df.head()
```

</div>|
After the dataset has been unified and all the key features are extracted, the last step is to apply the Machine Learning model. In this instance, multiple ML models are applied to

<div id="section-4-1" class="usecase-sub-section-heading">
<b>4.1 Feature Selection</b>
</div>
To apply the ML model we need to select the final features that will be utilised, as well as the target variable. 
The Features are maintained in a list, whilst the target variable is selected. 

```python
from sklearn.model_selection import train_test_split
import sklearn.preprocessing as preproc
from sklearn.ensemble import RandomForestRegressor 

#Features is a list of all the features that the ML model will be trained on 
features = [
    'distance_venue','sport','wedding','bookable','no_smoking', 'event','training','dog_prohib',
    'battery','windspeed','winddirection','gustspeed','vapourpressure','atmosphericpressure','relativehumidity',
    'airtemp','day_of_week'
]
#The X variable is the features 
X = model_df[features]
# The y variable is the target variable 
y = model_df['pedestriancount']

# Convert venue yes/no values to numeric values
yes_no_columns = [
    'sport',
    'wedding',
    'bookable',
    'no_smoking',
    'event',
    'training',
    'dog_prohib'
]

for column in yes_no_columns:
    X[column] = X[column].replace({'Y': 1, 'N': 0})

# Convert any remaining categorical values to numeric values
X = pd.get_dummies(X, dtype=int)

# Replace missing values with the median of each feature
X = X.fillna(X.median(numeric_only=True))
```

<div id="section-4-2" class="usecase-sub-section-heading">
<b>4.2 Polynomial Features</b>
</div>

Pairwise interaction (polynomial) features are generated to capture the combined effect of two predictor variables. In predictive modelling, the interaction between two features may provide additional information that cannot be captured when each feature is considered independently. By incorporating these interaction features, the models are able to learn more complex relationships within the data, which may improve the accuracy of pedestrian count predictions.

```python
# pairwise interaction features
X2 = preproc.PolynomialFeatures(degree=2,include_bias=False).fit_transform(X)
# Output the shape of the transformed features
print(X2.shape) 
```

We can see that the features have significantly increased within the dataset. We have gone from the initial 17 features to the final 170 features after pairwise interaction has been applied. 


<div id="section-4-3" class="usecase-sub-section-heading">
<b>4.3 Train-Test Split</b>
</div>

The modelling dataset is divided into separate training and testing datasets. The training dataset is used to develop the machine learning models, while the testing dataset consists of unseen data that is used to evaluate each model's predictive performance. This approach provides an unbiased assessment of how well the models generalise to new observations.

In this use case, a `test_size` of **0.2** is specified, meaning that **80%** of the data is allocated for model training and the remaining **20%** is reserved for model evaluation.

```python
# Split the original and polynomial feature sets into training and testing data
X1_train, X1_test, y_train, y_test = train_test_split(X, y,test_size=0.2, random_state=123)
X2_train, X2_test, _, _ = train_test_split(X2, y, test_size=0.2,random_state=123)
```
