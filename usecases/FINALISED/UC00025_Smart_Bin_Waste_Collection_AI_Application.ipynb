<div id="subsection-1-1" class="usecase-sub-section-heading"><b>1.1. Data Extraction and Preparation</b></div>

<div class="usecase-body">
The Waste Collected Per Month dataset is retrieved from the City of Melbourne Open Data API and prepared for analysis.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import requests
import datetime as dt
import plotly.express as px
from io import StringIO
```

```python
# https://data.melbourne.vic.gov.au/explore/dataset/waste-collected-per-month/information/
dataset_id = 'waste-collected-per-month'

base_url = 'https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/'
dataset_id = dataset_id
format = 'csv'

url = f'{base_url}{dataset_id}/exports/{format}'
params = {
    'select': '*',
    'limit': -1, # all records
    'lang': 'en',
    'timezone': 'UTC'
}

#GET request 
response = requests.get(url, params = params)

if response.status_code == 200:
    # StringIO to read the CSV data
    url_content = response.content.decode('utf-8')
    df = pd.read_csv(StringIO(url_content), delimiter= ';')
    print(df.sample(10, random_state = 999))
else: 
    print(f'Request failed with status code{response.status_code}')
```

```python
#check missing values
print(df.isna().sum())
print()
print("Dataset has",sum(df.isna().sum()),"missing values")
```

```python
#convert the date to date type
df.date=df.date.apply(pd.to_datetime)
df
```

```python
#make a copy of the data
df_waste=df.copy()
#extract year and month from date
df_waste['Year']=df_waste.date.dt.year
df_waste['Month']=df_waste.date.dt.month
df_waste
```

```python
df_waste=df_waste.drop(['date','month'],axis=1)
df_waste
```

<div id="subsection-2-1" class="usecase-sub-section-heading"><b>2.1. Data Extraction and Cleaning</b></div>

<div class="usecase-body">
The smart-bin dataset is retrieved through the DataVic API. Relevant sensor fields are retained and converted into suitable formats for analysis.
</div>

```python
resource_id = "07837d22-e224-40a1-a212-b8dcbcb45e98"

api_url = (
    "https://discover.data.vic.gov.au/en_AU/api/3/action/"
    "datastore_search"
)

records = []
offset = 0
limit = 10000

while True:
    params = {
        "resource_id": resource_id,
        "limit": limit,
        "offset": offset
    }

    response = requests.get(
        api_url,
        params=params,
        timeout=30
    )

    response.raise_for_status()

    data = response.json()

    if not data.get("success"):
        raise RuntimeError(
            "The DataVic API request was not successful."
        )

    batch = data["result"]["records"]
    records.extend(batch)

    total = data["result"]["total"]

    if len(records) >= total or len(batch) == 0:
        break

    offset += limit

df_bin = pd.DataFrame(records)

df_bin.drop(
    columns=["_id"],
    errors="ignore",
    inplace=True
)

print("Dataset shape:", df_bin.shape)

df_bin.head()
```

```python
df_bin
```

```python
#Data pre-processing
#iterate each column
print('Unique values of each columns except time column \n')
for col in df_bin.columns:
    #skip first column
    if col == 'time':
        pass
    #print out unique value of rest of the columns
    else:
        print(col,'unique value:',df_bin[col].unique())
        print('\n')
```

```python
#drop columns that are duplicated or provide no useful information
df_binClean = df_bin.copy() #make a copy of the dataset
df_binClean = df_binClean.drop(['sensor_name','last_update','product_type','description','address','wastebasket_size'], axis=1)
print("New dataset shape after dropping columns:",df_binClean.shape)
df_binClean
```

```python
# Create a clean copy of the smart-bin dataset
df_binC2 = df_binClean.copy()

# Convert time to datetime
df_binC2['time'] = pd.to_datetime(
    df_binC2['time'],
    utc=True,
    errors='coerce'
)
df_binC2['year'] = df_binC2['time'].dt.year
# Convert fill level to numeric
df_binC2['status_current_fill_level'] = pd.to_numeric(
    df_binC2['status_current_fill_level'],
    errors='coerce'
)

# Separate latitude and longitude
location_list = df_binC2['latlong'].str.split(
    ',',
    expand=True
)

df_binC2['lat'] = pd.to_numeric(
    location_list[0],
    errors='coerce'
)

df_binC2['long'] = pd.to_numeric(
    location_list[1],
    errors='coerce'
)

# Remove unusable rows
df_binC2 = df_binC2.dropna(
    subset=[
        'time',
        'serial',
        'status_current_fill_level',
        'bin_status'
    ]
).reset_index(drop=True)

# Sort readings chronologically for each bin
df_binC2 = df_binC2.sort_values(
    ['serial', 'time']
).reset_index(drop=True)

df_binC2.head()
```

```python
# Create one location record for each smart bin
df_serial = (
    df_binC2
    .dropna(
        subset=[
            'lat',
            'long'
        ]
    )
    [
        [
            'serial',
            'lat',
            'long'
        ]
    ]
    .drop_duplicates(
        subset='serial'
    )
    .reset_index(drop=True)
)

df_serial
```

<div id="section-3" class="usecase-section-heading"><b>3. Machine Learning for Next Bin Status Prediction</b></div>

<div class="usecase-body">
The machine-learning task predicts the status at the next recorded sensor observation. Current and historical fill-level information is retained so the models learn from recent bin behaviour rather than date and time alone.
</div>

<div id="subsection-3-1" class="usecase-sub-section-heading"><b>3.1. Historical Feature Engineering</b></div>

<div class="usecase-body">
Lagged fill-level values, a rolling mean, the elapsed time between readings and time-based variables are created from each bin's chronological sensor history.
</div>

```python
# Create machine-learning feature dataset
df_features = df_binC2.copy()

# Previous fill-level reading
df_features['fill_lag_1'] = (
    df_features.groupby('serial')[
        'status_current_fill_level'
    ].shift(1)
)

# Two readings ago
df_features['fill_lag_2'] = (
    df_features.groupby('serial')[
        'status_current_fill_level'
    ].shift(2)
)

# Average of recent historical readings
df_features['fill_rolling_mean_3'] = (
    df_features.groupby('serial')[
        'status_current_fill_level'
    ].transform(
        lambda x: x.shift(1)
        .rolling(
            window=3,
            min_periods=1
        )
        .mean()
    )
)

# Time since previous reading
df_features['minutes_since_previous'] = (
    df_features.groupby('serial')['time']
    .diff()
    .dt.total_seconds()
    .div(60)
)

# Time-based features
df_features['hour'] = df_features['time'].dt.hour
df_features['day_of_week'] = df_features['time'].dt.dayofweek
df_features['month'] = df_features['time'].dt.month

# Target: status at the NEXT reading
df_features['next_bin_status'] = (
    df_features.groupby('serial')[
        'bin_status'
    ].shift(-1)
)

df_features.head()
```

```python
from sklearn.preprocessing import LabelEncoder

serial_encoder = LabelEncoder()
status_encoder = LabelEncoder()

df_features['serial_encoded'] = (
    serial_encoder.fit_transform(
        df_features['serial'].astype(str)
    )
)

feature_columns = [
    'serial_encoded',
    'status_current_fill_level',
    'fill_lag_1',
    'fill_lag_2',
    'fill_rolling_mean_3',
    'minutes_since_previous',
    'hour',
    'day_of_week',
    'month'
]

df_model = df_features.dropna(
    subset=feature_columns + ['next_bin_status']
).copy()

df_model['target'] = (
    status_encoder.fit_transform(
        df_model['next_bin_status'].astype(str)
    )
)

df_model = df_model.sort_values(
    'time'
).reset_index(drop=True)

print("Model dataset shape:", df_model.shape)
print()
print("Target classes:")
print(status_encoder.classes_)

df_model[
    feature_columns + ['next_bin_status']
].head()
```

```python
X = df_model[feature_columns]
y = df_model['target']

n = len(df_model)

train_end = int(n * 0.60)
valid_end = int(n * 0.80)

Xtrain = X.iloc[:train_end]
ytrain = y.iloc[:train_end]

Xvalid = X.iloc[train_end:valid_end]
yvalid = y.iloc[train_end:valid_end]

Xtest = X.iloc[valid_end:]
ytest = y.iloc[valid_end:]

print("Training data:")
print(Xtrain.shape)
print(ytrain.shape)

print("\nValidation data:")
print(Xvalid.shape)
print(yvalid.shape)

print("\nTesting data:")
print(Xtest.shape)
print(ytest.shape)
```

<div id="subsection-3-2" class="usecase-sub-section-heading"><b>3.2. Compare Machine-Learning Models</b></div>

<div class="usecase-body">
Five classification models are evaluated using the chronological validation dataset. Accuracy, weighted precision, weighted recall and weighted F1-score are used to compare their performance when predicting the next recorded smart-bin status.
</div>

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.ensemble import (
    RandomForestClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier
)

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

```python
models = {
    'Decision Tree': DecisionTreeClassifier(
        random_state=16,
        class_weight='balanced'
    ),

    'KNN': make_pipeline(
        StandardScaler(),
        KNeighborsClassifier(
            n_neighbors=5
        )
    ),

    'Random Forest': RandomForestClassifier(
        random_state=16,
        class_weight='balanced'
    ),

    'Gradient Boosting': GradientBoostingClassifier(
        random_state=16
    ),

    'AdaBoost': AdaBoostClassifier(
        random_state=16
    )
}
```

```python
results = []

trained_models = {}

for name, model in models.items():

    model.fit(
        Xtrain,
        ytrain
    )

    prediction = model.predict(
        Xvalid
    )

    trained_models[name] = model

    results.append({
        'ML_Model': name,

        'Accuracy': accuracy_score(
            yvalid,
            prediction
        ),

        'Precision': precision_score(
            yvalid,
            prediction,
            average='weighted',
            zero_division=0
        ),

        'Recall': recall_score(
            yvalid,
            prediction,
            average='weighted',
            zero_division=0
        ),

        'F1_Score': f1_score(
            yvalid,
            prediction,
            average='weighted',
            zero_division=0
        )
    })

df_score = pd.DataFrame(results)

df_score = df_score.sort_values(
    'F1_Score',
    ascending=False
).reset_index(drop=True)

df_score
```

<div id="subsection-3-3" class="usecase-sub-section-heading"><b>3.3. Time-Aware Cross-Validation</b></div>

<div class="usecase-body">
TimeSeriesSplit is used to compare model performance across multiple chronological validation periods without randomly mixing future observations into earlier training data.
</div>

```python
from sklearn.model_selection import (
    TimeSeriesSplit,
    cross_validate
)

X_development = pd.concat(
    [Xtrain, Xvalid]
).reset_index(drop=True)

y_development = pd.concat(
    [ytrain, yvalid]
).reset_index(drop=True)

tscv = TimeSeriesSplit(
    n_splits=5
)

cv_results = []

for name, model in models.items():

    scores = cross_validate(
        model,
        X_development,
        y_development,
        cv=tscv,
        scoring={
            'accuracy': 'accuracy',
            'f1': 'f1_weighted'
        }
    )

    cv_results.append({
        'ML_Model': name,
        'Mean_Accuracy':
            scores['test_accuracy'].mean(),
        'Mean_F1':
            scores['test_f1'].mean(),
        'F1_Std':
            scores['test_f1'].std()
    })

df_cv = pd.DataFrame(
    cv_results
).sort_values(
    'Mean_F1',
    ascending=False
).reset_index(drop=True)

df_cv
```

```python
comparison_plot = (
    df_score.set_index('ML_Model')[
        ['Accuracy', 'Precision', 'Recall', 'F1_Score']
    ]
)

comparison_plot.plot(
    kind='bar',
    figsize=(11, 6)
)

plt.title(
    'Validation Performance of Machine-Learning Models'
)
plt.ylabel('Score')
plt.xlabel('Model')
plt.ylim(0, 1)
plt.xticks(rotation=30)
plt.legend(loc='lower right')
plt.show()
```

<div id="subsection-3-4" class="usecase-sub-section-heading"><b>3.4. Fine-Tune the Selected Model</b></div>

<div class="usecase-body">
Gradient Boosting achieved the highest weighted F1-score in both the chronological validation period and the time-aware cross-validation analysis. TimeSeriesSplit is therefore used with GridSearchCV to tune the model without randomly mixing future observations into earlier training periods.
</div>

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_estimators': [100, 200],
    'learning_rate': [0.05, 0.1],
    'max_depth': [2, 3],
    'min_samples_leaf': [1, 3]
}

gradient_model = GradientBoostingClassifier(
    random_state=16
)

grid_search = GridSearchCV(
    estimator=gradient_model,
    param_grid=param_grid,
    scoring='f1_weighted',
    cv=tscv,
    n_jobs=-1
)

grid_search.fit(
    X_development,
    y_development
)

best_model = grid_search.best_estimator_

print(
    "Best parameters:",
    grid_search.best_params_
)

print(
    "Best cross-validation F1:",
    grid_search.best_score_
)
```

<div id="subsection-3-5" class="usecase-sub-section-heading"><b>3.5. Final Test Evaluation</b></div>

<div class="usecase-body">
After model selection and tuning are complete, the held-out test period is used once to estimate final performance on later observations that were not used during training or parameter selection.
</div>

```python
from sklearn.metrics import (
    classification_report,
    ConfusionMatrixDisplay
)

y_test_pred = best_model.predict(
    Xtest
)

test_accuracy = accuracy_score(
    ytest,
    y_test_pred
)

test_precision = precision_score(
    ytest,
    y_test_pred,
    average='weighted',
    zero_division=0
)

test_recall = recall_score(
    ytest,
    y_test_pred,
    average='weighted',
    zero_division=0
)

test_f1 = f1_score(
    ytest,
    y_test_pred,
    average='weighted',
    zero_division=0
)

print(
    "Test Accuracy:",
    round(test_accuracy, 4)
)

print(
    "Test Precision:",
    round(test_precision, 4)
)

print(
    "Test Recall:",
    round(test_recall, 4)
)

print(
    "Test F1 Score:",
    round(test_f1, 4)
)
```

```python
print(
    classification_report(
        ytest,
        y_test_pred,
        labels=np.arange(
            len(status_encoder.classes_)
        ),
        target_names=status_encoder.classes_,
        zero_division=0
    )
)

ConfusionMatrixDisplay.from_predictions(
    ytest,
    y_test_pred,
    display_labels=status_encoder.classes_,
    cmap='Greens'
)

plt.title(
    'Confusion Matrix - Final Test Set'
)

plt.show()
```

Feature importance is examined to understand which historical and time-based variables contribute most strongly to the selected model's next-status predictions.

<div id="section-4" class="usecase-section-heading"><b>4. Next-Status Prediction and Visualisation</b></div>

<div class="usecase-body">
The final model uses the most recent available history for each smart bin to predict its status at the next recorded observation.
</div>

```python
def predict_next_status(model, feature_data):

    latest_data = (
        feature_data
        .dropna(
            subset=feature_columns
        )
        .sort_values('time')
        .groupby('serial')
        .tail(1)
        .copy()
    )

    predictions = model.predict(
        latest_data[
            feature_columns
        ]
    )

    latest_data[
        'predicted_next_status'
    ] = status_encoder.inverse_transform(
        predictions.astype(int)
    )

    return latest_data
```

```python
prediction_results = predict_next_status(
    best_model,
    df_features
)

prediction_results = prediction_results[
    [
        'serial',
        'time',
        'status_current_fill_level',
        'predicted_next_status',
        'lat',
        'long'
    ]
]

prediction_results
```

```python
prediction_map_data = (
    prediction_results
    .dropna(
        subset=[
            'lat',
            'long'
        ]
    )
)

status_colours = {
    'Low': '#059669',
    'Half Full': '#F97316',
    'Full': '#EC4899'
}

fig_prediction = px.scatter_map(
    prediction_map_data,
    lat='lat',
    lon='long',
    color='predicted_next_status',
    color_discrete_map=status_colours,
    hover_name='serial',
    hover_data={
        'status_current_fill_level': True,
        'predicted_next_status': True,
        'lat': False,
        'long': False
    },
    map_style='carto-positron',
    zoom=17,
    center={
        'lat': -37.8027,
        'lon': 144.9657
    },
    title='Predicted Next Smart-Bin Status',
    width=950,
    height=700
)

fig_prediction.update_layout(
    title_font=dict(
        color='#059669'
    )
)

fig_prediction.show()
```

<div class="usecase-body">
The map displays the predicted status at the next recorded observation for smart bins with known coordinates. Predictions are based on the most recent available historical fill-level and time-based information for each bin. They should not be interpreted as predictions for an arbitrary future date or time.
</div>

<div class="usecase-unnumbered-heading"><b>Conclusion</b></div>

<div class="usecase-body">
This use case explored historical City of Melbourne waste-collection patterns and used Argyle Square smart-bin sensor data to develop a machine-learning approach for predicting the status at the next recorded sensor observation.
<br><br>
The historical before-and-after analysis identified associations between the period following August 2015 and several waste measures. These associations should not be interpreted as evidence that smart-bin introduction directly caused the observed changes because other factors may also have influenced waste volumes.
<br><br>
For the prediction task, current fill level, recent historical fill levels, a rolling fill-level average and time-based information were used to predict the next observed bin status. Multiple classification models were compared using chronological validation and time-aware cross-validation before the selected model was tuned and evaluated on a held-out test period.
<br><br>
The analysis is limited by the available historical data, its focus on Argyle Square and the fact that the model predicts the next recorded observation rather than a status at an arbitrary future time. Future work could include additional smart-bin locations, longer historical periods and models designed for explicit forecasting horizons.
</div>
