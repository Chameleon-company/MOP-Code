```python
#Convert to Function

def API_Unlimited(datasetname): # pass in dataset name and api key
    dataset_id = datasetname

    base_url = 'https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/'
    #apikey = api_key
    dataset_id = dataset_id
    format = 'csv'

    url = f'{base_url}{dataset_id}/exports/{format}'
    params = {
        'select': '*',
        'limit': -1,  # all records
        'lang': 'en',
        'timezone': 'UTC'
    }

    # GET request
    response = requests.get(url, params=params)

    if response.status_code == 200:
        # StringIO to read the CSV data
        url_content = response.content.decode('utf-8')
        datasetname = pd.read_csv(StringIO(url_content), delimiter=';')
        print(datasetname.sample(10, random_state=999)) # Test
        return datasetname
    else:
        return (print(f'Request failed with status code {response.status_code}'))


"""
Get unlimited data from the API Function

Parameters:
datasetname (string): dataset name as from city of melbourn
apikey (string): the current api Key ( this should be gotton via the below if api stored in current workspace / google drive ( refer to Te API)

f = open("API.txt","r")
api_key = f.read()

Returns:
Csv : Returns the csv dataset of the dataset name
"""
```

```python
#Checking for empty data
missing_values = AirQuality_df.isnull().sum()
print("--------------------------Current Missing Values --------------------------")
print(missing_values)


#Lets see propotion
missing_proportion = (missing_values / len(AirQuality_df))*100
print("--------------------------Current Missing Values Propotion --------------------------")
print(missing_proportion)

#Function to call
def FindMissingVal(df):
  #now lets have a array to store the feature with number of NAN values
  MissingFeaturenValues = []
  #now we check each column
  for column in df.columns:
    missingVals = np.sum(df[column].isnull()) # sum the number of NAN values into variable
    MissingFeaturenValues.append({'Feature':column ,'Number of Missing Values':missingVals}) #the array consist of dictionary with feature and its missing values
  return MissingFeaturenValues
```

<div id="subsection-2-1-2" class="usecase-sub-sub-section-heading"><b>2.1.2 Missing Values</b></div>

<div class="usecase-body">
The dataset contains 9,765 missing values, representing approximately 6.99% of the data. Missing values need to be addressed before modelling because many machine learning models cannot process incomplete observations directly.

For this analysis, missing numerical values are replaced with the mean of the available values for that feature. Mean imputation allows the observations to remain in the dataset rather than removing rows containing missing data. However, replacing missing values with the mean can reduce the natural variation in a feature, so this should be considered when interpreting the results.
</div>

```python
AirQuality_df_copy = AirQuality_df.copy()
for item in FindMissingVal(AirQuality_df):
    if item['Number of Missing Values'] != 0 : # get the number of missing values > 0 we will use mean
        name = item['Feature'] # each column
        print(name + " : Has been replaced with its mean")
        AirQuality_df_copy.fillna(value=AirQuality_df[name].mean(), inplace=True)
```

<div id="subsection-2-1-3" class="usecase-sub-sub-section-heading"><b>2.1.3 Data Standardisation and Cleaning</b></div>

<div class="usecase-body">
Before the data can be analysed, some fields need to be cleaned and converted into a consistent format.

The date and time column is standardised so that each observation can be interpreted correctly and compared across time. The latitude and longitude values are also separated into individual columns so they can be used more easily for geographic visualisation and location-based analysis.

These cleaning steps help ensure that the dataset is structured consistently before it is used in later analysis and modelling.
</div>


```python
# Convert 'date_measure' to datetime remove any incontencies
AirQuality_df_copy['time'] = pd.to_datetime(AirQuality_df_copy['time'], utc=True)


# Split 'Lat/Long' into two separate columns and convert to float
AirQuality_df_copy[['latitude_air', 'longitude_air']] = AirQuality_df_copy['lat_long'].str.split(', ', expand=True).astype(float)
# Drop the original 'Lat Long' column
AirQuality_df_copy = AirQuality_df_copy.drop(columns=['lat_long'])

AirQuality_df_copy.head()
```

<div id="subsection-2-1-4" class="usecase-sub-sub-section-heading"><b>2.1.4 Feature Scaling</b></div>

<div class="usecase-body">
The air quality features have different numerical ranges, which can make it difficult to compare them directly and can affect how some models learn from the data. To address this, StandardScaler is used to standardise the numerical variables.

StandardScaler transforms each variable so that it is centred around a mean of 0 with a standard deviation of 1. The environmental features are scaled together, while the target variable, carbon monoxide, is scaled separately. The time and location columns are not scaled because they are not being used as continuous environmental measurements in this step.

The histograms below compare the distributions before and after standardisation. Notice that scaling changes the numerical range of each variable but does not change the overall shape of its distribution.
</div>


```python
# Lets add normalisation - Since the data has a large gap for each feature normalisation would help
#Using Standard Scaling
#We dont need norms for date_measure,latitude,longitude
AirQuality_df_copy_Normalized = AirQuality_df_copy.copy()
data_float_columns =['humidity','nitrogendioxide','ozone','particulateserr','particulatesvsn','temperature']
data_target = ['carbonmonoxide']

scaler = StandardScaler()
AirQuality_df_copy_Normalized[data_float_columns] = scaler.fit_transform(AirQuality_df_copy[data_float_columns])
target_scaler = StandardScaler()
AirQuality_df_copy_Normalized[data_target] = target_scaler.fit_transform(AirQuality_df_copy[data_target])
AirQuality_df_copy_Normalized.head(), AirQuality_df_copy.info()

AirQuality_df_copy_Normalized.head(), AirQuality_df_copy.info()
```

```python
#Lets see the scaling before and after
data_visuals =['carbonmonoxide','humidity','nitrogendioxide','ozone','particulateserr','particulatesvsn','temperature']

print("-------------- View BEFORE SCALING-----------------")
plt.figure(figsize=(10, 10))
AirQuality_df_copy[data_visuals].hist(color='#059669')
plt.tight_layout()
plt.show()


print("-------------- View AFTER SCALING-----------------")
plt.figure(figsize=(10, 10))
AirQuality_df_copy_Normalized[data_visuals].hist(color='#059669')
plt.tight_layout()
plt.show()
```

<div id="subsection-3-2" class="usecase-sub-section-heading"><b>3.2 Correlation Analysis</b></div>

<div class="usecase-body">
Correlation analysis is used to measure the strength and direction of the linear relationship between pairs of numerical variables. Correlation values range from -1 to +1. Values closer to +1 indicate a stronger positive relationship, values closer to -1 indicate a stronger negative relationship, and values close to 0 indicate little or no linear relationship.

The correlation matrix and heatmap below are used to explore relationships between the environmental features and carbon monoxide before building the prediction model. It is important to remember that correlation identifies associations between variables but does not demonstrate that one variable causes changes in another.
</div>

```python
# Calculate the correlation matrix
features = Air_df.drop(columns=['time', 'latitude_air', 'longitude_air'])
correlation_matrix = features.corr()
correlation_matrix
```

```python

plt.figure(figsize=(10, 8))

# Visualise the correlation matrix
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='Greens',
            square=True, cbar_kws={"shrink": .5}, linewidths=.5)

plt.title('Correlation HM')
plt.show()
```

<div class="usecase-body">
The heatmap shows that carbon monoxide has only weak linear relationships with the other environmental features in this dataset. For example, its correlations with humidity (-0.01), nitrogen dioxide (-0.08), ozone (-0.06) and temperature (approximately 0.00) are all close to zero.

In contrast, humidity, nitrogen dioxide, ozone, particulate measurements and temperature show very strong positive correlations with one another, with values close to +1. This means these variables tend to change in similar directions within this dataset.

These results are useful when interpreting the later modelling results. In particular, the weak individual correlations with carbon monoxide mean that strong predictive performance should not automatically be interpreted as evidence that these environmental variables directly explain changes in carbon monoxide.
</div>

<div id="subsection-4-2" class="usecase-sub-section-heading"><b>4.2 Testing LR Model</b></div>

<div class="usecase-body">
This section allows the user to test the trained linear regression model by entering their own environmental and development values. The model uses these inputs to estimate a carbon monoxide value.

To keep the test reasonably consistent with the data used to build the model, users should enter values within the ranges observed in this dataset:
</div>

<div class="usecase-list">

- <b>Humidity:</b> 16 to 100 – the amount of moisture in the air.
- <b>Temperature:</b> 2 to 41.4 – the recorded air temperature.
- <b>Nitrogen dioxide:</b> -212 to 1453 – the recorded nitrogen dioxide measurement.
- <b>Ozone:</b> -263 to 282 – the recorded ozone measurement.
- <b>Activity count:</b> 28 to 41 – the number of development activities associated with the year.

</div>

<div class="usecase-body">
These are the ranges observed in this dataset rather than recommended environmental limits. Values outside these ranges may be very different from the data on which the model was trained and may result in less reliable predictions.
</div>

```python
def get_numeric_input(prompt):
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Invalid input. Please enter a numeric value.")

def collect_and_scale_inputs(scaler):
    # Collect user inputs
    humidity = get_numeric_input("Enter humidity: ")
    temperature = get_numeric_input("Enter temperature: ")
    nitrogendioxide = get_numeric_input("Enter nitrogendioxide: ")
    ozone = get_numeric_input("Enter ozone: ")
    activity_count = get_numeric_input("Enter activity count: ")

    particulateserr = 0
    particulatesvsn = 144

    # Prepare the environmental features for scaling
    input_data_for_scaling = pd.DataFrame([{
        'humidity': humidity,
        'nitrogendioxide': nitrogendioxide,
        'ozone': ozone,
        'particulateserr': particulateserr,
        'particulatesvsn': particulatesvsn,
        'temperature' : temperature
    }])
    # Scale the environmental features using the fitted scaler
    scaled_features = scaler.transform(input_data_for_scaling)

    # Store the scaled features in a DataFrame
    scaled_features_df = pd.DataFrame(scaled_features, columns=input_data_for_scaling.columns)

    # Add 'activity_count' to the scaled_features_df
    scaled_features_df['activity_count'] = activity_count

    return scaled_features_df
```

<div id="subsection-4-3" class="usecase-sub-section-heading"><b>4.3 Neural Model Creation</b></div>

<div class="usecase-body">
A neural network is tested as an alternative to the linear regression model. Unlike linear regression, a neural network can learn more complex, non-linear relationships between the environmental and development features and carbon monoxide levels.

The model uses ReLU activation functions in its hidden layers to learn these relationships. A linear activation function is used in the output layer because carbon monoxide is being predicted as a continuous numerical value. The neural network can then be evaluated and compared with the simpler linear regression model.
</div>

```python
# Select the input features and target
X = combined_dataset[['humidity', 'temperature', 'nitrogendioxide', 'ozone',
                      'activity_count', 'particulateserr', 'particulatesvsn']]
y = combined_dataset['carbonmonoxide']

# Split the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Create the neural network
model = Sequential([
    Input(shape=(X_train.shape[1],)),
    Dense(11, activation='relu'),  # First hidden layer
    Dense(5, activation='relu'),   # Second hidden layer
    Dense(1, activation='linear')  # Continuous regression output
])

# Compile the model
model.compile(
    loss='mean_squared_error',
    optimizer=Adam(),
    metrics=['mean_squared_error']
)

# Display the model structure
model.summary()
```

```python
# Train the neural network
history = model.fit(X_train, y_train, epochs=30, batch_size=32, validation_split=0.2)
```

```python
import matplotlib.pyplot as plt

# Plotting training and validation loss
plt.figure(figsize=(10, 5))
plt.plot(
    history.history['loss'],
    label='Training Loss',
    color='#059669'
)
if 'val_loss' in history.history:
    plt.plot(
      history.history['val_loss'],
      label='Validation Loss',
      color='#9333EA'
    )

plt.title('Model Loss Progress During Training')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```

<div class="usecase-body">
The training and validation loss both decrease substantially during the first few epochs and then continue to decline more gradually. The two lines remain relatively close, suggesting that the model is learning similar patterns in the training and validation data rather than showing a large gap between them.

The next step is to evaluate the trained neural network on the separate test data. This provides a better indication of how the model performs on observations that were not used during training.
</div>

```python
# Evaluate the model on the test data

mse, _ = model.evaluate(X_test, y_test)
print(f"Test MSE: {mse}")
```

<div class="usecase-body">
The neural network produced a test Mean Squared Error (MSE) of <b>0.06271349638700485</b>. The linear regression model that included development activity produced a lower test MSE of <b>0.03944851163204824</b>.

For this test split, the lower MSE indicates that the linear regression model produced smaller squared prediction errors than the neural network. Therefore, the additional complexity of the neural network did not improve performance for this dataset and model configuration.

Based on these results, the linear regression model is retained for the final prediction demonstration. This comparison only reflects the models and test split used in this use case; it does not establish that linear regression will always outperform a neural network or that either model can reliably forecast future air quality.
</div>
