<div id="background" class="usecase-unnumbered-heading"><b>Background</b></div>

<div class="usecase-body">
This analysis combines population, residential dwelling, pedestrian-network, public-facility and café/restaurant seating data. It first visualises where people and dwellings are concentrated, then calculates three simple pressure indicators: residents per metre of pedestrian network, residents per public facility and residents per dining seat.

The indicators are normalised and combined with equal weights to produce an exploratory liveability score. The score is intended as a transparent demonstration of spatial analysis rather than a definitive measure of quality of life. Non-residential areas are excluded from the final scoring so that industrial or unpopulated zones do not distort the comparison.

<div class="usecase-section-heading">

<div class="usecase-body">
Load the Australian Statistical Geography Standard (ASGS) SA2 boundary file and retain the SA2 areas that belong to the Melbourne City SA3 region. The request is validated before the ZIP archive is read.
</div>   

```python
zip_url = (
    "https://www.abs.gov.au/statistics/standards/"
    "australian-statistical-geography-standard-asgs-edition-3/"
    "jul2021-jun2026/access-and-downloads/digital-boundary-files/"
    "SA2_2021_AUST_SHP_GDA2020.zip"
)

response = requests.get(zip_url, timeout=60)
response.raise_for_status()

with tempfile.TemporaryDirectory() as temp_dir:
    with ZipFile(BytesIO(response.content)) as zip_file:
        zip_file.extractall(temp_dir)

    shapefile_name = next(
        filename for filename in os.listdir(temp_dir) if filename.endswith(".shp")
    )
    LGAgdf = gpd.read_file(os.path.join(temp_dir, shapefile_name))

display(LGAgdf.head())
```

<div class="usecase-body">
Convert the SA3 code to numeric format so it can be compared consistently with the target Melbourne City SA3 code (20604). This ensures the correct SA2 boundary records are retained for the study area.

```python
LGAgdf["SA3_CODE21"] = pd.to_numeric(LGAgdf["SA3_CODE21"], errors="coerce")
filtered_LGAgdf = LGAgdf.loc[LGAgdf["SA3_CODE21"] == 20604].copy()

assert not filtered_LGAgdf.empty, "No SA2 boundaries were found for SA3 code 20604."
display(filtered_LGAgdf.head())
```

```python
# highlight on map
def highlight_function(feature):
    return {
        'color': 'black',
        'weight': 3,
    }
```

<div class="usecase-body">
    Define reusable functions for drawing consistently styled polygon layers and legends.
</div>

```python
def draw_map(existing_map, data, value_column, colour_scale, tooltip_formatter, add_colormap=True):
    """Add a consistently styled polygon layer to an existing Folium map."""
    valid_values = data[value_column].replace([np.inf, -np.inf], np.nan).dropna()
    if valid_values.empty:
        raise ValueError(f"No valid values are available for {value_column!r}.")

    colormap = LinearColormap(
        colors=colour_scale,
        vmin=float(valid_values.min()),
        vmax=float(valid_values.max()),
        caption=value_column.replace("_", " ").title(),
    )
    if add_colormap:
        colormap.add_to(existing_map)

    for _, row in data.iterrows():
        value = row[value_column]
        colour = "#ffffff00" if pd.isna(value) else colormap(value)
        folium.GeoJson(
            row["geometry"],
            style_function=lambda feature, colour=colour: {
                "fillColor": colour,
                "color": "black",
                "weight": 0.5,
                "fillOpacity": 0.7 if colour != "#ffffff00" else 0,
            },
            highlight_function=highlight_function,
            tooltip=tooltip_formatter(row),
        ).add_to(existing_map)
```

<div class="usecase-body">
    Add a marker cluster so large numbers of residential locations remain readable at different zoom levels.
</div>

```python
def add_marker_cluster(existing_map, data, geometry_column):
    """
    Add residential locations using FastMarkerCluster.

    FastMarkerCluster keeps the same point locations but stores them as one
    compact JavaScript coordinate array instead of embedding a separate
    Folium Marker object for every dwelling. This keeps the saved notebook
    small enough to review on GitHub.
    """
    points = [
        [point.y, point.x]
        for point in data[geometry_column]
        if point is not None and not point.is_empty
    ]

    FastMarkerCluster(points).add_to(existing_map)
```

Create a regular grid covering the study area for aggregating residential dwelling locations.
</div>

```python
def create_residential_heat_grid(points_gdf, n_side=50):
    """
    Aggregate residential points into a regular grid.

    Every residential record contributes to a grid-cell count, while the
    visualisation stores only one weighted heat point per occupied cell.
    """
    points = points_gdf.to_crs("EPSG:3857").copy()

    xmin, ymin, xmax, ymax = points.total_bounds
    x_edges = np.linspace(xmin, xmax, n_side + 1)
    y_edges = np.linspace(ymin, ymax, n_side + 1)

    # Bin every residential point into a grid cell.
    x_bin = np.clip(
        np.searchsorted(x_edges, points.geometry.x, side="right") - 1,
        0,
        n_side - 1,
    )
    y_bin = np.clip(
        np.searchsorted(y_edges, points.geometry.y, side="right") - 1,
        0,
        n_side - 1,
    )

    counts = (
        pd.DataFrame({"x_bin": x_bin, "y_bin": y_bin})
        .value_counts()
        .rename("dwelling_count")
        .reset_index()
    )

    # Use each occupied cell centre as the heat-map location.
    counts["x"] = (x_edges[counts["x_bin"]] + x_edges[counts["x_bin"] + 1]) / 2
    counts["y"] = (y_edges[counts["y_bin"]] + y_edges[counts["y_bin"] + 1]) / 2

    heat_points = gpd.GeoDataFrame(
        counts,
        geometry=gpd.points_from_xy(counts["x"], counts["y"]),
        crs="EPSG:3857",
    ).to_crs("EPSG:4326")

    return heat_points
```

```python
# Aggregate all residential records into a 50 × 50 grid.
residential_heat_grid = create_residential_heat_grid(
    residential_with_area[["geometry"]].copy(),
    n_side=50,
)

print(
    f"Residential records represented: {len(residential_with_area):,}\n"
    f"Occupied heat-map cells: {len(residential_heat_grid):,}"
)
```

<div class="usecase-body">
    Extract dwelling coordinates and retain only points that fall within the City of Melbourne boundaries.
</div>

```python
# Validate the aggregated heat-map data.
assert not residential_heat_grid.empty, "No residential heat-map cells were created."
assert residential_heat_grid["dwelling_count"].sum() == len(residential_with_area), (
    "The heat-map aggregation must preserve all residential records."
)

display(residential_heat_grid.head())
```

Compare population with pedestrian-network length to estimate pressure on pedestrian infrastructure within each SA2.

```python
# API URL of the GeoJSON file
ped_geojson_url = "https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/pedestrian-network/exports/geojson?lang=en&timezone=Australia%2FSydney"
# Read
ped_gdf = gpd.read_file(ped_geojson_url)
assert not ped_gdf.empty, "The pedestrian network dataset is empty."
```

```python
# Display
display(ped_gdf)
```

```python
ped_gdf = gpd.GeoDataFrame(ped_gdf, geometry="geometry")
if ped_gdf.crs is None:
    ped_gdf = ped_gdf.set_crs(merged_data_geo.crs)
else:
    ped_gdf = ped_gdf.to_crs(merged_data_geo.crs)

joined = gpd.sjoin(
    ped_gdf,
    merged_data_geo[["SA2 code", "geometry"]],
    how="inner",
    predicate="within",
)

projected_crs = "EPSG:3857"
joined = joined.to_crs(projected_crs)
merged_data_geo = merged_data_geo.to_crs(projected_crs)
```

<div class="usecase-body">
Calculate the pedestrian-network length within each SA2 and divide population by that length to estimate pressure.
</div>

<div class="usecase-body">

The pedestrian network must be projected into a metric coordinate reference system before measuring line length. Geographic latitude and longitude coordinates are measured in degrees, which are not appropriate for calculating accurate distances in metres.

</div>

```python
joined["street_length"] = joined.geometry.length
street_lengths = joined.groupby("SA2 code")["street_length"].sum()
merged_data_geo["Street_Length"] = (
    merged_data_geo["SA2 code"].map(street_lengths).fillna(0)
)
merged_data_geo["Street_Load"] = np.where(
    merged_data_geo["Street_Length"] > 0,
    merged_data_geo["Total Population"] / merged_data_geo["Street_Length"],
    np.nan,
)

merged_data_geo = merged_data_geo.to_crs(epsg=4326)
```

<div class="usecase-section-heading">
7. Calculate and validate the liveability score
</div>

<div class="usecase-body">
Normalise the infrastructure-pressure indicators and combine them into an exploratory liveability score for eligible residential areas.
</div>

```python
pressure_columns = ["Street_Load", "landmark_load", "seat_load"]
normalised_columns = [
    "Street_Load_norm", "landmark_load_norm", "seat_load_norm"
]

residential_counts = residential_with_area.groupby("CLUE small area").size()
merged_data_geo["residential_record_count"] = (
    merged_data_geo["SA2 name"].map(residential_counts).fillna(0).astype(int)
)

eligible_mask = (
    merged_data_geo["Total Population"].gt(0)
    & merged_data_geo["residential_record_count"].gt(0)
)

merged_data_geo[normalised_columns] = np.nan
eligible_pressure = merged_data_geo.loc[eligible_mask, pressure_columns].copy()

# A missing pressure value means that no mapped capacity was available.
# Replace it with a value slightly above the largest observed pressure so it
# receives the lowest normalised component score.
for column in pressure_columns:
    observed_max = eligible_pressure[column].replace([np.inf, -np.inf], np.nan).max()
    fallback = observed_max * 1.05 if pd.notna(observed_max) and observed_max > 0 else 1.0
    eligible_pressure[column] = (
        eligible_pressure[column].replace([np.inf, -np.inf], np.nan).fillna(fallback)
    )

scaler = MinMaxScaler()
normalised_values = 1 - scaler.fit_transform(eligible_pressure)
merged_data_geo.loc[eligible_mask, normalised_columns] = normalised_values

weights = np.array([1 / 3, 1 / 3, 1 / 3])
merged_data_geo["liveability_score"] = np.nan
merged_data_geo.loc[eligible_mask, "liveability_score"] = (
    merged_data_geo.loc[eligible_mask, normalised_columns].to_numpy() @ weights
)
```
