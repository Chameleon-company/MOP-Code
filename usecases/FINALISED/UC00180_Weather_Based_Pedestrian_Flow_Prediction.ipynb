<div id="section-2" class="usecase-section-heading">
<b>2. Importing Datasets</b>
</div>

<div class="usecase-body">
This section imports the datasets required for the pedestrian flow prediction model. The data is retrieved from the City of Melbourne Open Data API and loaded into pandas DataFrames for further cleaning, exploration and modelling.
</div>

```python
def fetch_data(base_url, dataset, api_key, num_records=99, offset=0):
    all_records = []
    max_offset = 9900  # Maximum number of requests

    while True:
        # Maximum limit check
        if offset > max_offset:
            break

        # Create API request URL
        filters = f'{dataset}/records?limit={num_records}&offset={offset}'
        url = f'{base_url}{filters}&api_key={api_key}'

        # Start request
        try:
            result = requests.get(url, timeout=10)
            result.raise_for_status()
            records = result.json().get('results')
        except requests.exceptions.RequestException as e:
            raise Exception(f"API request failed: {e}")
        if records is None:
            break
        all_records.extend(records)
        if len(records) < num_records:
            break

        # Next cycle offset
        offset += num_records

    # DataFrame all data
    df = pd.DataFrame(all_records)
    return df

#os.environ.get('MELBOURNE_API_KEY')

# Retrieve API key from environment variable
API_KEY = os.environ.get('MELBOURNE_API_KEY')

# Base URL for the City of Melbourne Open Data API
BASE_URL = 'https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/'
```

<div id="section-3-5" class="usecase-sub-section-heading">
<b>3.5 Calculating Distance from Venues to Pedestrian Sensors</b>
</div>
<div class="usecase-body">
To associate each event venue with nearby pedestrian monitoring infrastructure, the geographical distance between each venue and each unique pedestrian sensor location is calculated. Only unique sensor locations are used because the pedestrian dataset contains repeated observations from the same sensors across different time periods. Removing these duplicate locations prevents the same geographical distance from being calculated repeatedly and significantly improves execution time.

The Haversine formula is applied using vectorised operations to calculate distances in metres between venue and sensor coordinates. These distances are then used in the following section to identify the pedestrian sensor located closest to each venue.
</div>

```python
import numpy as np

# Keep only venues with valid coordinates
venues = venue_location[
    [
        'prop_id', 'full_name', 'latitude', 'longitude',
        'sport', 'wedding', 'bookable', 'no_smoking',
        'event', 'training', 'dog_prohib'
    ]
].dropna(subset=['latitude', 'longitude']).copy()

# Keep only unique pedestrian sensor locations
sensors = (
    pedestrian_counting[
        ['location_id', 'latitude', 'longitude']
    ]
    .dropna(subset=['latitude', 'longitude'])
    .drop_duplicates(subset=['location_id', 'latitude', 'longitude'])
    .copy()
)

# Rename coordinate columns before combining the datasets
venues = venues.rename(
    columns={
        'latitude': 'venue_latitude',
        'longitude': 'venue_longitude'
    }
)

sensors = sensors.rename(
    columns={
        'latitude': 'sensor_latitude',
        'longitude': 'sensor_longitude'
    }
)

# Create every venue-to-sensor combination
venue_sensor_distance = venues.merge(
    sensors,
    how='cross'
)

# Convert coordinates to radians
venue_lat = np.radians(venue_sensor_distance['venue_latitude'])
venue_lon = np.radians(venue_sensor_distance['venue_longitude'])
sensor_lat = np.radians(venue_sensor_distance['sensor_latitude'])
sensor_lon = np.radians(venue_sensor_distance['sensor_longitude'])

# Calculate distance using the Haversine formula
earth_radius = 6371000  # metres

lat_diff = sensor_lat - venue_lat
lon_diff = sensor_lon - venue_lon

a = (
    np.sin(lat_diff / 2) ** 2
    + np.cos(venue_lat)
    * np.cos(sensor_lat)
    * np.sin(lon_diff / 2) ** 2
)

venue_sensor_distance['distance_venue'] = (
    2 * earth_radius * np.arcsin(np.sqrt(a))
)

# Rename columns to match the existing use case structure
venue_sensor_distance = venue_sensor_distance.rename(
    columns={
        'prop_id': 'venue_id',
        'full_name': 'venue_name',
        'location_id': 'sensor_id'
    }
)

# Keep the required columns
venue_sensor_distance = venue_sensor_distance[
    [
        'venue_id',
        'venue_name',
        'sensor_id',
        'distance_venue',
        'sport',
        'wedding',
        'bookable',
        'no_smoking',
        'event',
        'training',
        'dog_prohib'
    ]
]

# Preview the calculated distances
venue_sensor_distance.head()
```

<div id="section-3-6" class="usecase-sub-section-heading">
<b>3.6 Nearest Sensor Calculation</b>
</div>

<div class="usecase-body">
The calculated distances are grouped by venue, and <code>idxmin()</code> is used to identify the pedestrian sensor closest to each venue. Linking each venue to its nearest sensor allows the most geographically relevant pedestrian count data to be used in the subsequent analysis and modelling.
</div>

```python
# Select the row with the minimum distance for each venue_id
nearest_sensors = venue_sensor_distance.loc[venue_sensor_distance.groupby('venue_id')['distance_venue'].idxmin()].reset_index(drop=True)

nearest_sensors.head()
```

</div>

<div class="usecase-body">
The final modelling dataset is created by combining the pedestrian count, nearest sensor and weather datasets. First, the pedestrian count data is merged with the nearest sensor data by matching <code>location_id</code> with <code>sensor_id</code> using an inner join. This associates each venue with pedestrian observations from its nearest sensor.

The resulting dataset is then merged with the weather station data using <code>date_time</code> as the common field. This combines the pedestrian, venue, spatial and environmental features into a single dataset that can be prepared for predictive modelling.

```python
# Check the original pedestrian and weather date ranges
print("Pedestrian range:")
print(
    pedestrian_counting['date_time'].min(),
    "to",
    pedestrian_counting['date_time'].max()
)

print("\nWeather range:")
print(
    weather_stations_df['date_time'].min(),
    "to",
    weather_stations_df['date_time'].max()
)
```

```python
# Convert pedestrian timestamps to hourly datetime values
pedestrian_counting['date_time'] = (
    pd.to_datetime(pedestrian_counting['date_time'])
      .dt.floor('h')
)

# Convert weather timestamps to timezone-naive hourly datetime values
weather_stations_df['date_time'] = (
    pd.to_datetime(weather_stations_df['date_time'])
      .dt.tz_localize(None)
      .dt.floor('h')
)

# Merge pedestrian data with the nearest sensor information
model_df = pd.merge(
    pedestrian_counting,
    nearest_sensors,
    left_on='location_id',
    right_on='sensor_id',
    how='inner'
)

# Merge the combined dataset with the weather data
model_df = pd.merge(
    model_df,
    weather_stations_df,
    on='date_time',
    how='inner'
)

# Preview the merged dataset
model_df.head()
```

<div id="section-4-1" class="usecase-sub-section-heading">
<b>4.1 Feature Selection</b>
</div>
<div class="usecase-body">
The final features are selected to provide the machine learning models with information that may help explain variations in pedestrian activity. These features include weather conditions, the distance between venues and pedestrian sensors, venue characteristics, and the day of the week. Together, they represent environmental, spatial and temporal factors that may influence pedestrian movement.

The target variable is <code>pedestriancount</code>, meaning that the models will learn from the selected features to predict the number of pedestrians recorded. Developing accurate pedestrian-count predictions can help city planners and emergency services anticipate periods of increased pedestrian activity and support decisions relating to crowd management and resource allocation.

Categorical venue characteristics are converted into numerical values so they can be processed by the machine learning models, while missing numerical values are replaced using the median of each feature to ensure that incomplete observations do not prevent model training.
</div>

```python
from sklearn.model_selection import train_test_split
import sklearn.preprocessing as preproc
from sklearn.ensemble import RandomForestRegressor

# Features used to train the machine learning models
features = [
    'distance_venue',
    'sport',
    'wedding',
    'bookable',
    'no_smoking',
    'event',
    'training',
    'dog_prohib',
    'battery',
    'windspeed',
    'winddirection',
    'gustspeed',
    'vapourpressure',
    'atmosphericpressure',
    'relativehumidity',
    'airtemp',
    'day_of_week'
]

# Create the feature dataset
X = model_df[features].copy()

# Select the target variable
y = model_df['pedestriancount']

# Convert venue yes/no values to numeric values
yes_no_columns = [
    'sport',
    'wedding',
    'bookable',
    'no_smoking',
    'event',
    'training',
    'dog_prohib'
]

for column in yes_no_columns:
    X[column] = X[column].map({'Y': 1, 'N': 0})

# Convert any remaining categorical values into numeric features
X = pd.get_dummies(X, dtype=int)

# Replace missing numerical values with the median of each feature
X = X.fillna(X.median(numeric_only=True))

# Preview the prepared feature dataset
X.head()
```

<div id="section-4-2" class="usecase-sub-section-heading">
<b>4.2 Polynomial Features</b>
</div>

<div class="usecase-body">
Pairwise interaction features are generated to examine whether combinations of predictor variables provide information that is not captured when each feature is considered independently. For example, pedestrian activity may respond differently to temperature under high-wind conditions than under calm conditions, meaning the combined effect of two features may be more informative than either feature alone.

Polynomial interaction features allow the models to represent these more complex relationships and test whether they improve pedestrian-count predictions. This is relevant to the scenario because pedestrian movement is likely influenced by several environmental, temporal and venue-related conditions acting together rather than by a single factor in isolation.
</div>

```python
# pairwise interaction features
X2 = preproc.PolynomialFeatures(degree=2,include_bias=False).fit_transform(X)
# Output the shape of the transformed features
print(X2.shape) 
```

<div class="usecase-body">
The polynomial transformation increases the feature set from 17 original features to 170 features by introducing pairwise interactions and squared terms. This provides the models with substantially more information about how combinations of environmental, temporal and venue-related factors may relate to pedestrian activity.

The increased number of features allows more complex relationships to be represented, which may improve pedestrian-count predictions if these interactions contain useful predictive information. However, the larger feature set also increases model complexity and does not guarantee improved performance. The original and interaction feature sets will therefore both be evaluated to determine whether the additional complexity produces more accurate and useful predictions for the pedestrian-flow scenario.
</div>

<div id="section-4-3" class="usecase-sub-section-heading">
<b>4.3 Train-Test Split</b>
</div>

<div class="usecase-body">
The modelling dataset is divided into separate training and testing datasets. The training dataset is used to develop the machine learning models, while the testing dataset contains unseen observations that are used to evaluate how accurately each model can predict pedestrian counts.

In this use case, 80% of the observations are allocated to training and the remaining 20% are reserved for testing. Evaluating the models on data that was not used during training provides a more realistic indication of how well they may generalise to new pedestrian conditions.

The same split is applied to both the original and polynomial feature sets so that their performance can be compared consistently. This helps determine whether the additional interaction features produce predictions that could provide more reliable information for city planners and emergency services when anticipating pedestrian activity.
</div>

```python
# Split the original feature set into training and testing data
X1_train, X1_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=123
)

# Apply the same split to the polynomial feature set
X2_train, X2_test, _, _ = train_test_split(
    X2,
    y,
    test_size=0.2,
    random_state=123
)
```

<div id="section-4-4" class="usecase-sub-section-heading">
<b>4.4 Applying Linear Regression Model</b>
</div>
