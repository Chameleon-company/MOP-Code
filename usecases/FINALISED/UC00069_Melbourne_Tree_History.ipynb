<div id="section-1-3" class="usecase-sub-section-heading"><b>1.3. Importing Datasets</b></div>

<div class="usecase-body">Retrieves both datasets in full using the Melbourne Open Data Platform's <code>/exports</code> endpoint, which returns the complete dataset in one request rather than paging through the <code>/records</code> endpoint (which is capped at 10,000 total records and was previously truncating the current-trees dataset to 9,999 rows).</div>

```python
def fetch_data(base_url, dataset, api_key):
    url = f'{base_url}{dataset}/exports/json'
    params = {}
    if api_key:
        params['api_key'] = api_key

    result = requests.get(url, params=params, timeout=60)
    result.raise_for_status()
    records = result.json()

    df = pd.DataFrame(records)
    return df

API_KEY = os.environ.get("API_KEY")  # note: removed trailing space in the env var name — was "API_KEY " before, which meant it always returned None
BASE_URL = 'https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/'
```

```python
# data set name
PRE_COLONIAL_TREES = 'pre-colonial-plant-list'

pre_colonial_trees = fetch_data(BASE_URL, PRE_COLONIAL_TREES, API_KEY)

pre_colonial_trees.head()
```

```python
# data set name
CURRENT_TREES = 'trees-with-species-and-dimensions-urban-forest'

current_trees = fetch_data(BASE_URL, CURRENT_TREES, API_KEY)

current_trees.head()
```

<div id="section-1-4" class="usecase-sub-section-heading"><b>1.4. Preprocessing the Data</b></div>

<div class="usecase-body">Reviews the available columns in both datasets, then removes columns that aren't needed for this species-diversity comparison. From the pre-colonial dataset, conservation-status columns (extinction risk and threat-category ratings) are dropped because this analysis compares which species are present, not their conservation status. From the current trees dataset, physical and administrative columns (age description, useful life expectancy, precinct, diameter, and location/coordinate fields other than what's needed later for mapping) are dropped because they describe individual trees rather than species-level presence, which is what this comparison needs. The remaining columns are then renamed so both datasets share a common species column, and reconciled so the pre-colonial list is filtered down to tree species only and species names are normalised for accurate comparison.</div>


```python
# Print available columns in the pre-colonial trees cleaned dataset
print("Available columns in pre-colonial trees dataset:")
print(pre_colonial_trees.columns.tolist())

# Print available columns in the current trees cleaned dataset
print("\nAvailable columns in current trees dataset:")
print(current_trees.columns.tolist())
```

```python
# Columns to remove from the pre-colonial dataset.
# These are conservation-status fields (extinction risk / threat category)
# and a certainty flag - not needed for a species-presence comparison.
columns_to_remove_pre_colonial = ['epbc_category_of_threat', 'ffg_extinction_risk', 'ffg_category_of_threat', 'certain']

# Removing the specified columns from the pre-colonial trees dataset
pre_colonial_trees_cleaned = pre_colonial_trees.drop(columns=columns_to_remove_pre_colonial)

# Print available columns in the pre-colonial trees cleaned dataset after removal
print("Available columns in pre-colonial trees dataset after removal:")
print(pre_colonial_trees_cleaned.columns.tolist())

# Columns to remove from the current trees dataset.
# These describe individual-tree physical/administrative attributes
# (age, life expectancy, precinct, diameter, duplicate location fields)
# rather than species identity, so they aren't needed for this comparison.
columns_to_remove_current = ['age_description', 'useful_life_expectency', 'useful_life_expectency_value', 'precinct', 'diameter_breast_height', 'coordinatelocation', 'easting', 'northing', 'geolocation']

# Removing the specified columns from the current trees dataset
current_trees_cleaned = current_trees.drop(columns=columns_to_remove_current)

# Print available columns in the current trees cleaned dataset after removal
print("\nAvailable columns in current trees dataset after removal:")
print(current_trees_cleaned.columns.tolist())
```

```python
# Print available columns in the pre-colonial trees cleaned dataset
print("Available columns in pre-colonial trees dataset:")
print(pre_colonial_trees_cleaned.columns.tolist())

# Print available columns in the current trees cleaned dataset
print("\nAvailable columns in current trees dataset:")
print(current_trees_cleaned.columns.tolist())
```

```python
# Rename "Scientific Name" column to "Species" in the current trees dataset
current_trees_renamed = current_trees_cleaned.rename(columns={"scientific_name": "species"})

# Verify the column renaming by printing the columns of the updated DataFrame
print("Available columns in current trees dataset after renaming:")
print(current_trees_renamed.columns.tolist())
```

<div class="usecase-body">The pre-colonial plant list includes shrubs, grasses and herbs alongside trees, since it is a general plant list rather than a trees-only dataset. The step below filters it down to genera that also appear in the current trees dataset (which is trees-only), so the comparison in later sections is between trees and trees, not trees and all pre-colonial plants. Species names are also normalised by removing variety and subspecies qualifiers (e.g. "var. longifolia", "subsp. lutea") so that near-identical species are not counted as separate entries due to naming differences alone.</div>

```python
import re

# Filter the pre-colonial plant list down to genera that are actually trees,
# using the current (trees-only) dataset's genus list as the reference
pre_colonial_trees_cleaned['genus'] = pre_colonial_trees_cleaned['species'].str.split().str[0]
tree_genera = set(current_trees_renamed['genus'].unique())

pre_colonial_trees_filtered = pre_colonial_trees_cleaned[
    pre_colonial_trees_cleaned['genus'].isin(tree_genera)
].copy()

print(f"Pre-colonial species before filtering to trees: {len(pre_colonial_trees_cleaned)}")
print(f"Pre-colonial species after filtering to trees:  {len(pre_colonial_trees_filtered)}")

# Reconcile naming — strip subspecies/variety qualifiers (e.g. "var. longifolia", "subsp. lutea")
# so near-identical species aren't counted as "lost" due to naming differences
def normalize_species_name(name):
    return re.sub(r'\s+(var\.|subsp\.).*$', '', str(name)).strip()

pre_colonial_trees_filtered['species_normalized'] = pre_colonial_trees_filtered['species'].apply(normalize_species_name)
current_trees_renamed['species_normalized'] = current_trees_renamed['species'].apply(normalize_species_name)
```

```python
current_trees_renamed.head()
```

```python
# Extract species sets — computed here (end of Section 1.4) because both
# Section 2 (Venn diagram) and Section 3 (abundance comparison) need them
pre_colonial_species = set(pre_colonial_trees_filtered['species_normalized'].unique())
current_species = set(current_trees_renamed['species_normalized'].unique())
```

<div id="section-2" class="usecase-section-heading"><b>2. Historical vs. Current Species Diversity (Venn Diagram)</b></div>

<div class="usecase-body">Compares the set of pre-colonial tree species against the set of current tree species and visualises the overlap. Note: only pre-colonial species sharing a genus with the current dataset are included in this comparison (see Section 1.4), so this diagram does not represent the full pre-colonial plant list.</div>

```python
plt.figure(figsize=(10, 7))
v = venn2([pre_colonial_species, current_species], ('Pre-Colonial Trees', 'Current Trees'))
v.get_patch_by_id('10').set_color('#059669')
v.get_patch_by_id('01').set_color('#9333EA')
v.get_patch_by_id('11').set_color('#F97316')
plt.title('Comparison of Tree Species Diversity: Pre-Colonial vs Current')
plt.show()
```

<div class="usecase-body">

**Insights**:

- Species only in the current dataset (not present in the pre-colonial tree list): 633 species. Many of these are introduced or non-native species that have become common in Melbourne's urban forest, such as the London Plane and English Elm.
- Species shared between both datasets: 34 species.

This comparison highlights which pre-colonial tree species are not currently recorded, which can help prioritise species for future native tree planting programs.

<div id="section-3" class="usecase-section-heading"><b>3. Change in Species Abundance Over Time</b></div>

<div class="usecase-body">Calculates the shared, absent and newly recorded species between the pre-colonial and current tree datasets.</div>

```python
# Determine shared and unique species (pre_colonial_species / current_species
# were already extracted at the end of Section 1.4)
shared_species = pre_colonial_species.intersection(current_species)
lost_species = pre_colonial_species.difference(current_species)
gained_species = current_species.difference(pre_colonial_species)

# Print results
print(f"Shared Species: {len(shared_species)}")
print(f"Lost Species: {len(lost_species)}")
print(f"Gained Species: {len(gained_species)}")
```

<div class="usecase-body">

**Insights**:

- 633 species are recorded in the current dataset but not in the pre-colonial tree list, largely reflecting introduced and cultivated species planted across Melbourne over time.
- 30 species are recorded in the pre-colonial tree list but not in current records. This does not necessarily mean these species are extinct; it means they are not present among the trees currently catalogued in the council's urban forest dataset, which only covers council-managed street and park trees.
- 34 species are recorded in both datasets.

These figures can help city planners and conservationists identify native species that may be worth prioritising in future planting programs.

</div>

<div id="section-4" class="usecase-section-heading"><b>4. Geographical Distribution of Trees (Map Visualisation)</b></div>

<div class="usecase-body">Builds an interactive map of Melbourne showing the location of every tree in the current dataset, grouped into clusters and split into Parks and Streets layers that can be toggled on and off.</div>

```python
# Convert DataFrame to GeoDataFrame
gdf = gpd.GeoDataFrame(
    current_trees_renamed,
    geometry=gpd.points_from_xy(current_trees_renamed.longitude, current_trees_renamed.latitude),
    crs="EPSG:4326"
)

# Initialize a map centred around Melbourne
melbourne_map = folium.Map(location=[-37.8136, 144.9631], zoom_start=12)

# Use MarkerCluster to manage multiple markers
marker_cluster = MarkerCluster().add_to(melbourne_map)

# Create FeatureGroupSubGroups for parks and streets within the MarkerCluster
park_group = FeatureGroupSubGroup(marker_cluster, 'Parks')
street_group = FeatureGroupSubGroup(marker_cluster, 'Streets')
park_group.add_to(melbourne_map)
street_group.add_to(melbourne_map)

# Add markers to the appropriate subgroup based on their location
for idx, row in gdf.iterrows():
    marker = folium.CircleMarker(
        location=[row.geometry.y, row.geometry.x],
        radius=3,
        color='#059669' if row['located_in'] == 'Park' else '#9333EA',
        fill=True,
        fill_color='#059669' if row['located_in'] == 'Park' else '#9333EA',
        popup=f"Species: {row['species']}, Planted: {row['date_planted']}"
    )
    if row['located_in'] == 'Park':
        marker.add_to(park_group)
    else:
        marker.add_to(street_group)

# Add LayerControl to toggle parks and streets
folium.LayerControl().add_to(melbourne_map)

# Display the map
melbourne_map
```

<div class="usecase-body">This map provides a view of the geographical distribution of trees across Melbourne. Users can zoom in for detail on individual clusters, zoom out for a broader view of distribution patterns, and use the layer control to filter by whether trees are located in Parks or on Streets.</div>

<div id="section-5" class="usecase-section-heading"><b>5. Time-Series Analysis of Tree Planting</b></div>

<div class="usecase-body">Plots the number of trees planted per year, after filtering out rows where the planting year is missing or invalid.</div>

```python
# Ensure the 'year_planted' column is treated as integers
current_trees_renamed['year_planted'] = pd.to_numeric(current_trees_renamed['year_planted'], errors='coerce')

# Filtering out any rows where 'year_planted' might be zero or invalid
current_trees_filtered = current_trees_renamed[current_trees_renamed['year_planted'] > 0]

# Create the plot
plt.figure(figsize=(15, 8))
ax = sns.countplot(x='year_planted', data=current_trees_filtered, color='#059669')
plt.xticks(rotation=90)
plt.title('Number of Trees Planted Per Year')
plt.ylabel('Number of Trees')
plt.xlabel('Year Planted')

# Adding counts above the bars
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', fontsize=10, color='black', xytext=(0, 5),
                textcoords='offset points')

plt.tight_layout()
plt.show()
```

<div class="usecase-body">This chart shows tree planting activity by year across Melbourne. Note that a large number of records use 1900 as a placeholder value where the true planting year is unknown; these are excluded from the ARIMA forecasting analysis in Section 10, but are still included here as they represent genuinely planted trees, just with an unknown planting date.</div>

<div id="section-6" class="usecase-section-heading"><b>6. Habitat Analysis for Pre-Colonial Species</b></div>

Understanding historical habitat associations can help inform which conditions to prioritise when selecting sites for native species reintroduction.

<div id="section-10" class="usecase-section-heading"><b>10. Time-Series Model Fitting (ARIMA Model Forecasting)</b></div>

<div class="usecase-body">Fits an ARIMA (AutoRegressive Integrated Moving Average) model to the annual tree planting counts to forecast future planting activity, and validates the model against held-out data — a portion of the historical data set aside and not shown to the model during training, so its forecasts can be compared against real values it hasn't seen.</div>


```python
# Print all column names in the DataFrame to verify the correct name of the year_planted column
print(current_trees_renamed.columns)
```

<div id="section-10-1" class="usecase-sub-section-heading"><b>10.1. Data Preparation</b></div>

<div class="usecase-body">Filters out invalid and placeholder planting years before building the time series. A value of 1900 is a common council default used where the true planting year is unknown, rather than a genuine planting date, so rows with a planting year at or before 1900 are excluded here. The remaining data is aggregated into an annual count of trees planted per year.</div>

```python
# Filter out invalid/placeholder years before building the time series
# (0 = missing, 1900 = a common council default/placeholder for "unknown")
current_trees_valid_years = current_trees_renamed[
    (current_trees_renamed['year_planted'] > 1900)
].copy()

current_trees_valid_years['year_planted'] = pd.to_datetime(
    current_trees_valid_years['year_planted'], format='%Y'
).dt.year

# Aggregate data by year to get the total count of trees planted each year
yearly_data = current_trees_valid_years.groupby('year_planted').size().reset_index(name='Tree_Count')
yearly_data.set_index('year_planted', inplace=True)  # Set the year as the index, ensuring it is a datetime index for time series modelling
yearly_data.index = pd.to_datetime(yearly_data.index, format='%Y')
```

<div id="section-10-2" class="usecase-sub-section-heading"><b>10.2. Visualise the Data</b></div>

<div class="usecase-body">Plots the annual tree planting counts as a line chart to visually check the trend before modelling.</div>

```python
yearly_data['Tree_Count'].plot(title='Annual Tree Plantings')
plt.ylabel('Number of Trees Planted')
plt.show()
```

<div class="usecase-body">The chart shows a gradual increase in planting activity from 2004, a distinct peak around 2012 (over 8,000 trees planted that year), and a fairly steady 2,000–3,500 trees planted per year since. The 2012 peak likely reflects a specific planting initiative rather than a data error, and is worth noting as it does influence the model fitted below.</div>

<div id="section-10-3" class="usecase-sub-section-heading"><b>10.3. Check for Stationarity</b></div>

<div class="usecase-body">Runs the Augmented Dickey-Fuller (ADF) test to check whether the annual planting counts are stationary (i.e. whether their statistical properties stay stable over time). A p-value below 0.05 suggests the series is stationary; if not, differencing is used within the ARIMA model to address this.</div>

```python
result = adfuller(yearly_data['Tree_Count'])
print('ADF Statistic:', result[0])
print('p-value:', result[1])

# Interpret the p-value
if result[1] < 0.05:
    print("Data is stationary")
else:
    print("Data is not stationary, differencing might be needed")
```

<div class="usecase-body">Please record the ADF statistic and p-value printed above, as they depend on the full dataset now being used and will differ from any figures seen in earlier drafts of this notebook. If the p-value is above 0.05, the data is not stationary; this is expected for a short, irregular series like this one, and is handled by the differencing (the "I" in ARIMA) term used in the model below.</div>

<div id="section-10-4" class="usecase-sub-section-heading"><b>10.4. Fit the ARIMA Model</b></div>

<div class="usecase-body">Fits an ARIMA(1,1,1) model to the annual planting counts. This configuration uses one autoregressive term, one order of differencing, and one moving average term a reasonable starting point for a short, non-stationary annual series like this one.</div>

```python
# Fitting the ARIMA model (parameters should be optimised based on your specific dataset)
model = ARIMA(yearly_data['Tree_Count'], order=(1,1,1))  # Example parameters
model_fit = model.fit()

# Model summary
print(model_fit.summary())
```

<div class="usecase-body">Review the model summary output above before relying on it. With a series this short (around 20 years of data) and one large outlier year (2012), the fitted coefficients may not all be statistically significant; check the reported p-values for each coefficient (values under 0.05 indicate the term is contributing meaningfully to the model) rather than assuming both terms are significant by default.</div>

<div id="section-10-5" class="usecase-sub-section-heading"><b>10.5. Forecast Future Plantings</b></div>

<div class="usecase-body">Uses the fitted model to forecast tree planting counts for the next five years, starting the year after the most recent year in the dataset (calculated automatically, rather than hardcoded, so this stays correct as new data is added in future).</div>

```python
# Example forecasts from the ARIMA model
forecast_values = model_fit.forecast(steps=5)
# Convert forecasted values to integers (rounded)
forecast_values_rounded = [int(round(value)) for value in forecast_values]
# Create a DataFrame to display the forecast with corresponding years
last_year = yearly_data.index.year.max()
forecast_years = range(last_year + 1, last_year + 6)
forecast_df = pd.DataFrame({
    'Year': forecast_years,
    'Forecasted Plantings': forecast_values_rounded
})
# Print the DataFrame to see the formatted output
print(forecast_df.to_string(index=False))
```

<div class="usecase-body">With a short, noisy series like this one, ARIMA forecasts can come out close to flat (a near-constant value across all forecast years) rather than showing a clear trend. Treat this as a limitation of forecasting from around 20 years of annual data with one large outlier, not as a confident prediction that planting activity will not change. A longer historical series would likely produce a more informative forecast.</div>

<div id="section-10-6" class="usecase-sub-section-heading"><b>10.6. Model Validation</b></div>

<div class="usecase-body">Splits the data into a training set (all but the last 5 years) and a test set (the last 5 years), refits the model on the training set only, and compares its forecast against the actual test-set values using Mean Absolute Error (MAE) and Root Mean Squared Error (RMSE).</div>

```python
# Splitting data into train and test sets
train = yearly_data['Tree_Count'][:-5]
test = yearly_data['Tree_Count'][-5:]

# Fit model on train data
model = ARIMA(train, order=(1,1,1))
model_fit = model.fit()

# Forecast
forecasts = model_fit.forecast(steps=5)

# Visualization of the forecast alongside actual data
plt.figure(figsize=(12,6))
plt.plot(yearly_data['Tree_Count'], label='Original')
plt.plot(pd.date_range(start=test.index[0], periods=5, freq='A'), forecasts, label='Forecasts', color='red')
plt.legend()
plt.show()

# Calculate accuracy metrics
from sklearn.metrics import mean_squared_error, mean_absolute_error

mae = mean_absolute_error(test, forecasts)
rmse = mean_squared_error(test, forecasts) ** 0.5  # RMSE = sqrt(MSE); avoids the removed 'squared' argument
print('MAE:', mae)
print('RMSE:', rmse)
```

<div class="usecase-body">Please record the MAE and RMSE printed above, as they will differ from any figures seen in earlier drafts of this notebook, since the model is now fitted on the full dataset. Compare the red forecast line against the blue actual line in the chart: the model should track the general level of the actual data reasonably well, even though it cannot capture sharp year-to-year jumps such as the 2012 peak.</div>
