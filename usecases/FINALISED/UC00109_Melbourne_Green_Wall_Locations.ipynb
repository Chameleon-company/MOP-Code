```python
# Import dependencies
import requests
from io import StringIO

import pandas as pd
pd.set_option("display.max_columns", None)

import matplotlib.pyplot as plt
import seaborn as sns
import folium
from folium.plugins import MarkerCluster
from wordcloud import WordCloud
import geopandas as gpd

from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

from sklearn import cluster
import scipy

# Template map helper functions
MOP_PRIMARY_COLOUR = "#059669"
MOP_SECONDARY_COLOUR = "#2DBE6C"

def create_template_map(location, zoom_start):
    """Create a Folium map without using the default OpenStreetMap tile server."""
    m = folium.Map(
        location=location,
        zoom_start=zoom_start,
        tiles=None
    )
    folium.TileLayer(
        tiles="https://{s}.basemaps.cartocdn.com/light_all/{z}/{x}/{y}{r}.png",
        attr="&copy; OpenStreetMap contributors &copy; CARTO",
        name="Light base map",
        control=False
    ).add_to(m)
    return m

def add_template_legend(m, title, items):
    """Add a template-coloured legend to a Folium map."""
    legend_rows = ""
    for label, colour in items:
        legend_rows += (
            f"<div style='margin-bottom:6px;'>"
            f"<span style='display:inline-block;width:12px;height:12px;"
            f"background:{colour};border-radius:50%;margin-right:8px;"
            f"border:1px solid #ffffff;box-shadow:0 0 0 1px #6B7280;'></span>"
            f"{label}</div>"
        )

    legend_html = f"""
    <div style="
        position: fixed;
        bottom: 40px;
        left: 40px;
        z-index: 9999;
        background: white;
        border: 2px solid {MOP_PRIMARY_COLOUR};
        border-radius: 6px;
        padding: 10px 12px;
        font-size: 13px;
        color: #121212;
        box-shadow: 0 2px 6px rgba(0,0,0,0.25);
    ">
        <div style="font-weight: 700; color: {MOP_PRIMARY_COLOUR}; margin-bottom: 8px;">
            {title}
        </div>
        {legend_rows}
    </div>
    """
    m.get_root().html.add_child(folium.Element(legend_html))
    return m
```

<div id="uc109-sub-2-1" class="usecase-sub-section-heading"><b>2.1. Microclimate Sensor Readings</b></div>
<div class="usecase-body">Retrieve microclimate readings and examine PM10 and PM2.5 measurements across the available monitoring sites.</div>


```python
# Function to collect data using City of Melbourne Open Data API v2.1.
def collect_data(dataset_id):
    base_url = "https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/"
    export_format = "csv"

    url = f"{base_url}{dataset_id}/exports/{export_format}"
    params = {
        "select": "*",
        "limit": -1,
        "lang": "en",
        "timezone": "UTC",
    }

    response = requests.get(url, params=params)
    response.raise_for_status()

    csv_content = response.content.decode("utf-8")
    return pd.read_csv(StringIO(csv_content), delimiter=";")
```

```python
# IMPORT MICROCLIMATE SENSOR READINGS DATASET
# Set dataset_id to query for the API call dataset name
dataset_id = 'microclimate-sensor-readings'
# Save dataset to DataFrame variable
micro_data = collect_data(dataset_id)
# Check number of records in df
print(f'The dataset contains {len(micro_data)} records.')
# View df
micro_data.head(3)
```

```python
# Convert local time in micro DataFrame to a datetime datatype
micro_data['local_time'] = pd.to_datetime(micro_data['local_time'] )
micro_data['local_time'].dtypes
```

<div class="usecase-body">

Investigate the measurements of Particulate Matter 10ug/m^3 in size at each location

</div>


<div class="usecase-body">

PM10 are common air pollutant particles in the size of 0.01mm or smaller. PM10 particles can impact the health of people living in the City of Melbourne, particularly those with heart or lung conditions <a href="#fn-4">[4]</a> .

</div>


```python
# Make a DataFrame with only the PM10 values
micro_PM10 = micro_data.loc[micro_data['type'] == 'PM10']
micro_PM10.shape
```

```python
# Check the ids of different sites PM10 values were recorded at
micro_PM10['site_id'].unique()
```

```python
# Split the PM10 DataFrame into 3 DataFrames, one for each site
site_ids = [1011, 1012, 1009]
site_dfs = []

for site_id in site_ids:
    # Select site id from list, add only values from one site id to side_id DataFrame
    site_df = micro_PM10.loc[micro_PM10['site_id'] == site_id]
    # Sort values into time order
    site_df = site_df.sort_values(by='local_time')
    # Select only the local_time and value variables from the DataFrame
    site_df = site_df[['local_time', 'value']]
    # Set local time to datetime object
    site_df['local_time'] = pd.to_datetime(site_df['local_time'] )
    # Round time to the nearest minute
    site_df["local_time"] = site_df["local_time"].dt.floor('1min')
    # Add site_df to the DataFrame
    site_dfs.append(site_df)
    
site1011_10 = site_dfs[0]
site1012_10 = site_dfs[1]
site1009_10 = site_dfs[2]
```

```python
# Plot the PM10 readings over time
sns.set_theme()
fig, ax = plt.subplots(figsize=(15, 7.5))

ax.plot(site1011_10['local_time'], site1011_10['value'], label='site_1011', color='#059669')
ax.plot(site1012_10['local_time'], site1012_10['value'], label='site_1012', color='#9333EA')
ax.plot(site1009_10['local_time'], site1009_10['value'], label='site_1009', color='#F97316')

ax.set_xlabel("Time")
ax.set_ylabel("PM$_{10}$ Reading")
ax.set_title("PM$_{10}$ Reading (µg/m\u00b3) Across an Hour", y=1.1, fontsize=15)

plt.legend()
plt.show()
```

<div class="usecase-body">

Three sites recorded PM10 measurements over a one-hour period. Across the three sites, PM10 values ranged from approximately 19 µg/m³ to 30.5 µg/m³. Site 1012 recorded the lowest measurement, approximately 19 µg/m³ at 10:30 pm, while site 1011 recorded the highest measurement, approximately 30.5 µg/m³ at 11:00 pm.

</div>

It would be useful to identify the sensor locations on a map, so an additional dataset will need to be loaded and combined to view this.

```python
# IMPORT MICROCLIMATE SENSOR LOCATIONS DATASET
# Set dataset_id to query for the API call dataset name
dataset_id = 'microclimate-sensor-locations'
# Save dataset to DataFrame variable
micro_loc = collect_data(dataset_id)
# Check number of records in df
print(f'The dataset contains {len(micro_loc)} records.')
# View df
micro_loc.head(3)
```

```python
# Merging sensor location data with sensor reading data.
# Merge DataFrames on site_id value
micro = pd.merge(micro_data, micro_loc, on='site_id')
# Drop duplicates and columns with NaN data
micro = micro.drop(['gatewayhub_id_y', 'end_reading', 'last_data_local', 'site_status_y'], axis=1)
# Automatically convert datatypes based on column values
micro = micro.convert_dtypes()
# Drop retired sensors
micro = micro[micro['site_status_x'] != 'R']
micro.head(3)
```

```python
# Get unique locations from microdata
micro_loc = micro.drop_duplicates(subset=["site_id"])

# Convert to GeoDataFrame using longitude as x and latitude as y.
gdf_micro_loc = gpd.GeoDataFrame(
    micro_loc,
    geometry=gpd.points_from_xy(
        micro_loc["longitude"],
        micro_loc["latitude"]
    ),
    crs="EPSG:4326"
)

gdf_micro_loc.head(3)
```

```python
# Initialise the map at Melbourne and store it in a m object
m = create_template_map(location=[-37.815, 144.96], zoom_start=14)

# Add microsensor locations.
for i in range(0, len(gdf_micro_loc)):
    folium.CircleMarker(
        location=[gdf_micro_loc.iloc[i]["latitude"], gdf_micro_loc.iloc[i]["longitude"]],
        radius=6,
        popup=gdf_micro_loc.iloc[i]["site_id"],
        color=MOP_PRIMARY_COLOUR,
        fill=True,
        fill_color=MOP_PRIMARY_COLOUR,
        fill_opacity=0.85,
    ).add_to(m)

# Add a title.
m.get_root().html.add_child(
    folium.Element("<h3 align='center'>Locations of microsensors</h3>")
)

# Add a legend using template colours.
add_template_legend(
    m,
    "Legend",
    [("Microsensor location", MOP_PRIMARY_COLOUR)]
)

# Show the map.
m
```

The microclimate sensor locations are displayed on the map. Selecting a marker shows the corresponding sensor ID. Site 1011 is located near Siddeley Street, site 1009 is in Flinders Lane and site 1012 is on Little Collins Street.

```python
# IMPORT INSECT DATASET
# Set dataset_id to query for the API call dataset name
dataset_id = 'insect-records-in-the-city-of-melbourne-from-little-things-that-run-the-city'
# Save dataset to DataFrame variable
insect = collect_data(dataset_id)
# Check number of records in df
print(f'The dataset contains {len(insect)} records.')
# View df
insect.head(3)
```

<div class="usecase-body">

As with the butterfly data, biodiversity is an important consideration. It is therefore useful to examine the number of observations for each insect group. Because the `species` column contains missing values, insects are counted by genus instead. A `sighting` column is added so observations can be aggregated consistently.

</div>


```python
# Add a column with default value 1 for insect sighting = True
insect = insect.assign(sighting = 1)

# Set grouping for aggregated DataFrame
groupbyfields = ['order','family','genus','identification_notes']
aggregatebyfields = ["sum"]

# Create new DataFrame with sum of each insect species.
insect_species = pd.DataFrame(insect.groupby(groupbyfields)['sighting'].agg(aggregatebyfields).astype(int).add_prefix('total'))
# Add in an index column
insect_species = insect_species.reset_index()

# Check number of records in df
print(f'The dataset contains {len(insect_species)} records.')
insect_species.head(3)
```

```python
# Plot of insect species
fig = plt.figure(figsize = (12, 5))
 
# creating the bar plot
plt.bar(insect_species["genus"], insect_species["totalsum"], color="#059669", width = 0.8)
 
plt.xlabel("Genus")
plt.xticks(rotation=90, fontsize=6)
plt.ylabel("Count")
plt.title("Number of Insects by Genus")
plt.show()
```

<div class="usecase-body">

Many insect genera in the dataset were observed only once. Because the large number of categories makes the genus names difficult to compare in a bar plot, the observations are also presented as a word cloud, where larger words indicate genera with more recorded observations.

</div>


```python
# Drop NA values
new = insect["genus"].dropna(how='any')
# Convert data to a string list
text = " ".join(review for review in new.astype(str))

# Generate the word cloud with no repetition of words
wordcloud = WordCloud(collocations=False).generate(text)

# Display the word cloud
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```

<div class="usecase-body">

The most frequently recorded insect genus was *Diomus*, followed by *Iridomyrmex*. These results describe the frequency of observations in the dataset rather than the total population of each genus.

</div>


<div class="usecase-body">

Location of insect sightings is also essential, as this task focuses on finding the optimal location for green wall locations to enhance existing biodiversity. To achieve this, insect data will be re-summed by geographical data, combined with butterfly data and then mapped.

</div>


```python
# Aggregating the insect data as count of individual species by monitoring location.
insect_counts = insect.groupby(["location"])["location"].count().reset_index(name="Insect Total")
# Creating id index
insect_counts.insert(0, 'id', range(1, 1 + len(insect_counts)))
# Check number of records in df
print(f'The dataset contains {len(insect_counts)} records.')
# View data
insect_counts
```

<div class="usecase-body">

The insect dataset identifies monitoring sites by location name but does not provide latitude and longitude fields. Rather than manually hard-coding coordinates, the location names are geocoded programmatically using OpenStreetMap's Nominatim service.

Each monitoring location is searched together with **Melbourne, Victoria, Australia** to improve geographic specificity. The resulting latitude and longitude values are then added to the insect-count dataset. <a href="#fn-7">[7]</a>

</div>


```python
# Geocode insect monitoring locations instead of manually hard-coding coordinates.
geolocator = Nominatim(user_agent="mop_uc00109_green_wall_locations")
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1)

# Alternate search names are used only to improve geocoding of dataset labels.
location_aliases = {
    "Canning/Neil Street Reserve": "Canning Street Carlton Victoria Australia",
    "Carlton Gardens South": "Carlton Gardens Melbourne Victoria Australia",
    "Fitzroy-Treasury Gardens": "Fitzroy Gardens East Melbourne Victoria Australia",
    "Garrard Street Reserve": "Garrard Street Parkville Victoria Australia",
    "Murchinson Square": "Murchison Square Carlton Victoria Australia",
    "Women's Peace Gardens": "Womens Peace Garden Kensington Victoria Australia",
}

def geocode_melbourne_location(location_name):
    queries = [
        f"{location_name}, Melbourne, Victoria, Australia",
        location_aliases.get(location_name),
    ]

    for query in queries:
        if not query:
            continue
        result = geocode(query)
        if result is not None:
            return result

    return None

insect_counts["geocode_result"] = insect_counts["location"].apply(
    geocode_melbourne_location
)

insect_counts["latitude"] = insect_counts["geocode_result"].apply(
    lambda result: result.latitude if result is not None else pd.NA
)
insect_counts["longitude"] = insect_counts["geocode_result"].apply(
    lambda result: result.longitude if result is not None else pd.NA
)

unresolved_locations = insect_counts.loc[
    insect_counts[["latitude", "longitude"]].isna().any(axis=1),
    "location",
].tolist()

if unresolved_locations:
    print(
        "The following monitoring locations could not be geocoded and will be excluded from maps:",
        ", ".join(unresolved_locations),
    )

insect_counts.drop(columns="geocode_result", inplace=True)
insect_counts.head(3)
```

```python
# Merge butterfly counts with insect counts by monitoring location.
insect_total = pd.merge(
    left=insect_counts,
    right=bfly_counts,
    how="left",
    on="location",
)

# Replace missing butterfly counts with zero.
insect_total["Butterfly Total"] = insect_total["Butterfly Total"].fillna(0)

# Calculate the combined insect and butterfly count at each location.
insect_total["Total Insect Count"] = (
    insect_total["Butterfly Total"] + insect_total["Insect Total"]
)

# Keep a complete analytical table and a map-ready table with valid coordinates.
insect_total_map = insect_total.dropna(
    subset=["latitude", "longitude"]
).copy()

# Convert valid coordinates to a GeoDataFrame for mapping.
gdf_insect_total = gpd.GeoDataFrame(
    insect_total_map,
    geometry=gpd.points_from_xy(
        insect_total_map["longitude"],
        insect_total_map["latitude"]
    ),
    crs="EPSG:4326",
)

gdf_insect_total.head(3)
```

```python
# Initialise the map at Melbourne.
m = create_template_map(location=[-37.815, 144.96], zoom_start=14)

# Add insect locations that have valid geocoded coordinates.
for i in range(len(insect_total_map)):
    folium.CircleMarker(
        location=[
            insect_total_map.iloc[i]["latitude"],
            insect_total_map.iloc[i]["longitude"],
        ],
        radius=6,
        popup="Insects = {}, Butterflies = {}, Total = {}".format(
            insect_total_map["Insect Total"].values[i],
            insect_total_map["Butterfly Total"].values[i],
            insect_total_map["Total Insect Count"].values[i],
        ),
        color=MOP_SECONDARY_COLOUR,
        fill=True,
        fill_color=MOP_SECONDARY_COLOUR,
        fill_opacity=0.85,
    ).add_to(m)

m.get_root().html.add_child(
    folium.Element("<h3 align='center'>Locations of insect sightings</h3>")
)

# Add a legend using template colours.
add_template_legend(
    m,
    "Legend",
    [("Insect sighting location", MOP_SECONDARY_COLOUR)]
)

m
```

Insect monitoring locations are shown on the map with green markers. Selecting a marker displays the insect count, butterfly count and combined total recorded at that location. The monitoring sites cover a broad geographic area across the City of Melbourne.

```python
# IMPORT PEDESTRIAN COUNTS DATASET
# Set dataset_id to query for the API call dataset name
dataset_id = 'pedestrian-counting-system-monthly-counts-per-hour'
# Save dataset to DataFrame variable
ped_count = collect_data(dataset_id)
# Check number of records in df
print(f'The dataset contains {len(ped_count)} records.')
# View df
ped_count.head(3)
```

```python
# Pedestrian data is prepared in the following cell.
```

```python
# Pedestrian data pre-processing.

# Remove the rows excluded in the original analysis.
# errors="ignore" allows the notebook cell to be rerun safely.
ped_count = ped_count.drop(
    index=[0, 3, 4],
    errors="ignore"
).copy()

# Rename current API column names to the names used throughout this notebook.
ped_count.rename(
    columns={
        "location_id": "locationid",
        "pedestriancount": "total_of_directions"
    },
    inplace=True
)

# Split location into separate latitude and longitude columns.
ped_count[["latitude", "longitude"]] = (
    ped_count["location"]
    .str.split(", ", expand=True)
)

# Convert required columns safely to numeric data types.
ped_count["locationid"] = pd.to_numeric(
    ped_count["locationid"],
    errors="coerce"
)

ped_count["total_of_directions"] = pd.to_numeric(
    ped_count["total_of_directions"],
    errors="coerce"
)

ped_count["latitude"] = pd.to_numeric(
    ped_count["latitude"],
    errors="coerce"
)

ped_count["longitude"] = pd.to_numeric(
    ped_count["longitude"],
    errors="coerce"
)

ped_count = ped_count.convert_dtypes()

ped_count.dtypes
```

<div class="usecase-body">

As location is of importance, an average of pedestrians at each location should be obtained.

</div>


```python
# Keep rows with the values required for the spatial analysis.
ped_count_valid = ped_count.dropna(
    subset=[
        "locationid",
        "total_of_directions",
        "latitude",
        "longitude"
    ]
).copy()

# Set grouping fields.
groupbyfields = [
    "locationid",
    "location",
    "latitude",
    "longitude"
]

# Create a DataFrame containing the mean pedestrian count at each sensor.
ped_count_mean = (
    ped_count_valid
    .groupby(groupbyfields)["total_of_directions"]
    .mean()
    .reset_index(name="hourly_counts_mean")
)

# Convert sensor IDs to strings for plotting.
ped_count_mean["locationid"] = ped_count_mean["locationid"].astype("string")

# Convert mean pedestrian counts to integers.
ped_count_mean["hourly_counts_mean"] = (
    ped_count_mean["hourly_counts_mean"].astype(int)
)

print(f"The dataset contains {len(ped_count_mean)} records.")
ped_count_mean.head(3)
```

<div id="uc109-section-3" class="usecase-section-heading"><b>3. Optimisation Modelling</b></div>
<div class="usecase-body">Apply K-means clustering to high pedestrian-count observations to identify 10 candidate green wall locations.</div>


<div class="usecase-body">

The purpose of this use case is to identify candidate locations for green walls in the City of Melbourne. K-means clustering is applied to high pedestrian-count observations. Because K-means relies on Euclidean distance, the WGS84 latitude and longitude coordinates are first projected to a metric coordinate reference system before clustering. GDA2020 / MGA Zone 55 (EPSG:7855) is used for the Melbourne study area so distances are measured in metres. The resulting cluster centres are then converted back to WGS84 (EPSG:4326) for mapping. Recorded insect locations are displayed alongside the candidate green wall locations to provide additional biodiversity context.

</div>


```python
# Keep only pedestrian records with valid values required for clustering.
max_ped_count = (
    ped_count
    .dropna(subset=["total_of_directions", "latitude", "longitude"])
    .sort_values(
        by="total_of_directions",
        ascending=False
    )
)

# Select the top 100,000 pedestrian-count observations.
max_ped = max_ped_count.head(100000).copy()

# Create a GeoDataFrame in WGS84 using longitude as x and latitude as y.
gdf_max_ped = gpd.GeoDataFrame(
    max_ped,
    geometry=gpd.points_from_xy(
        max_ped["longitude"],
        max_ped["latitude"]
    ),
    crs="EPSG:4326"
)

# Project to GDA2020 / MGA Zone 55 so K-means operates in metres.
gdf_max_ped_metric = gdf_max_ped.to_crs(epsg=7855)

# Extract projected x/y coordinates for clustering.
gdf_max_ped_metric["x_m"] = gdf_max_ped_metric.geometry.x
gdf_max_ped_metric["y_m"] = gdf_max_ped_metric.geometry.y

gdf_max_ped_metric.head(3)
```

<div class="usecase-body">

The top 100,000 pedestrian-count observations are selected for clustering. K-means is used to divide these observations into 10 clusters, corresponding to the 10 candidate green wall locations required in this use case. The resulting cluster centres are presented in a scatter plot, an interactive map and a table of coordinates.

</div>


```python
# Set k to the number of green walls to be located.
k = 10

# Initialise a reproducible K-means model.
model = cluster.KMeans(
    n_clusters=k,
    init="k-means++",
    random_state=42,
    n_init=10
)

# Cluster using projected metric coordinates only.
X = gdf_max_ped_metric[["x_m", "y_m"]].astype(float)

# Fit the model and assign clusters.
gdf_max_ped_metric["cluster"] = model.fit_predict(X)

# Find the observation nearest to each cluster centre.
closest, distances = scipy.cluster.vq.vq(
    model.cluster_centers_,
    X.values
)

gdf_max_ped_metric["centroids"] = 0
gdf_max_ped_metric.loc[
    gdf_max_ped_metric.index[closest],
    "centroids"
] = 1

gdf_max_ped_metric.head(3)
```

```python
# Plot clusters using projected coordinates in metres.
fig, ax = plt.subplots()

sns.scatterplot(
    x="x_m",
    y="y_m",
    data=gdf_max_ped_metric,
    hue="cluster",
    size="centroids",
    size_order=[1, 0],
    palette=[
        "#059669", "#9333EA", "#F97316", "#2563EB", "#EAB308",
        "#EC4899", "#0891B2", "#7C3AED", "#EA580C", "#4D7C0F"
    ],
    legend="brief",
    ax=ax
).set_title(f"Clustering (k={k})")

# Plot K-means cluster centres as black crosses.
cluster_centres = model.cluster_centers_
ax.scatter(
    cluster_centres[:, 0],
    cluster_centres[:, 1],
    s=50,
    c="black",
    marker="x"
)

ax.set_xlabel("Easting (metres)")
ax.set_ylabel("Northing (metres)")
plt.show()
```

<div class="usecase-body">

The K-means results show the geographic grouping of the selected pedestrian observations. Each colour represents a cluster, while black crosses show the cluster centres. These centres represent the candidate green wall locations generated by the model. The locations are also displayed on an interactive Folium map for easier geographic interpretation.

</div>


```python
# Create centroid points from the K-means cluster centres in the projected CRS.
centroid_metric = gpd.GeoDataFrame(
    {"cluster": range(k)},
    geometry=gpd.points_from_xy(
        model.cluster_centers_[:, 0],
        model.cluster_centers_[:, 1]
    ),
    crs="EPSG:7855"
)

# Convert centroid locations back to WGS84 for mapping.
ped_counts_centroid = centroid_metric.to_crs(epsg=4326).copy()
ped_counts_centroid["longitude"] = ped_counts_centroid.geometry.x
ped_counts_centroid["latitude"] = ped_counts_centroid.geometry.y

ped_counts_centroid.head(3)
```

```python
# Initialise the map at Melbourne.
m = create_template_map(
    location=[-37.815, 144.96],
    zoom_start=14
)

# Add candidate green wall centroid locations.
for i in range(len(ped_counts_centroid)):
    folium.CircleMarker(
        location=[
            ped_counts_centroid.iloc[i]["latitude"],
            ped_counts_centroid.iloc[i]["longitude"]
        ],
        radius=7,
        popup="Latitude = {}, Longitude = {}".format(
            ped_counts_centroid["latitude"].values[i],
            ped_counts_centroid["longitude"].values[i]
        ),
        color=MOP_PRIMARY_COLOUR,
        fill=True,
        fill_color=MOP_PRIMARY_COLOUR,
        fill_opacity=0.9,
    ).add_to(m)

# Add insect locations with valid coordinates only.
for i in range(len(insect_total_map)):
    folium.CircleMarker(
        location=[
            insect_total_map.iloc[i]["latitude"],
            insect_total_map.iloc[i]["longitude"]
        ],
        radius=6,
        color=MOP_SECONDARY_COLOUR,
        fill=True,
        fill_color=MOP_SECONDARY_COLOUR,
        fill_opacity=0.85,
    ).add_to(m)

# Add a title.
m.get_root().html.add_child(
    folium.Element(
        "<h3 align='center'>Optimal Locations of Green Walls</h3>"
    )
)

# Add a legend using template colours.
add_template_legend(
    m,
    "Legend",
    [
        ("Candidate green wall location", MOP_PRIMARY_COLOUR),
        ("Insect sighting location", MOP_SECONDARY_COLOUR),
    ]
)

m
```

<div id="conclusion" class="usecase-unnumbered-heading"><b>Conclusion</b></div>


<div class="usecase-body">

This use case combines microclimate, biodiversity and pedestrian activity data to identify candidate locations for green walls in the City of Melbourne. The analysis uses environmental observations to provide context and applies K-means clustering to areas with high pedestrian activity to identify locations where green-wall investment could have strong visibility and potential community value.

This links directly back to the scenario. The ecologist's goal is to increase urban biodiversity, while the city planner needs to identify suitable locations where green walls can support biodiversity and also be experienced by areas with high pedestrian activity. By combining biodiversity context with pedestrian-demand clustering, the use case provides an evidence-based way to narrow the city to candidate locations for further site assessment.

The resulting candidate locations should be treated as a starting point rather than final construction sites. Future work should include building suitability, wall orientation, sunlight, irrigation access, ownership, maintenance requirements, ecological connectivity and more recent environmental and pedestrian data before a final green-wall location is selected.

</div>


<div class="usecase-body">

**What we achieved in this analysis**

We imported and prepared multiple City of Melbourne datasets, visualised particulate readings and biodiversity observations, geocoded insect monitoring locations and mapped pedestrian activity. K-means clustering was then applied to the highest pedestrian-count observations to identify 10 candidate green wall locations. The pedestrian coordinates were projected from WGS84 to GDA2020 / MGA Zone 55 before clustering so that Euclidean distance was meaningful. The resulting candidate locations were presented in both map and table formats and compared visually with recorded insect monitoring locations.

**What we learned from this analysis**

Butterfly and insect observations varied across the recorded monitoring locations, while the available microclimate readings provided only a limited snapshot of particulate levels. Together, these datasets provide useful environmental and spatial context, but they vary in coverage and should be interpreted as exploratory evidence rather than as a complete assessment of biodiversity, air quality or final site suitability across the City of Melbourne.

**Observations for further opportunities**

Further analysis could incorporate building construction materials and footpath width. Building-material data could be used to exclude locations where structures may not be suitable for supporting a green wall, while footpath-width data could help remove locations where a green wall could obstruct pedestrian movement. Additional biodiversity surveys, more recent pedestrian data and other site-feasibility variables would further strengthen the location-selection process.
