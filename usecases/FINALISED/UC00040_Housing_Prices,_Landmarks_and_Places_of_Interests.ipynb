<div id="section-2" class="usecase-section-heading"><b>2. House Prices dataset — loading & cleaning</b></div>

<div class="usecase-body"> Loading the House Prices by Small Area Dataset </div>

```python
def API_unlimited(datasetName):

    dataset_id = datasetName
    # https://data.melbourne.vic.gov.au/explore/dataset/pedestrian-counting-system-monthly-counts-per-hour/information/
    #dataset_id = 'pedestrian-counting-system-monthly-counts-per-hour'

    base_url = 'https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/'
    #apikey = api_key
    dataset_id = dataset_id
    format = 'csv'

    url = f'{base_url}{dataset_id}/exports/{format}'
    params = {
        'select': '*',
        'limit': -1,  # all records
        'lang': 'en',
        'timezone': 'UTC',
       # 'api_key': apikey
    }

    # GET request
    response = requests.get(url, params=params)

    if response.status_code == 200:
        # StringIO to read the CSV data
        url_content = response.content.decode('utf-8')
        datasetName = pd.read_csv(StringIO(url_content), delimiter=';')
        print(datasetName.sample(10, random_state=999)) # Test
        return datasetName
    else:
        print(f'Request failed with status code {response.status_code}')
```

```python
dataset_id_1 = 'house-prices-by-small-area-sale-year'

dataset_id_2 ='landmarks-and-places-of-interest-including-schools-theatres-health-services-spor'

dataset1 = API_unlimited(dataset_id_1)
dataset2 = API_unlimited(dataset_id_2)
```

```python
house_price= dataset1
print(house_price.columns.tolist())
house_price = house_price.rename(columns={
    'median_price': 'Median_Price',
    'sale_year': 'Sale_Year',
    'transaction_count': 'Transaction_Count',
    'small_area': 'Small_Area',
    'type': 'Type',
})

house_price.head(10)
```

```python
# Displaying shape
print("Shape of the dataset:")
print(house_price.shape)

# Displaying data Types
print("\nData Types of the dataset:")
print(house_price.dtypes)
```

<div id="subsection-2-2" class="usecase-sub-section-heading"><b>2.2 Null / zero handling</b></div>

<div class="usecase-body">Cleaning the data by eliminating null values. This matters because missing or zero prices would distort the averages and regression results later in the analysis.</div>

```python
#Finding null and Zero values
null_values = house_price.isnull().sum()

zero_values = (house_price == 0).sum()

print("Null values:")
print(null_values)

print("\nZero values:")
print(zero_values)
```

<div class="usecase-body">Cleaning the House Prices dataset and displaying a cleaned version of the dataset.</div>

```python

# Dropping rows with NaN, 0, or NA values
housing_price = house_price.dropna(axis=0, how='any')
housing_price = housing_price.replace(0, pd.NA).dropna()
housing_price = housing_price[housing_price['Median_Price'] >= 100000].reset_index(drop=True)


print("Cleaned dataset shape:", housing_price.shape)
print("\n Cleaned data:")
housing_price

```

<div id="subsection-2-3" class="usecase-sub-section-heading"><b>2.3 Outlier trimming (1st/99th percentile)</b></div>

<div class="usecase-body">Trimming the top and bottom 1% of prices to remove extreme outliers. This matters because a few unusually high or low sales can skew the median, mean, and price trends we rely on.</div>

```python
import numpy as np

# Calculate the 1st and 99th percentiles of 'Median_Price'
price_1st_percentile = housing_price['Median_Price'].quantile(0.01)
price_99th_percentile = housing_price['Median_Price'].quantile(0.99)

# Filter out the extreme values based on percentiles
housing_price = housing_price[
    (housing_price['Median_Price'] >= price_1st_percentile) &
    (housing_price['Median_Price'] <= price_99th_percentile)
].reset_index(drop=True)


print("Cleaned dataset shape:", housing_price.shape)
print("\n Cleaned data:")
housing_price

```

<div id="subsection-3-1" class="usecase-sub-section-heading"><b>3.1 Transactions per year</b></div>

<div class="usecase-body">Here we total all property transactions per year to see the overall market trend, with a dashed line marking the yearly average so we can spot above- and below-average years.</div>

```python
import matplotlib.pyplot as plt

transaction_counts = housing_price.groupby('Sale_Year')['Transaction_Count'].sum()
average_transaction_count = transaction_counts.mean()

plt.figure(figsize=(8, 4))
plt.bar(transaction_counts.index, transaction_counts.values, color='#059669')
plt.axhline(average_transaction_count, color='#9333EA', linestyle='--',
            label=f'Average ({average_transaction_count:.2f} transactions/year)')
plt.title('Number of Transactions Each Year')
plt.xlabel('Year')
plt.ylabel('Transaction Count')
plt.legend()
plt.tight_layout()
plt.show()
```

<div id="section-5" class="usecase-section-heading"><b>5. Regression: price trend per suburb</b></div>

<div class="usecase-body">We fit a linear regression for each suburb to quantify its yearly price trend — how much the median price changes per year — so we can compare which suburbs are appreciating fastest.</div>

```python
from sklearn.linear_model import LinearRegression

regression_results = []
regression_models = {}

for suburb, data in suburb_datasets.items():
    X = data[['Sale_Year', 'Transaction_Count']]
    y = data['Median_Price']
    model = LinearRegression().fit(X, y)
    regression_models[suburb] = model
    regression_results.append({
        'Suburb': suburb,
        'Yearly_Price_Trend': model.coef_[0],          # $ change per year
        'Per_Transaction_Effect': model.coef_[1],
        'R2': model.score(X, y),
    })

regression_df = (pd.DataFrame(regression_results)
                 .sort_values('Yearly_Price_Trend', ascending=False)
                 .reset_index(drop=True))
regression_df
```

```python
import matplotlib.pyplot as plt

# The chart below is generated directly from the fitted models.
plt.figure(figsize=(9, 5))
plt.bar(regression_df['Suburb'], regression_df['Yearly_Price_Trend'], color='#059669')
plt.title('Estimated yearly change in median price by suburb ($)')
plt.xlabel('Suburb')
plt.ylabel('Yearly Price Trend ($)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

<div class="usecase-body">
For each suburb we fit a linear regression of median price on sale year and transaction count. The <b>Yearly_Price_Trend</b> coefficient (charted above) estimates how much a suburb's median price changes per year: a larger positive value means faster-rising prices, while a value near zero means prices are flat. The R² shows how well the model fits each suburb. Suburbs at the top of the chart are appreciating fastest; those at the bottom, slowest.
</div>

<div id="conclusion" class="usecase-unnumbered-heading"><b>Conclusion</b></div>

<div class="usecase-body">

**What we achieved in this analysis**

We combined three City of Melbourne open datasets — house prices, landmarks, and CLUE suburb boundaries — to examine how a suburb's amenities relate to its housing market. Along the way we cleaned and trimmed the price data, charted transaction and price trends over time and by suburb, spatially assigned every landmark to its suburb, measured landmark density, and fitted regressions linking amenities to prices.

**What we learned from this analysis**

- Homebuyer (Scenario 1): The per-suburb price trends and landmark-density comparison let a buyer see which suburbs are appreciating and how well-serviced each is, helping judge whether a suburb's prices reflect its access to amenities.
- Builder (Scenario 2): Transaction volumes by suburb and year, alongside median/mean prices and amenity density, highlight the active, higher-value, well-serviced areas that make strong candidates for development.
- Council (Scenario 3): The landmark counts, themes, and density per suburb reveal which areas are under-served with amenities, pointing to where infrastructure investment could most improve liveability and appeal.

Transaction activity was not constant over time — the 2009–2014 period saw above-average volumes before tapering off. Prices varied sharply between suburbs: the per-suburb regressions showed some areas appreciating strongly year on year while others stayed nearly flat, and the gap between mean and median prices revealed which suburbs are skewed by a handful of high-value sales. On the amenity side, landmarks are unevenly distributed, with the CBD and inner suburbs showing far higher landmark density. Bringing the two together, landmark density showed almost no relationship with median price (correlation 0.017, R² 0.000, across 11 suburbs), so amenity density alone does not explain the price differences between suburbs.

**Observations for further opportunities**

The absence of a relationship suggests price is driven by far more than amenity count. Future work could weight landmarks by type (a hospital or train station likely matters more than a minor site), control for suburb area, population, and dwelling mix, add distance-to-CBD as a feature, and try non-linear models to capture relationships a straight line misses.
