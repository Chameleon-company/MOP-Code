<div id="section-1" class="usecase-section-heading">
    1. Set up the Python environment
</div>

<div class="usecase-body">

In this section we import the libraries used throughout the use case. We use `requests` to call the API, `pandas` to hold the data, `folium` for the interactive maps, and `geopy` for the distance calculation in section 5.1. This cell also sets up three things the maps need later: the template colours, the basemap style and the zoom level it stops at, and two small helpers, one that draws a marker in an exact colour and one that draws a legend, since `folium` has neither built in.

</div>

```python
%pip install geopy
```

```python
import requests
import pandas as pd
from geopy.distance import geodesic, great_circle
import folium

MOP_GREEN = '#059669'
MOP_LIGHT_GREEN = '#2DBE6C'
MOP_DARK = '#121212'

EXTRA_BLUE = '#0B6FA4'
EXTRA_LIGHT_BLUE = '#4FA3D1'
EXTRA_SLATE = '#5D6D7E'
EXTRA_ORANGE = '#E07B00'
EXTRA_RUST = '#B23A00'
EXTRA_OLIVE = '#8C6D1F'
EXTRA_PURPLE = '#7B3FA0'
EXTRA_MAGENTA = '#C2185B'

MAP_TILES = (
    'https://server.arcgisonline.com/ArcGIS/rest/services/Canvas/'
    'World_Light_Gray_Base/MapServer/tile/{z}/{y}/{x}'
)
MAP_ATTR = 'Tiles &copy; Esri'
MAP_MAX_ZOOM = 16


def add_marker(folium_map, location, colour, popup):
    """Add a circular marker in an exact colour to a Folium map.

    Parameters
    ----------
    folium_map : folium.Map
        The map to add the marker to.
    location : list or tuple
        Latitude and longitude of the marker.
    colour : str
        Hex colour for the marker fill.
    popup : str
        Text shown when the marker is clicked.
    """
    folium.CircleMarker(
        location=location,
        radius=6,
        color=MOP_DARK,
        weight=1,
        fill=True,
        fill_color=colour,
        fill_opacity=0.9,
        popup=popup,
    ).add_to(folium_map)
    return folium_map


def add_legend(folium_map, title, entries):
    """Add a fixed legend box to a Folium map.

    Parameters
    ----------
    folium_map : folium.Map
        The map to add the legend to.
    title : str
        Heading shown at the top of the legend.
    entries : list of (str, str)
        Pairs of swatch colour and label.
    """
    rows = ''.join(
        f'<div style="margin-top:4px">'
        f'<span style="display:inline-block;width:12px;height:12px;'
        f'background-color:{colour};border-radius:50%;margin-right:8px"></span>'
        f'{label}</div>'
        for colour, label in entries
    )

    legend = f"""
    <div style="position:fixed; bottom:30px; left:30px; z-index:9999;
                background-color:#fff; border:2px solid {MOP_DARK};
                border-radius:6px; padding:10px 14px;
                font-family:'Inter',sans-serif; font-size:13px; color:{MOP_DARK};">
      <div style="font-weight:700; color:{MOP_DARK}; margin-bottom:4px">{title}</div>
      {rows}
    </div>
    """

    folium_map.get_root().html.add_child(folium.Element(legend))
    return folium_map
```

<div class="usecase-body">

`folium` comes with a standard Anaconda distribution. `geopy` often does not, which is why the cell above installs it into the kernel running this notebook. If you already have both, the install step will report that the requirement is satisfied and move on.

</div>

<div id="section-2" class="usecase-section-heading">
    2. Retrieve the facilities data
</div>

<div class="usecase-body">

The City of Melbourne Open Data Portal exposes every dataset through the Opendatasoft Explore API. A request is built from three parts: the catalogue base URL, the dataset name as it appears on the portal, and a query string that controls the response. The query string carries the two settings that matter here. `limit` is how many records to return in one call, and `offset` is where in the dataset to start. The API will not return the whole dataset at once, so collecting it means calling the API repeatedly and moving `offset` forward each time.

</div>

<div id="section-2-1" class="usecase-sub-section-heading">
    2.1. Define the retrieval function
</div>

<div class="usecase-body">

`get_data` fetches one page. It builds the URL, sends the request, and returns the records along with `total_count`, which is how many records exist on the portal rather than how many came back. That gap is what the loop in the next subsection uses to decide when to stop. A failed request raises rather than printing a message and carrying on. Returning an empty list instead would leave the caller looping without ever reaching `total_count`.

</div>

```python
BASE_URL = 'https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/'

SPECIFIC_PATH = 'landmarks-and-places-of-interest-including-schools-theatres-health-services-spor'

PAGE_SIZE = 50

def get_data(base, data_url, offset=0):
    """Fetch one page of records from the Explore API.

    Parameters
    ----------
    base : str
        Catalogue base URL.
    data_url : str
        Dataset name as it appears on the portal.
    offset : int
        Where in the dataset to start. Defaults to the first record.

    Returns
    -------
    tuple
        The records on this page, and the total number of records the portal
        reports for the dataset.

    Raises
    ------
    RuntimeError
        If the request fails, or the response carries no records.
    """
    
    filters = f'records?limit={PAGE_SIZE}&offset={offset}&timezone=UTC'
    url = f'{base}{data_url}/{filters}'

    response = requests.get(url, timeout=30)

    if response.status_code != 200:
        raise RuntimeError(
            f'Request failed with status {response.status_code}: {url}'
        )

    payload = response.json()

    if 'results' not in payload:
        raise RuntimeError(f'Unexpected response from the API: {payload}')

    return payload['results'], payload['total_count']
```

<div class="usecase-body">

Keeping the request and its checks inside one function means every call is made the same way, and a failure reports what actually went wrong rather than surfacing as a `KeyError` several steps later.

</div>

<div id="section-2-2" class="usecase-sub-section-heading">
    2.2. Retrieve the full dataset
</div>

<div class="usecase-body">

The loop below calls `get_data` repeatedly, moving `offset` forward by one page each time, until the collected records match the count the portal reported. The assembled records are then turned into the DataFrame `df`, which every later section works from.

</div>

```python
records, max_results = get_data(BASE_URL, SPECIFIC_PATH)
offset = PAGE_SIZE

while len(records) < max_results:
    page, _ = get_data(BASE_URL, SPECIFIC_PATH, offset)

    if not page:
        print(f'Stopped early at {len(records):,} of {max_results:,} records.')
        break

    records += page
    offset += PAGE_SIZE
df = pd.DataFrame(records)

print(f'Retrieved {len(df):,} rows and {df.shape[1]} columns.')
df.head()
```

<div class="usecase-body">

The row count should match the total the portal reports for the dataset. If it stops short, the message above says where, and everything below runs on the partial dataset.

</div>

<div id="section-3" class="usecase-section-heading">
    3. Inspect and prepare the data
</div>

<div class="usecase-body">

Two things stand between the retrieved data and a map: it may have gaps, and its coordinates are not yet in a form `folium` can read. This section deals with both.

</div>

<div id="section-3-1" class="usecase-sub-section-heading">
    3.1. Check for missing values
</div>

<div class="usecase-body">

`isnull()` marks every empty cell in the DataFrame as `True`, and `sum()` counts those marks column by column. Together they give a count of missing values per column, which is the quickest way to see whether a column can be relied on.

</div>

```python
null_vals = df.isnull().sum()

print(f'{len(df):,} rows and {df.shape[1]} columns\n')
print(null_vals)
```

<div class="usecase-body">

The columns this notebook depends on are `feature_name`, `sub_theme` and `co_ordinates`. A high count against any of those means the maps below will quietly leave sites out, so they are the numbers to read first. Counts against other columns matter less, because nothing here uses them.

</div>

<div id="section-3-2" class="usecase-sub-section-heading">
    3.2. Unpack the coordinates
</div>

<div class="usecase-body">

The coordinates arrive as a nested dictionary in the `co_ordinates` column, holding a `lat` and a `lon` key. `folium` needs two separate numbers, so `apply()` is used with a lambda function to pull each key out into a column of its own.

</div>

```python
df['latitude'] = df['co_ordinates'].apply(lambda point: point['lat'])
df['longitude'] = df['co_ordinates'].apply(lambda point: point['lon'])

df[['feature_name', 'sub_theme', 'latitude', 'longitude']].head()
```

<div class="usecase-body">

This gives the two columns every map and distance calculation from here on relies on. Printing the four columns that matter together is a quick check that the coordinates line up with the sites they belong to.

</div>

<div id="section-4" class="usecase-section-heading">
    4. Map the facilities
</div>

<div class="usecase-body">

With coordinates in place the sites can be mapped. We start with everything the dataset holds, then narrow it to the categories that answer the question in the scenario, and map the result so the two can be compared.

</div>

<div id="section-4-1" class="usecase-sub-section-heading">
    4.1. Map every location
</div>

<div class="usecase-body">

A `folium` map needs a centre point to open on, and a zoom level to open at. Rather than guessing either, the map is centred on the average of the coordinates and then told to fit the bounds of the data, so every marker is in frame whatever the dataset holds. The basemap is a light grey canvas rather than the default, which keeps the coloured markers legible and avoids the usage limits that leave the default OpenStreetMap tiles blank.

</div>

```python
all_map = folium.Map(
    location=[df['latitude'].mean(), df['longitude'].mean()],
    zoom_start=13,
    tiles=MAP_TILES,
    attr=MAP_ATTR,
    max_zoom=MAP_MAX_ZOOM,
)

for _, row in df.iterrows():
    add_marker(
        all_map,
        [row['latitude'], row['longitude']],
        MOP_GREEN,
        row['feature_name'],
    )

all_map.fit_bounds(
    [
        [df['latitude'].min(), df['longitude'].min()],
        [df['latitude'].max(), df['longitude'].max()],
    ],
    max_zoom=MAP_MAX_ZOOM,
)

add_legend(
    all_map,
    'All places of interest',
    [(MOP_GREEN, 'Landmark or place of interest (click for the name)')],
)

all_map
```

<div class="usecase-body">

Every site in the dataset is now on the map, which is exactly the problem. There are too many markers to read, and most of them are not places you would take a child. Two things are worth noticing before moving on: how densely the markers pack into the city centre, and how mixed they are. Narrowing them down is what the next two subsections do.

</div>

<div id="section-4-2" class="usecase-sub-section-heading">
    4.2. See which categories are available
</div>

<div class="usecase-body">

Before filtering on `sub_theme`, we need to know what it actually contains. `unique()` returns each distinct value once. Dropping the empty values first means the result can be sorted, and the loop prints one per line, which is what makes a long list readable.

</div>

```python
unique_cat = df['sub_theme'].dropna().unique()

print(f'{len(unique_cat):,} distinct sub_theme values\n')
for category in sorted(unique_cat):
    print(category)
```

<div class="usecase-body">

Reading this list is the step that decides the filter in the next subsection. Two things to watch for: categories that sound relevant but are not, such as a private club a family cannot enter, and categories that describe the same kind of site in two different ways. Matching a value that is not spelled exactly as it appears here returns nothing, without raising an error.

</div>

<div id="section-4-3" class="usecase-sub-section-heading">
    4.3. Filter to the child friendly categories
</div>

<div class="usecase-body">

The categories chosen from the list above are collected into `child_cat`: parks and reserves, sports and recreation facilities, schools, cinemas and the aquarium. `isin()` then keeps only the rows whose `sub_theme` matches one of them, into a new DataFrame called `filter_loc`. A new DataFrame is used rather than overwriting `df`, so the full dataset is still there to compare against. `filter_loc` is what sections 4.4 and 5.2 both work from.

</div>

```python
child_cat = [
    'Informal Outdoor Facility (Park/Garden/Reserve)',
    'Private Sports Club/Facility',
    'Major Sports & Recreation Facility',
    'Indoor Recreation Facility',
    'Gymnasium/Health Club',
    'Primary Schools',
    'Secondary Schools',
    'Outdoor Recreation Facility (Zoo, Golf Course)',
    'School - Primary and Secondary Education',
    'Cinema',
    'Aquarium',
]

filter_loc = df[df['sub_theme'].isin(child_cat)].copy()

print(f'{len(filter_loc):,} of {len(df):,} sites match these categories.\n')
print(filter_loc['sub_theme'].value_counts())
```

<div class="usecase-body">

This list is a judgement call rather than a property of the data, and it is the one choice in the notebook that most changes the answer. Adding or removing a category changes what appears on the map below, so it is worth revisiting once you have seen the result.

</div>

<div id="section-4-4" class="usecase-sub-section-heading">
    4.4. Map the filtered locations
</div>

<div class="usecase-body">

The same sites are mapped again, this time from `filter_loc`, with one addition: each marker is coloured by its `sub_theme` using the template palette, and a legend sits in the corner of the map explaining what each colour means. A family reading the map can then tell a park from a school without clicking anything, and without scrolling away from the map to find a key.

</div>

```python
CATEGORY_COLOURS = {
    'Informal Outdoor Facility (Park/Garden/Reserve)': MOP_GREEN,
    'Outdoor Recreation Facility (Zoo, Golf Course)': MOP_LIGHT_GREEN,
    'Major Sports & Recreation Facility': MOP_DARK,
    'Indoor Recreation Facility': EXTRA_BLUE,
    'Gymnasium/Health Club': EXTRA_LIGHT_BLUE,
    'Private Sports Club/Facility': EXTRA_SLATE,
    'Primary Schools': EXTRA_ORANGE,
    'Secondary Schools': EXTRA_RUST,
    'School - Primary and Secondary Education': EXTRA_OLIVE,
    'Cinema': EXTRA_PURPLE,
    'Aquarium': EXTRA_MAGENTA,
}

child_map = folium.Map(
    location=[filter_loc['latitude'].mean(), filter_loc['longitude'].mean()],
    zoom_start=13,
    tiles=MAP_TILES,
    attr=MAP_ATTR,
    max_zoom=MAP_MAX_ZOOM,
)

for _, row in filter_loc.iterrows():
    add_marker(
        child_map,
        [row['latitude'], row['longitude']],
        CATEGORY_COLOURS.get(row['sub_theme'], MOP_DARK),
        f"{row['feature_name']}<br>{row['sub_theme']}",
    )

child_map.fit_bounds(
    [
        [filter_loc['latitude'].min(), filter_loc['longitude'].min()],
        [filter_loc['latitude'].max(), filter_loc['longitude'].max()],
    ],
    max_zoom=MAP_MAX_ZOOM,
)

add_legend(
    child_map,
    'Child friendly categories',
    list(zip(CATEGORY_COLOURS.values(), CATEGORY_COLOURS.keys())),
)

child_map
```

<div class="usecase-body">

Against the first map this one is readable, and it is the first output that answers the scenario: these are the sites a family would consider, and where they sit. What it does not answer is which of them are near any particular address, since a marker gives no sense of distance. Section 5 takes that on.

</div>

    5. Find facilities near a location
</div>

<div class="usecase-body">

This section narrows the map to a radius around a point you choose. The distance is measured with `geopy`, and everything inside the radius is kept and mapped, nearest first.

</div>

<div id="section-5-1" class="usecase-sub-section-heading">
    5.1. Measure the distance between two points
</div>

<div class="usecase-body">

`geopy` measures the distance between two coordinates for us. Its `geodesic` function works across the WGS-84 ellipsoid, which is the model of the earth that GPS coordinates are recorded against, while `great_circle` treats the earth as a sphere, which is what a hand written haversine formula does. Printing both shows how little they differ over a couple of kilometres, and why a tested library is worth more than a formula of your own either way.

</div>

```python
flinders_street = (-37.8183, 144.9671)
mcg = (-37.8200, 144.9834)

print(f'geodesic:     {geodesic(flinders_street, mcg).km:.4f} km')
print(f'great_circle: {great_circle(flinders_street, mcg).km:.4f} km')
```

<div class="usecase-body">

The two answers sit a few metres apart over roughly 1.4 kilometres, so either would do here. `geodesic` is the one used below because it is `geopy`’s default and the more accurate of the two, and because distances given to a family are worth getting right on the first decimal place.

</div>

<div id="section-5-2" class="usecase-sub-section-heading">
    5.2. Search within a radius
</div>

<div class="usecase-body">

The point to search from is set as `USER_LAT` and `USER_LON` at the top of the cell, which means the notebook runs from top to bottom without stopping to ask for anything. It defaults to the city centre; change the two numbers to search from somewhere else. Applying `geodesic` row by row gives a distance from that point, and anything inside the radius is kept and sorted nearest first. The search runs over `filter_loc`, so the results are the child friendly sites mapped in section 4 rather than every site in the city.

</div>

```python
USER_LAT = -37.8136
USER_LON = 144.9631
SEARCH_RADIUS_KM = 2

user_loc = (USER_LAT, USER_LON)

filter_loc['dist'] = filter_loc.apply(
    lambda row: geodesic(user_loc, (row['latitude'], row['longitude'])).km,
    axis=1,
)
radii_locs = (
    filter_loc[filter_loc['dist'] <= SEARCH_RADIUS_KM]
    .sort_values('dist')
)

print(
    f'{len(radii_locs):,} sites within {SEARCH_RADIUS_KM} km, nearest first:\n'
)
for _, row in radii_locs.iterrows():
    print(f"  {row['dist']:.2f} km   {row['feature_name']}")

nearby_map = folium.Map(
    location=user_loc,
    zoom_start=15,
    tiles=MAP_TILES,
    attr=MAP_ATTR,
    max_zoom=MAP_MAX_ZOOM,
)

folium.Marker(
    user_loc,
    popup='Your location',
    icon=folium.Icon(color='black', icon='home'),
).add_to(nearby_map)

for _, row in radii_locs.iterrows():
    add_marker(
        nearby_map,
        [row['latitude'], row['longitude']],
        CATEGORY_COLOURS.get(row['sub_theme'], MOP_DARK),
        f"{row['feature_name']} ({row['dist']:.2f} km)",
    )

nearby_categories = set(radii_locs['sub_theme'])

add_legend(
    nearby_map,
    f'Within {SEARCH_RADIUS_KM} km',
    [(MOP_DARK, 'Your location')]
    + [(colour, category)
       for category, colour in CATEGORY_COLOURS.items()
       if category in nearby_categories],
)

nearby_map
```

<div class="usecase-body">

The list is now ordered, so the first line is the nearest child friendly site to the point you set, and each marker gives its distance when you click it. The red marker is the point being searched from, and the legend carries only the categories that actually turned up inside the radius. If the count comes back as zero, the coordinates sit outside the area the dataset covers, and something closer to the city centre will give you results. One thing to keep in mind when reading the map: this is a distance across open ground, so a site two kilometres away by this measure is further than that on foot. Section 6 gives the honest figure.

</div>

<div id="section-6" class="usecase-section-heading">
    6. Plan a route to a chosen facility
</div>

<div class="usecase-body">

The last step turns a nearby site into directions, using the OSRM API, a free routing service that returns the path between two points as a list of coordinates.

</div>

<div id="section-6-1" class="usecase-sub-section-heading">
    6.1. Define the routing function
</div>

<div class="usecase-body">

`route_finder` sends a start and an end coordinate to OSRM and returns the response. Passing the travel mode into the URL is what makes it a walking, cycling or driving route. OSRM reports some of its own failures in the body of an otherwise successful response, so the status code alone is not enough to know the call worked, and both are checked. One thing to be aware of: the free demo server at `router.project-osrm.org` only hosts the driving profile. It accepts `walking` and `cycling` in the URL without complaint, but the route it returns is the driving one, so treat the mode as illustrative unless you are pointing this at your own OSRM instance.

</div>

```python
def route_finder(user_loc, destination, mode_travel):
    """Return the OSRM route between two points.

    Parameters
    ----------
    user_loc : tuple
        Starting point as (latitude, longitude).
    destination : tuple
        Destination as (latitude, longitude).
    mode_travel : str
        One of 'walking', 'cycling' or 'driving'.

    Returns
    -------
    dict
        The parsed OSRM response, including the route geometry.

    Raises
    ------
    RuntimeError
        If the request fails, or OSRM reports that it could not route.
    """
    
    url = (
        f'http://router.project-osrm.org/route/v1/{mode_travel}/'
        f'{user_loc[1]},{user_loc[0]};{destination[1]},{destination[0]}'
        '?overview=full&geometries=geojson'
    )

    response = requests.get(url, timeout=30)

    if response.status_code != 200:
        raise RuntimeError(
            f'Routing request failed with status {response.status_code}'
        )

    data_route = response.json()

    if data_route.get('code') != 'Ok':
        raise RuntimeError(
            f"OSRM could not route this journey: {data_route.get('code')}"
        )

    return data_route
```

<div id="section-6-2" class="usecase-sub-section-heading">
    6.2. Request and draw the route
</div>

<div class="usecase-body">

The route runs from the point set in section 5.2 to the nearest site found there, which is what `DESTINATION_NAME` picks up by default. Set it to any other name from that list to route somewhere else. The cell also prints the routed distance against the straight line distance, which is the comparison the section is really about.

</div>

```python
if radii_locs.empty:
    raise ValueError(
        'Section 5.2 found no sites, so there is nothing to route to. Move '
        'USER_LAT and USER_LON closer to the city centre and run it again.'
    )

DESTINATION_NAME = radii_locs['feature_name'].iloc[0]
TRAVEL_MODE = 'walking'

matches = radii_locs[radii_locs['feature_name'] == DESTINATION_NAME]

if matches.empty:
    raise ValueError(
        f'"{DESTINATION_NAME}" is not one of the sites found in section 5.2.'
    )

destination = (matches['latitude'].iloc[0], matches['longitude'].iloc[0])

data_route = route_finder(user_loc, destination, TRAVEL_MODE)

routed_km = data_route['routes'][0]['distance'] / 1000
straight_km = matches['dist'].iloc[0]

print(f'{DESTINATION_NAME}')
print(f'  {straight_km:.2f} km in a straight line')
print(f'  {routed_km:.2f} km by the route below')

route_map = folium.Map(
    location=user_loc,
    zoom_start=15,
    tiles=MAP_TILES,
    attr=MAP_ATTR,
    max_zoom=MAP_MAX_ZOOM,
)

folium.Marker(
    user_loc,
    popup='Your location',
    icon=folium.Icon(color='black', icon='home'),
).add_to(route_map)

add_marker(route_map, destination, MOP_GREEN, DESTINATION_NAME)

route_points = [
    (point[1], point[0])
    for point in data_route['routes'][0]['geometry']['coordinates']
]

folium.PolyLine(
    locations=route_points,
    color=MOP_GREEN,
    weight=4,
    opacity=0.8,
).add_to(route_map)

route_map.fit_bounds(route_points, max_zoom=MAP_MAX_ZOOM)

add_legend(
    route_map,
    f'Route by {TRAVEL_MODE}',
    [(MOP_DARK, 'Your location'),
     (MOP_GREEN, DESTINATION_NAME),
     (MOP_GREEN, f'{routed_km:.2f} km {TRAVEL_MODE} route')],
)

route_map
```

<div class="usecase-body">

The two numbers printed above are the point of the whole section. The straight line distance is what section 5.2 sorted on; the routed distance is what somebody actually walks, and the gap between them is the rivers, rail lines and closed streets in between. The polyline shows where that gap comes from, and the map is framed on the route so both ends stay in view however far apart they are.

</div>

<div id="conclusion" class="usecase-unnumbered-heading"><b>Conclusion</b></div>

<div class="usecase-body">

The scenario at the top of this notebook asked a plain question: as a parent new to the city, which child friendly places are near a given address, and how long does it take to reach the one I pick. The notebook answers it end to end. It retrieves every landmark and place of interest the City of Melbourne publishes, narrows that register to the categories a family would actually use, and maps what is left with a colour and a legend so the result can be read at a glance rather than clicked through one marker at a time.

Setting a single pair of coordinates returns every child friendly site within two kilometres, ordered nearest first, and the routing step turns the closest of those into a drawn path with a distance attached. <div class="usecase-body"><b>What we achieved in this analysis</b></div>

<div class="usecase-body">

This analysis took the City of Melbourne landmarks and places of interest dataset from a live API through to a set of interactive maps and a point to point route. The records were paged out of the Explore API into a single DataFrame, checked for missing values, and the nested `co_ordinates` field was unpacked into usable coordinate columns. The sites were then mapped in full, narrowed to the sub-themes relevant to families with children, and mapped again with each marker coloured by category and a legend drawn onto the map. From a single pair of coordinates set in section 5.2, the notebook listed every child friendly site within two kilometres nearest first, and routed to the closest one, printing what that journey costs against the straight line distance.

</div>

<div class="usecase-body"><b>What we learned from this analysis</b></div>

<div class="usecase-body">

An open data API rarely returns everything in one call. Paging with `limit` and `offset` until the collected rows match `total_count` is what turns a partial response into the complete dataset. Coordinates are not always ready to plot. Here they arrive nested inside a single `co_ordinates` field, and they have to be unpacked into their own columns before `folium` can place a marker. Category columns are what make a general dataset answer a specific question. Listing the distinct `sub_theme` values first, then filtering on the ones that matter, is what turns a map of the whole city into a map of child friendly facilities. And a straight line distance is not a journey: the gap between the two figures in section 6.2 is the part a family actually feels.

</div>

<div class="usecase-body"><b>Limitations</b></div>

<div class="usecase-body">

The child friendly categories in `child_cat` are chosen by hand, so a site filed under an unexpected sub-theme is left out without any warning, and a category such as a private sports club may be counted in when a family cannot actually use it. The notebook also depends on two live services: both the Explore API and the public OSRM router can rate limit or be unavailable. That OSRM server only hosts the driving profile, so the travel mode in section 6.2 changes the URL but not the route that comes back.

</div>

<div class="usecase-body"><b>Observations for further opportunities</b></div>

<div class="usecase-body">

Geocoding a typed address would remove the need to set coordinates at all, and clustering the markers on the first map would make the full dataset readable rather than only illustrative. Beyond that, joining the facilities to population or public transport data would move the analysis from where facilities are to where they are missing, which is the question a council would ask rather than a parent.
