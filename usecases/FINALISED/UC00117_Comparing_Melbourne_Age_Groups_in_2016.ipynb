<div id="section-1" class="usecase-section-heading"><b>1. Import libraries</b></div>
<div class="usecase-body"><p>This step loads the Python tools needed to collect, prepare, visualise and compare the two 2016 profile datasets. Keeping the required libraries together at the start makes the rest of the analysis easier to follow and reproduce.</p></div>

```python
# Import the libraries required for this part of the analysis.
import requests
import pandas as pd
from io import StringIO
# Import the libraries required for this part of the analysis.
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
import seaborn as sns

# Import the libraries required for this part of the analysis.
from IPython.display import display
from scipy.stats import ttest_ind
import statsmodels.api as sm
# Import the libraries required for this part of the analysis.
from statsmodels.stats.outliers_influence import variance_inflation_factor
import patsy

# Import the libraries required for this part of the analysis.
from sklearn.preprocessing import MinMaxScaler, LabelEncoder, PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
# Import the libraries required for this part of the analysis.
from sklearn.metrics import mean_squared_error, r2_score

# Official MOP plot colour order from the latest use case template.
MOP_COLOURS = [
    "#059669",  # Primary Green
    "#9333EA",  # Distinct Purple
    "#F97316",  # Distinct Orange
    "#2563EB",  # Distinct Blue
    "#EAB308",  # Distinct Yellow
    "#EC4899",  # Distinct Pink
]

PRIMARY_GREEN = MOP_COLOURS[0]

# Correlation plots require a diverging scale. This uses only colours from
# the approved template palette, with white as the neutral midpoint.
MOP_CORRELATION_CMAP = LinearSegmentedColormap.from_list(
    "mop_correlation",
    [MOP_COLOURS[3], "#FFFFFF", MOP_COLOURS[2]],
)
```

<div id="section-2" class="usecase-section-heading"><b>2. Collect and inspect the data</b></div>
<div class="usecase-body"><p>This step retrieves the two City of Melbourne profile datasets and checks that they have loaded correctly. Inspecting the data first helps confirm what information is available before comparing younger and older population profiles.</p></div>

```python
# Define the City of Melbourne API base URL used to retrieve the datasets.
BASE_URL = "https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets"

# Create a reusable function to download and validate a City of Melbourne dataset.
def collect_data(dataset_id):
    url = f"{BASE_URL}/{dataset_id}/exports/csv"
    params = {"select": "*", "limit": -1, "lang": "en", "timezone": "Australia/Melbourne"}
    try:
        # Send the API request and stop clearly if the download fails.
        response = requests.get(url, params=params, timeout=30)
        response.raise_for_status()
        data = pd.read_csv(StringIO(response.content.decode("utf-8-sig")), sep=";", on_bad_lines="skip")
    except requests.RequestException as exc:
        raise RuntimeError(f"Unable to download dataset: {dataset_id}") from exc
    except pd.errors.ParserError as exc:
        raise ValueError(f"Unable to parse dataset: {dataset_id}") from exc
    if data.empty:
        raise ValueError(f"Dataset returned no records: {dataset_id}")
    return data

# Load the older people profile dataset.
older_people_data = collect_data("older-people-profile-2016-60-years-and-over")
# Load the younger people profile dataset.
younger_people_data = collect_data("younger-people-profile-2016-15-24-years")

# Create a compact summary to check the size, missing values and duplicate rows.
def dataset_summary(data, name):
    display(pd.DataFrame({
        "Rows": [data.shape[0]],
        "Columns": [data.shape[1]],
        "Missing values": [int(data.isna().sum().sum())],
        "Duplicate rows": [int(data.duplicated().sum())],
    }, index=[name]))
    display(data.head())

dataset_summary(older_people_data, "Older People Profile")
dataset_summary(younger_people_data, "Younger People Profile")
```

<div id="section-3" class="usecase-section-heading"><b>3. Prepare and clean the data</b></div>
<div class="usecase-body"><p>This step places both datasets into a consistent format so they can be compared fairly. Adding a profile label, standardising column names and checking duplicates or missing values reduces avoidable inconsistencies before the analysis begins.</p></div>

```python
# Work on copies of the source data so the original objects remain unchanged.
older_people_data = older_people_data.copy()
younger_people_data = younger_people_data.copy()
older_people_data["profile_type"] = "Older People"
younger_people_data["profile_type"] = "Younger People"

# Combine the younger and older datasets into one table for comparison.
combined_data = pd.concat([older_people_data, younger_people_data], ignore_index=True)
# Standardise column names for consistent downstream analysis.
combined_data.columns = [c.strip().lower().replace(" ", "_") for c in combined_data.columns]
# Remove duplicate records before analysis.
combined_data = combined_data.drop_duplicates().reset_index(drop=True)

display(pd.DataFrame({
    "Rows": [combined_data.shape[0]],
    "Columns": [combined_data.shape[1]],
    "Missing values": [int(combined_data.isna().sum().sum())],
    "Duplicate rows": [int(combined_data.duplicated().sum())],
}))
display(combined_data.head())
```

<div id="section-5" class="usecase-section-heading"><b>5. Prepare features for comparison</b></div>
<div class="usecase-body"><p>This section prepares selected fields so they can be used consistently in the statistical and modelling steps. The aim is to make the younger and older profiles easier to compare without changing the meaning of the source data.</p></div>

<div id="section-5-1" class="usecase-sub-section-heading"><b>5.1. Create an age-group feature</b></div>
<div class="usecase-body"><p>This step groups age-related sub-categories into broader labels. Doing this makes age patterns easier to summarise and compare across the profile data.</p></div>

```python
# Group age-related sub-categories into broader age-band labels.
def classify_age_group(value):
    text = str(value)
    for label in ["Under 60", "60-69", "70-79", "80-89", "90+"]:
        if label in text:
            return label
    return "Other"

combined_data["age_group"] = combined_data["sub_category"].apply(classify_age_group)
display(combined_data[["sub_category", "age_group"]].head(10))
```

<div id="section-5-2" class="usecase-sub-section-heading"><b>5.2. Scale numerical features</b></div>
<div class="usecase-body"><p>Min-max scaling places <code>value</code> and <code>sub_order</code> on the same 0–1 range. This makes their numerical scales more consistent for later comparisons while preserving their relative ordering.</p></div>

```python
numerical_columns = ["value", "sub_order"]
# Scale the selected numerical features to a common 0–1 range.
scaler = MinMaxScaler()
combined_data[numerical_columns] = scaler.fit_transform(combined_data[numerical_columns])
display(combined_data[numerical_columns].describe())
```

<div id="section-5-3" class="usecase-sub-section-heading"><b>5.3. Encode categorical variables</b></div>
<div class="usecase-body"><p>This step converts text categories into numeric codes so the existing modelling workflow can use them. The codes are only computational labels and should not be interpreted as a natural ranking of the categories.</p></div>

```python
# List categorical fields that need numeric encoding for the existing modelling workflow.
categorical_columns = ["profile", "category", "sub_category", "profile_type", "age_group"]
for column in categorical_columns:
    encoder = LabelEncoder()
    combined_data[f"{column}_encoded"] = encoder.fit_transform(combined_data[column].astype(str))

display(combined_data[[c for c in combined_data.columns if c.endswith("_encoded")]].head())
```

<div id="section-6-1" class="usecase-sub-section-heading"><b>6.1. Conduct independent-samples t-tests</b></div>
<div class="usecase-body"><p>An independent-samples t-test compares the average value of a field between two separate groups. Here it is used to check whether the younger and older profiles have statistically different averages for <code>value</code> and <code>sub_order</code>.</p></div>

```python
# Separate the combined data into older and younger profile groups.
older_group = combined_data.loc[combined_data["profile_type"] == "Older People"].copy()
younger_group = combined_data.loc[combined_data["profile_type"] == "Younger People"].copy()

results = []
for variable in ["value", "sub_order"]:
    t_stat, p_value = ttest_ind(older_group[variable].dropna(), younger_group[variable].dropna(), equal_var=False)
    results.append({"Variable": variable, "t-statistic": t_stat, "p-value": p_value, "Significant at 0.05": p_value < 0.05})
display(pd.DataFrame(results))
```

The earlier notebook found no statistically significant difference in `value`, while `sub_order` differed between the two datasets. Because `sub_order` is an ordering/index field, this should not be described as proof that one group has higher or lower wellbeing.


<div id="section-6-2" class="usecase-sub-section-heading"><b>6.2. Compare <code>sub_order</code> distributions</b></div>
<div class="usecase-body"><p>This boxplot provides a visual comparison of <code>sub_order</code> across the two profile groups. It helps show whether the difference identified statistically is also visible in the spread and central tendency of the data.</p></div>

```python
# Keep the older and younger profile groups in a fixed comparison order.
profile_type_order = ["Older People", "Younger People"]

# Compare the distribution of sub_order between the two profile groups.
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=combined_data,
    x="profile_type",
    y="sub_order",
    order=profile_type_order,
    hue="profile_type",
    palette=MOP_COLOURS[:2],
    legend=False,
)
plt.title("Sub Order by Profile Type")
plt.xlabel("Profile Type")
plt.ylabel("Scaled Sub Order")
plt.tight_layout()
plt.show()
```

<div id="section-6-3" class="usecase-sub-section-heading"><b>6.3. Inspect possible outliers</b></div>
<div class="usecase-body"><p>This step checks unusually high or low observations in the younger profile. The records are inspected rather than automatically removed or reclassified so that potentially valid source data is not changed without evidence.</p></div>

```python
# Calculate the interquartile range to identify possible outliers.
q1 = younger_group["sub_order"].quantile(0.25)
q3 = younger_group["sub_order"].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

younger_outliers = younger_group.loc[(younger_group["sub_order"] < lower_bound) | (younger_group["sub_order"] > upper_bound)]
print(f"Potential younger-group outliers: {len(younger_outliers)}")
display(younger_outliers.head(10))
```

<div id="section-7" class="usecase-section-heading"><b>7. Investigate relationships using regression</b></div>
<div class="usecase-body"><p>This section looks beyond simple group differences and examines how several fields are related to <code>sub_order</code>. The goal is to understand which variables are associated with changes in the response while keeping the findings exploratory.</p></div>

<div id="section-7-1" class="usecase-sub-section-heading"><b>7.1. Compare group-specific correlation matrices</b></div>
<div class="usecase-body"><p>Separate correlation matrices are used to see whether relationships between variables look different for younger and older profiles. This provides context before fitting regression models to each group.</p></div>

```python
# Separate the combined data into older and younger profile groups.
older_group = older_group.copy()
younger_group = younger_group.copy()

# Encode selected categorical fields within each profile group for the existing analysis.
for group_data in (older_group, younger_group):
    for column in ["profile", "category", "sub_category"]:
        encoder = LabelEncoder()
        group_data[f"{column}_encoded"] = encoder.fit_transform(
            group_data[column].astype(str)
        )

# Choose the fields used in the group-specific correlation analysis.
corr_features = [
    "profile_encoded",
    "census_year",
    "category_encoded",
    "sub_category_encoded",
    "sub_order",
    "value",
]

# Calculate separate correlation matrices for older and younger profiles.
older_corr_matrix = older_group[corr_features].corr()
younger_corr_matrix = younger_group[corr_features].corr()

# Visualise correlations within the older profile dataset.
plt.figure(figsize=(10, 7))
sns.heatmap(
    older_corr_matrix,
    annot=True,
    cmap=MOP_CORRELATION_CMAP,
    vmin=-1,
    vmax=1,
    center=0,
)
plt.title("Correlation Matrix for Older People")
plt.xlabel("Variable")
plt.ylabel("Variable")
plt.tight_layout()
plt.show()

# Visualise correlations within the younger profile dataset.
plt.figure(figsize=(10, 7))
sns.heatmap(
    younger_corr_matrix,
    annot=True,
    cmap=MOP_CORRELATION_CMAP,
    vmin=-1,
    vmax=1,
    center=0,
)
plt.title("Correlation Matrix for Younger People")
plt.xlabel("Variable")
plt.ylabel("Variable")
plt.tight_layout()
plt.show()
```

<div id="section-7-2" class="usecase-sub-section-heading"><b>7.2. Fit OLS regression models with encoded features</b></div>
<div class="usecase-body"><p><b>Ordinary Least Squares (OLS)</b> is a regression method that estimates the line or combination of predictors that best fits the observed data by minimising squared prediction errors. Here, separate OLS models are used to explore which available fields are associated with <code>sub_order</code> for younger and older profiles.</p></div>

```python
# Define the predictor fields used in the OLS regression models.
predictors = ["profile_encoded", "census_year", "category_encoded", "sub_category_encoded", "value"]
X_older = sm.add_constant(older_group[predictors])
X_younger = sm.add_constant(younger_group[predictors])
model_older = sm.OLS(older_group["sub_order"], X_older).fit()
model_younger = sm.OLS(younger_group["sub_order"], X_younger).fit()
print("Older People OLS model")
print(model_older.summary())
print("\nYounger People OLS model")
print(model_younger.summary())
```

<div id="section-7-3" class="usecase-sub-section-heading"><b>7.3. Check multicollinearity</b></div>
<div class="usecase-body"><p><b>Variance Inflation Factor (VIF)</b> measures how strongly one predictor overlaps with the information provided by other predictors. High VIF values can make regression coefficients less stable, so this check helps identify variables that may be too closely related to one another.</p></div>

```python
# Calculate VIF values to check whether predictors overlap too strongly.
def calculate_vif(data, features):
    values = data[features].dropna()
    return pd.DataFrame({"Feature": features, "VIF": [variance_inflation_factor(values.values, i) for i in range(len(features))]})

print("VIF — Older People")
display(calculate_vif(older_group, predictors))
print("VIF — Younger People")
display(calculate_vif(younger_group, predictors))
```

<div id="section-7-4" class="usecase-sub-section-heading"><b>7.4. Refit regression models without <code>census_year</code></b></div>
<div class="usecase-body"><p>This step refits the models after removing <code>census_year</code> following the multicollinearity check. It helps assess whether the remaining predictors provide a clearer and more stable view of the relationships in each profile group.</p></div>

```python
# Refit the regression using the reduced predictor set after the VIF check.
reduced_predictors = ["profile_encoded", "category_encoded", "sub_category_encoded", "value"]
model_older_reduced = sm.OLS(older_group["sub_order"], sm.add_constant(older_group[reduced_predictors])).fit()
model_younger_reduced = sm.OLS(younger_group["sub_order"], sm.add_constant(younger_group[reduced_predictors])).fit()
print("Older People OLS model — census_year excluded")
print(model_older_reduced.summary())
print("\nYounger People OLS model — census_year excluded")
print(model_younger_reduced.summary())
```

<div id="section-7-5" class="usecase-sub-section-heading"><b>7.5. Fit an interaction model</b></div>
<div class="usecase-body"><p>The interaction model checks whether relationships between predictors and <code>sub_order</code> differ between younger and older profiles. This helps determine whether the same factor may be associated differently across the two groups, without treating the result as evidence of causation.</p></div>

```python
# Prepare copies of both profile groups for the combined interaction analysis.
older_model_data = older_group.copy()
younger_model_data = younger_group.copy()
# Add group labels so each observation retains its age-profile identity.
older_model_data["group"] = "Older"
younger_model_data["group"] = "Younger"
# Combine the prepared groups into one dataset for the interaction model.
model_data = pd.concat([older_model_data, younger_model_data], ignore_index=True)
# Encode the profile group so it can be included in the regression model.
model_data["group_encoded"] = LabelEncoder().fit_transform(model_data["group"])
display(model_data.head())

# Define an interaction model to test whether relationships differ by age group.
formula = "sub_order ~ group_encoded * (profile_encoded + census_year + category_encoded + sub_category_encoded + value)"
y_interaction, X_interaction = patsy.dmatrices(formula, model_data, return_type="dataframe")
interaction_model = sm.OLS(y_interaction, X_interaction).fit()
print(interaction_model.summary())
```

<div id="section-8" class="usecase-section-heading"><b>8. Compare non-linear models</b></div>
<div class="usecase-body"><p>This section keeps the notebook's existing non-linear models to compare how closely different approaches fit the observed data. The comparison is exploratory because the reported results are measured on the same data used to fit the models.</p></div>

<div id="section-8-1" class="usecase-sub-section-heading"><b>8.1. Polynomial regression</b></div>
<div class="usecase-body"><p>Polynomial regression allows curved relationships between the predictors and <code>sub_order</code>. This provides a simple way to check whether the data may contain patterns that a straight-line relationship does not capture well.</p></div>

```python
# Define the features used by the non-linear models.
features = ["profile_encoded", "census_year", "category_encoded", "sub_category_encoded", "value", "group_encoded"]
X_model = model_data[features]
y_model = model_data["sub_order"]
# Create polynomial features to capture curved relationships.
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X_model)
model_poly = LinearRegression().fit(X_poly, y_model)
y_pred_poly = model_poly.predict(X_poly)
poly_metrics = {"Model": "Polynomial Regression", "MSE": mean_squared_error(y_model, y_pred_poly), "R_squared": r2_score(y_model, y_pred_poly)}
display(pd.DataFrame([poly_metrics]))
```

<div id="section-8-2" class="usecase-sub-section-heading"><b>8.2. Random forest regression</b></div>
<div class="usecase-body"><p>Random forest regression combines many decision trees to model more complex patterns in the data. It is included to compare how a flexible tree-based approach fits the same profile information.</p></div>

```python
# Fit a random forest model to capture more complex patterns.
model_rf = RandomForestRegressor(n_estimators=100, random_state=42).fit(X_model, y_model)
y_pred_rf = model_rf.predict(X_model)
rf_metrics = {"Model": "Random Forest", "MSE": mean_squared_error(y_model, y_pred_rf), "R_squared": r2_score(y_model, y_pred_rf)}
display(pd.DataFrame([rf_metrics]))
```

<div id="section-8-3" class="usecase-sub-section-heading"><b>8.3. Gradient boosting regression</b></div>
<div class="usecase-body"><p>Gradient boosting builds a sequence of small decision trees, with each new tree attempting to improve the errors made by the previous ones. This gives another way to explore complex relationships in the two profile datasets.</p></div>

```python
# Fit a gradient boosting model for comparison with the other approaches.
model_gbm = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, random_state=42).fit(X_model, y_model)
y_pred_gbm = model_gbm.predict(X_model)
gbm_metrics = {"Model": "Gradient Boosting", "MSE": mean_squared_error(y_model, y_pred_gbm), "R_squared": r2_score(y_model, y_pred_gbm)}
display(pd.DataFrame([gbm_metrics]))
```

<div id="section-8-4" class="usecase-sub-section-heading"><b>8.4. Compare in-sample model performance</b></div>
<div class="usecase-body"><p><b>Mean Squared Error (MSE)</b> is the average squared difference between predicted and observed values, so lower values indicate a closer fit. <b>R² (Coefficient of Determination)</b> shows how much of the variation in the response is explained by the model, with values closer to 1 indicating a closer fit to the observed data. Because these metrics are calculated on the training data, they do not show how well the models would perform on new data.</p></div>

```python
# Combine model metrics into one table for easier comparison.
model_comparison = pd.DataFrame([poly_metrics, rf_metrics, gbm_metrics]).sort_values("R_squared", ascending=False)
display(model_comparison)
```

<div id="section-8-5" class="usecase-sub-section-heading"><b>8.5. Review random forest feature importance</b></div>
<div class="usecase-body"><p>Feature importance shows which input fields the fitted random forest relied on most when making its predictions. This can highlight variables worth further investigation, but it does not prove that those variables cause the observed differences.</p></div>

```python
# Summarise which features the fitted random forest relied on most.
importance_df = pd.DataFrame({
    "Feature": X_model.columns,
    "Importance": model_rf.feature_importances_,
}).sort_values("Importance", ascending=False)

# Display the ranked feature-importance values before plotting them.
display(importance_df)

# Visualise feature importance so the strongest contributors are easy to compare.
plt.figure(figsize=(10, 6))
sns.barplot(
    data=importance_df,
    x="Importance",
    y="Feature",
    color=PRIMARY_GREEN,
)
plt.title("Random Forest Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()
```

<div id="conclusion" class="usecase-unnumbered-heading"><b>Conclusion</b></div>

<div class="usecase-body">

<p>This analysis addressed the scenario by comparing the 2016 younger and older population profiles using Census categories, response sub-categories, their source ordering and the recorded source values. The comparison shows that the two age profiles are not simply different across every field; the result depends on which part of the profile is being examined.</p>

<p>For the source <code>value</code> field, the independent-samples t-test produced a p-value of <code>0.6655</code>. A p-value measures how compatible the observed difference is with the assumption that the two groups have the same average. Using the common significance threshold of <code>0.05</code>, <code>0.6655</code> is high, so this analysis does not provide evidence of a statistically significant difference in the average recorded <code>value</code> between the younger and older profile datasets. This finding must be interpreted cautiously because <code>value</code> is not one single demographic measure: depending on the category, it can represent a count of people, a median, an average or another source summary value.</p>

<p>For <code>sub_order</code>, the p-value was approximately <code>0.000010</code>, which is well below <code>0.05</code>. This indicates a statistically significant difference in the distribution of the source response-category ordering between the two profile datasets. However, <code>sub_order</code> is an ordering field rather than a direct demographic outcome, so the result should not be interpreted as showing that one age group has better or worse wellbeing.</p>

</div>

<p><b>What we achieved in this analysis</b></p>

<div class="usecase-body">

<p>The analysis also found a weak negative correlation of approximately <code>-0.26</code> between <code>value</code> and <code>sub_order</code>. This means there was only a slight tendency for higher source values to appear with lower response-category order numbers, and the relationship was not strong.</p>

<p>Among the exploratory models, the random forest produced the strongest in-sample fit, with an R² of approximately <code>0.9583</code> and MSE of approximately <code>0.001426</code>. Its feature-importance results assigned the greatest importance to <code>category</code> (about 41.5%), followed by <code>sub_category</code> (about 32.8%) and <code>value</code> (about 19.8%). This suggests that the type of Census characteristic and its response category were important to the fitted model, not just the numeric source value.</p>

</div>

<p><b>What we learned from this analysis</b></p>

<div class="usecase-body">

<p>The main finding for the scenario is that the younger and older 2016 profiles should be compared in context rather than treated as two groups with a single overall score. The broad <code>value</code> field did not show a statistically significant difference when aggregated across the datasets, while the source ordering of response categories did differ. The modelling results also indicate that the demographic <code>category</code> and <code>sub_category</code> structure contributes substantially to how the profiles differ.</p>

<p>For planners and decision-makers, this means the useful differences are found in the composition of demographic characteristics and response categories rather than in a simple claim that one age group has a higher overall value. The analysis can therefore help direct attention to specific Census themes and responses that warrant closer investigation when considering the needs of younger and older Melbourne populations.</p>

</div>

<p><b>Limitations</b></p>

<div class="usecase-body">

<p>The datasets describe Melbourne in 2016, so the findings may not represent current population conditions. The <code>value</code> field combines different kinds of source measures across categories, which limits the meaning of an overall average comparison. The <code>sub_order</code> field records response-category order and is not a direct outcome measure. Some categorical fields are converted into numeric labels for modelling, and the non-linear model metrics are calculated on the same observations used for training, so those model results should be treated as exploratory rather than evidence of performance on unseen data. The <code>census_year</code> field also provides no variation because all observations are from 2016.</p>

</div>

<p><b>Observations for further opportunities</b></p>

<div class="usecase-body">

<p>A useful next step would be to compare individual Census categories and sub-categories directly between the two age profiles rather than relying on an overall mixed <code>value</code> field. Where newer equivalent datasets are available, the same comparison could also be repeated to determine whether the demographic patterns identified in 2016 have changed over time. Future predictive modelling should use suitable encoding for nominal categories and separate training and test data.</p>
