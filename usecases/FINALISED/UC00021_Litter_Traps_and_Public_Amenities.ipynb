<div class="usecase-body">The function below queries the City of Melbourne Open Data API for a given dataset identifier and returns the result as a pandas DataFrame. Passing <code>limit=-1</code> requests every record for the dataset rather than a paginated subset.</div>

```python
def get_dataset(dataset_id):
    """
    Retrieve a full dataset from the City of Melbourne Open Data API as a DataFrame.

    Parameters:
    dataset_id (str): The dataset identifier on the City of Melbourne Open Data Platform.

    Returns:
    pandas.DataFrame: The dataset, or None if the request was unsuccessful.
    """
    base_url = 'https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/'
    export_format = 'csv'

    url = f'{base_url}{dataset_id}/exports/{export_format}'
    params = {
        'select': '*',
        'limit': -1,  # request all records
        'lang': 'en',
        'timezone': 'UTC'
    }

    response = requests.get(url, params=params)

    if response.status_code == 200:
        content = response.content.decode('utf-8')
        df = pd.read_csv(StringIO(content), delimiter=';')
        return df
    else:
        print(f'Request for "{dataset_id}" failed with status code {response.status_code}')
        return None
```

<div id="section-3" class="usecase-section-heading"><b>3. Proximity Calculation Functions</b></div>

<div class="usecase-body">Vectorised functions to find the nearest amenity, and count/list amenities within a radius, for every litter trap at once.</div>

<div class="usecase-body">

A naive way to calculate "the nearest café to each litter trap" is to loop over every litter trap and, for each one, loop over every café and measure the distance. This is an O(N&times;M) operation: with 63 litter traps and a few hundred cafes and BBQs this is manageable, but it does not scale, and it is far slower than it needs to be.

Instead, this notebook uses a <a href="https://scikit-learn.org/stable/modules/generated/sklearn.neighbors.BallTree.html" target="_blank">BallTree</a> with the <code>haversine</code> distance metric. A BallTree is a spatial index &mdash; it organises the amenity coordinates once, and then finds nearest neighbours or counts neighbours within a radius in roughly O(log M) time per query, so all 63 litter traps can be queried against the tree in a single vectorised call rather than a nested loop. The haversine metric also correctly accounts for the curvature of the Earth when converting latitude/longitude degrees into real-world distances, which a simple Euclidean approximation does not.

</div>

```python
EARTH_RADIUS_M = 6_371_000  # mean radius of the Earth, in metres

def build_haversine_tree(coords):
    """
    Build a BallTree (haversine metric) from an array of (lat, lon) coordinates,
    for fast vectorised nearest-neighbour and radius queries.

    Parameters:
    coords (array-like of shape (n, 2)): (latitude, longitude) pairs, in degrees.

    Returns:
    sklearn.neighbors.BallTree: A tree built on the coordinates, in radians.
    """
    coords_rad = np.radians(np.asarray(coords, dtype=float))
    return BallTree(coords_rad, metric='haversine')


def nearest_distance_m(points, tree):
    """
    Find the distance in metres from each point to the nearest point in a BallTree.

    Parameters:
    points (array-like of shape (n, 2)): (latitude, longitude) pairs, in degrees.
    tree (sklearn.neighbors.BallTree): A tree built with build_haversine_tree.

    Returns:
    numpy.ndarray: The distance, in metres, from each point to its nearest neighbour.
    """
    points_rad = np.radians(np.asarray(points, dtype=float))
    distances_rad, _ = tree.query(points_rad, k=1)
    return distances_rad[:, 0] * EARTH_RADIUS_M


def count_within_radius(points, tree, radius_m):
    """
    Count how many points in a BallTree fall within radius_m of each input point.

    Parameters:
    points (array-like of shape (n, 2)): (latitude, longitude) pairs, in degrees.
    tree (sklearn.neighbors.BallTree): A tree built with build_haversine_tree.
    radius_m (float): Search radius, in metres.

    Returns:
    numpy.ndarray: The number of neighbours within radius_m of each input point.
    """
    points_rad = np.radians(np.asarray(points, dtype=float))
    radius_rad = radius_m / EARTH_RADIUS_M
    return tree.query_radius(points_rad, r=radius_rad, count_only=True)


def points_within_radius(points, tree, tree_coords, radius_m):
    """
    List which points in a BallTree fall within radius_m of each input point.

    Parameters:
    points (array-like of shape (n, 2)): (latitude, longitude) pairs, in degrees.
    tree (sklearn.neighbors.BallTree): A tree built with build_haversine_tree.
    tree_coords (array-like of shape (m, 2)): The original (latitude, longitude) coordinates the tree was built from.
    radius_m (float): Search radius, in metres.

    Returns:
    list of lists of tuples: For each input point, the (lat, lon) coordinates of its neighbours within radius_m.
    """
    points_rad = np.radians(np.asarray(points, dtype=float))
    radius_rad = radius_m / EARTH_RADIUS_M
    tree_coords = np.asarray(tree_coords, dtype=float)
    indices = tree.query_radius(points_rad, r=radius_rad)
    return [[tuple(tree_coords[i]) for i in idx] for idx in indices]
```

```python
# Extract coordinates as (lat, lon) arrays
litter_coords = litter_df[['lat', 'lon']].to_numpy()
bbq_coords = np.array([[float(c.split(',')[0]), float(c.split(',')[1])] for c in bbq_df['co_ordinates']])
cafe_coords = cafe_df[['latitude', 'longitude']].to_numpy()
amenity_coords = np.vstack([bbq_coords, cafe_coords])  # BBQs and cafes combined

# Build a BallTree for each set of amenities we need to search against
bbq_tree = build_haversine_tree(bbq_coords)
cafe_tree = build_haversine_tree(cafe_coords)
amenity_tree = build_haversine_tree(amenity_coords)
```

<div id="section-4" class="usecase-section-heading"><b>4. Proximity Analysis</b></div>

<div class="usecase-body">For every litter trap, calculate the distance to the nearest BBQ and café, and count how many amenities are within 200m.</div>

<div class="usecase-body">A 200m radius is used as a walkable proxy for "the area a litter trap is likely to be serving". This threshold is a simplifying assumption for this proof-of-concept analysis, and could be refined using pedestrian catchment data in a future iteration.</div>

```python
radius_m = 200

# Nearest BBQ and nearest cafe distance for every litter trap
litter_df['Nearest BBQ Distance (m)'] = nearest_distance_m(litter_coords, bbq_tree)
litter_df['Nearest Cafe Distance (m)'] = nearest_distance_m(litter_coords, cafe_tree)

# Number of BBQs + cafes within 200m of each litter trap
litter_df['Number of Nearby Points in 200m Radius'] = count_within_radius(litter_coords, amenity_tree, radius_m)

# The actual coordinates of those nearby BBQs/cafes, for use in the map below
litter_df['Nearby Points in 200m Radius'] = points_within_radius(litter_coords, amenity_tree, amenity_coords, radius_m)

litter_df[['asset_number', 'Nearest BBQ Distance (m)', 'Nearest Cafe Distance (m)', 'Number of Nearby Points in 200m Radius']].head()
```

<div id="section-6" class="usecase-section-heading"><b>6. Network Analysis</b></div>

<div class="usecase-body">Model the litter traps as a spatial network to identify which traps are structurally critical to network coverage.</div>

<div id="subsection-6-1" class="usecase-sub-section-heading"><b>6.1. Building the Litter Trap Network</b></div>

<div class="usecase-body">Connect any two litter traps that are within 700m of each other, using the haversine formula for accurate real-world distances.</div>

<div class="usecase-body">A distance of 700m is assumed to represent "the same local network" of coverage. Each pair of traps is only compared once (using <code>itertools.combinations</code>), and distances are calculated with the haversine formula rather than a flat Euclidean approximation on latitude/longitude degrees, so that the distances reflect real-world metres rather than an approximation that distorts with latitude. This threshold is illustrative rather than an official City of Melbourne standard, and should ideally be validated against operational maintenance policies or pedestrian-network distances before being used to guide real asset decisions. The network is plotted on a Folium map (rather than a blank matplotlib canvas) so its structure can be read alongside the actual streets and geography it covers.</div>

```python
def haversine_distance_m(lat1, lon1, lat2, lon2):
    """Great-circle distance in metres between two (lat, lon) points, in degrees."""
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    return 2 * EARTH_RADIUS_M * np.arcsin(np.sqrt(a))


threshold_distance_m = 700

G = nx.Graph()
for _, row in litter_df.iterrows():
    G.add_node(row['asset_number'], pos=(row['lon'], row['lat']))

nodes = list(G.nodes(data=True))
for (id1, data1), (id2, data2) in combinations(nodes, 2):
    lon1, lat1 = data1['pos']
    lon2, lat2 = data2['pos']
    dist = haversine_distance_m(lat1, lon1, lat2, lon2)
    if dist < threshold_distance_m:
        G.add_edge(id1, id2, weight=dist)

# Visualise the network on a real Melbourne basemap (Folium), rather than a blank
# matplotlib canvas, so the network's geography is visible alongside its structure
pos = nx.get_node_attributes(G, 'pos')  # {asset_number: (lon, lat)}
map_center = [litter_df['lat'].mean(), litter_df['lon'].mean()]
network_map = folium.Map(location=map_center, zoom_start=13)

# Draw each edge (a connection between two litter traps within 700m) as a line
for id1, id2 in G.edges():
    lon1, lat1 = pos[id1]
    lon2, lat2 = pos[id2]
    folium.PolyLine(
        locations=[(lat1, lon1), (lat2, lon2)],
        color='#059669',
        weight=1.5,
        opacity=0.6
    ).add_to(network_map)

# Draw each litter trap as a node on top of the edges
for node_id, (lon, lat) in pos.items():
    folium.CircleMarker(
        location=(lat, lon),
        radius=5,
        color='#059669',
        fill=True,
        fill_color='#059669',
        fill_opacity=0.9,
        popup=f'Litter Trap: {node_id}'
    ).add_to(network_map)

print('Network of Litter Traps (edges = within 700m)')
network_map
```

<div id="subsection-6-2" class="usecase-sub-section-heading"><b>6.2. Robustness Analysis &amp; Critical Nodes</b></div>

<div class="usecase-body">Remove each litter trap in turn and check whether the network's largest connected group of traps shrinks as a result.</div>

<div class="usecase-body">A litter trap is treated as "critical" if removing it (e.g. to take it offline for maintenance) reduces the size of the largest connected group of traps &mdash; meaning some traps would end up more than 700m from any other trap without it.</div>

```python
largest_cc = max(nx.connected_components(G), key=len)
original_size = len(largest_cc)

removal_effects = {}
for node in G.nodes:
    H = G.copy()
    H.remove_node(node)
    if H.number_of_nodes() == 0:
        removal_effects[node] = original_size
        continue
    new_largest_cc = max(nx.connected_components(H), key=len)
    removal_effects[node] = original_size - len(new_largest_cc)

# A litter trap is "critical" if removing it shrinks the largest connected group
critical_nodes = {k: v for k, v in removal_effects.items() if v > 0}

litter_df['IsCritical'] = litter_df['asset_number'].isin(critical_nodes)
litter_df['IsCritical'].value_counts()
```

<div class="usecase-body">The map below highlights the critical litter traps identified above in orange, and the non-critical traps in green.</div>

```python
map_center = [litter_df['lat'].mean(), litter_df['lon'].mean()]
critical_map = folium.Map(location=map_center, zoom_start=13)

for _, row in litter_df.iterrows():
    color = '#F97316' if row['IsCritical'] else '#059669'  # Distinct Orange vs Primary Green
    folium.CircleMarker(
        location=[row['lat'], row['lon']],
        radius=5,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.6,
        popup=f"Asset Number: {row['asset_number']}, IsCritical: {row['IsCritical']}"
    ).add_to(critical_map)

critical_map
```

<div id="subsection-6-3" class="usecase-sub-section-heading"><b>6.3. Inspection Frequency: Critical vs. Non-Critical Nodes</b></div>

<div class="usecase-body">Compare how often critical and non-critical litter traps are inspected.</div>

```python
inspection_frequencies = litter_df.groupby('IsCritical')['inspection_frequency'].value_counts().unstack(fill_value=0)
inspection_frequencies_pct = inspection_frequencies.div(inspection_frequencies.sum(axis=1), axis=0) * 100

inspection_frequencies_pct.plot.bar(stacked=True, color=['#059669', '#9333EA', '#F97316', '#2563EB', '#EAB308', '#EC4899'])
plt.ylabel('% of traps')
plt.title('Inspection Frequency by Critical Status')
plt.show()
```

<div class="usecase-body">

<b>Discussion</b>

The network above represents the litter traps within the municipality, connected wherever two traps are within 700m of one another. A few observations stand out:

- The traps are reasonably well distributed, with notably high density in the CBD and riverside areas. This aligns with appropriate asset allocation based on increased pedestrian, commercial, and leisure activity.
- By removing nodes one at a time, we can examine which are critical to the network &mdash; that is, whether their removal would leave nearby traps more than 700m from any other trap.

<div id="section-7" class="usecase-section-heading"><b>7. Correlation Analysis</b></div>

<div class="usecase-body">Check how strongly nearby-amenity density correlates with proximity to BBQs versus proximity to cafes.</div>

```python
selected_columns = [
    'Number of Nearby Points in 200m Radius',
    'Nearest BBQ Distance (m)',
    'Nearest Cafe Distance (m)'
]

# Scale so all three columns are comparable despite different units/ranges
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(litter_df[selected_columns])
scaled_df = pd.DataFrame(scaled_data, columns=selected_columns)

correlation_matrix = scaled_df.corr()

# A diverging colour scale is needed here to show negative vs. positive correlation,
# so this uses a custom map built from the template's Distinct Blue -> white -> Distinct Orange
brand_diverging = mcolors.LinearSegmentedColormap.from_list(
    'brand_diverging', ['#2563EB', '#ffffff', '#F97316']
)

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap=brand_diverging, center=0, linewidths=.5)
plt.title('Correlation Heatmap (Scaled Data)')
plt.show()

# Print the specific coefficients discussed below, so the discussion is tied to this run's actual numbers
r_density_cafe = correlation_matrix.loc['Number of Nearby Points in 200m Radius', 'Nearest Cafe Distance (m)']
r_density_bbq = correlation_matrix.loc['Number of Nearby Points in 200m Radius', 'Nearest BBQ Distance (m)']
print(f"r (nearby-amenity density vs. distance to nearest cafe): {r_density_cafe:.2f}")
print(f"r (nearby-amenity density vs. distance to nearest BBQ):  {r_density_bbq:.2f}")
```

The two coefficients printed above are both negative, which is expected: a shorter distance to the nearest amenity should coincide with a higher count of nearby amenities. The magnitude of <code>r</code> for distance to the nearest café is noticeably larger (more negative) than for distance to the nearest BBQ area &mdash; i.e. amenity density is more tightly tied to café proximity than to BBQ proximity in this run. This is in keeping with the intuitive understanding from the map above: litter traps and cafes both tend to be concentrated in denser, more commercially active areas of the city. Because this dataset is refreshed from the live API, the exact <code>r</code> values above will vary slightly each time this notebook is run &mdash; read the printed numbers alongside this discussion rather than the discussion alone.

<div id="section-8" class="usecase-section-heading"><b>8. Clustering Analysis</b></div>

<div class="usecase-body">Use K-Means to check whether litter traps naturally group by density, inspection frequency, construction material, and trap type.</div>

```python
cat_columns = ['object_type_lupvalue', 'construct_material_lupvalue', 'inspection_frequency']
num_columns = ['Number of Nearby Points in 200m Radius']

# One-hot encode the categorical columns, and combine with the numeric column
cluster_data = pd.get_dummies(litter_df[cat_columns])
cluster_data[num_columns] = litter_df[num_columns]

num_clusters = 3
kmeans = KMeans(n_clusters=num_clusters, random_state=42, n_init=10)
litter_df['Cluster'] = kmeans.fit_predict(cluster_data)

litter_df['Cluster'].value_counts()
```

```python
litter_df['inspection_frequency_code'] = litter_df['inspection_frequency'].astype('category').cat.codes
litter_df['object_type_code'] = litter_df['object_type_lupvalue'].astype('category').cat.codes

# Template palette, one distinct colour per cluster (num_clusters == 3)
cluster_palette = ['#059669', '#9333EA', '#F97316']
cluster_cmap = mcolors.ListedColormap(cluster_palette[:num_clusters])

plt.figure(figsize=(8, 6))
scatter = plt.scatter(
    litter_df['inspection_frequency_code'],
    litter_df['object_type_code'],
    c=litter_df['Cluster'],
    cmap=cluster_cmap
)
plt.xticks(litter_df['inspection_frequency_code'].unique(), litter_df['inspection_frequency'].unique(), rotation='vertical')
plt.yticks(litter_df['object_type_code'].unique(), litter_df['object_type_lupvalue'].unique())
plt.xlabel('Inspection Frequency')
plt.ylabel('Object Type')
plt.title('Litter Traps by Cluster')
plt.colorbar(scatter, label='Cluster')
plt.tight_layout()
plt.show()
```

<div class="usecase-body">

<b>Discussion</b>

The clustering does not show a strong visible pattern against inspection frequency and object type in this scatter plot &mdash; the points appear largely scattered rather than forming clean groups. This suggests these two categorical features alone are not the main driver of the clusters K-Means is finding (which also include nearby-amenity density and construction material). It's worth being clear that K-Means is being used here as an <i>exploratory</i> technique to see whether a grouping structure is visible &mdash; fixing <code>num_clusters = 3</code> in advance does not itself demonstrate that three natural groups exist in the data, and a further step (e.g. an elbow plot or silhouette score across different values of <code>k</code>) would be needed to support that claim. Further exploration &mdash; for example, plotting the clusters against amenity density directly &mdash; would be a reasonable next step.

</div>

<div id="section-9" class="usecase-section-heading"><b>9. Predictive Modelling: Classifying Critical Litter Traps</b></div>

<div class="usecase-body">A proof-of-concept classification model that predicts whether a litter trap is network-critical from its proximity features.</div>

<div class="usecase-body">

This section trains a logistic regression classifier to predict the <code>IsCritical</code> label calculated in the network analysis above, using each litter trap's proximity features as inputs. This is a meaningful prediction target: whether a trap is critical to network coverage is a real planning question, and a model like this could, with more data, help flag likely-critical traps for new assets before a full network analysis is run.

<i>Note: with only 63 litter traps in this dataset, this is a small-sample proof of concept rather than a production-ready model &mdash; the results below should be read as illustrative, not as a finished recommendation.</i>

</div>

```python
features = ['Number of Nearby Points in 200m Radius', 'Nearest BBQ Distance (m)', 'Nearest Cafe Distance (m)']
target = 'IsCritical'

X = litter_df[features]
y = litter_df[target].astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print(f'Accuracy: {accuracy_score(y_test, y_pred):.2f}')
print()
print('Classification report:')
print(classification_report(y_test, y_pred, zero_division=0))
print('Confusion matrix (rows = actual, cols = predicted):')
print(confusion_matrix(y_test, y_pred))
```

<div class="usecase-body">To demonstrate how the trained model would be used, the cell below scores three example litter traps with different levels of nearby amenity density, rather than requiring interactive input (which would block a static export of this notebook).</div>

```python
example_traps = pd.DataFrame({
    'Number of Nearby Points in 200m Radius': [2, 30, 120],
    'Nearest BBQ Distance (m)': [500, 250, 80],
    'Nearest Cafe Distance (m)': [400, 150, 40]
})

example_traps['Predicted IsCritical'] = model.predict(example_traps[features])
example_traps['P(Critical)'] = model.predict_proba(example_traps[features])[:, 1].round(2)

example_traps
```

<div class="usecase-body">

<b>Discussion</b>

The model performs reasonably well at separating critical from non-critical traps using only proximity features, which supports the earlier correlation finding that trap placement is linked to nearby amenity density. Given the small dataset size and class imbalance (critical traps outnumber non-critical ones), these results are a proof of concept rather than a deployable model &mdash; a useful extension would be to retrain this model periodically as new litter traps, cafes, or BBQ areas are added to the underlying datasets. The train/test split and K-Means clustering above both use a fixed <code>random_state</code>, so re-running this notebook against the <i>same</i> pulled data will reproduce the same accuracy and cluster assignments; however, because the underlying datasets are refreshed live from the API, the actual accuracy figure will still drift over time as new litter traps, cafes, and BBQs are added &mdash; a fixed seed controls randomness in the split, not variation in the input data itself.
