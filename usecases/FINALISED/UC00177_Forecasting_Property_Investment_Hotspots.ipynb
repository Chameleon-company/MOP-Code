<div id="section-1" class="usecase-section-heading"><b>1. Importing Libraries and Setting Up the Environment</b></div>

<div class="usecase-body">The setup cell imports Requests and StringIO for data retrieval; Pandas and NumPy for data preparation; Matplotlib and Seaborn for visualisation; and scikit-learn for standardisation, KMeans clustering, and silhouette assessment. IPython supports the display of HTML and results. The cell also defines the chart colours, property type, comparison periods, and clustering settings. Executing this cell before downloading the datasets establishes the environment required for the subsequent analysis.</div>

```python
# Data retrieval and supporting utilities
from io import StringIO
from datetime import datetime, timezone
import requests

# Data preparation and numerical calculations
import pandas as pd
import numpy as np

# Data visualisation
import matplotlib.pyplot as plt
import seaborn as sns

# Clustering, feature standardisation, and cluster assessment
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

# Notebook styling and table display
from IPython.display import HTML, display

# Apply the use case template's chart colour palette
PALETTE = [
    '#059669',  # Primary green
    '#9333EA',  # Distinct purple
    '#F97316',  # Distinct orange
    '#2563EB',  # Distinct blue
    '#EAB308',  # Distinct yellow
    '#EC4899',  # Distinct pink
]

sns.set_theme(style='whitegrid', palette=PALETTE)

plt.rcParams.update({
    'figure.dpi': 110,
    'axes.spines.top': False,
    'axes.spines.right': False,
})

pd.set_option('display.max_columns', 12)

# Define the property type and comparison periods
PROPERTY_TYPE = 'Residential Apartment'
PRICE_START_YEAR, PRICE_END_YEAR = 2000, 2016
BASELINE_YEAR, FUTURE_YEAR = 2023, 2043

# Keep clustering results reproducible
RANDOM_STATE = 42
N_INIT = 20
```

<div id="section-2" class="usecase-section-heading"><b>2. Downloading Datasets</b></div>

<div class="usecase-body">The <code>collect_data</code> function downloads CSV data from the City of Melbourne Open Data Platform, checks the HTTP response, and loads each dataset into a DataFrame. After both downloads, the record counts, retrieval time, and first few rows provide an initial check of the retrieved data.</div>

```python
def collect_data(dataset_id):
    url = f'https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/{dataset_id}/exports/csv'
    response = requests.get(url, params={'select': '*', 'limit': -1, 'lang': 'en', 'timezone': 'UTC'}, timeout=60)
    response.raise_for_status()
    data = pd.read_csv(StringIO(response.content.decode('utf-8-sig')), sep=';')
    if data.empty:
        raise ValueError(f'No records returned for {dataset_id}.')
    return data

prices_raw = collect_data('house-prices-by-small-area-sale-year')
pop_raw = collect_data('city-of-melbourne-population-forecasts-by-small-area-2020-2040')
print('Retrieved (UTC):', datetime.now(timezone.utc).strftime('%Y-%m-%d %H:%M'))
for label, frame in [('House prices', prices_raw), ('Population forecasts', pop_raw)]:
    print(f'{label}: {len(frame):,} records')
    display(frame.head(3))
```

<div id="section-3" class="usecase-section-heading"><b>3. Data Cleaning and Preparation</b></div>

<div class="usecase-body">Data preparation involves standardising column names, inspecting missing values, and converting fields to suitable data types. Invalid prices are excluded from growth calculations without filling gaps across property types or years. Population records are restricted to the small areas shared by both datasets to support comparisons between matching locations.</div>

<div id="subsection-3-1" class="usecase-sub-section-heading"><b>3.1. Standardising names and checking missing values</b></div>

<div class="usecase-body">Consistent column names improve readability and simplify subsequent operations. For example, <code>sale_year</code> becomes <code>Sale_Year</code>. Renaming the population field <code>Geography</code> to <code>Small_Area</code> provides a common field for matching areas across the datasets. A missing-value summary then identifies gaps in each field.</div>

```python
prices, pop = prices_raw.copy(), pop_raw.copy()
for frame in (prices, pop):
    frame.columns = ['_'.join(word.capitalize() for word in col.split('_')) for col in frame.columns]
pop = pop.rename(columns={'Geography': 'Small_Area'})
required_prices = {'Small_Area', 'Sale_Year', 'Type', 'Median_Price', 'Transaction_Count'}
required_pop = {'Small_Area', 'Year', 'Gender', 'Age', 'Value'}
if not required_prices.issubset(prices.columns) or not required_pop.issubset(pop.columns):
    raise ValueError('The source schema has changed; review the required fields.')
display(pd.concat({'House prices': prices.isna().sum(), 'Population': pop.isna().sum()}, axis=1).fillna('—'))
```

<div id="subsection-3-2" class="usecase-sub-section-heading"><b>3.2. Preparing valid observations</b></div>

<div class="usecase-body">Preparing valid observations begins with removing exact duplicate records. Rows with missing area names, invalid years, or non-positive median prices are excluded from price growth calculations, and missing prices are not filled. Conflicting records for the same area, year, and property type trigger an error. Population age and gender categories are checked before aggregation and must represent disjoint groups to avoid double-counting.</div>

```python
initial_price_rows = len(prices)
prices = prices.drop_duplicates().copy()
for col in ['Small_Area', 'Type']:
    prices[col] = prices[col].astype('string').str.strip().replace('', pd.NA)
for col in ['Sale_Year', 'Median_Price', 'Transaction_Count']:
    prices[col] = pd.to_numeric(prices[col], errors='coerce')
prices = prices.dropna(subset=['Small_Area', 'Type', 'Sale_Year']).copy()
prices = prices.loc[prices['Sale_Year'].mod(1).eq(0)].copy()
prices['Sale_Year'] = prices['Sale_Year'].astype(int)
# Retain valid transaction counts for EDA, even when a price is unavailable.
prices.loc[prices['Transaction_Count'] < 0, 'Transaction_Count'] = np.nan
prices.loc[prices['Median_Price'] <= 0, 'Median_Price'] = np.nan
if prices.duplicated(['Small_Area', 'Sale_Year', 'Type']).any():
    raise ValueError('Conflicting price records require investigation.')
valid_prices = prices.dropna(subset=['Median_Price']).copy()

pop = pop.drop_duplicates().copy()
pop['Small_Area'] = pop['Small_Area'].astype('string').str.strip().replace('', pd.NA)
for col in ['Year', 'Value']:
    pop[col] = pd.to_numeric(pop[col], errors='coerce')
if pop[list(required_pop)].isna().any().any() or (pop['Value'] < 0).any() or pop['Year'].mod(1).ne(0).any():
    raise ValueError('Incomplete or invalid population records require investigation.')
pop['Year'] = pop['Year'].astype(int)
if pop.duplicated(['Small_Area', 'Year', 'Gender', 'Age']).any():
    raise ValueError('Conflicting population records require investigation.')
if set(pop['Gender']) != {'Female', 'Male'} or not pop['Age'].str.match(r'^Age \d+(?:-\d+|\+)$').all():
    raise ValueError('Review age/gender categories before summing: totals may overlap.')
common_areas = sorted(set(prices['Small_Area']) & set(pop['Small_Area']))
if not common_areas:
    raise ValueError('No matching small areas found.')
prices = prices.loc[prices['Small_Area'].isin(common_areas)].copy()
valid_prices = valid_prices.loc[valid_prices['Small_Area'].isin(common_areas)].copy()
pop = pop.loc[pop['Small_Area'].isin(common_areas)].copy()
print(f'Original price rows: {initial_price_rows}; valid price observations for shared areas: {len(valid_prices)}')
print(f'Price rows excluded from growth analysis within shared areas: {prices.Median_Price.isna().sum()}')
print(f'Shared small areas ({len(common_areas)}):', ', '.join(common_areas))
print(f'Price years: {prices.Sale_Year.min()}–{prices.Sale_Year.max()}')
print(f'Population years: {pop.Year.min()}–{pop.Year.max()}')
```

<div id="section-4" class="usecase-section-heading"><b>4. Exploratory Data Analysis</b></div>

<div class="usecase-body">Summary statistics describe valid median prices by area, year, and property type. The histogram illustrates the distribution of these aggregated median prices rather than individual dwelling prices. Transaction totals provide context on recorded sales activity, although these counts alone do not measure liquidity or future returns.</div>

```python
display(valid_prices[['Sale_Year', 'Median_Price', 'Transaction_Count']].describe().round(2))
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
property_types = sorted(valid_prices['Type'].unique())
type_colours = dict(zip(property_types, PALETTE))
sns.histplot(data=valid_prices, x='Median_Price', hue='Type', hue_order=property_types, bins=30, multiple='stack', palette=type_colours, alpha=1, ax=axes[0])
axes[0].set(title='Distribution of recorded median prices', xlabel='Median price (AUD)', ylabel='Observations')
transactions = prices.groupby('Small_Area')['Transaction_Count'].sum(min_count=1).sort_values()
transactions_by_type = prices.groupby(['Small_Area', 'Type'])['Transaction_Count'].sum(min_count=1).unstack('Type').reindex(index=transactions.index, columns=property_types).fillna(0)
transactions_by_type.plot.barh(stacked=True, ax=axes[1], color=[type_colours[t] for t in property_types])
axes[1].legend(title='Property type', fontsize=8)
axes[1].set(title='Recorded transactions by small area', xlabel='Transactions across available years', ylabel='Small area')
plt.tight_layout()
plt.show()
```

<div id="subsection-4-1" class="usecase-sub-section-heading"><b>4.1. Reading the exploratory results</b></div>

<div class="usecase-body">Unusually large median prices warrant closer examination, particularly where transaction counts are low. Differences may reflect property type, sales composition, or unusual transactions. Separate growth calculations for the two property types maintain consistency in the subsequent comparisons.</div>

<div id="section-5" class="usecase-section-heading"><b>5. Population Growth Analysis</b></div>

<div class="usecase-body">Summing population counts across disjoint age and gender categories produces totals for each area and year. Projected growth compares the specified baseline and future years, using the same interval as the later annualised growth calculation. Reporting missing endpoint coverage identifies areas without the observations required for comparison.</div>

```python
if FUTURE_YEAR <= BASELINE_YEAR:
    raise ValueError('The future year must follow the baseline year.')
# Check that every area/year contains the same set of demographic categories.
expected_categories = set(zip(pop['Gender'], pop['Age']))
for key, group in pop.groupby(['Small_Area', 'Year']):
    if set(zip(group['Gender'], group['Age'])) != expected_categories:
        raise ValueError(f'Incomplete demographic coverage for {key}.')
pop_agg = pop.groupby(['Small_Area', 'Year'])['Value'].sum().unstack('Year')
if not {BASELINE_YEAR, FUTURE_YEAR}.issubset(pop_agg.columns):
    raise ValueError('Requested population years are unavailable; review the source coverage.')
pop_endpoints = pop_agg[[BASELINE_YEAR, FUTURE_YEAR]].dropna()
pop_endpoints = pop_endpoints.loc[pop_endpoints[BASELINE_YEAR] > 0]
print('Areas without valid population endpoints:', sorted(set(common_areas) - set(pop_endpoints.index)))
pop_growth_df = pd.DataFrame(index=pop_endpoints.index)
pop_growth_df['Baseline_Population'] = pop_endpoints[BASELINE_YEAR]
pop_growth_df['Future_Population'] = pop_endpoints[FUTURE_YEAR]
ratio = pop_endpoints[FUTURE_YEAR] / pop_endpoints[BASELINE_YEAR]
pop_growth_df['Pop_Growth_Pct'] = (ratio - 1) * 100
pop_growth_df['Population_CAGR'] = ratio ** (1 / (FUTURE_YEAR - BASELINE_YEAR)) - 1
display(pop_growth_df.sort_values('Pop_Growth_Pct', ascending=False).round(3))
leader = pop_growth_df['Pop_Growth_Pct'].idxmax()
print(f'Largest projected population increase: {leader} ({pop_growth_df.loc[leader, "Pop_Growth_Pct"]:.1f}%, {BASELINE_YEAR}–{FUTURE_YEAR}).')
```

<div id="subsection-5-1" class="usecase-sub-section-heading"><b>5.1. Interpreting population growth</b></div>

<div class="usecase-body">These figures represent projected changes rather than observed future outcomes. Additional residents may increase housing demand, but comparisons between areas also require consideration of housing supply, household size, affordability, and local policy.</div>

<div id="section-6" class="usecase-section-heading"><b>6. Historical Property Price Growth</b></div>

<div class="usecase-body">CAGR is calculated separately for each small area and property type over the same 2000–2016 interval. Both endpoint prices must be observed and positive. The ranking uses Residential Apartment observations because this type has broader area coverage in the supplied dataset. Houses/townhouses remain in the comparison table. Analysing that property type in the ranking requires changing <code>PROPERTY_TYPE</code> in the setup cell and rerunning the notebook.</div>

<div class="usecase-body">CAGR = (end price ÷ start price)<sup>1 ÷ number of years</sup> − 1. This formula summarises the change between endpoint prices. CAGR alone does not establish steady growth, low volatility, or investment safety.</div>

```python
if PRICE_END_YEAR <= PRICE_START_YEAR:
    raise ValueError('The ending price year must follow the starting year.')
price_panel = valid_prices.pivot(index=['Small_Area', 'Type'], columns='Sale_Year', values='Median_Price')
if not {PRICE_START_YEAR, PRICE_END_YEAR}.issubset(price_panel.columns):
    raise ValueError('Requested price years are unavailable.')
endpoints = price_panel[[PRICE_START_YEAR, PRICE_END_YEAR]].dropna().copy()
price_cagr_all = pd.DataFrame(index=endpoints.index)
price_cagr_all['Start_Price'] = endpoints[PRICE_START_YEAR]
price_cagr_all['End_Price'] = endpoints[PRICE_END_YEAR]
price_cagr_all['Price_CAGR'] = (endpoints[PRICE_END_YEAR] / endpoints[PRICE_START_YEAR]) ** (1 / (PRICE_END_YEAR - PRICE_START_YEAR)) - 1
display(price_cagr_all.assign(Price_CAGR_Pct=price_cagr_all['Price_CAGR'] * 100).drop(columns='Price_CAGR').round(3))
price_cagr_df = price_cagr_all.xs(PROPERTY_TYPE, level='Type').copy()
print('Ranking property type:', PROPERTY_TYPE)
print('Areas excluded because selected-type endpoints are unavailable:', sorted(set(common_areas) - set(price_cagr_df.index)))
```

<div id="subsection-6-1" class="usecase-sub-section-heading"><b>6.1. Method correction</b></div>

<div class="usecase-body">Selecting explicit area, property type, and year endpoints ensures comparisons between like properties. Sorting only by year and selecting the first and last rows can mix property types because several types can share a year, making the result depend on row order. Identifying both endpoints directly removes this ambiguity. Missing or zero prices are not replaced with a suburb-wide median.</div>

<div id="section-7" class="usecase-section-heading"><b>7. Investment Index and Visualisation</b></div>

<div class="usecase-body">The exploratory investment index multiplies historical price CAGR by total projected population growth expressed as a fraction. The result is a dimensionless screening indicator rather than an expected return. Plotting the two growth components separately shows how each contributes to the comparison.</div>

```python
investment_df = price_cagr_df.join(pop_growth_df, how='inner').reset_index()
if investment_df.empty:
    raise ValueError('No areas have valid price and population endpoints.')
investment_df['Investment_Index'] = investment_df['Price_CAGR'] * investment_df['Pop_Growth_Pct'] / 100
index_table = investment_df[['Small_Area', 'Price_CAGR', 'Pop_Growth_Pct', 'Investment_Index']].copy()
index_table['Price_CAGR'] *= 100
index_table = index_table.rename(columns={'Price_CAGR': 'Price_CAGR_Pct'})
display(index_table.sort_values('Investment_Index', ascending=False).round(4))
fig, ax = plt.subplots(figsize=(10, 7))
ax.scatter(investment_df['Price_CAGR'] * 100, investment_df['Pop_Growth_Pct'], color=PALETTE[0])
for row in investment_df.itertuples():
    ax.annotate(row.Small_Area, (row.Price_CAGR * 100, row.Pop_Growth_Pct), xytext=(5, 5), textcoords='offset points', fontsize=8)
ax.margins(x=0.22, y=0.15)
ax.set(xlabel=f'Historical price CAGR (%) — {PROPERTY_TYPE}', ylabel=f'Projected population growth (%) — {BASELINE_YEAR}–{FUTURE_YEAR}', title='Historical price growth and projected population growth')
plt.tight_layout()
plt.show()
print(f'Descriptive Pearson correlation across included areas: {investment_df["Price_CAGR"].corr(investment_df["Pop_Growth_Pct"]):.3f}')
```

<div id="subsection-7-1" class="usecase-sub-section-heading"><b>7.1. Interpreting the comparison</b></div>

<div class="usecase-body">The correlation provides a descriptive comparison across a small number of areas and different time periods; it does not establish causation. The individual growth measures also require inspection because multiplying two negative inputs can produce a positive index and give a misleading impression.</div>

<div id="section-8" class="usecase-section-heading"><b>8. Combined Investment Score</b></div>

<div class="usecase-body">A second screening score combines historical price growth with an illustrative population-related uplift. Explicit assumptions explain the calculation and support interpretation of the score as a screening measure rather than a forecast of investment returns.</div>

<div id="subsection-8-1" class="usecase-sub-section-heading"><b>8.1. Population–House Price Elasticity</b></div>

<div class="usecase-body">The multiplier of 2.81 serves as a scenario assumption rather than an estimate of a suburb-level relationship from these datasets. Domain’s research examines a long-run relationship. Applying a single multiplier to annualised suburb population growth is a simplifying assumption that limits interpretation of the score.</div>

<div class="usecase-body">Source context: <a href="https://www.domain.com.au/research/the-influence-of-mortgage-rates-on-property-prices-1199988/">Domain Research, The influence of mortgage rates on property prices (9 March 2023)</a>; also discussed by <a href="https://shorefinancial.com.au/how-australias-fast-population-growth-is-affecting-property-prices-and-rents/">Shore Financial (23 January 2024)</a>. These sources do not validate this notebook’s suburb-level score or its 60/40 weighting.</div>

<div id="subsection-8-2" class="usecase-sub-section-heading"><b>8.2. Calculating and Ranking the Investment Score</b></div>

<div class="usecase-body">The population scenario component equals annualised population growth × 2.81 × 100. The combined score assigns weights of 60% to historical price CAGR in percentage terms and 40% to the population component. These illustrative weights support comparisons between areas. The resulting score is expressed in points rather than as a predicted percentage return.</div>

```python
VIC_ELASTICITY = 2.81
PRICE_WEIGHT = 0.60
investment_df['Pop_Driven_Price_Uplift'] = investment_df['Population_CAGR'] * VIC_ELASTICITY * 100
investment_df['Investment_Score'] = PRICE_WEIGHT * investment_df['Price_CAGR'] * 100 + (1 - PRICE_WEIGHT) * investment_df['Pop_Driven_Price_Uplift']
top_suburbs = investment_df.sort_values(['Investment_Score', 'Small_Area'], ascending=[False, True]).copy()
display(top_suburbs[['Small_Area', 'Investment_Score', 'Pop_Driven_Price_Uplift']].round(3))
fig, ax = plt.subplots(figsize=(10, 6))
positions = np.arange(len(top_suburbs))
price_component = PRICE_WEIGHT * top_suburbs['Price_CAGR'] * 100
population_component = (1 - PRICE_WEIGHT) * top_suburbs['Pop_Driven_Price_Uplift']
ax.barh(positions - 0.18, price_component, height=0.34, color=PALETTE[0], label='Historical price contribution')
ax.barh(positions + 0.18, population_component, height=0.34, color=PALETTE[1], label='Population scenario contribution')
ax.scatter(top_suburbs['Investment_Score'], positions, color=PALETTE[2], marker='D', s=40, label='Combined score', zorder=3)
ax.set_yticks(positions, labels=top_suburbs['Small_Area'])
ax.legend(loc='lower right', fontsize=8)
ax.invert_yaxis()
ax.set(title=f'Illustrative combined score — {PROPERTY_TYPE}', xlabel='Investment score (points)', ylabel='Small area')
plt.tight_layout()
plt.show()
print('Highest scores in this run:', ', '.join(top_suburbs.head(3)['Small_Area']))
```

<div id="sub-subsection-8-2-1" class="usecase-sub-sub-section-heading"><b>8.2.1. Score sensitivity</b></div>

<div class="usecase-body">Varying the historical-price weight while holding the population multiplier constant reveals how the rankings respond to this modelling choice. Changes in an area’s ranking indicate sensitivity to the chosen weight. This comparison does not identify a uniquely correct weight.</div>

```python
sensitivity = investment_df[['Small_Area']].copy()
for weight in [0.4, 0.6, 0.8]:
    scores = weight * investment_df['Price_CAGR'] * 100 + (1 - weight) * investment_df['Pop_Driven_Price_Uplift']
    sensitivity[f'Rank_at_{int(weight * 100)}pct_price_weight'] = scores.rank(ascending=False, method='min').astype(int)
display(sensitivity.sort_values('Rank_at_60pct_price_weight'))
```

<div id="section-9" class="usecase-section-heading"><b>9. Clustering Suburbs by Investment Potential</b></div>

<div class="usecase-body">KMeans groups areas by historical price CAGR and annualised population growth. Standardising both features before fitting the model prevents their different scales from determining the grouping. The resulting clusters describe relative growth patterns rather than validated investment risk categories.</div>

<div id="subsection-9-1" class="usecase-sub-section-heading"><b>9.1. Assessing the Number of Clusters</b></div>

<div class="usecase-body">Candidate cluster counts from 2 to at most 6 are compared using an elbow curve and average silhouette scores. The elbow curve shows within-cluster inertia, while silhouette scores describe cohesion and separation. The candidate with the highest average silhouette score is selected, with fewer clusters preferred if scores tie. The smallest cluster size remains visible to support interpretation, but single-area clusters are not excluded. The same selection rule applies to every candidate.</div>

```python
features = investment_df[['Price_CAGR', 'Population_CAGR']]
if len(features) < 4 or not np.isfinite(features.to_numpy()).all():
    raise ValueError('Clustering requires at least four areas with finite growth measures.')
scaler = StandardScaler()
data_scaled = scaler.fit_transform(features)
max_k = min(6, len(features) - 1, len(np.unique(data_scaled, axis=0)) - 1)
if max_k < 2:
    raise ValueError('Insufficient distinct feature values for cluster assessment.')
records = []
for k in range(2, max_k + 1):
    model = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=N_INIT)
    labels = model.fit_predict(data_scaled)
    counts = np.bincount(labels)
    records.append({'Clusters': k, 'Inertia': model.inertia_, 'Silhouette': silhouette_score(data_scaled, labels), 'Smallest_Cluster': counts.min()})
cluster_checks = pd.DataFrame(records)
display(cluster_checks.round(4))
# Apply the same silhouette-based selection rule to every candidate.
best_k = int(cluster_checks.sort_values(['Silhouette', 'Clusters'], ascending=[False, True]).iloc[0]['Clusters'])
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
for ax, metric, title, colour in zip(axes, ['Inertia', 'Silhouette'], ['Elbow check', 'Silhouette check'], PALETTE[:2]):
    ax.plot(cluster_checks['Clusters'], cluster_checks[metric], marker='o', color=colour)
    ax.axvline(best_k, linestyle='--', color=PALETTE[2], label=f'Selected k = {best_k}')
    ax.set(xlabel='Number of clusters', ylabel=metric, title=title, xticks=cluster_checks['Clusters'])
    ax.legend()
plt.tight_layout()
plt.show()
print(f'Selected {best_k} clusters using the highest average silhouette score among the tested candidates.')
print('Three clusters retained.' if best_k == 3 else 'Three clusters were not selected; the following analysis uses the assessed count.')
```

<div id="sub-subsection-9-1-1" class="usecase-sub-sub-section-heading"><b>9.1.1. Interpreting the selection</b></div>

<div class="usecase-body">Average silhouette scores are interpreted alongside the elbow curve and cluster sizes. The highest-scoring candidate is the best of the tested solutions under this criterion; it does not establish a uniquely correct grouping. A cluster containing one area may highlight an unusual growth profile. Its measured values and source observations therefore require inspection before assigning meaning to the grouping. The small number of areas calls for cautious interpretation.</div>

<div id="subsection-9-2" class="usecase-sub-section-heading"><b>9.2. Applying and Visualising KMeans Clustering</b></div>

<div class="usecase-body">KMeans is fitted using the selected cluster count and the same initialisation settings as the assessment. The chart presents the growth measures in percentage terms and distinguishes each cluster using the template colours, allowing visual comparison between groups.</div>

```python
kmeans = KMeans(n_clusters=best_k, random_state=RANDOM_STATE, n_init=N_INIT)
investment_df['Cluster'] = kmeans.fit_predict(data_scaled)
fig, ax = plt.subplots(figsize=(11, 7))
for cluster, group in investment_df.groupby('Cluster'):
    ax.scatter(group['Price_CAGR'] * 100, group['Population_CAGR'] * 100, color=PALETTE[int(cluster)], label=f'Cluster {cluster}', marker=['o', 's', '^', 'D', 'P', 'X'][int(cluster)], s=65)
    for row in group.itertuples():
        ax.annotate(row.Small_Area, (row.Price_CAGR * 100, row.Population_CAGR * 100), xytext=(5, 5), textcoords='offset points', fontsize=8)
ax.margins(x=0.22, y=0.15)
ax.set(xlabel='Historical price CAGR (%)', ylabel='Projected population CAGR (%)', title=f'Growth patterns across small areas — {best_k} clusters')
ax.legend(title='Cluster')
plt.tight_layout()
plt.show()
```

<div id="subsection-9-3" class="usecase-sub-section-heading"><b>9.3. Cluster Interpretation</b></div>

<div class="usecase-body">The cluster summary describes current memberships and compares each group’s growth measures with the overall means. Cluster numbers are arbitrary identifiers, so interpretation depends on the current profiles rather than labels from earlier runs. For any single-area cluster, the endpoint prices, population counts, and growth rates provide additional detail for inspection. Isolation alone does not establish a meaningful investment category.</div>

```python
cluster_summary = investment_df.groupby('Cluster').agg(Areas=('Small_Area', 'size'), Mean_Price_CAGR=('Price_CAGR', 'mean'), Mean_Population_CAGR=('Population_CAGR', 'mean'))
cluster_summary[['Mean_Price_CAGR', 'Mean_Population_CAGR']] *= 100
display(cluster_summary.rename(columns={'Mean_Price_CAGR': 'Mean_Price_CAGR_Pct', 'Mean_Population_CAGR': 'Mean_Population_CAGR_Pct'}).round(3))
mean_price = investment_df['Price_CAGR'].mean()
mean_pop = investment_df['Population_CAGR'].mean()
for cluster, group in investment_df.groupby('Cluster'):
    price_label = 'higher' if group['Price_CAGR'].mean() >= mean_price else 'lower'
    pop_label = 'higher' if group['Population_CAGR'].mean() >= mean_pop else 'lower'
    print(f'Cluster {cluster}: {price_label} historical price growth and {pop_label} projected population growth relative to the overall means.')
    print('Areas:', ', '.join(sorted(group['Small_Area'])))

# Inspect any single-area clusters using the underlying observations.
cluster_sizes = investment_df.groupby('Cluster')['Small_Area'].transform('size')
singleton_areas = investment_df.loc[cluster_sizes.eq(1)].copy()
if not singleton_areas.empty:
    print('Single-area clusters: inspect these observations before interpreting their significance.')
    singleton_review = singleton_areas[['Small_Area', 'Cluster', 'Start_Price', 'End_Price', 'Price_CAGR', 'Population_CAGR']].copy()
    singleton_review['Price_CAGR'] *= 100
    singleton_review['Population_CAGR'] *= 100
    singleton_review = singleton_review.rename(columns={'Price_CAGR': 'Price_CAGR_Pct', 'Population_CAGR': 'Population_CAGR_Pct'})
    singleton_review = singleton_review.join(pop_agg[[BASELINE_YEAR, FUTURE_YEAR]], on='Small_Area')
    display(singleton_review.round(3))
    print('A single-area cluster is a profile to investigate, not evidence of investment quality or a stable group.')
```
