<div id="Dataset-Imported-through-API" class="usecase-sub-section-heading"><b>1.1 Dataset Imported through API</b></div>

```python
import requests
import pandas as pd

def fetch_data(base_url, dataset, num_records=99, offset=0):
    all_records = []
    max_offset = 9900  # Maximum number of requests

    while True:
        # maximum limit check
        if offset > max_offset:
            break

        # Create API request URL
        filters = f'{dataset}/records?limit={num_records}&offset={offset}'
        url = f'{base_url}{filters}'

        # Start request
        try:
            result = requests.get(url, timeout=10)
            result.raise_for_status()
            records = result.json().get('results')
        except requests.exceptions.RequestException as e:
            raise Exception(f"API request failed: {e}")
        if records is None:
            break
        all_records.extend(records)
        if len(records) < num_records:
            break

        # next cycle offset
        offset += num_records

    # DataFrame all data
    df = pd.DataFrame(all_records)
    return df    
```

<div id="Optimising-Data-Types" class="usecase-sub-section-heading"><b>2.2 Optimising Data Types for Analysis</b></div>

<div class="usecase-body">
To ensure our dataset is primed for analysis, we undertake a comprehensive data type optimisation process. This involves adjusting the data types of various columns to more accurately reflect their content and optimise memory usage. Here's an overview of the steps taken:
</div>
<div class="usecase-list">

- **Timestamp Conversion**: Convert the `time` and `last_update` columns to `datetime` objects to facilitate time series analysis.
- **String Specification**: Change `serial`, `sensor_name`, `description`, and `address` columns to the `string` data type for consistency in text data handling.
- **Category Optimisation**: For columns like `bin_status` and `product_type`, we use the `category` data type when we have a limited set of possible values, enhancing memory efficiency and performance during grouping operations.
- **Numerical Adjustments**: Ensure numerical data, such as `wastebasket_size`, is correctly typed (`float`) for accurate mathematical computations.
- **Geolocation Handling**: Split the `Latlong` column into separate `latitude` and `longitude` columns and convert them to `float` data types, dropping the original `Latitude and Longitude` column to simplify geographical analysis.

These adjustments are crucial for the following reasons:
- **Efficiency**: Optimising data types can significantly reduce memory usage, making data manipulation more efficient.
- **Accuracy**: Ensuring data types match the nature of the data prevents errors during analysis.
- **Convenience**: Having the right data type (e.g., datetime) simplifies operations like date range filtering or time-based aggregation.
</div>

<div class="usecase-body"> By refining our dataset's structure through these changes, we lay a solid foundation for robust and error-free data analysis. </div>


```python
# Let's change the datatypes to more appropriate types:
data['time'] = pd.to_datetime(data['time'])
data['serial'] = data['serial'].astype('string')
data['sensor_name'] = data['sensor_name'].astype('string')
# 'status_current_fill_level' is already an integer, no change needed.
data['bin_status'] = data['bin_status'].astype('category')  # Or 'string' if it's not a finite set of statuses.
data['last_update'] = pd.to_datetime(data['last_update'])
data['product_type'] = data['product_type'].astype('category')  # Or 'string' if it's not a finite set of types.
data['description'] = data['description'].astype('string')
data['address'] = data['address'].astype('string')
# Assuming 'wastebasket_size' refers to a numerical measurement and not a category.
data['wastebasket_size'] = data['wastebasket_size'].astype('float')  # Or 'category' if it's not numerical.
# Split the latlong column into separate latitude and longitude columns
# Convert both values to numeric format for geographic analysis.
if not data.empty and 'latlong' in data.columns and data['latlong'].str.contains(',').any():
    latlong_split = data['latlong'].str.split(',', expand=True)

    data['latitude'] = pd.to_numeric(latlong_split[0], errors='coerce')
    data['longitude'] = pd.to_numeric(latlong_split[1], errors='coerce')

    data.drop('latlong', axis=1, inplace=True)
# Display the DataFrame with new datatypes
data.dtypes
```

<div class="usecase-body">
This segment focuses on refining our dataset's temporal data by splitting the 'time' column into distinct 'date' and 'time_only' columns. This transformation aids in detailed time-based analysis and enables more flexible data manipulation. Here are the key steps involved:
</div>

<div class="usecase-list">

- **Datetime Parsing**: Initially, we ensure the 'time' column is correctly parsed into datetime format with `pd.to_datetime(data['time'], errors='coerce')`. The `errors='coerce'` parameter converts any parsing errors into NaT (Not a Time), preventing data corruption.

- **Column Creation**: Post successful datetime conversion, we proceed to split the datetime column:
   - The `date` column is created to store only the date component of each entry.
   - The `time_only` column is created to store only the time component.

- **Error Handling**: In cases where datetime parsing results in all null values (indicating parsing issues), both `date` and `time_only` columns are set to `None`. This precaution ensures the dataset's integrity remains uncompromised.

<div id="visualisation-part-2" class="usecase-section-heading"><b>4. Visualisation Part 2: Finding the Best Route</b></div>

<div class="usecase-body">This section uses bin fill levels and location data to determine an efficient collection route.</div>

<div id="Preparing-Fill-Levels" class="usecase-sub-section-heading"><b>4.1 Preparing Fill Levels for Route Prioritisation</b></div>

<div class="usecase-body">
The `status_current_fill_level` values are already represented as percentages from 0 to 100, so no additional normalisation is required.

These original fill-level percentages are used directly to rank the smart bins, with higher fill levels receiving greater collection priority during route planning.
</div>

```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import haversine_distances
```

<div id="Calculating-Distance-Matrix" class="usecase-sub-section-heading"><b>4.2 Calculating the Pairwise Distance Matrix</b></div>
<div class="usecase-body">
To compare the locations of the smart bins, the latitude and longitude coordinates are first converted from degrees to radians.

The Haversine method is then used to calculate the approximate great-circle distance between every pair of bin locations. This is more suitable than applying Euclidean distance directly to latitude and longitude because geographic coordinates represent positions on the Earth's curved surface.

The resulting `dist_matrix` stores the pairwise distances in kilometres and is used in the following route-planning steps.
</div>

```python

# Convert latitude and longitude from degrees to radians
coords_radians = np.radians(new_DF[['latitude', 'longitude']].to_numpy())

# Calculate pairwise Haversine distances
earth_radius_km = 6371.0

dist_matrix = (haversine_distances(coords_radians) * earth_radius_km)

print(np.round(dist_matrix, 3))
```

<div id="Route-Optimisation" class="usecase-sub-section-heading"><b>4.3 Route Optimisation Based on Fill Levels and Proximity</b></div>

<div class="usecase-body">
To enhance operational efficiency in servicing bins, we employ a strategy that prioritises bins with higher fill levels and minimises travel distance. This two-pronged approach ensures that the most critical bins are serviced first while also optimising the route taken to reduce time and resources spent. Here’s how we implement this strategy: </div>

<div id="Prioritising-by-Fill-Level" class="usecase-sub-section-heading"><b> 4.3.1 Prioritising by Fill Level:</b></div>

<div class="usecase-list">

- **Sorting Locations**: We begin by sorting the locations based on their fill levels in descending order. This is achieved by using `np.argsort` on the negative values of the `status_current_fill_level` column in `new_DF`. The result, `priority_indices`, represents the indices of locations sorted by their priority, with higher fill levels first.</div>


```python
# Sort locations by descending fill level (prioritize higher fill levels)
new_DF['status_current_fill_level'] = pd.to_numeric(new_DF['status_current_fill_level'],errors='coerce')

new_DF = new_DF.dropna(subset=['status_current_fill_level'])

priority_indices = np.argsort(-new_DF['status_current_fill_level'].to_numpy())
```

<div id="Route-Initialisation" class="usecase-sub-section-heading"><b>4.3.2 Route Initialisation:</b></div>
<div class="usecase-list">

- **Initialising the Route**: The route list is initialised with the starting point, chosen as the first location in our prioritised list. This represents the location with the highest fill level.
</div>

```python
# Initialize route list with the starting point (assuming first location as start)
route = [priority_indices[0]]
```

<div id="Determine-Next-Stop" class="usecase-sub-section-heading"><b>4.3.3 Determine Next Stop:</b></div>

<div class="usecase-body"> To determine the next stop, we follow a simple heuristic: from the current location, move to the nearest unvisited location. For each subsequent location: </div>

<div class="usecase-list">

- Calculate the distances from the last visited location to all others using the precomputed distance matrix, dist_matrix.

- Temporarily set the distance to already visited locations to infinity to exclude them from consideration.

- Identify the next stop as the location with the minimum distance to the last visited location. This location is then added to the route.
</div>

<div class="usecase-body"> This approach prioritises the bin with the highest fill level and then selects the nearest unvisited locations. This helps reduce unnecessary travel distance while ensuring the highest-priority bin is serviced first.</div>

```python
for _ in range(1, len(priority_indices)):
    last_visited = route[-1]
    distances_to_last_visited = dist_matrix[last_visited].copy()
    # Filter out already visited locations
    distances_to_last_visited[route] = np.inf
    next_stop = np.argmin(distances_to_last_visited)
    route.append(next_stop)
```

```python
# Calculate the total distance of the proposed route
total_route_distance = 0

for i in range(len(route) - 1):
    total_route_distance += dist_matrix[route[i], route[i + 1]]

print(f"Total proposed route distance: {total_route_distance:.3f} km")
```

<div class="usecase-body">
The total distance of the proposed collection sequence is calculated using the Haversine distances between consecutive stops. This provides a simple quantitative measure for evaluating the generated route rather than relying only on the map visualisation.</div>

```python
# Calculate a baseline route using fill-level priority order
baseline_route = list(priority_indices)

baseline_distance = 0

for i in range(len(baseline_route) - 1):
    baseline_distance += dist_matrix[
        baseline_route[i],
        baseline_route[i + 1]
    ]

print(f"Baseline route distance: {baseline_distance:.3f} km")
print(f"Proposed route distance: {total_route_distance:.3f} km")

if baseline_distance > 0:
    improvement = (
        (baseline_distance - total_route_distance)
        / baseline_distance
    ) * 100

    print(f"Distance reduction: {improvement:.2f}%")
```

<div class="usecase-body">
The proposed route reduces the total estimated travel distance from 1.229 km to 0.591 km, representing a distance reduction of approximately 51.92% compared with the fill-level-only baseline route.

For comparison, a baseline route follows the bins in fill-level priority order without considering proximity between consecutive stops. The proposed route is compared against this baseline using total Haversine distance. This provides a simple way to evaluate whether considering proximity reduces the overall collection distance.
</div>

<div id="Visualising-the-Optimised-Route" class="usecase-sub-section-heading"><b>4.4 Visualisation of the Optimised Route</b></div>

<div class="usecase-body">
To further refine our visualisation and provide a comprehensive view of the optimised route, we employ Folium to map out each stop with markers and visually connect these stops with lines. This approach not only highlights the individual locations but also delineates the sequence and path of the route, offering a holistic view of the journey. Here's the step-by-step process:
</div>
<div id="Setting-Up-the-Map" class="usecase-sub-section-heading"><b>4.4.1 Setting Up the Map:</b></div>
<div class="usecase-list">

- **Centring the Map**: We calculate the central point of our route by averaging the latitude and longitude of all stops. This ensures the map is optimally centred, making all route points easily viewable.</div>

```python
# Visualisation
map_center = [new_DF['latitude'].mean(), new_DF['longitude'].mean()]
m = folium.Map(location=map_center, zoom_start=20)
```

<div id="Marking-Stops" class="usecase-sub-section-heading"><b>4.4.2 Marking Stops:</b></div>
<div class="usecase-list">

- **Placing Marker**: Each location along the route is marked using a colour-coded marker based on its fill level. The markers include popups displaying the fill-level percentage, providing immediate insight into the priority of each stop.
</div>

```python
# Add a marker for the starting point
start_idx = route[0]
folium.Marker(
    [new_DF.iloc[start_idx]['latitude'], new_DF.iloc[start_idx]['longitude']],
    popup="Start",
    icon=folium.Icon(color="purple", icon="play")
).add_to(m)

# Add a marker for the end point
end_idx = route[-1]
folium.Marker(
    [new_DF.iloc[end_idx]['latitude'], new_DF.iloc[end_idx]['longitude']],
    popup="End",
    icon=folium.Icon(color="blue", icon="stop")
).add_to(m)

def get_marker_color(fill_level):
    """Determine the marker color based on fill level."""
    if fill_level <= 30:
        return "green"
    elif fill_level <= 70:
        return "orange"
    else:
        return "red"

for idx in route:
    fill_level = new_DF.iloc[idx]['status_current_fill_level']
    folium.Marker(
        [new_DF.iloc[idx]['latitude'], new_DF.iloc[idx]['longitude']],
        popup=f"Fill Level: {fill_level}%",
        icon=folium.Icon(color=get_marker_color(fill_level), icon="info-sign")
    ).add_to(m)
```
