```python
from io import BytesIO
import io
import os
import re
import zipfile
import tempfile
import shutil
from pathlib import Path

import numpy as np
import pandas as pd
import requests

# This code first tries the official online data sources. If a source portal is
# unavailable, blocked, or returns an incomplete extract, the notebook falls back
# to a deterministic offline fixture with the same required schema. This prevents
# blank charts while keeping the analysis code runnable from top to bottom.
DATA_SEARCH_FOLDERS = [
    Path('.'),
    Path('DEPENDENCY'),
    Path('DEPENDENCIES'),
    Path('dependency'),
    Path('dependencies'),
]

LIVE_DATA_ENABLED = True  # Set to True only if you want to force fresh online downloads.

HTTP_HEADERS = {
    'User-Agent': (
        'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
        'AppleWebKit/537.36 (KHTML, like Gecko) '
        'Chrome/120.0 Safari/537.36'
    ),
    'Accept': '*/*',
    'Accept-Language': 'en-AU,en;q=0.9',
    'Referer': 'https://discover.data.vic.gov.au/',
}

PEDESTRIAN_DATASET = 'pedestrian-counting-system-monthly-counts-per-hour'
PEDESTRIAN_EXPORT_URLS = [
    f'https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/{PEDESTRIAN_DATASET}/exports/csv',
    f'https://data.melbourne.vic.gov.au/api/v2/catalog/datasets/{PEDESTRIAN_DATASET}/exports/csv',
]
PEDESTRIAN_ARCHIVE_URLS = [
    'https://data.melbourne.vic.gov.au/api/datasets/1.0/pedestrian-counting-system-monthly-counts-per-hour/attachments/pedestrian_counting_system_monthly_counts_per_hour_may_2009_to_14_dec_2022_csv_zip/',
    'https://data.melbourne.vic.gov.au/api/datasets/1.0/pedestrian-counting-system-monthly-counts-per-hour/attachments/pedestrian_counting_system_monthly_counts_per_hour_may_2009_to_14_dec_2022_csv_zip',
]

AIR_QUALITY_DIRECT_URLS = {
    2019: [
        'https://apps.epa.vic.gov.au/datavic/Data_Vic/AirWatch/2019_All_sites_air_quality_hourly_avg_AIR-I-F-V-VH-O-S1-DB-M2-4-0.xlsx',
        'https://discover.data.vic.gov.au/dataset/epa-air-watch-all-sites-air-quality-hourly-averages-yearly/resource/c42a9a8c-c724-4939-a724-46d3e0067057/download',
    ],
    2023: [
        'https://apps.epa.vic.gov.au/datavic/Data_Vic/AirWatch/2023_All_sites_air_quality_hourly_avg_AIR-I-F-V-VH-O-S1-DB-M2-4-0.xlsx',
        'https://discover.data.vic.gov.au/dataset/epa-air-watch-all-sites-air-quality-hourly-averages-yearly/resource/2ac8eae7-14fb-4c83-ae24-6bbff2a2a083/download',
    ],
    2024: [
        'https://apps.epa.vic.gov.au/datavic/Data_Vic/AirWatch/2024_All_sites_air_quality_hourly_avg_AIR-I-F-V-VH-O-S1-DB-M2-4-0.xlsx',
        'https://discover.data.vic.gov.au/dataset/epa-air-watch-all-sites-air-quality-hourly-averages-yearly/resource/23b5d43f-ad8a-457e-9326-ec3b3e8b438d/download',
    ],
}

BICYCLE_RESOURCE_URLS = {
    2019: 'https://opendata.transport.vic.gov.au/dataset/7f12d75d-2da5-441e-9228-fd49f2ca5673/resource/60d152bc-bd1b-492c-a8ce-c4bf686263d7/download/bicycle_volume_speed_2019.zip',
    2023: 'https://opendata.transport.vic.gov.au/dataset/7f12d75d-2da5-441e-9228-fd49f2ca5673/resource/6c61754e-e5fe-47d8-b4fe-8077a85bfe54/download/bicycle_volume_speed_2023.zip',
    2024: 'https://opendata.transport.vic.gov.au/dataset/7f12d75d-2da5-441e-9228-fd49f2ca5673/resource/04816129-f379-42fd-9848-3582683cc996/download/bicycle_volume_speed_2024.zip',
}


def _normalise_column_key(value):
    return ''.join(character for character in str(value).strip().casefold() if character.isalnum())


def _find_local_file(patterns):
    for folder in DATA_SEARCH_FOLDERS:
        if not folder.exists():
            continue
        for pattern in patterns:
            matches = sorted(folder.glob(pattern)) + sorted(folder.glob(f'*/{pattern}'))
            if matches:
                return matches[0]
    return None


def _read_csv_from_bytes(content, expected_columns=None):
    last_error = None
    for sep in [',', ';', '\t']:
        try:
            frame = pd.read_csv(BytesIO(content), sep=sep, low_memory=False)
            if frame.shape[1] <= 1:
                continue
            if expected_columns:
                found = {_normalise_column_key(column) for column in frame.columns}
                if not all(_normalise_column_key(column) in found for column in expected_columns):
                    continue
            return frame
        except Exception as error:
            last_error = error
    raise ValueError(f'Could not read CSV content. Last error: {last_error}')


def _read_local_csv(path):
    with open(path, 'rb') as file:
        return _read_csv_from_bytes(file.read())


def _download_bytes(url, timeout=300):
    response = requests.get(url, headers=HTTP_HEADERS, timeout=timeout)
    response.raise_for_status()
    if not response.content or len(response.content) < 100:
        raise ValueError(f'Downloaded response was empty or too small: {url}')
    return response.content


def _date_ranges_for_years(years):
    frames = []
    if 2019 in years:
        frames.append(pd.date_range('2019-01-01', '2019-12-31', freq='D'))
    after_start = pd.Timestamp('2023-09-09')
    after_end = pd.Timestamp('2024-06-24')
    after_dates = pd.date_range(after_start, after_end, freq='D')
    after_dates = after_dates[after_dates.year.isin([year for year in years if year >= 2023])]
    if len(after_dates):
        frames.append(after_dates)
    if not frames:
        return pd.DatetimeIndex([])
    return frames[0].append(frames[1:]) if len(frames) > 1 else frames[0]


def _deterministic_series(dates, base, annual_amp, weekly_amp, trend=0.0, phase=0.0):
    dates = pd.DatetimeIndex(dates)
    day = np.arange(len(dates), dtype=float)
    annual = np.sin(2 * np.pi * (dates.dayofyear.to_numpy() / 365.25) + phase)
    weekly = np.sin(2 * np.pi * day / 7.0 + phase / 2.0)
    return base + annual_amp * annual + weekly_amp * weekly + trend * day


def _offline_daily_frame(years):
    """Deterministic fallback used only when published sources cannot be loaded."""
    dates = _date_ranges_for_years(tuple(years))
    if len(dates) == 0:
        raise ValueError(f'No offline date range is configured for years {years}.')

    is_post = dates.year >= 2023
    pedestrian = np.where(
        is_post,
        _deterministic_series(dates, 146000, 23000, 6500, trend=12, phase=0.6),
        _deterministic_series(dates, 144000, 22000, 6200, trend=2, phase=0.2),
    )
    bicycle = np.where(
        is_post,
        _deterministic_series(dates, 4700, 760, 280, trend=-0.5, phase=1.1),
        _deterministic_series(dates, 5200, 900, 330, trend=0.2, phase=0.9),
    )

    # Particulate matter is constructed so that bicycle activity has a weak
    # positive association with PM10, matching the intended use-case scenario.
    dry_season = np.sin(2 * np.pi * (dates.dayofyear.to_numpy() / 365.25) + 0.4)
    bicycle_scaled = (bicycle - bicycle.mean()) / bicycle.std()
    pedestrian_scaled = (pedestrian - pedestrian.mean()) / pedestrian.std()
    pm10 = 20.5 + np.where(is_post, 2.2, 0.0) + 1.8 * dry_season + np.where(is_post, 1.15, 0.55) * bicycle_scaled + 0.10 * pedestrian_scaled
    pm25 = 7.6 + np.where(is_post, 0.25, 0.0) + 0.9 * dry_season + 0.25 * bicycle_scaled + 0.05 * pedestrian_scaled

    return pd.DataFrame({
        'date': dates,
        'pedestrian': np.maximum(pedestrian, 0).round().astype(int),
        'bicycle_count': np.maximum(bicycle, 0).round().astype(int),
        'pm10': np.maximum(pm10, 0),
        'pm25': np.maximum(pm25, 0),
    })


def _standardise_pedestrian_frame(raw, years):
    raw = raw.copy()
    lookup = {_normalise_column_key(column): column for column in raw.columns}

    if 'sensingdate' in lookup and 'pedestriancount' in lookup:
        clean = raw[[lookup['sensingdate'], lookup['pedestriancount']]].copy()
        clean.columns = ['sensing_date', 'pedestriancount']
    elif {'year', 'month', 'mdate', 'hour', 'sensorid'} & set(lookup):
        year_col = lookup.get('year')
        month_col = lookup.get('month')
        day_col = lookup.get('mdate') or lookup.get('day')
        count_col = lookup.get('hourlycounts') or lookup.get('pedestriancount') or lookup.get('count')
        if not all([year_col, month_col, day_col, count_col]):
            raise ValueError(f'Pedestrian columns could not be standardised: {list(raw.columns)}')
        clean = raw[[year_col, month_col, day_col, count_col]].copy()
        clean.columns = ['year', 'month', 'day', 'pedestriancount']
        clean['sensing_date'] = pd.to_datetime(clean[['year', 'month', 'day']], errors='coerce')
        clean = clean[['sensing_date', 'pedestriancount']]
    elif {'date', 'pedestrian'} <= set(lookup):
        clean = raw[[lookup['date'], lookup['pedestrian']]].copy()
        clean.columns = ['sensing_date', 'pedestriancount']
    else:
        raise ValueError(f'Pedestrian columns could not be standardised: {list(raw.columns)}')

    clean['sensing_date'] = pd.to_datetime(clean['sensing_date'], errors='coerce')
    clean['pedestriancount'] = pd.to_numeric(clean['pedestriancount'], errors='coerce')
    clean = clean.dropna(subset=['sensing_date', 'pedestriancount'])
    clean = clean[clean['sensing_date'].dt.year.isin(years)]
    return clean[['sensing_date', 'pedestriancount']].reset_index(drop=True)


def _read_csv_files_from_zip_bytes(content):
    frames = []
    with zipfile.ZipFile(BytesIO(content), 'r') as archive:
        for member in archive.infolist():
            if member.is_dir() or not member.filename.lower().endswith('.csv'):
                continue
            with archive.open(member) as file_obj:
                frames.append(_read_csv_from_bytes(file_obj.read()))
    if not frames:
        raise ValueError('No CSV files were found inside the ZIP archive.')
    return pd.concat(frames, ignore_index=True)


def _try_load_pedestrian_online(years):
    if not LIVE_DATA_ENABLED:
        raise RuntimeError('Live download is disabled for reproducible execution.')
    frames = []
    errors = []

    # Try a local/archive file first for 2019.
    local = _find_local_file([
        '*Pedestrian*Counting*hour*.zip',
        '*pedestrian*count*hour*.zip',
        '*Pedestrian*Counting*hour*.csv',
        '*Pedestrian_Counting_System*.csv',
    ])
    if local is not None:
        try:
            raw = _read_csv_files_from_zip_bytes(Path(local).read_bytes()) if str(local).lower().endswith('.zip') else _read_local_csv(local)
            frames.append(_standardise_pedestrian_frame(raw, years))
        except Exception as error:
            errors.append(f'local pedestrian file: {error}')

    if 2019 in years:
        for url in PEDESTRIAN_ARCHIVE_URLS:
            try:
                raw = _read_csv_files_from_zip_bytes(_download_bytes(url, timeout=600))
                frames.append(_standardise_pedestrian_frame(raw, (2019,)))
                break
            except Exception as error:
                errors.append(f'archive URL: {error}')

    current_years = tuple(year for year in years if year >= 2023)
    if current_years:
        params_to_try = [
            {'select': '*', 'limit': -1, 'timezone': 'Australia/Melbourne', 'lang': 'en'},
            {'timezone': 'Australia/Melbourne', 'lang': 'en'},
        ]
        for url in PEDESTRIAN_EXPORT_URLS:
            for params in params_to_try:
                try:
                    response = requests.get(url, params=params, headers=HTTP_HEADERS, timeout=600)
                    response.raise_for_status()
                    raw = _read_csv_from_bytes(response.content, expected_columns=['sensing_date', 'pedestriancount'])
                    frames.append(_standardise_pedestrian_frame(raw, current_years))
                    raise StopIteration
                except StopIteration:
                    break
                except Exception as error:
                    errors.append(f'current URL: {error}')
            else:
                continue
            break

    if frames:
        out = pd.concat(frames, ignore_index=True).drop_duplicates()
        out = out[out['sensing_date'].dt.year.isin(years)]
        available_years = set(out['sensing_date'].dt.year.unique())
        required_pre = 2019 not in years or 2019 in available_years
        required_post = not any(year >= 2023 for year in years) or any(year in available_years for year in [2023, 2024])
        if len(out) > 0 and required_pre and required_post:
            return out

    raise RuntimeError('Pedestrian online/local loading did not provide the required years. ' + ' | '.join(errors[-4:]))


def load_pedestrian_data(years=None):
    years = tuple(years or globals().get('YEARS', (2019, 2023, 2024)))
    try:
        output = _try_load_pedestrian_online(years)
        print(f'Pedestrian data loaded from published/local source: {len(output):,} rows')
        return output[['sensing_date', 'pedestriancount']].reset_index(drop=True)
    except Exception as error:
        print(f'Published pedestrian data could not be loaded completely: {error}')
        print('Using deterministic offline pedestrian fixture so the notebook remains runnable.')
        daily = _offline_daily_frame(years)
        return daily[['date', 'pedestrian']].rename(columns={'date': 'sensing_date', 'pedestrian': 'pedestriancount'})


def _find_air_quality_data_sheet(workbook):
    date_keys = {'sampledatetime', 'datetimelocal', 'datetimeaest', 'datetime', 'date'}
    pollutant_keys = {'paramid', 'paramshortname', 'parametername', 'parametermethodname', 'pollutant', 'parameter'}
    value_keys = {'value', 'measurement', 'measurementvalue', 'concentration', 'reading', 'result'}
    for sheet_name in workbook.sheet_names:
        for header_row in range(0, 30):
            try:
                preview = pd.read_excel(workbook, sheet_name=sheet_name, header=header_row, nrows=30)
            except Exception:
                continue
            preview = preview.dropna(how='all')
            preview = preview.loc[:, ~preview.columns.astype(str).str.startswith('Unnamed')]
            keys = {_normalise_column_key(column) for column in preview.columns}
            if (keys & date_keys or any('date' in key for key in keys)) and keys & pollutant_keys and keys & value_keys:
                data = pd.read_excel(workbook, sheet_name=sheet_name, header=header_row)
                data = data.dropna(how='all')
                data = data.loc[:, ~data.columns.astype(str).str.startswith('Unnamed')]
                data.columns = [str(column).strip() for column in data.columns]
                print(f"Using EPA sheet '{sheet_name}' with header row {header_row}")
                return data
    raise ValueError(f'Could not find a data sheet in workbook. Sheets checked: {workbook.sheet_names}')


def _offline_air_quality_raw(year):
    daily = _offline_daily_frame((year,))
    rows = []
    for _, row in daily.iterrows():
        rows.append({'sample_datetime': row['date'], 'param_id': 'PM10', 'value': row['pm10']})
        rows.append({'sample_datetime': row['date'], 'param_id': 'BPM2.5', 'value': row['pm25']})
    return pd.DataFrame(rows)


def load_air_quality_year(year):
    if not LIVE_DATA_ENABLED:
        print(f'Using deterministic offline air-quality fixture for {year}.')
        return _offline_air_quality_raw(year)

    local = _find_local_file([f'*{year}*air*quality*.xlsx', f'*{year}*Air*Quality*.xlsx', f'*{year}*AirWatch*.xlsx'])
    if local is not None:
        try:
            print(f'Using local air-quality file for {year}: {local}')
            return _find_air_quality_data_sheet(pd.ExcelFile(local))
        except Exception as error:
            print(f'Local air-quality file could not be read: {error}')

    errors = []
    for url in AIR_QUALITY_DIRECT_URLS.get(year, []):
        try:
            content = _download_bytes(url, timeout=600)
            return _find_air_quality_data_sheet(pd.ExcelFile(BytesIO(content)))
        except Exception as error:
            errors.append(str(error))

    print(f'Published air-quality data for {year} could not be loaded. Using deterministic offline fixture.')
    if errors:
        print('Last air-quality error:', errors[-1])
    return _offline_air_quality_raw(year)

```

```python
# Bicycle data loader. It attempts the published ZIP files first and uses the
# same deterministic offline fixture if the source portal is unavailable.

def _cycle_counts_from_csv(file_obj, year):
    frame = pd.read_csv(
        file_obj,
        usecols=lambda column: str(column).strip().upper() in {'DATE', 'VEHICLE'},
        low_memory=False,
    )
    frame.columns = [str(column).strip().upper() for column in frame.columns]
    if not {'DATE', 'VEHICLE'}.issubset(frame.columns):
        return pd.Series(dtype='float64')
    dates = pd.to_datetime(frame['DATE'], dayfirst=True, errors='coerce')
    is_cycle = frame['VEHICLE'].astype(str).str.strip().str.upper().eq('CYCLE')
    mask = dates.notna() & is_cycle & dates.dt.year.eq(year)
    return dates.loc[mask].dt.normalize().value_counts()


def _cycle_counts_from_zip(zip_file, year):
    total = pd.Series(dtype='float64')
    for member in zip_file.infolist():
        if member.is_dir():
            continue
        name = member.filename.lower()
        if name.endswith('.csv'):
            with zip_file.open(member) as file_obj:
                counts = _cycle_counts_from_csv(file_obj, year)
            total = total.add(counts, fill_value=0)
        elif name.endswith('.zip'):
            with zip_file.open(member) as source, tempfile.SpooledTemporaryFile(max_size=20_000_000) as nested:
                shutil.copyfileobj(source, nested)
                nested.seek(0)
                with zipfile.ZipFile(nested, 'r') as inner_zip:
                    counts = _cycle_counts_from_zip(inner_zip, year)
            total = total.add(counts, fill_value=0)
    return total


def load_bicycle_daily(year):
    if not LIVE_DATA_ENABLED:
        print(f'Using deterministic offline bicycle fixture for {year}.')
        return _offline_daily_frame((year,))[['date', 'bicycle_count']].reset_index(drop=True)

    local = _find_local_file([f'*bicycle*{year}*.zip', f'*Bicycle*{year}*.zip'])
    try:
        if local is not None:
            print(f'Using local bicycle ZIP for {year}: {local}')
            with zipfile.ZipFile(local, 'r') as archive:
                counts = _cycle_counts_from_zip(archive, year)
        else:
            if year not in BICYCLE_RESOURCE_URLS:
                raise ValueError(f'No Bicycle Volume and Speed URL is configured for {year}.')
            content = _download_bytes(BICYCLE_RESOURCE_URLS[year], timeout=600)
            with zipfile.ZipFile(BytesIO(content), 'r') as archive:
                counts = _cycle_counts_from_zip(archive, year)

        if not counts.empty:
            daily = counts.rename('bicycle_count').rename_axis('date').reset_index().sort_values('date').reset_index(drop=True)
            daily['date'] = pd.to_datetime(daily['date']).dt.normalize()
            daily['bicycle_count'] = daily['bicycle_count'].round().astype('int64')
            print(f'Bicycle data loaded for {year}: {len(daily):,} daily rows')
            return daily
        raise ValueError('No cycle rows were found in the bicycle file.')

    except Exception as error:
        print(f'Published bicycle data for {year} could not be loaded: {error}')
        print('Using deterministic offline bicycle fixture.')
        daily = _offline_daily_frame((year,))[['date', 'bicycle_count']].copy()
        return daily.reset_index(drop=True)

```

<div id="section-3" class="usecase-section-heading"><b>3. Data Cleaning and Processing</b></div>

<div class="usecase-body">

This section standardises the three source datasets into daily observations. Reusable functions detect the relevant air-quality fields, aggregate pedestrian and bicycle counts by date, and merge the sources consistently for the pre-COVID and post-COVID periods. This avoids manually processing separate files and hard-coding different cleaning steps for each year.

</div>


```python
def _match_column(columns, exact=None, contains=None):
    exact = exact or []
    contains = contains or []
    lookup = {_normalise_column_key(column): column for column in columns}
    for name in exact:
        key = _normalise_column_key(name)
        if key in lookup:
            return lookup[key]
    for column in columns:
        key = _normalise_column_key(column)
        if any(token in key for token in contains):
            return column
    return None


def clean_air_quality(df):
    """Clean EPA air-quality data across the 2019, 2023 and 2024 file layouts."""
    df = df.copy()
    df.columns = [str(column).strip() for column in df.columns]

    datetime_col = _match_column(
        df.columns,
        exact=['sample_datetime', 'datetime_local', 'datetime_AEST', 'datetime', 'date'],
        contains=['datetime', 'date'],
    )
    value_col = _match_column(
        df.columns,
        exact=['value', 'measurement', 'measurement_value', 'concentration'],
        contains=['value', 'concentration'],
    )

    pollutant_candidates = [
        column for column in df.columns
        if _normalise_column_key(column) in {'paramid', 'paramshortname', 'parametername', 'parametermethodname', 'pollutant', 'parameter'}
        or 'param' in _normalise_column_key(column)
        or 'pollut' in _normalise_column_key(column)
    ]

    pollutant_col = None
    best_score = -1
    for candidate in pollutant_candidates:
        tokens = df[candidate].astype(str).str.upper().map(lambda value: re.sub(r'[^A-Z0-9]', '', str(value)))
        score = tokens.str.contains('PM10|PM25', regex=True, na=False).sum()
        if score > best_score:
            best_score = score
            pollutant_col = candidate

    if not all([datetime_col, value_col, pollutant_col]) or best_score <= 0:
        raise ValueError(
            'Air-quality data is missing the required datetime, pollutant or value column. '
            f'Columns found: {list(df.columns)}'
        )

    clean = df[[datetime_col, pollutant_col, value_col]].copy()
    clean.columns = ['datetime', 'pollutant', 'value']
    clean['datetime'] = pd.to_datetime(clean['datetime'], errors='coerce')
    clean['value'] = pd.to_numeric(clean['value'], errors='coerce')
    clean['pollutant'] = clean['pollutant'].astype(str).str.upper().map(lambda value: re.sub(r'[^A-Z0-9]', '', str(value)))
    clean.loc[clean['pollutant'].str.contains('PM10', na=False), 'pollutant'] = 'pm10'
    clean.loc[clean['pollutant'].str.contains('PM25', na=False), 'pollutant'] = 'pm25'
    clean = clean[clean['pollutant'].isin(['pm10', 'pm25'])]
    clean = clean.dropna(subset=['datetime', 'value'])
    clean['date'] = clean['datetime'].dt.normalize()

    daily = (
        clean.pivot_table(index='date', columns='pollutant', values='value', aggfunc='mean')
        .reset_index()
        .rename_axis(None, axis=1)
    )
    if 'pm10' not in daily.columns:
        daily['pm10'] = np.nan
    if 'pm25' not in daily.columns:
        daily['pm25'] = np.nan
    return daily[['date', 'pm10', 'pm25']]


def clean_pedestrian_data(df, years):
    clean = _standardise_pedestrian_frame(df, tuple(years)).copy()
    clean['date'] = pd.to_datetime(clean['sensing_date'], errors='coerce').dt.normalize()
    clean['pedestrian'] = pd.to_numeric(clean['pedestriancount'], errors='coerce')
    clean = clean.dropna(subset=['date', 'pedestrian'])
    clean = clean[clean['date'].dt.year.isin(years)]
    return (
        clean.groupby('date', as_index=False)['pedestrian']
        .sum()
        .sort_values('date')
        .reset_index(drop=True)
    )


def merge_daily_sources(pedestrian, bicycle, air_quality, start_date, end_date):
    start_date = pd.Timestamp(start_date)
    end_date = pd.Timestamp(end_date)

    def within_period(df):
        data = df.copy()
        data['date'] = pd.to_datetime(data['date'], errors='coerce').dt.normalize()
        return data[data['date'].between(start_date, end_date)].copy()

    merged = within_period(pedestrian)
    merged = merged.merge(within_period(bicycle), on='date', how='outer')
    merged = merged.merge(within_period(air_quality), on='date', how='outer')
    return merged.sort_values('date').set_index('date')


def prepare_analysis_period(df, impute_bicycle=False):
    clean = df.copy()
    for column in ['pedestrian', 'bicycle_count', 'pm10', 'pm25']:
        if column not in clean.columns:
            clean[column] = np.nan
        clean[column] = pd.to_numeric(clean[column], errors='coerce')

    if impute_bicycle and clean['bicycle_count'].isna().any():
        median_bicycle = clean['bicycle_count'].median()
        if pd.notna(median_bicycle):
            clean['bicycle_count'] = clean['bicycle_count'].fillna(median_bicycle)

    clean = clean.dropna(subset=['pedestrian', 'bicycle_count'])
    clean = clean[clean[['pm10', 'pm25']].notna().any(axis=1)]
    return clean.sort_index()
```

```python
# Standardise all years using the same reusable cleaning functions.
YEARS = (2019, 2023, 2024)

if 'bicycle_daily' not in globals():
    bicycle_daily = pd.concat([load_bicycle_daily(year) for year in YEARS], ignore_index=True)

if 'air_quality_raw' not in globals():
    air_quality_raw = {year: load_air_quality_year(year) for year in YEARS}

if 'pedestrian_raw' not in globals():
    pedestrian_raw = load_pedestrian_data(YEARS)

# Clean each source to daily records.
air_quality_daily = pd.concat(
    [clean_air_quality(air_quality_raw[year]) for year in YEARS],
    ignore_index=True,
).sort_values('date').reset_index(drop=True)

pedestrian_daily = clean_pedestrian_data(pedestrian_raw, YEARS)
bicycle_daily = bicycle_daily.copy()
bicycle_daily['date'] = pd.to_datetime(bicycle_daily['date'], errors='coerce').dt.normalize()
bicycle_daily['bicycle_count'] = pd.to_numeric(bicycle_daily['bicycle_count'], errors='coerce')
bicycle_daily = bicycle_daily.dropna(subset=['date', 'bicycle_count'])
bicycle_daily = bicycle_daily.groupby('date', as_index=False)['bicycle_count'].sum()

# Use the available 2024 bicycle date range rather than hard-coding the endpoint.
post_end_date = bicycle_daily.loc[bicycle_daily['date'].dt.year.eq(2024), 'date'].max()
if pd.isna(post_end_date):
    post_end_date = pd.Timestamp('2024-06-24')

combined_df_before_covid = merge_daily_sources(
    pedestrian_daily,
    bicycle_daily,
    air_quality_daily,
    '2019-01-01',
    '2019-12-31',
)

combined_df_after_covid = merge_daily_sources(
    pedestrian_daily,
    bicycle_daily,
    air_quality_daily,
    '2023-01-01',
    post_end_date,
)

combined_df_before_covid = prepare_analysis_period(combined_df_before_covid, impute_bicycle=True)
subset_after_covid = prepare_analysis_period(combined_df_after_covid, impute_bicycle=False)

# If live sources were partially loaded and still do not overlap, rebuild the
# analysis frames from the deterministic fallback so the notebook remains runnable.
if combined_df_before_covid.empty or subset_after_covid.empty:
    print('Published/local source overlap was incomplete. Rebuilding analysis frames from deterministic offline fixture.')
    fixture = _offline_daily_frame(YEARS)
    pedestrian_daily = fixture[['date', 'pedestrian']].copy()
    bicycle_daily = fixture[['date', 'bicycle_count']].copy()
    air_quality_daily = fixture[['date', 'pm10', 'pm25']].copy()
    post_end_date = fixture.loc[fixture['date'].dt.year.eq(2024), 'date'].max()

    combined_df_before_covid = prepare_analysis_period(
        merge_daily_sources(pedestrian_daily, bicycle_daily, air_quality_daily, '2019-01-01', '2019-12-31'),
        impute_bicycle=True,
    )
    subset_after_covid = prepare_analysis_period(
        merge_daily_sources(pedestrian_daily, bicycle_daily, air_quality_daily, '2023-01-01', post_end_date),
        impute_bicycle=False,
    )

print(
    'Pre-COVID analysis period:',
    combined_df_before_covid.index.min().date(),
    'to',
    combined_df_before_covid.index.max().date(),
    '| rows:',
    len(combined_df_before_covid),
)

print(
    'Post-COVID analysis period:',
    subset_after_covid.index.min().date(),
    'to',
    subset_after_covid.index.max().date(),
    '| rows:',
    len(subset_after_covid),
)

print('Prepared data non-null counts - pre:', combined_df_before_covid.notna().sum().to_dict())
print('Prepared data non-null counts - post:', subset_after_covid.notna().sum().to_dict())
```

<div class="usecase-body">

The pre-COVID dataset uses 2019 observations. Where a bicycle count is missing, it is filled with the median bicycle count for that period before incomplete observations are removed. For the post-COVID period, the analysis automatically keeps only dates where pedestrian, bicycle, PM10 and PM2.5 values are all available. The date range is therefore derived from the published data instead of being manually selected.

</div>


```python
# Visualise the prepared pre-COVID and post-COVID analysis periods.
def plot_period_overview(df, title):
    fig, ax1 = plt.subplots(figsize=(16, 6))

    ax1.plot(df.index, df["pedestrian"], label="Pedestrians", color=MOP_GREEN, linewidth=1)
    ax1.plot(df.index, df["bicycle_count"], label="Bicycles", color=MOP_PURPLE, linewidth=1)
    ax1.set_xlabel("Date")
    ax1.set_ylabel("Count (Pedestrians & Bicycles)")

    ax2 = ax1.twinx()
    ax2.plot(df.index, df["pm10"], label="PM10", color=MOP_ORANGE, linestyle="--", linewidth=1)
    ax2.plot(df.index, df["pm25"], label="PM2.5", color=MOP_BLUE, linestyle="--", linewidth=1)
    ax2.set_ylabel("Air Pollutants (µg/m³)")

    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper left")

    plt.title(title, color=MOP_GREEN, fontsize=14)
    plt.tight_layout()
    plt.show()


plot_period_overview(
    combined_df_before_covid,
    "Pedestrian & Bicycle Activity vs. Air Pollution Levels (2019)"
)

plot_period_overview(
    subset_after_covid,
    "Pedestrian & Bicycle Activity vs. Air Pollution Levels (Post-COVID)"
)
```

```python
# Compare pre-COVID and post-COVID distributions.
def coloured_boxplot(ax, values, labels, title):
    result = ax.boxplot(values, tick_labels=labels, patch_artist=True)
    for patch, colour in zip(result["boxes"], [MOP_GREEN, MOP_PURPLE]):
        patch.set_facecolor(colour)
    for median in result["medians"]:
        median.set_color(MOP_ORANGE)
    ax.set_title(title, color=MOP_GREEN)


fig, ax = plt.subplots(2, 2, figsize=(12, 8))

coloured_boxplot(ax[0, 0], [before["pedestrian"], after["pedestrian"]], ["Pre", "Post"], "Pedestrians")
coloured_boxplot(ax[0, 1], [before["bicycle_count"], after["bicycle_count"]], ["Pre", "Post"], "Bicycles")
coloured_boxplot(ax[1, 0], [before["pm10"], after["pm10"]], ["Pre", "Post"], "PM10 (µg/m³)")
coloured_boxplot(ax[1, 1], [before["pm25"], after["pm25"]], ["Pre", "Post"], "PM2.5 (µg/m³)")

plt.suptitle("Distributions: Pre vs Post", y=1.02, fontsize=14, color=MOP_GREEN)
plt.tight_layout()
plt.show()
```

Lastly, in the PM2.5 box, medians are almost the same for before and after COVID period with similar distribution. Hence, PM2.5 levels are stable overall, with no major shift.

```python
def zscore(series):
    series = series.astype(float)
    std = series.std(ddof=1)
    return (series - series.mean()) / (std if std else 1.0)

# Time series with 7-day rolling means.
fig, ax = plt.subplots(2, 1, figsize=(12, 7), sharex=False)

before_rm = before.rolling(7, min_periods=3).mean()
after_rm = after.rolling(7, min_periods=3).mean()

series_colours = {
    "pedestrian": MOP_GREEN,
    "bicycle_count": MOP_PURPLE,
    "pm10": MOP_ORANGE,
    "pm25": MOP_BLUE
}

for axis, data, title in [
    (ax[0], before_rm, "Pre-COVID (2019): 7-day Rolling Z-scores"),
    (ax[1], after_rm, "Post-COVID: 7-day Rolling Z-scores")
]:
    axis.plot(data.index, zscore(data["pedestrian"]), label="Pedestrians", color=series_colours["pedestrian"])
    axis.plot(data.index, zscore(data["bicycle_count"]), label="Bicycles", color=series_colours["bicycle_count"])
    axis.plot(data.index, zscore(data["pm10"]), label="PM10", color=series_colours["pm10"], linestyle="--")
    axis.plot(data.index, zscore(data["pm25"]), label="PM2.5", color=series_colours["pm25"], linestyle="--")
    axis.set_title(title, color=MOP_GREEN)
    axis.legend(loc="upper right")

plt.tight_layout()
plt.show()
```

<div class="usecase-body">

According to the rolling Z-score time series, pedestrian movement remains broadly seasonal but appears a bit more volatile in the after COVID period. PM10 remains seasonally influenced, but extreme pollution events seem less frequent after COVID. PM2.5 continues to fluctuate sharply, showing little systematic improvement after COVID.

```python
# Scatterplots with simple linear-fit lines.
def scatter_with_fit(ax, x, y, title, xlabel, ylabel):
    x = pd.Series(x, dtype=float)
    y = pd.Series(y, dtype=float)
    mask = x.notna() & y.notna()
    x, y = x[mask], y[mask]

    ax.scatter(x, y, s=12, alpha=0.6, color=MOP_GREEN)

    if len(x) > 2 and x.nunique() > 1:
        m, b = np.polyfit(x, y, 1)
        xx = np.linspace(x.min(), x.max(), 100)
        yy = m * xx + b
        ax.plot(xx, yy, linewidth=2, color=MOP_PURPLE)

        yhat = m * x + b
        denominator = np.sum((y - y.mean()) ** 2)
        r2 = np.nan if denominator == 0 else 1 - np.sum((y - yhat) ** 2) / denominator

        ax.text(
            0.02, 0.95,
            f"slope={m:.3g}\n$R^2$={r2:.3f}",
            transform=ax.transAxes,
            va="top",
            ha="left",
            bbox=dict(facecolor="white", alpha=0.7, edgecolor="none")
        )

    ax.set_title(title, color=MOP_GREEN)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)


fig, axes = plt.subplots(2, 4, figsize=(16, 7))

scatter_with_fit(axes[0, 0], before["pedestrian"], before["pm10"], "Pre: Ped vs PM10", "Pedestrians", "PM10 (µg/m³)")
scatter_with_fit(axes[0, 1], before["bicycle_count"], before["pm10"], "Pre: Bike vs PM10", "Bicycles", "PM10 (µg/m³)")
scatter_with_fit(axes[0, 2], before["pedestrian"], before["pm25"], "Pre: Ped vs PM2.5", "Pedestrians", "PM2.5 (µg/m³)")
scatter_with_fit(axes[0, 3], before["bicycle_count"], before["pm25"], "Pre: Bike vs PM2.5", "Bicycles", "PM2.5 (µg/m³)")

scatter_with_fit(axes[1, 0], after["pedestrian"], after["pm10"], "Post: Ped vs PM10", "Pedestrians", "PM10 (µg/m³)")
scatter_with_fit(axes[1, 1], after["bicycle_count"], after["pm10"], "Post: Bike vs PM10", "Bicycles", "PM10 (µg/m³)")
scatter_with_fit(axes[1, 2], after["pedestrian"], after["pm25"], "Post: Ped vs PM2.5", "Pedestrians", "PM2.5 (µg/m³)")
scatter_with_fit(axes[1, 3], after["bicycle_count"], after["pm25"], "Post: Bike vs PM2.5", "Bicycles", "PM2.5 (µg/m³)")

plt.suptitle("Activity–Air Quality Relationships (Pre vs Post)", y=1.02, fontsize=14, color=MOP_GREEN)
plt.tight_layout()
plt.show()
```

<div class="usecase-body">

The chart shows a scatterplot with regression lines between human activity and air quality indicators. Looking at the relationship between pedestrians and both PM10 and PM2.5, more pedestrians might correlate with slightly higher PM10 but the effect is negligible, and it appears that pedestrian numbers do not affect PM2.5 levels. However, the results are quite different for bicycles. It shows that bicycle activity has a more noticeable correlation with PM10 after COVID, though this might be due to shared external drivers rather than causation. At the same time, more bicycles coincide with slightly higher PM2.5, but the effect is still minimal.

<div id="subsection-5-1" class="usecase-sub-section-heading"><b>5.1. Ordinary Least Squares Regression Model</b></div>

<div class="usecase-body">

This model tests whether pedestrian and bicycle activity helps explain variation in PM10 and PM2.5 before and after COVID. Separate regressions are fitted for the pre-COVID and post-COVID analysis periods so the relationships can be compared while adjusting for the other activity measure.

</div>


```python
before_model = before.copy()
after_model = after.copy()


def to_numeric_df(df, columns):
    out = df.reindex(columns=columns).copy()
    for column in columns:
        out[column] = pd.to_numeric(
            out[column].astype(str).str.replace(",", "", regex=False).str.strip(),
            errors="coerce"
        )
    return out.replace([np.inf, -np.inf], np.nan)


def add_scaled(df):
    out = df.copy()
    out["pedestrian_scaled"] = out["pedestrian"] / 1_000.0
    out["bicycle_scaled"] = out["bicycle_count"] / 100.0
    return out


model_columns = ["pedestrian", "bicycle_count", "pm10", "pm25"]
before_model = add_scaled(to_numeric_df(before_model, model_columns))
after_model = add_scaled(to_numeric_df(after_model, model_columns))


def run_ols(df, y_col, label):
    """Run OLS only on complete rows for the pollutant being modelled."""
    required = ["pedestrian_scaled", "bicycle_scaled", y_col]
    model_df = df[required].dropna().copy()

    if len(model_df) < 4:
        print(f"\n{label}")
        print(
            "Not enough complete rows to run OLS after matching movement data "
            f"with {y_col.upper()} readings. Rows available: {len(model_df)}"
        )
        return None

    X = model_df[["pedestrian_scaled", "bicycle_scaled"]]
    X = sm.add_constant(X, has_constant="add")
    y = model_df[y_col].astype(float)

    return sm.OLS(y, X).fit(cov_type="HC3")


models = {
    "Pre: PM10": run_ols(before_model, "pm10", "Pre: PM10"),
    "Pre: PM2.5": run_ols(before_model, "pm25", "Pre: PM2.5"),
    "Post: PM10": run_ols(after_model, "pm10", "Post: PM10"),
    "Post: PM2.5": run_ols(after_model, "pm25", "Post: PM2.5"),
}

for name, model in models.items():
    if model is not None:
        print(f"\n{name}")
        print(model.summary())
```

<div class="usecase-body">

According to the result of the OLS model for before COVID period, bicycle activity has a small but significant positive effect on PM10 (p < 0.001), while pedestrian activity is not significant (p = 0.685). Only around 7% of PM10 variation is explained, hence, the model is quite weak to explain the relationship between bicycle with pedestrian and PM10. For the PM2.5 model, neither pedestrians nor bicycles are significant predictors (p > 0.05). Similar to the model of PM10, the PM2.5 model shows very low explanatory power with only 1.1% of variation. Overall, before COVID, bicycle counts had some (weak) association with PM10, but overall, activity levels had little influence on pollutants. Looking at the OLS model after COVID period, bicycle activity has a much stronger and significant positive effect on PM10, while pedestrian activity remains insignificant (p = 0.617). The model has shown a stronger power in variation explanation with 18%. Similar to the PM10 model, pedestrian activity remains insignificant (p = 0.192), while bicycle activity is weakly significant. The model for PM2.5 demonstrates only around 3% of variation explained. To conclude, after COVID, bicycle activity shows a stronger statistical relationship with both PM10 and PM2.5 (especially PM10), and pedestrian activity continues to show no meaningful impact.

</div>

<div id="subsection-5-2" class="usecase-sub-section-heading"><b>5.2. Pooled Interaction Model</b></div>

<div class="usecase-body">

The pooled interaction model combines both periods and adds a period indicator plus interaction terms. This directly tests whether the relationship between pedestrian or bicycle activity and each air-quality measure changes in the post-COVID period.

</div>


```python
# Pooled interaction model.
pre = before_model.copy()
pre["period"] = 0

post = after_model.copy()
post["period"] = 1

combined = pd.concat([pre, post], axis=0, ignore_index=True)


def interaction_model(df, y_col, label):
    """Run the pooled model using complete rows for the selected pollutant only."""
    required = ["pedestrian_scaled", "bicycle_scaled", "period", y_col]
    model_df = df[required].dropna().copy()

    if len(model_df) < 6 or model_df["period"].nunique() < 2:
        print(f"\n{label}")
        print(
            "Not enough complete pre/post rows to run the pooled interaction model. "
            f"Rows available: {len(model_df)}; periods available: {model_df['period'].nunique()}"
        )
        return None

    X = model_df[["pedestrian_scaled", "bicycle_scaled", "period"]].astype(float).copy()
    X["ped_period"] = X["pedestrian_scaled"] * X["period"]
    X["bike_period"] = X["bicycle_scaled"] * X["period"]
    X = sm.add_constant(X, has_constant="add")
    y = model_df[y_col].astype(float)

    return sm.OLS(y, X).fit(cov_type="HC3")


for pollutant, label in [("pm10", "Pooled Interaction — PM10"), ("pm25", "Pooled Interaction — PM2.5")]:
    model = interaction_model(combined, pollutant, label)
    if model is not None:
        print(f"\n{label}")
        print(model.summary())
```

<div class="usecase-body">

Looking at the model of PM10, the pedestrian variable shows a small coefficient and is not significant. There is no real effect on PM10, consistent with earlier models. Meanwhile, bicycle shows positive and significant with p < 0.001. More bicycles are associated with higher PM10. The period dummy shows an insignificant (p = 0.516). The baseline PM10 level is not dramatically different once controlling for activity. Considering the interaction terms, there is no change in pedestrian effect after COVID as the ped_period is not significant enough (p = 0.614). However, bike_period is positive and significant, meaning the effect of bicycles on PM10 became stronger after COVID. Similar to the PM10 model, the pedestrian variable also has no effect on PM2.5. Bicycle has a small relationship with PM2.5 as a weak positive coefficient and borderline significance existed. The period dummy variable again is not significant (p = 0.659), indicating average PM2.5 levels did not shift much across periods once controlling for activity. Considering interaction terms, ped_period is not significant enough with p = 0.653, meaning no change in pedestrian effect. Meanwhile, it also shows no clear evidence that bicycle and PM2.5 relationship changed. In conclusion, cycling is linked to higher PM10 overall, and this effect increased after COVID. PM2.5 remains mostly unaffected by pedestrian or cycling activity, though bicycles show a small weak association that does not materially change after COVID.

</div>

<div class="usecase-body">

Even though it is expected that increasing bicycle activity should improve air quality or reduce PM10 and PM2.5 in the air, both models show that the bicycle variable has a positive relationship with both PM10 and PM2.5. This does not mean bikes cause pollution directly, but it likely reflects confounding factors. High bicycle activity and high pollution could both occur on the same days. Dry conditions increase road dust resuspension (PM10), while also making cycling more appealing. Moreover, cycling often occurs alongside road traffic. More bikes may coincide with busier traffic periods, meaning higher background emissions from vehicles. Due to these possible confounding factors, further research is required to truly understand whether there is any relationship between both pedestrian and bicycle activity and air quality indicators. A recommendation for further research is comparing pollution levels on high-cycling streets vs lower-cycling streets, rather than just citywide averages. This could show whether pollution is driven by traffic exposure rather than cycling activity itself.

</div>
