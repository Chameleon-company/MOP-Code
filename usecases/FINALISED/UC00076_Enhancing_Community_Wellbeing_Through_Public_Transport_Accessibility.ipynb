<div class="usecase-section-heading"><b>2. Downloading Datasets</b></div>

<div class="usecase-body">
This section automates the process of downloading the datasets required for the analysis from the City of Melbourne Open Data Portal.
</div>


```python
def download_dataset(
    api_key,
    dataset_id,
    base_url="https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/"
):
    """
    Download a dataset from the City of Melbourne Open Data Portal.

    Args:
        api_key (str): API key for authentication.
        dataset_id (str): Dataset identifier.
        base_url (str): Base API endpoint.

    Returns:
        pd.DataFrame: Downloaded dataset, or None if the request fails.
    """
    file_format = "csv"
    url = f"{base_url}{dataset_id}/exports/{file_format}"

    params = {
        "select": "*",
        "limit": -1,
        "lang": "en",
        "timezone": "UTC",
        "api_key": api_key
    }

    with requests.get(url, params=params, stream=True) as response:
        if response.status_code == 200:
            total_size = int(response.headers.get("content-length", 0))
            chunk_size = 1024

            progress_bar = tqdm(
                total=total_size,
                unit="iB",
                unit_scale=True,
                desc=f"Downloading {dataset_id}"
            )

            content = bytearray()

            for chunk in response.iter_content(chunk_size=chunk_size):
                if chunk:
                    content.extend(chunk)
                    progress_bar.update(len(chunk))

            progress_bar.close()

            data = pd.read_csv(
                StringIO(content.decode("utf-8")),
                delimiter=";"
            )

            return data

        print(f"Request failed with status code {response.status_code}")
        return None


# Dataset IDs
dataset_ids = [
    "social-indicators-for-city-of-melbourne-residents-2023",
    "bus-stops",
    "tram-tracks"
]

# Initialise a dictionary to hold the datasets
datasets = {}

# Download each dataset
for dataset_id in dataset_ids:
    datasets[dataset_id] = download_dataset("", dataset_id)

    if datasets[dataset_id] is not None:
        print(f"{dataset_id} downloaded successfully.")


# Store each dataset as a DataFrame
social_indicators_df = datasets[
    "social-indicators-for-city-of-melbourne-residents-2023"
]  # Dataset 1

bus_stops_df = datasets["bus-stops"]  # Dataset 2

tram_tracks_df = datasets["tram-tracks"]  # Dataset 3
```

<div class="usecase-body">
3.1.5 Seperating the respondent_group into age_group and location
</div>

```python
# 5. Separate respondent_group into age_group and location

# 1. Create age_group (keep values containing 'years')
social_indicators_df['age_group'] = social_indicators_df['respondent_group'].apply(
    lambda x: x if 'years' in str(x).lower() else None
)

# 2. Create location (ONLY actual Melbourne suburbs)
location_keywords = [
    'city of melbourne',
    'melbourne 3000',
    'south yarra',
    'docklands 3008',
    'south wharf',
    'southbank',
    'carlton 3053',
    'east melbourne 3002',
    'north melbourne',
    'west melbourne',
    'kensington',
    'flemington',
    'parkville 3052'
]

def identify_location(x):
    if pd.isna(x):
        return None
    x = str(x).strip().lower()
    
    # Special case: combine north melbourne and west melbourne
    if 'north melbourne' in x and 'west melbourne' in x:
        return 'north melbourne 3051'
    
    if any(keyword in x for keyword in location_keywords):
        return x
    return None

social_indicators_df['location'] = social_indicators_df['respondent_group'].apply(identify_location)

# 3. Drop the original respondent_group column
social_indicators_df.drop('respondent_group', axis=1, inplace=True)

print("Created 'age_group' and 'location' columns")
print("Dropped 'respondent_group' column")

# Display cleaned data
display(social_indicators_df.head(500))
```

<div class="usecase-body">
3.1.6 Geocoding the locations
</div>

```python
# 6. Geocoding locations (with rate limiting to prevent timeouts)
print("Geocoding locations (this may take a while)...")

# Initialise geolocator
geolocator = Nominatim(user_agent="my_geocoder")

def get_lat_lon(address):
    """
    Get latitude and longitude for a given address.

    Args:
        address (str): Location address

    Returns:
        tuple: (latitude, longitude) or (None, None) if failed
    """
    if pd.isna(address) or address is None:
        return None, None

    try:
        location = geolocator.geocode(address)

        if location:
            return location.latitude, location.longitude
        else:
            return None, None

    except Exception as e:
        print(f"   Warning: Error geocoding '{address}': {e}")
        return None, None


# Create temporary columns for latitude and longitude
social_indicators_df['latitude'] = None
social_indicators_df['longitude'] = None

# Geocode ONLY valid locations (skip age groups)
valid_locations = (
    social_indicators_df['location'].notna() &
    ~social_indicators_df['location'].str.contains('years', na=False)
)

locations_to_geocode = social_indicators_df.loc[
    valid_locations,
    'location'
]

# Geocode with rate limiting
for idx, address in locations_to_geocode.items():
    if pd.notna(address) and address is not None:
        lat, lon = get_lat_lon(address)

        social_indicators_df.at[idx, 'latitude'] = lat
        social_indicators_df.at[idx, 'longitude'] = lon

        time.sleep(1)

print("Geocoding complete!")

# DROP rows with missing coordinates (age groups and failed geocoding)
social_indicators_df = social_indicators_df.dropna(
    subset=['latitude', 'longitude']
)

# DROP rows with 0,0 coordinates (if any)
social_indicators_df = social_indicators_df[
    (social_indicators_df['latitude'] != 0) &
    (social_indicators_df['longitude'] != 0)
]

print(
    f"\nRows remaining after dropping missing/0,0 coordinates: "
    f"{len(social_indicators_df)}"
)

print("\nFINAL CLEANED SOCIAL INDICATORS DATASET")
display(social_indicators_df.head())

print(
    f"\nSocial Indicators cleaning complete! "
    f"Shape: {social_indicators_df.shape}"
)
```

<div class="usecase-section-heading"><b>4. Data Integration</b></div>

<div class="usecase-body">
This section combines the cleaned datasets to create a unified dataset for analysis. The integration process includes:
</div>

<div class="usecase-list">

- <b>Linking Social Indicators with geographic coordinates:</b> Using the geocoded respondent locations prepared during data cleaning.

- <b>Calculating public transport proximity:</b> Measuring the distance from each respondent location to the nearest bus stop and tram infrastructure.

- <b>Creating an integrated analysis dataset:</b> Combining respondent information, geographic coordinates, and public transport accessibility measures.

</div>

<div class="usecase-body">
The integrated dataset is then used for statistical and spatial analysis to examine the relationship between public transport accessibility and community wellbeing.
</div>

```python
social_indicators_df = social_indicators_df.copy()

# Split geo_point_2d into latitude and longitude for tram_tracks_df
tram_tracks_df[['latitude', 'longitude']] = (
    tram_tracks_df['geo_point_2d']
    .str.split(', ', expand=True)
    .astype(float)
)

# Split geo_point_2d into latitude and longitude for bus_stops_df
bus_stops_df[['latitude', 'longitude']] = (
    bus_stops_df['geo_point_2d']
    .str.split(', ', expand=True)
    .astype(float)
)


def calculate_nearest_distance(df, stops_df):
    distances = []

    for _, row in df.iterrows():

        # Skip rows with missing or 0,0 coordinates
        if (
            pd.isna(row['latitude'])
            or pd.isna(row['longitude'])
            or row['latitude'] == 0
            or row['longitude'] == 0
        ):
            distances.append(None)
            continue

        respondent_location = (
            row['latitude'],
            row['longitude']
        )

        min_distance = float('inf')

        for _, stop in stops_df.iterrows():

            # Skip transport locations with missing or 0,0 coordinates
            if (
                pd.isna(stop['latitude'])
                or pd.isna(stop['longitude'])
                or stop['latitude'] == 0
                or stop['longitude'] == 0
            ):
                continue

            stop_location = (
                stop['latitude'],
                stop['longitude']
            )

            distance = geodesic(
                respondent_location,
                stop_location
            ).meters

            if distance < min_distance:
                min_distance = distance

        distances.append(
            min_distance if min_distance != float('inf') else None
        )

    return distances


# Calculate distances to the nearest bus stop and tram infrastructure
social_indicators_df['nearest_bus_stop_distance'] = (
    calculate_nearest_distance(
        social_indicators_df,
        bus_stops_df
    )
)

social_indicators_df['nearest_tram_stop_distance'] = (
    calculate_nearest_distance(
        social_indicators_df,
        tram_tracks_df
    )
)


# Create an accessibility category based on average distance
social_indicators_df['accessibility'] = pd.cut(
    social_indicators_df[
        [
            'nearest_bus_stop_distance',
            'nearest_tram_stop_distance'
        ]
    ].mean(axis=1),
    bins=[0, 1000, 2000, 3000, float('inf')],
    labels=['Very Good', 'Good', 'Fair', 'Poor']
)
```

```python
# Ensure we're working on a clean copy
social_indicators_df = social_indicators_df.copy()

# Travel Time Estimates based on distances
def distance_to_travel_time(distance, average_speed_kmh=30):
    """
    Convert distance to travel time in seconds.
    If distance is None or invalid, return None.
    """
    if pd.isna(distance) or distance is None:
        return None
    return distance / (average_speed_kmh * 1000 / 3600)  # converting speed to metres per second

# Adding travel time estimates to the DataFrame
social_indicators_df['bus_stop_travel_time'] = social_indicators_df['nearest_bus_stop_distance'].apply(distance_to_travel_time)
social_indicators_df['tram_stop_travel_time'] = social_indicators_df['nearest_tram_stop_distance'].apply(distance_to_travel_time)
```

<b>5.4 Categorising transport distance</b><br>
Distances are grouped into three categories: Very Close (500 metres or less), Moderate (more than 500 metres and up to 1,000 metres), and Far (more than 1,000 metres). These categories make the continuous accessibility measures easier to understand and compare in the later analysis.

```python
# Categorise distance
def categorise_distance(distance):
    """
    Categorise distance as 'Very Close', 'Moderate', or 'Far'.

    Returns None if the distance is missing.
    """
    if distance is None or pd.isna(distance):
        return None

    if distance <= 500:
        return 'Very Close'
    elif distance <= 1000:
        return 'Moderate'
    else:
        return 'Far'


# Apply distance categories to bus and tram accessibility measures
social_indicators_df['bus_distance_category'] = (
    social_indicators_df['nearest_bus_stop_distance']
    .apply(categorise_distance)
)

social_indicators_df['tram_distance_category'] = (
    social_indicators_df['nearest_tram_stop_distance']
    .apply(categorise_distance)
)

# Display the updated dataset
display(social_indicators_df.head())
```

```python
# Linear regression analysis

# Select accessibility features and target variable
X = merged_df[
    [
        'nearest_bus_stop_distance',
        'nearest_tram_stop_distance'
    ]
]

y = merged_df['result']


# Remove rows with missing values before regression
combined = pd.concat([X, y], axis=1)
combined = combined.dropna()

X = combined[
    [
        'nearest_bus_stop_distance',
        'nearest_tram_stop_distance'
    ]
]

y = combined['result']


# Standardise the accessibility features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)


# Initialise and fit the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)


# Make predictions
y_pred = model.predict(X_test)


# Evaluate the model
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error: {mse:.2f}")
print(f"R-squared: {r2:.3f}")
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```

The Mean Squared Error was 781.56, further indicating substantial prediction error. Although the model estimated small coefficients for both accessibility measures, the poor overall model performance means these coefficients should not be interpreted as strong relationships. In relation to the scenario, this suggests that community wellbeing is likely influenced by factors beyond public transport proximity alone, and additional social, demographic, and spatial characteristics may need to be considered.

<div id="sub-section-6-10" class="usecase-body"><b>6.10 Compare distances across groups</b><br>ANOVA tests whether average bus and tram distances differ across locations and types. A significant result would indicate that at least one group has a different mean distance.</div>

```python
# ANOVA by location and respondent type

# Prepare data for location-based comparisons
anova_loc = merged_df[
    [
        'nearest_bus_stop_distance',
        'nearest_tram_stop_distance',
        'location'
    ]
].dropna()


# ANOVA 1: Bus-stop distance by location
model_loc_bus = ols(
    'nearest_bus_stop_distance ~ C(location)',
    data=anova_loc
).fit()

anova_loc_bus = sm.stats.anova_lm(
    model_loc_bus,
    typ=2
)

print("ANOVA for Bus Stop Distance by Location:")
display(anova_loc_bus)


# ANOVA 2: Tram-infrastructure distance by location
model_loc_tram = ols(
    'nearest_tram_stop_distance ~ C(location)',
    data=anova_loc
).fit()

anova_loc_tram = sm.stats.anova_lm(
    model_loc_tram,
    typ=2
)

print("ANOVA for Tram Infrastructure Distance by Location:")
display(anova_loc_tram)


# Prepare data for respondent-type comparisons
anova_type = merged_df[
    [
        'nearest_bus_stop_distance',
        'nearest_tram_stop_distance',
        'type'
    ]
].dropna()


# ANOVA 3: Bus-stop distance by respondent type
model_type_bus = ols(
    'nearest_bus_stop_distance ~ C(type)',
    data=anova_type
).fit()

anova_type_bus = sm.stats.anova_lm(
    model_type_bus,
    typ=2
)

print("ANOVA for Bus Stop Distance by Type:")
display(anova_type_bus)


# ANOVA 4: Tram-infrastructure distance by respondent type
model_type_tram = ols(
    'nearest_tram_stop_distance ~ C(type)',
    data=anova_type
).fit()

anova_type_tram = sm.stats.anova_lm(
    model_type_tram,
    typ=2
)

print("ANOVA for Tram Infrastructure Distance by Type:")
display(anova_type_tram)
```

In contrast, the respondent-type comparisons are not statistically significant for either bus-stop or tram-infrastructure distance (p = 1.00). <div id="sub-section-6-13" class="usecase-body"><b>6.13 Test for spatial autocorrelation</b><br>Moran’s I tests whether nearby observations have similar bus-stop distances. The maps then show where bus and tram distances are concentrated geographically.</div>

```python
# Spatial analysis - test spatial autocorrelation and visualise geographic patterns

# Aggregate repeated observations at the same geographic coordinates
spatial_data = (
    merged_df
    .groupby(
        ['longitude', 'latitude'],
        as_index=False
    )
    .agg(
        nearest_bus_stop_distance=(
            'nearest_bus_stop_distance',
            'mean'
        ),
        nearest_tram_stop_distance=(
            'nearest_tram_stop_distance',
            'mean'
        )
    )
)


# Create a GeoDataFrame using unique respondent locations
geo_df = gpd.GeoDataFrame(
    spatial_data,
    geometry=gpd.points_from_xy(
        spatial_data['longitude'],
        spatial_data['latitude']
    ),
    crs='EPSG:4326'
)

n_locations = len(geo_df)

print(f"Unique respondent locations: {n_locations}")


# Find the smallest k that creates one connected spatial network
for k in range(1, n_locations):

    # Suppress warnings only while testing candidate k values
    with warnings.catch_warnings():
        warnings.simplefilter('ignore')

        candidate_w = weights.KNN.from_dataframe(
            geo_df,
            k=k
        )

    if candidate_w.n_components == 1:
        w = candidate_w
        break


# Row-standardise the final connected weights matrix
w.transform = 'r'

print(f"K-nearest neighbours used: {k}")
print(f"Connected components: {w.n_components}")


# Calculate Moran's I for bus-stop distance
y = geo_df[
    'nearest_bus_stop_distance'
].to_numpy()

moran = Moran(
    y,
    w
)

print(f"Moran's I: {moran.I:.3f}")
print(
    f"Permutation p-value: "
    f"{moran.p_sim:.3f}"
)


# Visualise Moran's I
plot_moran(moran)

plt.suptitle(
    'Spatial Autocorrelation of Bus-stop Distance'
)

plt.tight_layout()
plt.show()


# Create a continuous colour map using Mollie's requested palette
accessibility_cmap = LinearSegmentedColormap.from_list(
    'accessibility_palette',
    [
        '#059669',  # Green
        '#9333EA',  # Purple
        '#F97316',  # Orange
        '#2563EB',  # Blue
        '#EAB308',  # Yellow
        '#EC4899'   # Pink
    ]
)


# Map distance to the nearest bus stop
fig, ax = plt.subplots(figsize=(6, 6))

geo_df.plot(
    ax=ax,
    column='nearest_bus_stop_distance',
    legend=True,
    cmap=accessibility_cmap,
    markersize=80,
    legend_kwds={
        'label':
        'Distance to Nearest Bus Stop (metres)'
    }
)

plt.title(
    'Distance to the Nearest Bus Stop'
)

plt.xlabel('Longitude')
plt.ylabel('Latitude')

plt.tight_layout()
plt.show()


# Map distance to the nearest tram infrastructure
fig, ax = plt.subplots(figsize=(6, 6))

geo_df.plot(
    ax=ax,
    column='nearest_tram_stop_distance',
    legend=True,
    cmap=accessibility_cmap,
    markersize=80,
    legend_kwds={
        'label':
        'Distance to Nearest Tram Infrastructure (metres)'
    }
)

plt.title(
    'Distance to the Nearest Tram Infrastructure'
)

plt.xlabel('Longitude')
plt.ylabel('Latitude')

plt.tight_layout()
plt.show()
```

<div class="usecase-body">
The spatial autocorrelation analysis was conducted using the nine unique respondent locations to avoid repeated coordinates disproportionately influencing the spatial weights structure. A connected spatial network was achieved using two nearest neighbours for each location.

Moran’s I was -0.189 with a permutation p-value of 0.451. This indicates a weak negative spatial pattern in bus-stop distance, but the result is not statistically significant. Therefore, there is insufficient evidence to conclude that bus-stop proximity is spatially clustered or dispersed across the locations represented in this dataset.

The accompanying maps nevertheless show that distances to bus stops and tram infrastructure vary geographically across the sampled Melbourne locations. Some locations are comparatively closer to public transport infrastructure than others. In relation to the scenario, this geographic variation remains useful for identifying differences in accessibility, although the Moran’s I result suggests that these differences do not form a statistically significant overall spatial pattern within the nine locations analysed.

The spatial findings should be interpreted cautiously because the analysis is based on only nine unique respondent locations, which limits the amount of independent geographic information available.

```python
# Advanced Modelling - Feature engineering, feature selection, PCA, and model evaluation

# Data Preparation
features = ['nearest_bus_stop_distance', 'nearest_tram_stop_distance']
target = 'result'

# Select features and target variable
X = merged_df[features]
y = merged_df[target]

# Impute missing values
imputer = SimpleImputer(strategy='mean')
X_imputed = imputer.fit_transform(X)

# Create interaction term
merged_df['bus_tram_interaction'] = merged_df['nearest_bus_stop_distance'] * merged_df['nearest_tram_stop_distance']

# Apply log transformation
log_transformer = FunctionTransformer(np.log1p, validate=True)
merged_df[['log_bus_distance', 'log_tram_distance']] = log_transformer.fit_transform(
    merged_df[['nearest_bus_stop_distance', 'nearest_tram_stop_distance']]
)

# Feature Selection using RFE
rfe = RFE(estimator=LinearRegression(), n_features_to_select=2)
rfe.fit(X_imputed, y)
print("Selected Features:", rfe.support_)
print("Feature Ranking:", rfe.ranking_)

# PCA (Principal Component Analysis)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_imputed)
print(f'Explained variance ratio: {pca.explained_variance_ratio_}')

# Grid Search for Ridge and Lasso
param_grid = {'alpha': [1, 10, 100, 1000]}

grid_ridge = GridSearchCV(Ridge(), param_grid, cv=5)
grid_ridge.fit(X_imputed, y)
print(f'Best Ridge Alpha: {grid_ridge.best_params_["alpha"]}')

grid_lasso = GridSearchCV(Lasso(max_iter=10000), param_grid, cv=5)
grid_lasso.fit(X_imputed, y)
print(f'Best Lasso Alpha: {grid_lasso.best_params_["alpha"]}')

# Train-Test Split and Model Evaluation
X_train, X_test, y_train, y_test = train_test_split(X_imputed, y, test_size=0.2, random_state=42)

# Scale features for Ridge and Lasso
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

models = {
    'Random Forest': RandomForestRegressor(n_estimators=100, random_state=42),
    'Gradient Boosting': GradientBoostingRegressor(n_estimators=100, random_state=42),
    'Ridge': Ridge(alpha=grid_ridge.best_params_["alpha"]),
    'Lasso': Lasso(alpha=grid_lasso.best_params_["alpha"])
}

print("\nModel Performance (MSE):")
print("-" * 40)
for name, model in models.items():
    # Use scaled data for Ridge and Lasso, original for tree-based models
    if name in ['Ridge', 'Lasso']:
        X_train_use = X_train_scaled
        X_test_use = X_test_scaled
    else:
        X_train_use = X_train
        X_test_use = X_test
    
    model.fit(X_train_use, y_train)
    predictions = model.predict(X_test_use)
    mse = mean_squared_error(y_test, predictions)
    r2 = r2_score(y_test, predictions)
    print(f'{name} - MSE: {mse:.2f}, R²: {r2:.4f}')

# Histograms
merged_df[['nearest_bus_stop_distance', 'nearest_tram_stop_distance', 'bus_tram_interaction']].hist(figsize=(12, 6))
plt.suptitle('Distribution of Transport Accessibility Features')
plt.show()

# Pair Plot
sns.pairplot(merged_df[['nearest_bus_stop_distance', 'nearest_tram_stop_distance', 'bus_tram_interaction']])
plt.suptitle('Pairwise Relationships Between Transport Features')
plt.show()

# Scatter plot: Bus Stop Distance vs. Result
plt.figure(figsize=(10, 6))
sns.scatterplot(x='nearest_bus_stop_distance', y='result', data=merged_df, color='#059669')
plt.title('Bus Stop Distance vs. wellbeing Indicator')
plt.xlabel('Distance to Nearest Bus Stop (metres)')
plt.ylabel('wellbeing Indicator Result (%)')
plt.show()

# Scatter plot: Tram Stop Distance vs. Result
plt.figure(figsize=(10, 6))
sns.scatterplot(x='nearest_tram_stop_distance', y='result', data=merged_df, color='#9333EA')
plt.title('Tram Stop Distance vs. wellbeing Indicator')
plt.xlabel('Distance to Nearest Tram Stop (metres)')
plt.ylabel('wellbeing Indicator Result (%)')
plt.show()
```

<div class="usecase-section-heading"><b>8. Recommendations</b></div>

<div class="usecase-body">
This section converts the analysis into practical areas for attention. The recommendations are generated from the accessibility score, health outcome, and spatial location so the reader can see how each recommendation follows from the preceding analysis.
</div>


<div id="sub-section-8-1" class="usecase-body"><b>8.1 Identify lower-accessibility areas</b><br>The accessibility score combines bus and tram distance into a single measure. The zero-distance safeguard prevents a division-by-zero error before the score is normalised.</div>

```python
# Replace 0 distances with small value to avoid division by zero
social_indicators_df['nearest_bus_stop_distance'] = social_indicators_df['nearest_bus_stop_distance'].replace(0, 1e-6)
social_indicators_df['nearest_tram_stop_distance'] = social_indicators_df['nearest_tram_stop_distance'].replace(0, 1e-6)

# Calculate accessibility score based on inverse of combined distances
social_indicators_df['accessibility_score'] = 1 / (social_indicators_df['nearest_bus_stop_distance'] + social_indicators_df['nearest_tram_stop_distance'])

# Normalise the accessibility score to a range between 0 and 1
social_indicators_df['accessibility_score'] = (social_indicators_df['accessibility_score'] - social_indicators_df['accessibility_score'].min()) / \
                                      (social_indicators_df['accessibility_score'].max() - social_indicators_df['accessibility_score'].min())

print(social_indicators_df)
```

<div id="sub-section-8-2" class="usecase-body"><b>8.2 Filter priority areas</b><br>The 0.3 threshold isolates observations with lower accessibility scores. These observations become the starting point for the recommendation logic.</div>

```python
# Set a threshold for poor accessibility
threshold = 0.3

# Filter areas with poor accessibility
poor_accessibility_df = social_indicators_df[social_indicators_df['accessibility_score'] < threshold]

print("Areas with Poor Accessibility:")
print(poor_accessibility_df)
```

<div id="sub-section-8-3" class="usecase-body"><b>8.3 Connect accessibility with health outcomes</b><br>The priority observations are merged with health indicators so that recommendations can consider transport access and community wellbeing together.</div>

```python
# Merge poor accessibility data with health data
poor_accessibility_unique = poor_accessibility_df.drop_duplicates(subset=['location'])
social_indicators_unique = social_indicators_df.drop_duplicates(subset=['location'])

merged_df_2 = pd.merge(
    poor_accessibility_unique,
    social_indicators_unique,
    on='location',
    how='inner'
)

print("Merged Data with Health Indicators:")
print(merged_df_2)
```

<div id="sub-section-8-4" class="usecase-body"><b>8.4 Measure the strength of the relationship</b><br>Correlations between transport distances and the health outcome provide a statistical check of whether poorer access is associated with different health results.</div>

```python
# Use all rows from merged_df_2 (no filter)
health_data = merged_df_2.dropna(subset=['nearest_bus_stop_distance_y', 'nearest_tram_stop_distance_y', 'result_x'])

print(f"Found {len(health_data)} rows for analysis")

if len(health_data) > 1:
    # Correlation between distances and health outcomes
    correlation_bus = health_data['nearest_bus_stop_distance_y'].corr(health_data['result_x'])
    correlation_tram = health_data['nearest_tram_stop_distance_y'].corr(health_data['result_x'])
    
    print(f"\nCorrelation between bus stop distance and health outcome: {correlation_bus:.4f}")
    print(f"Correlation between tram stop distance and health outcome: {correlation_tram:.4f}")
    
    # Quick interpretation
    if abs(correlation_bus) > 0.5:
        print(f"Bus distance has a strong {'positive' if correlation_bus > 0 else 'negative'} correlation with health outcomes")
    elif abs(correlation_bus) > 0.3:
        print(f"Bus distance has a moderate {'positive' if correlation_bus > 0 else 'negative'} correlation with health outcomes")
    else:
        print("Bus distance has a weak correlation with health outcomes")
else:
    print("Not enough data to compute correlation")
```

```python
# Select features
X = health_data[
    [
        'nearest_bus_stop_distance_y',
        'nearest_tram_stop_distance_y',
        'accessibility_score_y'
    ]
]

y = health_data['result_x']

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = sm.add_constant(X_scaled)

# Fit the regression model
model = sm.OLS(y, X_scaled).fit()

# Display regression results without the invalid small-sample diagnostic warning
with warnings.catch_warnings():
    warnings.filterwarnings(
        'ignore',
        message='omni_normtest is not valid with less than 8 observations'
    )
    print(model.summary(slim=True))
```

<div id="sub-section-8-5" class="usecase-body"><b>8.5 Visualise the relationship directly</b><br>The scatter plot provides a simple visual check of the relationship between bus-stop distance and the health outcome, complementing the regression statistics.</div>

```python

#visualisation

# Scatter plot for bus stop distance vs life satisfaction
plt.scatter(health_data['nearest_bus_stop_distance_y'], health_data['result_x'])
plt.title('Bus Stop Distance vs Life Satisfaction')
plt.xlabel('Bus Stop Distance (metres)')
plt.ylabel('Life Satisfaction Score')
plt.show()
```

<div id="sub-section-8-6" class="usecase-body"><b>8.6 Compare the combined accessibility score with health outcomes</b><br>This focuses on the overall accessibility score rather than one transport mode at a time, linking the recommendation logic back to the central scenario.</div>

```python
# Calculate correlation between accessibility score and health indicators
correlation_matrix = merged_df_2[['accessibility_score_y', 'result_x']].corr()
print("Correlation between Accessibility Score and Health Outcome:")
print(correlation_matrix)
```

<div id="sub-section-8-7" class="usecase-body"><b>8.7 Generate recommendations from combined indicators</b><br>Priority areas are filtered using the accessibility threshold and then compared with the median health score. This creates a consistent rule for generating location-level recommendations.</div>

```python
# Filter Data Based on Accessibility and Health Scores

# Create a DataFrame for areas with poor accessibility
poor_accessibility_df = merged_df_2[merged_df_2['accessibility_score_y'] < 0.3].copy()

# Calculate the median health score for comparison
median_health_score = merged_df_2['result_x'].median()

poor_accessibility_df['health_score'] = pd.to_numeric(poor_accessibility_df['result_x'], errors='coerce')

# Create recommendations based on accessibility and health scores
recommendations = []

for index, row in poor_accessibility_df.iterrows():
    area = row['topic_x'] 
    accessibility_score = row['accessibility_score_y']
    health_score = row['health_score']
    
    # Logic for generating recommendations
    if accessibility_score < 0.3 and health_score < median_health_score:
        recommendations.append({
            'area': area,
            'recommendation': 'URGENT: Expand bus and tram routes, increase service frequency, and improve infrastructure for vulnerable groups.'
        })
    elif accessibility_score < 0.5:
        recommendations.append({
            'area': area,
            'recommendation': 'PRIORITY: Promote active transportation by developing cycling and walking paths, and integrate transport with health services.'
        })
    else:
        recommendations.append({
            'area': area,
            'recommendation': 'LOW: Monitor and evaluate current transport services and consider pilot programs for improvements.'
        })

recommendations_df = pd.DataFrame(recommendations)

print("Recommendations Summary:")
print(recommendations_df)
print("\nColumns in recommendations_df:")
print(recommendations_df.columns)
```

<div id="sub-section-8-8" class="usecase-body"><b>8.8 Attach coordinates to the recommendations</b><br>The generated recommendations are matched back to location coordinates so they can be displayed spatially.</div>

```python
# Added lat, long to recommendations dataframe
recommendations_df['area_key'] = recommendations_df['area'].str.lower().str.strip()
social_indicators_df['area_key'] = social_indicators_df['topic'].str.lower().str.strip()

# Merge the DataFrames based on the common key
final_recommendations_df = pd.merge(
    recommendations_df,
    social_indicators_df[['area_key', 'latitude', 'longitude']],
    on='area_key',
    how='left'
)

final_recommendations_df = final_recommendations_df.drop(columns=['area_key'])

print(final_recommendations_df)

# Check how many rows got valid coordinates
print("\n--- Merge Summary ---")
print(f"Total recommendations: {len(final_recommendations_df)}")
print(f"Rows with valid latitude: {final_recommendations_df['latitude'].notna().sum()}")
print(f"Rows with valid longitude: {final_recommendations_df['longitude'].notna().sum()}")
```
