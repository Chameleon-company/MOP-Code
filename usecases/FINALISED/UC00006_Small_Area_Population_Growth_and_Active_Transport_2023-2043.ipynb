<div id="section-1" class="usecase-section-heading"><b>1. Setting up and loading the data</b></div>

<div class="usecase-body">We import the libraries, then download all four datasets straight from the API.</div>

<div id="sub-1-1" class="usecase-sub-section-heading"><b>1.1. Importing the libraries</b></div>

<div class="usecase-body">pandas handles the tables, GeoPandas and Shapely handle anything with a shape or location, folium draws the interactive map, and matplotlib and seaborn draw the charts.</div>

```python
import json                         # reading shapes stored as GeoJSON text
import time                         # short pauses between download retries
from io import StringIO             # lets pandas read the API response as if it were a file

import numpy as np                  # numbers and arrays
import pandas as pd                 # tables
import geopandas as gpd             # tables that understand location
import matplotlib.pyplot as plt     # charts
from matplotlib.ticker import FuncFormatter
import seaborn as sns               # nicer looking charts
import requests                     # calling the API
import folium                       # interactive maps
import branca.colormap as cm        # colour scale for the map
from shapely.geometry import shape  # turns GeoJSON text into a shape

# The colour order used throughout this notebook, taken from the MOP style guide.
GREEN, PURPLE, ORANGE, BLUE, YELLOW = "#059669", "#9333EA", "#F97316", "#2563EB", "#EAB308"

# Each lane protection level keeps the same colour in the chart and on the map.
PROTECTION_COLOURS = {
    "Protected": GREEN,              # physically separated from traffic
    "Painted (Buffered)": PURPLE,    # painted lane with a painted buffer
    "Painted (Unbuffered)": ORANGE,  # painted lane only
    "Peak": BLUE,                    # lane that only operates at peak times
}

# A Coordinate Reference System (CRS) is the system used to say where
# something is on Earth. EPSG numbers are the catalogue codes for them.
CRS_DEGREES = "EPSG:4326"   # latitude and longitude, as the API supplies them
CRS_METRES = "EPSG:7855"    # Melbourne's local grid, measured in metres

FIRST_YEAR, LAST_YEAR = 2023, 2043   # the forecast period this use case covers

sns.set_style("whitegrid")
thousands = FuncFormatter(lambda value, _: f"{value:,.0f}")   # 12000 -> 12,000 on chart axes
```

<div id="sub-1-2" class="usecase-sub-section-heading"><b>1.2. Using the API to call the data from the Melbourne Open Data Site</b></div>

<div class="usecase-body">
One function handles all four downloads. Downloads can fail for many reasons, so the function deals with the common ones clearly instead of crashing with a confusing message:
</div>

<div class="usecase-list">

- <b>Network problems</b> such as a timeout or a dropped connection are retried up to three times, with a short wait in between.
- <b>Requests the server rejects</b>, such as a dataset that has been renamed or removed, are not retried, because trying again will not help.
- <b>Data that has changed shape</b> is caught by checking that the columns this notebook needs are still there.

</div>

<div class="usecase-body">
In each case the error names the dataset and links to its page, so the reader knows where to look.
</div>

```python
def fetch_melbourne_dataset(dataset_id, required_columns, retries=3):
    """Download a full City of Melbourne dataset and return it as a DataFrame.

    dataset_id is the name shown in the dataset's URL. required_columns lists
    the columns this notebook relies on, so a changed dataset is caught early.
    """
    base_url = "https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/"
    export_url = f"{base_url}{dataset_id}/exports/csv"
    dataset_page = f"https://data.melbourne.vic.gov.au/explore/dataset/{dataset_id}/"
    params = {"select": "*", "limit": -1, "lang": "en", "timezone": "UTC"}   # -1 = every record

    for attempt in range(1, retries + 1):
        try:
            response = requests.get(export_url, params=params, timeout=120)
            response.raise_for_status()   # turns an error reply (such as 404) into an exception
            break
        except requests.RequestException as error:
            print(f"Attempt {attempt} of {retries} to download '{dataset_id}' failed: {error}")
            # A 4xx reply means the request itself is wrong, so retrying will not help.
            rejected = error.response is not None and 400 <= error.response.status_code < 500
            if rejected or attempt == retries:
                raise RuntimeError(
                    f"Could not download '{dataset_id}'. Check your internet connection "
                    f"and whether the dataset is still available at {dataset_page}"
                ) from error
            time.sleep(5 * attempt)   # wait a little longer before each new attempt

    # The export endpoint returns semicolon-separated CSV.
    try:
        data = pd.read_csv(StringIO(response.content.decode("utf-8")), delimiter=";")
    except pd.errors.EmptyDataError:
        data = pd.DataFrame()

    missing = sorted(set(required_columns) - set(data.columns))
    if data.empty or missing:
        raise ValueError(
            f"'{dataset_id}' downloaded but is empty or missing the columns {missing}. "
            f"The dataset may have changed; see {dataset_page}"
        )
    return data
```

<div class="usecase-body">A quick test shows what a failure looks like. We ask for a dataset that does not exist and catch the error, so the notebook carries on.</div>

```python
try:
    fetch_melbourne_dataset("a-dataset-that-does-not-exist", required_columns=[])
except RuntimeError as error:
    print(f"\nHandled: {error}")
```

<div class="usecase-body">
<b>What this shows:</b> the server rejects the request, the function stops after one attempt instead of retrying, and the error explains what to check.
</div>

<div class="usecase-body">
<b>Why it matters:</b> a notebook that fails with a clear message is far easier to fix than one that fails several cells later with an unrelated error.
</div>

<div class="usecase-body">Now we load the four real datasets. This usually takes less than a minute.</div>

```python
DATASETS = {
    "Population forecasts": ("city-of-melbourne-population-forecasts-by-small-area-2020-2040",
                             ["geography", "year", "gender", "age", "value"]),
    "Small area boundaries": ("small-areas-for-census-of-land-use-and-employment-clue",
                              ["featurenam", "geo_shape"]),
    "Cycle lanes": ("bicycle-routes-including-informal-on-road-and-off-road-routes",
                    ["clueregion", "protection", "lanetype", "length", "json_geometry"]),
    "Bike counts": ("annual-bike-counts-super-tuesday",
                    ["site_id", "description", "latitude", "longitude", "year", "total"]),
}

raw = {name: fetch_melbourne_dataset(dataset_id, columns)
       for name, (dataset_id, columns) in DATASETS.items()}

for name, table in raw.items():
    print(f"{name:<22}: {len(table):>6,} rows x {table.shape[1]} columns")

population_raw = raw["Population forecasts"]
small_areas_raw = raw["Small area boundaries"]
cycle_lanes_raw = raw["Cycle lanes"]
bike_counts_raw = raw["Bike counts"]
```

<div class="usecase-body">
<b>What this shows:</b> the size of each dataset.
</div>

<div class="usecase-body">
<b>Why it matters:</b> the population table has one row per area, year, gender and age group, so it has thousands of rows even though there are only 13 areas. Section 3 adds those rows up.
</div>

<div id="section-2" class="usecase-section-heading"><b>2. Checking data quality</b></div>

<div class="usecase-body">Before analysing anything we check the columns we rely on for duplicates and gaps, and look at a few things that affect how the results should be read.</div>

```python
for name, (_, columns) in DATASETS.items():
    table = raw[name]
    missing = table[columns].isna().sum()
    missing_text = ", ".join(f"{column} ({count:,})" for column, count in missing[missing > 0].items()) or "none"
    print(f"{name:<22} duplicate rows: {table.duplicated().sum()}   missing values: {missing_text}")

# Population: the forecast period, and whether the 13 areas add up to the city total.
area_totals = population_raw.groupby(["geography", "year"])["value"].sum().unstack()
city_total = area_totals.loc["City of Melbourne"]
sum_of_areas = area_totals.drop(index="City of Melbourne").sum()
print(f"\nForecast years: {population_raw['year'].min()} to {population_raw['year'].max()}")
print(f"Largest gap between the city total and the sum of its areas: {(city_total - sum_of_areas).abs().max():.0f} people")
print(f"Residents in {FIRST_YEAR} for the two smallest areas:")
print(area_totals[FIRST_YEAR].nsmallest(2).to_string())

# Cycle lanes: how many segments have no protection level recorded, and what they are.
no_level = cycle_lanes_raw[cycle_lanes_raw["protection"].isna()]
print(f"\nLane segments with no protection level: {len(no_level):,} of {len(cycle_lanes_raw):,}")
print(no_level["lanetype"].value_counts().to_string())

# Bike counts: which years were counted, and whether the same sites were used each year.
sites_per_year = bike_counts_raw.groupby("year")["site_id"].nunique()
print(f"\nBike count years: {sites_per_year.index.min()} to {sites_per_year.index.max()}")
print(f"Sites counted per year: between {sites_per_year.min()} and {sites_per_year.max()}")
```

<div class="usecase-body">
<b>What this shows:</b> there are no duplicate rows and very few gaps. Two cycle lane segments have no shape, about a third have no protection level, and one bike count record has no site name, which only affects its label. The checks also point to a few limits worth keeping in mind:
</div>

<div class="usecase-list">

- <b>The population figures are forecasts</b>, not head counts, so they are best estimates that will change as plans change. The 13 areas add up to the city total to within a handful of people, which is just rounding.
- <b>Two areas have almost no residents in 2023.</b> Port Melbourne (the Fishermans Bend renewal area) starts from single figures, and West Melbourne (Industrial) has none at all. Any percentage growth from such a small start is meaningless, so both are left out of the growth rate comparisons.
- <b>Not every cycle lane segment is a lane you can ride in.</b> Segments with no protection level are chevron markings, bike boxes at intersections and unsignposted refuge lanes. We keep only the segments with a protection level.
- <b>The bike counts are old and narrow.</b> They stop in 2017 and cover one two-hour morning count a year, so they reflect commuter riding rather than all cycling. The set of sites also changes slightly between years, so section 5.2 compares years using only the sites counted every time.

</div>

<div class="usecase-body">
<b>Why it matters:</b> none of these problems stop the analysis, but each one limits how far the results can be pushed. The conclusion comes back to them.
</div>

<div id="section-3" class="usecase-section-heading"><b>3. Preparing the data for analysis</b></div>

<div class="usecase-body">We turn each dataset into a clean table: one population total per area and year, one shape per area, the riding lanes, and the bike count sites tagged with their area.</div>

<div id="sub-3-1" class="usecase-sub-section-heading"><b>3.1. Adding up the population for each area</b></div>

<div class="usecase-body">The forecast splits every area and year by gender and age group. Adding those rows together gives one total per area per year. The <i>City of Melbourne</i> rows describe the whole municipality, so we keep them separately.</div>

```python
assert {FIRST_YEAR, LAST_YEAR} <= set(population_raw["year"]), "The forecast no longer covers 2023 to 2043"

FORECAST_STEPS = list(range(FIRST_YEAR, LAST_YEAR + 1, 5))   # 2023, 2028, 2033, 2038, 2043

is_city = population_raw["geography"] == "City of Melbourne"
city_population = population_raw[is_city].groupby("year")["value"].sum()

# One row per small area and one column per year.
population = population_raw[~is_city].groupby(["geography", "year"])["value"].sum().unstack()
population.index.name = "small_area"

# Areas that start too small for a percentage change to mean anything (see section 2).
TINY_AREAS = population.index[population[FIRST_YEAR] < 100].tolist()

print(f"{len(population)} small areas, {population.shape[1]} forecast years")
print(f"City of Melbourne: {city_population[FIRST_YEAR]:,} residents in {FIRST_YEAR}, "
      f"{city_population[LAST_YEAR]:,} in {LAST_YEAR}")
print("Left out of growth rate comparisons:", ", ".join(TINY_AREAS))
```

<div class="usecase-body">
<b>What this shows:</b> the size of the tidy population table, the city-wide forecast, and the two areas set aside for growth rates.
</div>

<div class="usecase-body">
<b>Why it matters:</b> every later figure uses this table, so it needs exactly one number per area and year.
</div>

<div id="sub-3-2" class="usecase-sub-section-heading"><b>3.2. Reading the small area boundaries</b></div>

<div class="usecase-body">Each boundary is stored as GeoJSON text. <code>json.loads</code> reads the text and <code>shape</code> turns it into a polygon that GeoPandas can draw. These boundaries use the same area names as the population forecast, so the two join without any renaming.</div>

```python
small_areas = gpd.GeoDataFrame(
    small_areas_raw[["featurenam"]].rename(columns={"featurenam": "small_area"}),
    geometry=[shape(json.loads(text)) for text in small_areas_raw["geo_shape"]],
    crs=CRS_DEGREES,
)

unmatched = set(population.index) ^ set(small_areas["small_area"])
print(f"Boundaries read: {len(small_areas)}")
print(f"Names that differ from the population data: {unmatched or 'none'}")
```

<div class="usecase-body">
<b>What this shows:</b> that all 13 boundaries were read and every name matches the population data.
</div>

<div class="usecase-body">
<b>Why it matters:</b> a name that did not match would quietly drop an area from the map.
</div>

<div id="sub-3-3" class="usecase-sub-section-heading"><b>3.3. Keeping the riding lanes</b></div>

<div class="usecase-body">We keep only segments that have a protection level and a shape (see section 2). The lane data writes a few area names without brackets, for example <i>Melbourne CBD</i>, so we change them to match the population data.</div>

<div class="usecase-body">The <code>length</code> column is in metres. Each lane is its own segment, so a street with a lane on both sides counts twice. The figures below are therefore lane length, not street length.</div>

```python
RIDING_LANE_LEVELS = list(PROTECTION_COLOURS)   # Protected, Painted (Buffered), Painted (Unbuffered), Peak

keep = cycle_lanes_raw["protection"].isin(RIDING_LANE_LEVELS) & cycle_lanes_raw["json_geometry"].notna()
lanes = cycle_lanes_raw[keep].copy()

# "Melbourne CBD" -> "Melbourne (CBD)", "West Melbourne Residential" -> "West Melbourne (Residential)", and so on.
lanes["small_area"] = lanes["clueregion"].str.replace(
    r"^(Melbourne|West Melbourne) (CBD|Remainder|Residential|Industrial)$", r"\1 (\2)", regex=True)

cycle_lanes = gpd.GeoDataFrame(
    lanes[["small_area", "protection", "length"]],
    geometry=[shape(json.loads(text)) for text in lanes["json_geometry"]],
    crs=CRS_DEGREES,
)

print(f"Riding lane segments kept: {len(cycle_lanes):,} of {len(cycle_lanes_raw):,}")
print(f"Total riding lane length : {cycle_lanes['length'].sum() / 1000:.0f} km")
print(f"Area names that still differ: {set(cycle_lanes['small_area']) - set(population.index) or 'none'}")
```

<div class="usecase-body">
<b>What this shows:</b> how many segments remain, their total length, and that every area name now matches.
</div>

<div class="usecase-body">
<b>Why it matters:</b> removing the markings keeps the lane totals honest. Without this step, painted chevron markings alone would add more than 40 km of lane that nobody can ride in.
</div>

<div id="sub-3-4" class="usecase-sub-section-heading"><b>3.4. Placing the bike count sites</b></div>

<div class="usecase-body">Each count site has a latitude and longitude. A spatial join tags every site with the small area it sits inside, so counts can be compared with the population of that area.</div>

```python
bike_counts = gpd.GeoDataFrame(
    bike_counts_raw[["site_id", "year", "total"]].assign(
        description=bike_counts_raw["description"].fillna("Unnamed site")),   # one record has no name
    geometry=gpd.points_from_xy(bike_counts_raw["longitude"], bike_counts_raw["latitude"]),
    crs=CRS_DEGREES,
)
bike_counts = gpd.sjoin(bike_counts, small_areas, how="left", predicate="within").drop(columns="index_right")

LATEST_COUNT_YEAR = bike_counts["year"].max()
latest_counts = bike_counts[bike_counts["year"] == LATEST_COUNT_YEAR]

print(f"Count records: {len(bike_counts)} from {bike_counts['site_id'].nunique()} sites")
print(f"Sites counted in {LATEST_COUNT_YEAR}: {len(latest_counts)}, "
      f"of which {latest_counts['small_area'].isna().sum()} sit just outside the 13 small areas")
```

<div class="usecase-body">
<b>What this shows:</b> how many count records and sites there are, and how many of the latest sites fall outside the small areas.
</div>

<div class="usecase-body">
<b>Why it matters:</b> sites outside the areas stay on the map, but they cannot be counted towards any one area's riders.
</div>

<div id="section-4" class="usecase-section-heading"><b>4. Population growth, 2023&ndash;2043</b></div>

<div class="usecase-body">First we look at where and when the population is forecast to grow.</div>

<div id="sub-4-1" class="usecase-sub-section-heading"><b>4.1. How the population changes over time</b></div>

<div class="usecase-body">The forecast population of each area in five-year steps.</div>

```python
steps = population.loc[population[LAST_YEAR] > 0, FORECAST_STEPS].sort_values(LAST_YEAR, ascending=False)

ax = steps.plot(kind="bar", figsize=(12, 5.5), width=0.8, color=[GREEN, PURPLE, ORANGE, BLUE, YELLOW])
ax.yaxis.set_major_formatter(thousands)
plt.title(f"Forecast population by small area, {FIRST_YEAR} to {LAST_YEAR}")
plt.xlabel("")
plt.ylabel("Residents")
plt.xticks(rotation=35, ha="right")
plt.legend(title="Year")
plt.tight_layout()
plt.show()

print("City of Melbourne: " + " -> ".join(f"{city_population[year]:,} ({year})" for year in FORECAST_STEPS))
```

<div class="usecase-body">
<b>What this shows:</b> the forecast population of each area at five-year intervals, largest areas first, with the city total underneath.
</div>

<div class="usecase-body">
<b>Why it matters:</b> the CBD is by far the largest area, and almost every area gets bigger. The chart shows the scale of each area, which matters when reading the percentages in the next step.
</div>

<div id="sub-4-2" class="usecase-sub-section-heading"><b>4.2. Which areas grow the most?</b></div>

<div class="usecase-body">The number of residents each area is forecast to add over the 20 years, with the percentage growth written beside each bar.</div>

```python
growth = pd.DataFrame({
    f"residents_{FIRST_YEAR}": population[FIRST_YEAR],
    f"residents_{LAST_YEAR}": population[LAST_YEAR],
})
growth["added_residents"] = growth[f"residents_{LAST_YEAR}"] - growth[f"residents_{FIRST_YEAR}"]
growth["growth_%"] = (100 * growth["added_residents"] / growth[f"residents_{FIRST_YEAR}"]).round(0)
growth.loc[TINY_AREAS, "growth_%"] = np.nan   # meaningless from a near-zero start

city_growth = 100 * (city_population[LAST_YEAR] / city_population[FIRST_YEAR] - 1)

chart = growth[growth["added_residents"] > 0].sort_values("added_residents")
plt.figure(figsize=(10, 5.5))
ax = chart["added_residents"].plot(kind="barh", color=GREEN)
for bar, pct in zip(ax.patches, chart["growth_%"]):
    label = f"+{pct:.0f}%" if pd.notna(pct) else "from almost none"
    ax.text(bar.get_width() + 300, bar.get_y() + bar.get_height() / 2, label, va="center")
ax.xaxis.set_major_formatter(thousands)
plt.title(f"Residents added by {LAST_YEAR}")
plt.xlabel("Additional residents")
plt.ylabel("")
plt.xlim(0, chart["added_residents"].max() * 1.15)
plt.tight_layout()
plt.show()

print(f"City-wide growth: +{city_growth:.0f}%")
```

<div class="usecase-body">
<b>What this shows:</b> how many residents each area gains and how fast that is compared with its current size.
</div>

<div class="usecase-body">
<b>Why it matters:</b> both views matter. The CBD adds the most people, while North Melbourne grows the fastest and roughly doubles. An area that doubles needs its transport network to grow just as quickly.
</div>

<div id="sub-4-3" class="usecase-sub-section-heading"><b>4.3. When does the growth happen?</b></div>

<div class="usecase-body">The percentage growth of each area in each five-year period. Darker cells mean faster growth.</div>

```python
period_growth = pd.DataFrame({
    f"{start}-{start + 5}": 100 * (population[start + 5] / population[start] - 1)
    for start in FORECAST_STEPS[:-1]
}).drop(index=TINY_AREAS)
period_growth = period_growth.loc[growth["growth_%"].dropna().sort_values(ascending=False).index]

plt.figure(figsize=(9, 5.5))
sns.heatmap(period_growth, annot=True, fmt=".0f", cmap="Greens", linewidths=0.5,
            cbar_kws={"label": "% growth in the period"})
plt.title("Population growth in each five-year period (%)")
plt.xlabel("Period")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

<div class="usecase-body">
<b>What this shows:</b> whether each area's growth is spread over the 20 years or bunched into one part of it. Areas are listed from fastest to slowest overall growth.
</div>

<div class="usecase-body">
<b>Why it matters:</b> timing changes what should be built first. Carlton, East Melbourne, Parkville and South Yarra grow quickly in the first ten years and then almost stop, while North Melbourne, Kensington and West Melbourne (Residential) keep growing strongly into the 2030s. Those areas will need infrastructure that keeps up for longer.
</div>

<div id="section-5" class="usecase-section-heading"><b>5. Active transport today</b></div>

<div class="usecase-body">Next we look at the cycling side: how many bike lanes each area has, how well protected they are, and where people are counted riding.</div>

<div id="sub-5-1" class="usecase-sub-section-heading"><b>5.1. Bike lanes in each area</b></div>

<div class="usecase-body">The length of riding lane in each area, split by protection level.</div>

```python
lane_km = cycle_lanes.pivot_table(index="small_area", columns="protection", values="length",
                                  aggfunc="sum", fill_value=0) / 1000
lane_km = lane_km[[level for level in RIDING_LANE_LEVELS if level in lane_km.columns]]
lane_km = lane_km.loc[lane_km.sum(axis=1).sort_values().index]

ax = lane_km.plot(kind="barh", stacked=True, figsize=(10, 6),
                  color=[PROTECTION_COLOURS[level] for level in lane_km.columns])
plt.title("Riding lane length by small area and protection level")
plt.xlim(0, lane_km.sum(axis=1).max() * 1.05)
plt.xlabel("Lane length (km)")
plt.ylabel("")
plt.legend(title="Protection")
plt.tight_layout()
plt.show()

protected_share = 100 * lane_km["Protected"].sum() / lane_km.to_numpy().sum()
print(f"Across the city, {protected_share:.0f}% of riding lane length is physically protected.")
```

<div class="usecase-body">
<b>What this shows:</b> how much riding lane each area has and what kind it is.
</div>

<div class="usecase-body">
<b>Why it matters:</b> North Melbourne and Carlton have far more lanes than anywhere else, but most of that is painted rather than protected. The CBD and Southbank have less lane length, yet a much higher share of it is physically separated from traffic. Protection matters because separated lanes are the ones most people feel safe riding in.
</div>

<div id="sub-5-2" class="usecase-sub-section-heading"><b>5.2. Where people ride</b></div>

<div class="usecase-body">On the left, the change in riders over time, using only the sites counted every year so that each year is compared fairly. On the right, the riders counted in each area in the most recent count.</div>

```python
years_counted = bike_counts.groupby("site_id")["year"].nunique()
every_year_sites = years_counted[years_counted == bike_counts["year"].nunique()].index
trend = bike_counts[bike_counts["site_id"].isin(every_year_sites)].groupby("year")["total"].sum()

riders_by_area = (
    latest_counts.dropna(subset=["small_area"])
    .groupby("small_area")
    .agg(sites=("site_id", "size"), riders=("total", "sum"))
    .sort_values("riders")
)

fig, (left, right) = plt.subplots(1, 2, figsize=(14, 5))

left.plot(trend.index, trend.values, color=GREEN, linewidth=2.5, marker="o")
left.yaxis.set_major_formatter(thousands)
left.set_ylim(0, trend.max() * 1.15)
left.set_title(f"Riders counted at the {len(every_year_sites)} sites counted every year")
left.set_xlabel("Year")
left.set_ylabel("Riders (7-9 am)")

right.barh(riders_by_area.index, riders_by_area["riders"], color=PURPLE)
for position, (sites, riders) in enumerate(zip(riders_by_area["sites"], riders_by_area["riders"])):
    right.text(riders + 150, position, f"{sites} site{'s' if sites > 1 else ''}", va="center")
right.xaxis.set_major_formatter(thousands)
right.set_xlim(0, riders_by_area["riders"].max() * 1.2)
right.set_title(f"Riders counted by area, {LATEST_COUNT_YEAR}")
right.set_xlabel("Riders (7-9 am)")

plt.tight_layout()
plt.show()

change = 100 * (trend.iloc[-1] / trend.iloc[0] - 1)
print(f"Riders at the {len(every_year_sites)} sites counted every year rose {change:.0f}% "
      f"from {trend.index[0]} to {trend.index[-1]}.")
print(f"\nBusiest sites in {LATEST_COUNT_YEAR}:")
print(latest_counts.nlargest(3, "total")[["description", "small_area", "total"]].to_string(index=False))
```

<div class="usecase-body">
<b>What this shows:</b> how commuter cycling changed between 2010 and 2017, and where riders were counted in 2017. The label beside each bar shows how many sites that area had.
</div>

<div class="usecase-body">
<b>Why it matters:</b> morning bike traffic was already growing strongly before the forecast period starts. The most riders were counted in the CBD and Docklands, including along the Capital City Trail. The area totals depend on how many sites each area had, so they show where riding was observed rather than a full count.
</div>

<div id="section-6" class="usecase-section-heading"><b>6. Comparing growth with cycling infrastructure</b></div>

<div class="usecase-body">This is the part that answers the scenario. We put the growth from section 4 next to the lanes from section 5 to see where cycling infrastructure is most likely to fall behind.</div>

<div id="sub-6-1" class="usecase-sub-section-heading"><b>6.1. Bike lanes per 1,000 residents, now and in 2043</b></div>

<div class="usecase-body">Dividing each area's lane length by its population gives a simple, comparable measure: metres of riding lane per 1,000 residents. We work it out for 2023 and again for 2043, assuming no new lanes are built. The dashed line marks the city-wide figure for 2043.</div>

```python
comparison = growth.drop(index=TINY_AREAS).join(lane_km.sum(axis=1).rename("lane_km"))
comparison["growth_%"] = comparison["growth_%"].astype(int)
comparison["protected_%"] = (100 * lane_km["Protected"] / lane_km.sum(axis=1)).round().astype(int)

for year in (FIRST_YEAR, LAST_YEAR):
    comparison[f"lane_m_per_1000_{year}"] = (
        1_000_000 * comparison["lane_km"] / comparison[f"residents_{year}"]).round().astype(int)

total_lane_m = 1000 * lane_km.to_numpy().sum()
city_lane_rate = {year: 1000 * total_lane_m / city_population[year] for year in (FIRST_YEAR, LAST_YEAR)}

chart = comparison.sort_values(f"lane_m_per_1000_{LAST_YEAR}").rename(columns={
    f"lane_m_per_1000_{FIRST_YEAR}": str(FIRST_YEAR),
    f"lane_m_per_1000_{LAST_YEAR}": f"{LAST_YEAR} (no new lanes)",
})
ax = chart[[str(FIRST_YEAR), f"{LAST_YEAR} (no new lanes)"]].plot(
    kind="barh", figsize=(10, 6), color=[GREEN, PURPLE], width=0.8)
ax.axvline(city_lane_rate[LAST_YEAR], color="#121212", linestyle="--", linewidth=1,
           label=f"City average {LAST_YEAR}")
ax.xaxis.set_major_formatter(thousands)
plt.title("Metres of riding lane per 1,000 residents")
plt.xlabel("Metres of lane per 1,000 residents")
plt.ylabel("")
plt.legend()
plt.tight_layout()
plt.show()

print(f"City-wide: {city_lane_rate[FIRST_YEAR]:.0f} m per 1,000 residents in {FIRST_YEAR}, "
      f"falling to {city_lane_rate[LAST_YEAR]:.0f} m in {LAST_YEAR} if no lanes are added.")
```

<div class="usecase-body">
<b>What this shows:</b> how much riding lane each area has for every 1,000 residents today, and how far that falls by 2043 if nothing is added.
</div>

<div class="usecase-body">
<b>Why it matters:</b> raw kilometres favour large, lane-rich areas. Dividing by population shows how thin the network is for the people who live there. The CBD, Southbank and South Yarra already have far less lane per resident than the rest of the city, and growth pushes them lower still. North Melbourne keeps a lot of lane per resident, but its figure halves because its population doubles.
</div>

<div id="sub-6-2" class="usecase-sub-section-heading"><b>6.2. Areas that need the most attention</b></div>

<div class="usecase-body">
Finally we bring the measures into one table and flag areas using two simple rules, each compared with the city as a whole:
</div>

<div class="usecase-list">

- <b>Thin network:</b> fewer metres of riding lane per 1,000 residents in 2043 than the city average.
- <b>Fast growth, little protection:</b> population growth at least as fast as the city's, but a smaller share of protected lanes than the city's.

</div>

<div class="usecase-body">
The rules are deliberately simple, so anyone can check them against the table.
</div>

```python
comparison[f"riders_counted_{LATEST_COUNT_YEAR}"] = riders_by_area["riders"].astype("Int64")   # blank where no site

thin_network = comparison[f"lane_m_per_1000_{LAST_YEAR}"] < city_lane_rate[LAST_YEAR]
fast_unprotected = (comparison["growth_%"] >= city_growth) & (comparison["protected_%"] < protected_share)

comparison["attention"] = np.select(
    [thin_network & fast_unprotected, thin_network, fast_unprotected],
    ["Thin network; fast growth, little protection", "Thin network", "Fast growth, little protection"],
    default="",
)

print(f"City benchmarks: {city_lane_rate[LAST_YEAR]:.0f} m of lane per 1,000 residents in {LAST_YEAR}, "
      f"{city_growth:.0f}% growth, {protected_share:.0f}% of lane length protected\n")

columns = ["added_residents", "growth_%", "lane_km", "protected_%",
           f"lane_m_per_1000_{FIRST_YEAR}", f"lane_m_per_1000_{LAST_YEAR}",
           f"riders_counted_{LATEST_COUNT_YEAR}", "attention"]
comparison.sort_values("attention", ascending=False)[columns].round(1)
```

<div class="usecase-body">
<b>What this shows:</b> every area's growth, lanes, protection and 2017 riders side by side, with the areas that break either rule flagged. Blank rider cells are areas with no count site in 2017.
</div>

<div class="usecase-body">
<b>Why it matters:</b> this is the practical output of the notebook. The flagged areas are where new or better protected bike lanes would do the most to keep up with growth. Areas close to a threshold, such as Kensington, deserve a look too, because the rules are a guide rather than a hard line.
</div>

<div id="section-7" class="usecase-section-heading"><b>7. Mapping the results</b></div>

<div class="usecase-body">
The map brings the three layers together. Each area is shaded by its population growth, with darker grey meaning faster growth. Coloured lines are the riding lanes by protection level, and yellow circles are the 2017 bike count sites, sized by the number of riders.
</div>

<div class="usecase-body">
Hover over an area or a count site to see its details, and use the layer control in the top right to switch layers on and off. To keep the notebook small, the lanes are merged by protection level and their outlines simplified to about 1 m, which is too small to see at city scale.
</div>

```python
area_layer = small_areas.merge(growth.reset_index(), on="small_area")
area_layer["growth_text"] = area_layer["growth_%"].map(
    lambda value: "not meaningful (starts near zero)" if pd.isna(value) else f"+{value:.0f}%")
area_layer = area_layer[["small_area", f"residents_{FIRST_YEAR}", f"residents_{LAST_YEAR}", "growth_%", "growth_text", "geometry"]]

grey_scale = cm.LinearColormap(["#F2F2F2", "#636363"],
                               vmin=area_layer["growth_%"].min(), vmax=area_layer["growth_%"].max(),
                               caption=f"Population growth {FIRST_YEAR}-{LAST_YEAR} (%)")

centre = small_areas.to_crs(CRS_METRES).union_all().centroid
centre = gpd.GeoSeries([centre], crs=CRS_METRES).to_crs(CRS_DEGREES).iloc[0]
growth_map = folium.Map(location=[centre.y, centre.x], zoom_start=13)

# Layer 1: areas shaded by growth. Areas without a meaningful rate are shown in white.
folium.GeoJson(
    area_layer.to_json(),
    name="Population growth",
    style_function=lambda feature: {
        "fillColor": grey_scale(feature["properties"]["growth_%"])
        if feature["properties"]["growth_%"] is not None else "#FFFFFF",
        "color": "#121212", "weight": 1, "fillOpacity": 0.6,
    },
    tooltip=folium.GeoJsonTooltip(
        fields=["small_area", f"residents_{FIRST_YEAR}", f"residents_{LAST_YEAR}", "growth_text"],
        aliases=["Area", f"Residents {FIRST_YEAR}", f"Residents {LAST_YEAR}", "Growth"],
    ),
).add_to(growth_map)
grey_scale.add_to(growth_map)

# Layer 2: lanes, merged by protection level and simplified to about 1 m.
merged_lanes = cycle_lanes.to_crs(CRS_METRES).dissolve(by="protection")[["geometry"]]
merged_lanes["geometry"] = merged_lanes.simplify(1)
merged_lanes = merged_lanes.to_crs(CRS_DEGREES)
merged_lanes["geometry"] = merged_lanes.geometry.set_precision(0.00001)
for level, row in merged_lanes.iterrows():
    colour = PROTECTION_COLOURS[level]
    folium.GeoJson(
        gpd.GeoSeries([row.geometry], crs=CRS_DEGREES).to_json(),
        name=f"{level} lanes",
        style_function=lambda feature, colour=colour: {"color": colour, "fillColor": colour,
                                                       "weight": 2, "fillOpacity": 1},
    ).add_to(growth_map)

# Layer 3: bike count sites, sized by riders.
sites_layer = folium.FeatureGroup(name=f"Bike counts {LATEST_COUNT_YEAR}")
for site in latest_counts.itertuples():
    folium.CircleMarker(
        location=[site.geometry.y, site.geometry.x],
        radius=2 + site.total ** 0.5 / 5,   # square root keeps the busiest sites from covering the map
        color="#121212", weight=1, fill=True, fill_color=YELLOW, fill_opacity=0.85,
        tooltip=f"{site.description}<br>{site.total:,} riders (7-9 am)",
    ).add_to(sites_layer)
sites_layer.add_to(growth_map)

legend = "".join(
    f'<div><span style="display:inline-block;width:14px;height:4px;background:{colour};'
    f'margin-right:6px;vertical-align:middle;"></span>{level}</div>'
    for level, colour in PROTECTION_COLOURS.items()
)
growth_map.get_root().html.add_child(folium.Element(
    '<div style="position:fixed;bottom:30px;left:30px;z-index:9999;background:white;'
    'padding:8px 12px;border-radius:4px;font-size:13px;box-shadow:0 1px 4px rgba(0,0,0,.3);">'
    f'<b>Bike lanes</b>{legend}'
    f'<div style="margin-top:4px;"><span style="color:{YELLOW};font-size:16px;">&#9679;</span> '
    f'Bike count site</div></div>'
))

folium.LayerControl().add_to(growth_map)
growth_map
```

<div class="usecase-body">
<b>What this shows:</b> population growth, bike lanes and observed riding on one interactive map.
</div>

<div class="usecase-body">
<b>Why it matters:</b> the map shows the table from section 6 in its real street context. North Melbourne is the darkest area and is crossed by many lanes, but most of them are painted (purple and orange) rather than protected (green). The CBD and Docklands have the busiest count sites and, with Southbank, take much of the new population, yet their lane networks are sparse for the number of residents.
</div>

<div id="conclusion" class="usecase-unnumbered-heading"><b>Conclusion</b></div>

**What we achieved in this analysis**

We turned the City of Melbourne's 2023&ndash;2043 population forecast into one total per small area and year, and compared it with 136 km of riding lanes and eight years of Super Tuesday bike counts. For every area we measured how fast it grows, how much of its lane network is protected, and how many metres of lane it has for every 1,000 residents now and in 2043. We then flagged the areas where cycling infrastructure is most likely to fall behind.

**What we learned from this analysis**

<div class="usecase-body">
The city's bike lane network will be spread much more thinly by 2043. The pressure falls in two different places: the central areas, which already have little lane for their population, and the fast-growing northern areas, whose lanes are mostly painted rather than protected.
</div>

<div class="usecase-list">

- **The population grows by about 115,000, but unevenly.** The city is forecast to grow 65%, from 177,399 to 292,133 residents. The CBD adds the most people (32,542), while North Melbourne grows the fastest and doubles (+102%). Carlton, East Melbourne, Parkville and South Yarra stop growing after about 2033, while North Melbourne, Kensington and West Melbourne (Residential) keep growing into the late 2030s.
- **Lanes per resident fall sharply everywhere.** If no lanes are added, the city goes from 765 m of riding lane per 1,000 residents in 2023 to 465 m in 2043.
- **The central areas have a thin network.** The CBD (120 m per 1,000 residents by 2043), Southbank (124 m), South Yarra (122 m) and Docklands (322 m) all fall well below the city average. The CBD and Southbank also absorb over 50,000 of the new residents.
- **The fast-growing north relies on painted lanes.** North Melbourne and West Melbourne (Residential) grow faster than the city as a whole, but only 12% and 8% of their lane length is protected, against 19% city-wide. Kensington only just misses both flags: it grows 63% (just under the city's 65%) with only 5% protection, and its 480 m per 1,000 residents is barely above the city average.
- **Demand for cycling was already rising.** Riders at the 34 sites counted every year rose 47% between 2010 and 2017. The busiest sites were Flinders Street at Swanston Street in the CBD and two points on the Capital City Trail in Docklands.

</div>

**What this does not tell us**

<div class="usecase-body">
The population figures are forecasts, and the bike counts stop in 2017 and cover only two morning hours on one day a year, so they show commuter riding rather than all cycling. Lane length is only one measure of a good network: it says nothing about how well lanes connect, and the lane data is almost entirely on-road, so off-road shared paths such as the Capital City Trail are largely missing. Metres per resident also ignores the many people who ride into the city from elsewhere.
</div>

**Observations for further opportunities**

<div class="usecase-body">
The method is easy to reuse: add up a forecast to one row per area, divide infrastructure by residents, and flag the areas that fall behind a city-wide benchmark. The same steps would work for parks, libraries or public toilets.
</div>

<div class="usecase-list">

- Add public transport stops, tram routes and off-road shared paths for a fuller picture of low-emission travel.
- Use the forecast's age groups to see where growth is among children or older residents, who need safer, protected lanes.
- Replace the Super Tuesday counts with more recent cycling data as it becomes available.
- Look at how well the lanes connect, not just how long they are, by analysing them as a network.

</div>
