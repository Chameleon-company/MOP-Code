<div id="section-2" class="usecase-section-heading"><b>2. Fetching the Wellbeing Indicators Dataset</b></div>

<div class="usecase-body">This section downloads the first dataset, Indicators of wellbeing by year (Future Melbourne), from the [Melbourne Open Data Portal](https://data.melbourne.vic.gov.au/pages/home/). The portal offers a REST API, which lets code request data directly from a web address instead of downloading a file by hand. The code below builds that address from the dataset ID, asks for every record in CSV format, and reads the response into a pandas DataFrame called <code>fm_df</code>. It also checks that the request succeeded, so that a failed download shows a clear message rather than causing a confusing error later on.</div>

```python
# https://data.melbourne.vic.gov.au/explore/dataset/indicators-of-wellbeing-by-year-future-melbourne/information/
dataset_id = 'indicators-of-wellbeing-by-year-future-melbourne'

base_url = 'https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/'
# apikey = ''
dataset_id = dataset_id
format = 'csv'

url = f'{base_url}{dataset_id}/exports/{format}'
params = {
    'select': '*',
    'limit': -1, # all records
    'lang': 'en',
    'timezone': 'UTC'
    #'api_key': apikey
}

#GET request 
response = requests.get(url, params = params)

if response.status_code == 200:
    # StringIO to read the CSV data
    url_content = response.content.decode('utf-8')
    fm_df = pd.read_csv(StringIO(url_content), delimiter= ';')
else: 
    print(f'Request failed with status code{response.status_code}')
```

<div class="usecase-body">The first five rows are shown below to check that the dataset loaded correctly. Each row is one wellbeing indicator, with a label and description, the unit (a percentage), and a result for each survey year from 2012 to 2016.</div>

```python
# printing out the first 5 lines of Future Melbourne Data set fm_df
fm_df.head(5)
```

<div id="section-4" class="usecase-section-heading"><b>4. Exploring Wellbeing Indicator Trends</b></div>

<div class="usecase-body">This section explores how each wellbeing indicator changed between 2012 and 2016. A line chart is a simple way to see whether a value is rising, falling, or staying steady over time. The function below draws the trend for one indicator, and a dropdown menu built with ipywidgets lets you pick any indicator in the dataset without writing new code. See the [ipywidgets documentation](https://ipywidgets.readthedocs.io/en/stable/) to learn more about building interactive controls like this. The dropdown requires a live kernel to explore other options, so a static chart for total food insecurity, the indicator this analysis focuses on, is also shown directly below it.</div>

```python
# Creating an interactive plot, so plot will update according to the indicator selected by the user from the dropdown

# Function to plot the observations
def plot_trend(indicator_label):
    
    # Pick up the row selected by the user
    selected_data = fm_df[fm_df['indicator_label'] == indicator_label] 
    
    if not selected_data.empty:  # Checking to ensure selected data is not empty
        years = ['2012', '2013', '2014', '2015', '2016']
        values = selected_data[years].values[0]
        
        # Plot is then created 
        plt.figure(figsize=(10, 6))
        plt.plot(years, values, marker='o')
        plt.title(f'Trend for {indicator_label}')
        plt.xlabel('Year')
        plt.ylabel('Percentage')
        plt.grid(True)
        plt.show()
    else:
        print(f"No data available for {indicator_label}")

# Create the dropdown widget, allowing user to pick an indicator.  Options in the drop down are unique values from the indicator_label column 
indicator_dropdown = widgets.Dropdown(
    options=fm_df['indicator_label'].unique(),
    description='Indicator:')

# Use interact to create the interactive plot, linking a function and a widget
interact(plot_trend, indicator_label=indicator_dropdown)

# A static view of one indicator is shown directly below, so a chart is visible even
# when this notebook is viewed as a static export without a live kernel
plot_trend('Total Food insecurity')
```

<div class="usecase-body">The static chart shows how total food insecurity moved from year to year. This is the measure that the rest of the analysis builds on, and Section 8 looks at it in more detail alongside the related food security indicators.</div>

<div class="usecase-body">The second dataset, Free and cheap support services, with opening hours, public transport and parking options (Helping Out), lists services in Melbourne that offer free or low cost help, such as food, accommodation, health, and legal support. <div id="subsection-6-1" class="usecase-sub-section-heading"><b>6.1. Handling Missing Geocoded Locations</b></div>

<div class="usecase-body">Entries without a geocoded location have limited information and cannot be plotted on a map, so most will be removed. However, some of them are helplines, such as crisis and counselling phone lines, which have a phone number but no street address. These are still important, so the code below picks out the 11 entries with the category Helpful phone number and stores their names and phone numbers in a separate table before they are removed from the data used for mapping.</div>

```python
# Filter the DataFrame to include only rows where category_1 is 'Helpful phone number'
filtered_data = no_geocode[no_geocode['category_1'] == 'Helpful phone number']

# Select specific columns
helpful_number = filtered_data[['name', 'phone', 'free_call', 'category_1']]

# Reset the index
helpful_number_reset = helpful_number.reset_index(drop=True)

# Display the DataFrame with selected columns and reset index
helpful_number_reset
```

```python
# showing the single name observation with null values 
no_name = ho_df[ho_df['name'].isnull()]

no_name
```

<div class="usecase-body">The one observation with a missing name is also removed, as it does not contain enough information to be useful. In the next step, category_6 is dropped because it is completely empty, and the rows with no geocoded location or no name are removed, since they cannot be plotted on the map. The first five rows of the result are then displayed to confirm the change.</div>

```python
# Category_6 is then dropped as there was not any data in category_6
ho_df = ho_df.drop(columns=['category_6'])

# Dropping observations with no "geocoded_location" and "name" as these will not be useful for our future mapping
ho_df = ho_df.dropna(subset=['geocoded_location', 'name'])
ho_df.head(5)
```

```python
# In order to do a mapping later, category_1 is going to be used as categorisation.  However, as there is 1 entry with null value, this will also be removed

nan_rows = ho_df[ho_df['category_1'].isna()]

ho_df = ho_df.dropna(subset=['category_1'])
```

<div class="usecase-body">The category_1 column is used for grouping services later in this analysis. As one entry has a missing category_1 value, this row is also removed. The cleaned dataset shape is shown below and should now have 87 rows and 33 columns, compared with the 119 rows and 34 columns we started with.</div>

```python
# Showing new shape of ho_df.shape 
num_rows, num_columns = ho_df.shape

print(f'The Cleaned Helping Out Dataset has {num_rows} rows and {num_columns} columns.')
```

<div id="subsection-7-3" class="usecase-sub-section-heading"><b>7.3. Mapping Services Across Melbourne</b></div>

<div class="usecase-body">A map shows where services are located, which a table or a bar chart cannot show. The code below uses the folium library to draw a map of Melbourne and to place a marker for each service using its latitude and longitude. Each marker is coloured by category, and hovering over it shows the opening hours and a phone number. A dropdown menu built with ipywidgets can filter the map by category, and this requires a live kernel to interact with, so a static map showing all categories is also rendered directly below it. See the [Folium documentation](https://python-visualization.github.io/folium/) for more on building maps, and the [ipywidgets documentation](https://ipywidgets.readthedocs.io/en/stable/) for the dropdown.</div>

```python
# Create interactive folium map visualising services categories in Melb
# Markers to provide information such as opening hours and phone number of each service   

# Function to define colors for each category
# Category colours are drawn from the use case template palette, cycling through it
# since there are more service categories than palette colours
category_colors = {
    'Food': PALETTE[0],
    'Accommodation': PALETTE[1],
    'Health Services / Pharmacy': PALETTE[2],
    'Legal / Financial Advice': PALETTE[3],
    'Hospitals / Emergency': PALETTE[4],
    'Counselling and Psychiatric Services': PALETTE[5],
    'Drug and Alcohol': PALETTE[0],
    'Travel Assistance': PALETTE[1],
    'Tenancy Assistance': PALETTE[2],
    'Helpful phone number': PALETTE[3],
    'Clothes and Blankets': PALETTE[4],
    'Employment Assistance': PALETTE[5],
    'Needle Exchange': PALETTE[0],
    'Showers / Laundry': PALETTE[1],
}

# Function to create a map centered at a specific location
def create_map():
    # Somewhere in Melbourne so the map is more centered.
    # The OpenStreetMap default tiles started returning 403 Access blocked errors
    # because the app was not following OSM's tile usage policy for its volunteer-run
    # servers, so the map uses the keyless Esri Light Gray Canvas tiles instead, matching
    # the fix already applied elsewhere in the project.
    my_map = folium.Map(location=[-37.81534, 144.97215], zoom_start=13, tiles=None)
    folium.TileLayer(
        tiles="https://server.arcgisonline.com/ArcGIS/rest/services/Canvas/World_Light_Gray_Base/MapServer/tile/{z}/{y}/{x}",
        attr="Esri",
        name="Esri Light Gray Base",
        control=False
    ).add_to(my_map)
    folium.TileLayer(
        tiles="https://server.arcgisonline.com/ArcGIS/rest/services/Canvas/World_Light_Gray_Reference/MapServer/tile/{z}/{y}/{x}",
        attr="Esri",
        name="Esri Light Gray Labels",
        overlay=True,
        control=False
    ).add_to(my_map)
    return my_map

# Function to format opening hours
def format_hours(row):
    # Creating "hours" List - If value in each row is not NA then print out the value 
    hours = [ 
        f"Monday: {row['monday']}" if pd.notna(row['monday']) else "Monday: Time Not Available", 
        f"Tuesday: {row['tuesday']}" if pd.notna(row['tuesday']) else "Tuesday: Time Not Available",
        f"Wednesday: {row['wednesday']}" if pd.notna(row['wednesday']) else "Wednesday: Time Not Available",
        f"Thursday: {row['thursday']}" if pd.notna(row['thursday']) else "Thursday: Time Not Available",
        f"Friday: {row['friday']}" if pd.notna(row['friday']) else "Friday: Time Not Available",
        f"Saturday: {row['saturday']}" if pd.notna(row['saturday']) else "Saturday: Time Not Available",
        f"Sunday: {row['sunday']}" if pd.notna(row['sunday']) else "Sunday: Time Not Available",
        f"Public Holiday: {row['public_holidays']}" if pd.notna(row['public_holidays']) else "Public Holiday: Time Not Available"
    ]
    return "<br>".join(hours)

# Function to add markers to the map
def add_markers(data, map_obj):
    for index, row in data.iterrows():
        name = row['name']
        phone = row['phone']
        latitude = row['latitude']
        longitude = row['longitude']
        category = row['category_1']
        hours_info = format_hours(row)  # Get formatted opening hours
        tooltip_text = f'<b>Name: {name}</b><br><b>Phone: {phone}</b><br>{hours_info}'  # Display name, phone, and hours in the tooltip
        
        # Obtain category colour from category_colours
        color = category_colors.get(category)  
        
        # Create a custom HTML marker using DivIcon
        html = f'<div style="background-color: {color}; width: 12px; height: 12px; border-radius: 50%; display: flex;"> </div>'
        icon = folium.DivIcon(html=html)
        
        # Use folium.DivIcon to create colored markers
        folium.Marker(
            [latitude, longitude], # Coordinates for the marker
            tooltip=tooltip_text, # Tooltip
            icon=icon # custom icon from above
        ).add_to(map_obj)

# Adding legend to the map
def add_legend(map_obj):
    
    # html defining what the legend would look like
    legend_html = '''
    <div style="
        position: absolute; 
        left: 450px; 
        width: 200px; 
        background: white; 
        border: 0.5px solid black; 
        z-index:9999; 
        font-size: 11px;  
        ">
        <b>Legend</b><br>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #059669; width: 12px; height: 12px; border-radius: 50%;"></div> Food<br> 
        </div>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #9333EA; width: 12px; height: 12px; border-radius: 50%;"></div> Accommodation<br>
        </div>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #F97316; width: 12px; height: 12px; border-radius: 50%;"></div> Health Services / Pharmacy<br>
        </div>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #2563EB; width: 12px; height: 12px; border-radius: 50%;"></div> Legal / Financial Advice<br>
        </div>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #EAB308; width: 12px; height: 12px; border-radius: 50%;"></div> Hospitals / Emergency<br>
        </div>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #EC4899; width: 12px; height: 12px; border-radius: 50%;"></div> Counselling and Psychiatric Services<br>
        </div>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #059669; width: 12px; height: 12px; border-radius: 50%;"></div> Drug and Alcohol<br>
        </div>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #9333EA; width: 12px; height: 12px; border-radius: 50%;"></div> Travel Assistance<br>
        </div>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #F97316; width: 12px; height: 12px; border-radius: 50%;"></div> Tenancy Assistance<br>
        </div>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #2563EB; width: 12px; height: 12px; border-radius: 50%;"></div> Helpful phone number<br>
        </div>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #EAB308; width: 12px; height: 12px; border-radius: 50%;"></div> Clothes and Blankets<br>
        </div>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #EC4899; width: 12px; height: 12px; border-radius: 50%;"></div> Employment Assistance<br>
        </div>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #059669; width: 12px; height: 12px; border-radius: 50%;"></div> Needle Exchange<br>
        </div>
        <div style="display: flex; align-items: center;">
            <div style="background-color: #9333EA; width: 12px; height: 12px; border-radius: 50%;"></div> Showers / Laundry<br>
        </div>
    </div>
    '''
    folium.Marker(
        location=[-37.81534, 144.97215],  # Where the legend would be placed
        icon=folium.DivIcon(html=legend_html)
    ).add_to(map_obj)

# Function to update the map based on selected filters
def update_map(category):
    my_map = create_map()  # Create a new map instance
    
    # Filter data based on selected category
    if category == 'All Categories':
        filtered_data = ho_df
    else:
        filtered_data = ho_df[ho_df['category_1'] == category]
    
    add_markers(filtered_data, my_map)
    add_legend(my_map)
    folium.plugins.MousePosition().add_to(my_map)
    display(my_map)

# Sorting the categories from a-z
sorted_categories = sorted(ho_df['category_1'].dropna().unique())

# Create dropdown widget for categories
categories_dropdown = widgets.Dropdown(
    options=['All Categories'] + sorted_categories,
    description='Category:',
    disabled=False,
    layout={'width': '25%'}
)

categories_dropdown.style.description_width = '60px'

# Use interact to create the interactive plot
widgets.interactive(update_map, category=categories_dropdown)  

# The map is also rendered directly below for the default category, so it is visible even
# when this notebook is viewed as a static export without a live kernel
update_map('All Categories')
```

<div id="section-8" class="usecase-section-heading"><b>8. Focusing on Food Security Indicators</b></div>

<div id="subsection-8-1" class="usecase-sub-section-heading"><b>8.1. Filtering Food Security Indicators</b></div>

<div class="usecase-body">The wellbeing dataset holds many kinds of indicators, so it is filtered down to the ones relating to food, since this analysis focuses specifically on food security. The code keeps every row whose indicator_label contains the word food, ignoring capital letters, and stores them in a new table called <code>fs_indicator</code>.</div>

```python
# Create the food_indicator dataset by filtering rows containing "food" in the indicator_label column
fs_indicator = fm_df[fm_df['indicator_label'].str.contains('food', case=False, na=False)]
fs_indicator = fs_indicator.reset_index(drop=True)

# Display the resulting dataset
fs_indicator
```

<div class="usecase-body">Some of these indicators relate to growing and producing food rather than food security itself, so these are removed. What remains, in the table called <code>fs</code>, is 20 indicators that measure food security directly, such as running out of food, eating less, worrying about running out of food, and total food insecurity, for all residents and for groups by gender and age.</div>

```python
# Filter to rows where the indicator_label does not contains "growing" or "producing"
fs = fs_indicator[~fs_indicator['indicator_label'].str.contains('growing|producing', case=False)]
fs = fs.reset_index(drop=True)

# Display the resulting dataset
fs
```

<div id="subsection-8-2" class="usecase-sub-section-heading"><b>8.2. Forecasting Food Security Trends with ARIMA</b></div>

<div class="usecase-body">Having identified the key indicators related to food insecurity in the City of Melbourne, the yearly mean of these indicators is calculated to explore future trends. Combining the 20 indicators into one average for each year gives a single series that is easy to forecast. Understanding whether these indicators are increasing or decreasing helps inform whether food service assistance should be expanded or reduced in different parts of Melbourne.</div>

```python
# Working out Yearly mean
year_columns = ['2012', '2013', '2014', '2015', '2016']
average_per_year = fs[year_columns].mean()

print("Average values per year, in percentage:")
average_per_year
```

<div class="usecase-body">The yearly averages are 6.4 percent in 2012, 5.3 percent in 2013, 6.6 percent in 2014, 5.5 percent in 2015, and 6.2 percent in 2016. Food insecurity moves up and down between about 5 and 7 percent, rather than rising or falling steadily in one direction.</div>

<div class="usecase-body">ARIMA, short for AutoRegressive Integrated Moving Average, is a common method for forecasting values in a time series, such as a yearly average measured over several years. It combines three simple ideas: using past values to predict future ones (the AR, or autoregressive, part), working with the differences between consecutive values to make the trend easier to model (the I, or integrated, part), and using past forecast errors to improve the next prediction (the MA, or moving average, part). Each of these three ideas has a setting, generally written as p, d, and q, and the best combination is not known in advance. A grid search is used below to try a range of combinations and keep the one that fits the data best. See the [statsmodels ARIMA documentation](https://www.statsmodels.org/stable/generated/statsmodels.tsa.arima.model.ARIMA.html) and this [How to Create an ARIMA Model for Time Series Forecasting in Python](https://machinelearningmastery.com/arima-for-time-series-forecasting-with-python/) tutorial from Machine Learning Mastery for more detail.</div>

<div class="usecase-body">This dataset only has five yearly data points, from 2012 to 2016. A model with several settings fitted to only five observations is at real risk of overfitting, meaning it may fit the historical values closely without actually capturing a genuine underlying trend. The forecast below should therefore be read as an exploratory estimate rather than a statistically robust prediction, and some of the parameter combinations tried in the grid search may fail to fit or produce unreliable results, which would not be unexpected given how little data is available.</div>

<div class="usecase-body">The code below sets up the yearly averages as a time series and tries every combination of p, d, and q from 0 to 2, which is 27 combinations in total. For each one it fits an ARIMA model and records its AIC score. AIC rewards a model that fits the data well but penalises unnecessary complexity, so the combination with the lowest AIC is kept. That model is then used to forecast the next 10 years, and the historical values and the forecast are plotted together.</div>

```python
# Using ARIMA with a grid search to find the best p, d, q combination to forecast food security for the next 10 years
import warnings as _warnings

# Prepare data
data = average_per_year
data.index = pd.date_range(start='2012', periods=len(data), freq='YE')

# Define parameter grid for ARIMA (p, d, q)
p = d = q = range(0, 3)
pdq = list(product(p, d, q))

# Grid Search for ARIMA Parameters
def grid_search_arima(data, pdq):
    best_aic = np.inf  # Initialising positive infinity
    best_pdq = None
    best_model = None
    n_converged = 0

    # Warnings from failed or unstable fits are expected with this little data,
    # so they are suppressed only for the duration of the model fitting below
    with _warnings.catch_warnings():
        _warnings.simplefilter('ignore')
        for param in pdq:
            try:
                model = ARIMA(data, order=param)
                results = model.fit()
                n_converged += 1
                aic = results.aic
                if aic < best_aic:
                    best_aic = aic
                    best_pdq = param
                    best_model = results
            except Exception:
                continue

    print(f"{n_converged} of {len(pdq)} parameter combinations converged.")
    return best_pdq, best_model

# Perform Grid Search and print
best_pdq, best_model = grid_search_arima(data, pdq)
print('Best ARIMA parameters:', best_pdq)

# Forecast the next 10 years using the best model
forecast_steps = 10
forecast = best_model.forecast(steps=forecast_steps)
forecast_index = pd.date_range(start=data.index[-1] + pd.DateOffset(years=1), periods=forecast_steps, freq='YE')

# Combine historical and forecasted data for plotting
plt.figure(figsize=(12, 6))
plt.plot(data.index, data, label='Historical Data', color=PALETTE[0], marker='o')
plt.plot(forecast_index, forecast, label='Forecast', color=PALETTE[1], marker='o')

# Plot formatting
plt.title('Food Insecurity Forecast for the 10 Years Following the Last Surveyed Year')
plt.xlabel('Year')
plt.ylabel('Average Value')
plt.legend()
plt.grid(True)
plt.show()
```

<div class="usecase-body">All 27 combinations were fitted successfully, and the best one was (2, 0, 0). This means the model predicts each year mainly from the two years before it, with no differencing needed. The chart shows the five historical points and the forecast for the following ten years, which stays close to the historical range of roughly 5.5 to 6.5 percent.</div>

```python
# Calculate evaluation metrics (MAE, MSE, RMSE)
# Note: with only 5 data points in total, these metrics compare the model's fitted
# values against the same data it was trained on, rather than genuine held out data.
# They give a rough sense of in sample fit only, not a true measure of forecast accuracy.
actual_values = data.values[-5:]  # Last 5 years of actual values
predicted_values = best_model.fittedvalues[-5:]  # Fitted values from the ARIMA model over the same years

# Mean Absolute Error
mae = mean_absolute_error(actual_values, predicted_values)

# Mean Squared Error
mse = mean_squared_error(actual_values, predicted_values)

# Root Mean Squared Error
rmse = np.sqrt(mse)

# Print evaluation metrics
print("Model Evaluation Metrics (in-sample fit, not held-out validation):")
print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
```

<div class="usecase-body">The error values above are relatively low. The mean absolute error of 0.17 means the fitted values differ from the actual yearly averages by about 0.17 percentage points on average, so the model fits the five historical points closely. Because these figures come from the same small sample the model was trained on, they show that the model can reproduce the historical trend, not that it can reliably predict future years. The forecast should be treated as an exploratory estimate rather than a statistically validated prediction.</div>
