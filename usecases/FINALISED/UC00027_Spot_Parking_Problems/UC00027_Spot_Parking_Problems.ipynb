We then define a helper function to pull the full CLUE datasets from the City of Melbourne Open Data API, and use it to fetch and preview the first of our three datasets.</div>

```python
# All imports used throughout this notebook are collected here
import warnings
import json

import requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import folium
import ipywidgets as widgets
from ipywidgets import interact_manual
from IPython.display import display

from sklearn import linear_model
from sklearn.ensemble import IsolationForest
from sklearn.metrics import r2_score, mean_absolute_error

warnings.filterwarnings("ignore")  # suppress noisy pandas/sklearn warnings

%matplotlib inline

# MOP template colour palette (Primary Green, then the distinct accent colours, in order)
MOP_COLOURS = ['#059669', '#9333EA', '#F97316', '#2563EB', '#EAB308', '#EC4899']
sns.set_theme(color_codes=True)
sns.set_palette(sns.color_palette(MOP_COLOURS))
```

```python
def API_Unlimited(dataset_name):
    """
    Fetch the full (unlimited) dataset from the City of Melbourne Open Data API.

    Parameters
    ----------
    dataset_name : str
        The dataset identifier as listed on the City of Melbourne Open Data portal.

    Returns
    -------
    pandas.DataFrame or None
        The requested dataset as a DataFrame, or None if the request fails.
    """
    base_url = 'https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/'
    dataset_format = 'csv'
    url = f'{base_url}{dataset_name}/exports/{dataset_format}'
    params = {
        'select': '*',
        'limit': -1,  # all records
        'lang': 'en',
        'timezone': 'UTC'
    }

    response = requests.get(url, params=params, stream=True)
    response.raw.decode_content = True  # handle gzip if the server compresses it

    if response.status_code == 200:
        # Stream straight into pandas instead of loading the whole file into a string first
        dataset = pd.read_csv(response.raw, delimiter=';')
        print(dataset.sample(min(10, len(dataset)), random_state=999))  # quick sanity check
        return dataset
    else:
        print(f'Request failed with status code {response.status_code}')
        return None
```

```python
#Residential dwellings data
dataset_1 = "residential-dwellings"
resident = API_Unlimited(dataset_1)
resident = resident[resident["census_year"] != 2021]
resident.info()
```

<div id="section-2" class="usecase-section-heading"><b>2. Clean and merge the datasets</b></div>

<div class="usecase-body">CLUE datasets are mainly clean, although there are some issues with missing values and duplicated records. The process below is used to eliminate these problems and to drop unnecessary columns, one dataset at a time, before merging all three together.</div>

<div id="subsection-2-1" class="usecase-sub-section-heading"><b>2.1. Residential Dwellings dataset</b></div>

<div class="usecase-body">We remove duplicate records, keep only the columns we need, then sum the dwelling numbers by block and census year.</div>

```python
#Total records before cleaning
print('Total records before cleaning is',len(resident))
#Data cleaning (remove duplicated values)
resident.drop_duplicates(inplace = True)
#Total records after cleaning
print('Total records after cleaning is',len(resident))
#Select only necessary columns
selected_columns =['census_year','block_id','clue_small_area','dwelling_number']
resident = resident[selected_columns]
#Sum the [Dwelling number] based on [Block ID] and [Census Year], but keep one [CLUE small area] for each Block ID
resident_by_block = resident.groupby(['census_year','block_id']).agg({'clue_small_area':'max','dwelling_number':'sum'})
print('Total records after grouping (by Block ID) is',len(resident_by_block))
#Detail of the residential dwellings by block
resident_by_block.head()
```

<div class="usecase-body">As a result, we obtain the <i>resident_by_block</i> table containing two columns, "clue_small_area" and "dwelling_number", with "census_year" and "block_id" as a multi-index. We repeat similar steps for the <i>Employment by city blocks</i> and <i>Off-street car parking</i> datasets below.</div>

<div id="subsection-2-2" class="usecase-sub-section-heading"><b>2.2. Employment by city blocks dataset</b></div>

<div class="usecase-body">We fetch the employment dataset, drop rows with no employment total recorded, keep only the columns we need, then sum total employment by block and census year.</div>

```python
dataset_2 = "employment-by-block-by-anzsic"
employment = API_Unlimited(dataset_2)
#Total records before cleaning
print('Total records before cleaning is',len(employment))
#Cleaning the data (remove missing values at [total_jobs_in_block])
employment.dropna(subset=['total_jobs_in_block'],inplace=True)
#Total records after cleaning
print('Total records after cleaning is',len(employment))
#Select only necessary columns
selected_columns =['census_year','block_id','total_jobs_in_block']
employment = employment[selected_columns]
#Sum the employment based on Block ID and Year
employment_by_block = employment.groupby(['census_year','block_id']).sum()
print('Total records after grouping (by Block ID) is',len(employment_by_block))
#Detail of the employment_by_block
employment_by_block.head()
```

<div id="subsection-2-3" class="usecase-sub-section-heading"><b>2.3. Off-street car parking dataset</b></div>

<div class="usecase-body">We fetch the off-street parking dataset and apply the same cleaning steps: remove duplicates, keep only the columns we need, then sum the number of parking spaces by block and census year.</div>

```python
#Off-street parking data
dataset_3 = "off-street-car-parks-with-capacity-and-type"
parking = API_Unlimited(dataset_3)
#Total records before cleaning
print('Total records before cleaning is',len(parking))
#Data cleaning (remove duplicate values)
parking.drop_duplicates(inplace=True)
#Total records after cleaning
print('Total records after cleaning is',len(parking))
#Select only necessary columns
selected_columns =['census_year','block_id','parking_spaces']
parking = parking[selected_columns]
#Sum the parking based on Block ID and Year
parking_by_block = parking.groupby(['census_year','block_id']).sum()
print('Total records after grouping (by Block ID) is',len(parking_by_block))
#Detail of the parking_by_block
parking_by_block.head()
```

<div id="subsection-2-4" class="usecase-sub-section-heading"><b>2.4. Merge the clean datasets and plot the scattered distributions</b></div>

<div class="usecase-body">We merge the three cleaned tables on their shared (census_year, block_id) index, keeping only the blocks present in all three datasets. The scatter plots below give insight into the linear relationship between the dependent variable (off-street car parking) and the independent variables (residential dwellings, employment by city block). Points that fall well below the regression line indicate a block with less parking than its employment or dwelling numbers would suggest, which is the first sign of a potential shortfall.  The merged <code>data</code> table produced here is kept in memory and reused directly by the interactive tool in Section 5.</div>

```python
#Combine the datasets (only take the blocks that match between the three datasets)
data = resident_by_block.merge(employment_by_block, left_index= True, right_index=True)
data = data.merge(parking_by_block,left_index=True,right_index=True)
print(data.head())
#Prepare data for scatter plots
x1 = data['total_jobs_in_block']
x2 = data['dwelling_number']
y = data['parking_spaces']
#Plot the data points (Parking Spaces vs. Total Employment) and regression line
plt.plot(x1,y,'x', color=MOP_COLOURS[0], label='Blocks')
a, b = np.polyfit(x1,y,1)
plt.plot(x1, a*x1 + b, color=MOP_COLOURS[1], label='Regression line')
plt.title('Parking Spaces vs. Total Employment')
plt.ylabel('Parking Spaces')
plt.xlabel('Total Employment')
plt.legend()
plt.show()
#Plot the data points (Parking Spaces vs. Dwelling Number) and regression line
plt.plot(x2,y,'+', color=MOP_COLOURS[0], label='Blocks')
a, b = np.polyfit(x2,y,1)
plt.plot(x2, a*x2 + b, color=MOP_COLOURS[2], label='Regression line')
plt.title('Parking Spaces vs. Dwelling Number')
plt.ylabel('Parking Spaces')
plt.xlabel('Dwelling Number')
plt.legend()
plt.show()
```

<div id="section-3" class="usecase-section-heading"><b>3. Build analysis model</b></div>

<div class="usecase-body">This model uses multiple-variable linear regression. The dependent variable <b>Y</b> is "off-street car parking", and the independent variables <b>X1</b> and <b>X2</b> are "employment by city block" and "residential dwellings" respectively.</div>

<div id="subsection-3-1" class="usecase-sub-section-heading"><b>3.1. Prepare the data for model training</b></div>

<div class="usecase-body">This function selects the requested feature columns and year range from the merged data, and returns them as the X (feature) and Y (target) arrays that the regression model will be trained on.</div>

```python
#Prepare X and Y data for model training
def Get_data(Data, Year_to_include, Employment, Resident_dwellings):
    features = np.array(['total_jobs_in_block', 'dwelling_number'])
    features_to_include = features[[Employment, Resident_dwellings]]  #Define the data features for the study
    data_to_include = Data.loc[min(Year_to_include):max(Year_to_include)]  #Get the wanted rows
    X = np.array(data_to_include[features_to_include])  #Get the independent variable(s) based on the selected features
    Y = np.array(data_to_include['parking_spaces'])  #Get the dependent variable
    return X, Y
```

<div id="subsection-3-2" class="usecase-sub-section-heading"><b>3.2. Build and evaluate the linear regression model</b></div>

<div class="usecase-body">Before training, we use Isolation Forest to flag and remove outlier blocks so that a handful of unusual blocks do not skew the fitted line. We then fit a linear regression model on the remaining (inlier) blocks and print its R² and MAE (mean absolute error) on the training data. R² (0 to 1) tells us how much of the block-to-block variation in parking spaces is explained by employment and dwelling numbers together — closer to 1 means the two features are strong predictors of parking supply. MAE is in the same units as the target (parking spaces), so an MAE of, say, 15 means predictions are off by about 15 parking spaces on average — this gives a direct sense of how much to trust the Impact index below, since a high MAE relative to typical block sizes means the flagged shortfalls carry more uncertainty.</div>

```python
def lm_model(X, Y):
    """Fit a linear regression model after removing outliers with Isolation Forest,
    and print the R² and MAE of the fitted model on its training data."""
    iso = IsolationForest(contamination=0.1, random_state=42)  #outlier detector
    is_inlier = iso.fit_predict(X) != -1  #search for outliers

    lm = linear_model.LinearRegression()  #linear regression model
    lm.fit(X[is_inlier], Y[is_inlier])  #train the model excluding outliers

    #Evaluate the model on the (inlier) training data
    Y_pred = lm.predict(X[is_inlier])
    r2 = r2_score(Y[is_inlier], Y_pred)
    mae = mean_absolute_error(Y[is_inlier], Y_pred)
    print(f'Model evaluation on training data -> R²: {r2:.3f}, MAE: {mae:.1f} parking spaces')

    return lm
```

<div id="subsection-3-3" class="usecase-sub-section-heading"><b>3.3. Predict the parking problem based on the latest year (2020)</b></div>

<div class="usecase-body">Using the trained model, we predict how many parking spaces each block in 2020 should have given its employment and dwelling numbers, and compare that to the parking spaces the block actually has. The shortfall is normalised into an Impact index from 0-100, so blocks can be ranked and compared on a common scale.</div>

```python
#Predict parking issue in 2020
def Get_predict_result(Data, Model, Employment, Residential_dwellings):
    data_latest = Data.loc[years.max]  #Get data of 2020
    features = np.array(['total_jobs_in_block', 'dwelling_number'])
    features_to_include = features[[Employment, Residential_dwellings]]  #Define the data features for the study
    result = np.round(Model.predict(np.array(data_latest[features_to_include])))  #Predict based on the selected features
    result = np.maximum(result, 0)  #a block can't have a negative number of parking spaces
    data_latest['Predict_parking'] = result.astype(int)  #Add result to the data
    impact = result - data_latest['parking_spaces']
    impact[impact>=0] = 100*(impact[impact>=0]/(max(impact[impact>=0])))  #Normalise the impact index (0-100)
    data_latest['Impact_index'] = impact.astype(int)  #Add impact index to the data
    return data_latest
```

<div id="section-4" class="usecase-section-heading"><b>4. Visualise the analytical results</b></div>

<div id="subsection-4-1" class="usecase-sub-section-heading"><b>4.1. Plot a bar chart of the predicted results</b></div>

<div id="subsection-4-2" class="usecase-sub-section-heading"><b>4.2. Plot the map using Folium Choropleth</b></div>

<div class="usecase-body">This function fetches the CLUE block boundaries directly from the City of Melbourne Open Data API (as GeoJSON, using the same API pattern as the other datasets in this notebook, so no boundary file needs to be stored in the repository), joins them to the predicted Impact index for each block, and shades each block on a green colour scale (darker green = higher impact) so a city planner can immediately see where the most impacted blocks are located. The map uses CartoDB Positron tiles, which need no API key.</div>

```python
def get_clue_blocks_geojson():
    """
    Fetch the CLUE block boundaries directly from the City of Melbourne Open Data API as GeoJSON,
    using the same API pattern as API_Unlimited() above, so no boundary file needs to be stored
    in the repository.

    Returns
    -------
    dict or None
        The GeoJSON FeatureCollection of CLUE block boundaries, or None if the request fails.
    """
    base_url = 'https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/'
    dataset_name = 'blocks-for-census-of-land-use-and-employment-clue'
    url = f'{base_url}{dataset_name}/exports/geojson'
    params = {'lang': 'en', 'timezone': 'UTC'}

    response = requests.get(url, params=params)

    if response.status_code == 200:
        return response.json()
    else:
        print(f'Request failed with status code {response.status_code}')
        return None
```

```python
def plot_map(clue_data):
    #Fetch the CLUE block boundaries live from the Melbourne Open Data API
    clue_geo = get_clue_blocks_geojson()

    clue_data.reset_index(inplace=True)  #turn the index (block_id) into a column in the dataframe (clue_data)
    clue_data['block_id'] = clue_data['block_id'].astype(str)  #convert block_id to str, so that it is compatible with the geojson content

    #Create the initial map (CartoDB positron tiles - no API key required)
    fmap = folium.Map(location=[-37.811600, 144.964610],
              tiles='CartoDB positron',
              width='70%',
              height='100%',
              zoom_start=13)

    #Create the choropleth layer and add it to the map above (Greens scale to match the MOP Primary Green)
    choropleth = folium.Choropleth(
        geo_data=clue_geo,
        name='choropleth',
        data=clue_data,
        columns=['block_id','Impact_index'],
        key_on='feature.properties.block_id',
        fill_color='Greens',
        fill_opacity=1,
        line_opacity=0.5,
        nan_fill_color='lightgray',
        nan_fill_opacity=0.2,
        highlight=True,
        legend_name='Impact index'
    ).add_to(fmap)

    #Add more layers and tooltips (block_id and area_name are the property names on the CLUE blocks dataset;
    #if the API ever changes its property names, print(clue_geo['features'][0]['properties'].keys()) to check)
    choropleth.geojson.add_child(folium.features.GeoJsonTooltip(['block_id','area_name'],aliases=['Block ID:','CLUE area:'],labels=True))
    return fmap
```

<div class="usecase-body">This function ties everything together: it takes a copy of the merged data already in memory from Section 2.4, filters it to the selected years and features, builds and evaluates the regression model, predicts the 2020 parking shortfall, then displays a bar chart, a results table and the choropleth map for the blocks above the selected impact threshold.</div>

```python
#Define function to execute when using the interactive tools
def show_interaction(Year_to_include, Employment, Residential_dwellings, Impact_level):
    #Reuse the merged dataset already built in Section 2.4 (no file read/write needed) - work on a copy
    #so repeated interactions never mutate the original in-memory 'data' table
    interactive_data = data.copy()
    print('='*23)
    print('|DATA ANALYSIS RESULTS|')
    print('='*23)
    print('(NOTE: The city blocks without parking information available are ignored by our analysing model.)\n\n')
    if (not(Employment)) & (not(Residential_dwellings)):
        Employment = True  #Prevent both data features being unticked ==> Error
    #Get the filtered data for analysis
    X, Y = Get_data(interactive_data,Year_to_include,Employment,Residential_dwellings)
    #Build regression model
    lm = lm_model(X,Y)
    #Get the predicted results
    predict_results = Get_predict_result(interactive_data,lm, Employment, Residential_dwellings)
    predict_results = predict_results[predict_results['Impact_index']>=Impact_level]  #only plot city blocks that have a higher impact level

    #Plot the bar chart of the predicted results
    bar_plot(predict_results, Impact_level);
    #Show table of the impacted blocks in a detail table
    print('\n\n The table below shows the impacted blocks (limited parking) in order of the impact index.\n')
    display(predict_results.sort_values('Impact_index',ascending = False))
    #Plot the folium choropleth map
    print('The impacted city blocks can be seen via the map here:')
    display(plot_map(predict_results))
```
