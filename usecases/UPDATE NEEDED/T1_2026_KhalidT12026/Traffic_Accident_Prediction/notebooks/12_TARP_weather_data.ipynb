The project now uses a single consolidated weather file instead of separate monthly and historical extracts. The file is loaded first so its structure can be inspected before any standardisation or cleaning is applied.

```python
def read_weather_csv(filename):
    path = raw_data_dir / filename
    if not path.exists():
        raise FileNotFoundError(f"Could not find file: {path}")

    try:
        df = pd.read_csv(path, delimiter=",", encoding="ISO-8859-1", header=5)
        if "Date" in df.columns:
            return df
    except Exception:
        pass

    for header_row in range(0, 10):
        try:
            df = pd.read_csv(path, delimiter=",", encoding="ISO-8859-1", header=header_row)
            if "Date" in df.columns:
                return df
        except Exception:
            continue

    raw_df = pd.read_csv(path, delimiter=",", encoding="ISO-8859-1", header=None)
    for idx in range(min(10, len(raw_df))):
        candidate = raw_df.iloc[idx].astype(str).tolist()
        if "Date" in candidate:
            raw_df.columns = raw_df.iloc[idx]
            return raw_df.iloc[idx + 1 :].reset_index(drop=True)

    raise ValueError(f"Header row not found in {path}")
```

After loading, the dataset is placed into a dictionary for inspection. Even though the project now uses one source file, keeping it in a named collection makes the later auditing and cleaning steps easier to follow.

<div class="usecase-section-header">Cleaning</div>

The cleaning function below standardises the consolidated weather archive into one consistent structure. It keeps only the required fields, resolves small naming differences, and drops unsupported columns such as cloud amount and other detail fields that are not part of the final specification.

```python
def standardise_weather_file(df, source_name):
    renamed = df.rename(columns={
        "Minimum temperature (°C)": "Min temperature (°C)",
        "Maximum temperature (°C)": "Max temperature (°C)",
        "Rainfall (mm)": "Total Rainfall (mm)",
        "Time of maximum  gust": "Time of maximum gust",
        "Time of maximum wind gust": "Time of maximum gust",
        "Direction of maximum wind gust ": "Direction of wind",
        "Direction of maximum wind gust": "Direction of wind",
        "Speed of maximum wind gust (km/h)": "max gust speed(km/h)",
        "Maximum wind gust speed (km/h)": "max gust speed(km/h)",
        "max  gust speed(km/h)": "max gust speed(km/h)",
    }).copy()

    cleaned = pd.DataFrame()
    cleaned["Date"] = renamed["Date"] if "Date" in renamed.columns else np.nan
    cleaned["Min temperature (°C)"] = renamed["Min temperature (°C)"] if "Min temperature (°C)" in renamed.columns else np.nan
    cleaned["Max temperature (°C)"] = renamed["Max temperature (°C)"] if "Max temperature (°C)" in renamed.columns else np.nan
    cleaned["Total Rainfall (mm)"] = renamed["Total Rainfall (mm)"] if "Total Rainfall (mm)" in renamed.columns else np.nan
    cleaned["Time of maximum gust"] = renamed["Time of maximum gust"] if "Time of maximum gust" in renamed.columns else np.nan
    cleaned["Direction of wind"] = renamed["Direction of wind"] if "Direction of wind" in renamed.columns else np.nan

    if "max wind speed (km/h)" in renamed.columns:
        cleaned["max wind speed (km/h)"] = renamed["max wind speed (km/h)"]
    elif "9am wind speed (km/h)" in renamed.columns:
        cleaned["max wind speed (km/h)"] = renamed["9am wind speed (km/h)"]
    elif "3pm wind speed (km/h)" in renamed.columns:
        cleaned["max wind speed (km/h)"] = renamed["3pm wind speed (km/h)"]
    else:
        cleaned["max wind speed (km/h)"] = np.nan

    cleaned["max gust speed(km/h)"] = renamed["max gust speed(km/h)"] if "max gust speed(km/h)" in renamed.columns else np.nan

    if "max pressure (hPa)" in renamed.columns:
        cleaned["max pressure (hPa)"] = renamed["max pressure (hPa)"]
    elif "9am MSL pressure (hPa)" in renamed.columns:
        cleaned["max pressure (hPa)"] = renamed["9am MSL pressure (hPa)"]
    elif "3pm MSL pressure (hPa)" in renamed.columns:
        cleaned["max pressure (hPa)"] = renamed["3pm MSL pressure (hPa)"]
    else:
        cleaned["max pressure (hPa)"] = np.nan

    if "Max Humidity (%)" in renamed.columns:
        cleaned["Max Humidity (%)"] = renamed["Max Humidity (%)"]
    elif "9am relative humidity (%)" in renamed.columns:
        cleaned["Max Humidity (%)"] = renamed["9am relative humidity (%)"]
    else:
        cleaned["Max Humidity (%)"] = np.nan

    if "Min Humidity (%)" in renamed.columns:
        cleaned["Min Humidity (%)"] = renamed["Min Humidity (%)"]
    elif "3pm relative humidity (%)" in renamed.columns:
        cleaned["Min Humidity (%)"] = renamed["3pm relative humidity (%)"]
    else:
        cleaned["Min Humidity (%)"] = np.nan

    cleaned["source_file"] = source_name
    return cleaned
```

<div class="usecase-section-header">Handle missing weather observations</div>

Once the archive has been cleaned, the dataset needs a final quality pass. This step removes blank dates, converts numeric weather values safely, and keeps missing observations as null values rather than forcing potentially misleading replacements.

```python
weather_df["Date"] = weather_df["Date"].astype(str).str.strip()
weather_df = weather_df[weather_df["Date"].notna() & (weather_df["Date"] != "")].copy()

numeric_columns = [
    "Min temperature (°C)",
    "Max temperature (°C)",
    "Total Rainfall (mm)",
    "max wind speed (km/h)",
    "max gust speed(km/h)",
    "max pressure (hPa)",
    "Max Humidity (%)",
    "Min Humidity (%)",
]

for col in numeric_columns:
    weather_df[col] = pd.to_numeric(weather_df[col], errors="coerce")

weather_df.isna().sum()
```

<div class="usecase-section-header">Convert weather timestamps</div>

The original weather date is converted into a true datetime field and then expanded into separate day, month, and year columns. These extra calendar fields make later analysis much easier to organise.

```python
weather_df["Date"] = pd.to_datetime(weather_df["Date"], errors="coerce", dayfirst=True)
weather_df = weather_df.dropna(subset=["Date"]).copy()

weather_df["day"] = weather_df["Date"].dt.day
weather_df["month"] = weather_df["Date"].dt.month
weather_df["year"] = weather_df["Date"].dt.year
weather_df["month_name"] = weather_df["Date"].dt.month_name()
weather_df["weekday"] = weather_df["Date"].dt.day_name()

weather_df = weather_df.sort_values("Date").reset_index(drop=True)
weather_df[["Date", "day", "month", "year", "month_name", "weekday"]].head()
```

<div class="usecase-section-header">Create rain indicator variable</div>

A rain indicator variable is added so that each date can be classified quickly as a rainy or non-rainy day. This helps with both descriptive summaries and later modelling work.

```python
weather_df["rain_indicator"] = np.where(weather_df["Total Rainfall (mm)"].fillna(0) > 0, 1, 0)
weather_df[["Date", "Total Rainfall (mm)", "rain_indicator"]].head()
```

The cleaned observations are now exported as one merged file covering the full 2015 to 2026 study period. This cleaned weather file is saved into the `cleaned` folder and becomes the official input for the exploratory analysis and feature engineering sections that follow.

```python
cleaned_file = cleaned_dir / "cleaned_merged_weather_2015_2026.csv"
weather_df.to_csv(cleaned_file, index=False)
print(f"Saved cleaned merged dataset to: {cleaned_file.resolve()}")
```

To make the workflow explicit, the notebook now reads the exported merged file back into memory. This confirms that the later analysis uses the saved cleaned dataset rather than an intermediate dataframe in memory.

<div class="usecase-section-header">Exploratory Data Analysis</div>

This summary gives a first overview of the cleaned weather dataset exported from the consolidated 2015 to 2026 archive. It helps confirm the spread of temperature, rainfall, humidity, pressure, and wind measurements across the full study period before the engineered features are added.

```python
analysis_df.describe(include="all")
```

Average monthly minimum and maximum temperatures are calculated from the merged weather file. This provides a straightforward view of seasonal temperature patterns across all available years.

```python
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

monthly_temperature = (
    analysis_df.groupby("month_name", as_index=False)
    .agg(
        avg_min_temp=("Min temperature (°C)", "mean"),
        avg_max_temp=("Max temperature (°C)", "mean")
    )
)

monthly_temperature["month_name"] = pd.Categorical(monthly_temperature["month_name"], categories=month_order, ordered=True)
monthly_temperature = monthly_temperature.sort_values("month_name")
monthly_temperature
```

The temperature chart below gives a clearer visual comparison between average monthly minimum and maximum temperatures.

```python
plt.figure(figsize=(10, 5))
plt.plot(monthly_temperature["month_name"], monthly_temperature["avg_min_temp"], marker="o", label="Average minimum temperature")
plt.plot(monthly_temperature["month_name"], monthly_temperature["avg_max_temp"], marker="o", label="Average maximum temperature")
plt.xticks(rotation=45)
plt.ylabel("Temperature (°C)")
plt.title("Average temperature patterns by month (2015–2026)")
plt.legend()
plt.tight_layout()
plt.show()
```

Rainfall is summarised in two ways: the total rainfall recorded in each month and the number of rainy days in each month. Together, these show both intensity and frequency.

```python
monthly_rainfall = (
    analysis_df.groupby("month_name", as_index=False)
    .agg(
        total_rainfall_mm=("Total Rainfall (mm)", "sum"),
        rainy_day_count=("rain_indicator", "sum")
    )
)

monthly_rainfall["month_name"] = pd.Categorical(monthly_rainfall["month_name"], categories=month_order, ordered=True)
monthly_rainfall = monthly_rainfall.sort_values("month_name")
monthly_rainfall
```

The rainfall chart highlights which months are wetter across the full merged study period.

```python
plt.figure(figsize=(10, 5))
plt.bar(monthly_rainfall["month_name"], monthly_rainfall["total_rainfall_mm"])
plt.xticks(rotation=45)
plt.ylabel("Rainfall (mm)")
plt.title("Rainfall totals by month (2015–2026)")
plt.tight_layout()
plt.show()
```

Grouping by weekday gives a simple consistency check across the cleaned dataset and also provides another way to profile typical daily conditions.

```python
weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

weekday_profile = (
    analysis_df.groupby("weekday", as_index=False)
    .agg(
        avg_min_temp=("Min temperature (°C)", "mean"),
        avg_max_temp=("Max temperature (°C)", "mean"),
        avg_rainfall=("Total Rainfall (mm)", "mean"),
        rainy_days=("rain_indicator", "sum")
    )
)

weekday_profile["weekday"] = pd.Categorical(weekday_profile["weekday"], categories=weekday_order, ordered=True)
weekday_profile = weekday_profile.sort_values("weekday")
weekday_profile
```

The next ranking highlights extreme weather days in the merged file by looking at high temperature, rainfall, strong gusts, and large daily temperature range.

```python
extreme_weather_days = analysis_df.copy()
extreme_weather_days["temperature_range"] = extreme_weather_days["Max temperature (°C)"] - extreme_weather_days["Min temperature (°C)"]

extreme_weather_days = extreme_weather_days.sort_values(
    by=["Max temperature (°C)", "Total Rainfall (mm)", "max gust speed(km/h)", "temperature_range"],
    ascending=False
)

extreme_weather_days[
    ["Date", "Min temperature (°C)", "Max temperature (°C)", "Total Rainfall (mm)", "max gust speed(km/h)", "temperature_range", "source_file"]
].head(10)
```

<div class="usecase-section-header">Feature Engineering</div>

The final stage creates a modelling-ready version of the cleaned weather file. In addition to numeric derived features, the wind direction variable is one-hot encoded so that directional categories such as N, SW, and E can be used more effectively in later machine learning workflows.

```python
analysis_df["temperature_range"] = analysis_df["Max temperature (°C)"] - analysis_df["Min temperature (°C)"]
analysis_df["humidity_range"] = analysis_df["Max Humidity (%)"] - analysis_df["Min Humidity (%)"]
analysis_df["gust_to_wind_ratio"] = analysis_df["max gust speed(km/h)"] / analysis_df["max wind speed (km/h)"]
analysis_df["is_hot_day"] = np.where(analysis_df["Max temperature (°C)"] >= 30, 1, 0)
analysis_df["is_heavy_rain_day"] = np.where(analysis_df["Total Rainfall (mm)"] >= 10, 1, 0)

analysis_df["Direction of wind"] = analysis_df["Direction of wind"].astype(str).str.strip().replace({"nan": np.nan, "": np.nan})
wind_direction_dummies = pd.get_dummies(
    analysis_df["Direction of wind"],
    prefix="wind_dir",
    dummy_na=False,
    dtype=int
)

analysis_df = pd.concat([analysis_df, wind_direction_dummies], axis=1)
analysis_df.head()
```

The completed analysis dataset is exported into the `cleaned` folder as the final submission file. This version includes the cleaned weather data from the single 2015 to 2026 archive plus the engineered features and one-hot encoded wind-direction fields.

```python
final_file = cleaned_dir / "cleaned_merged_weather_2015_2026_with_features.csv"
analysis_df.to_csv(final_file, index=False)
print(f"Saved final analysis dataset to: {final_file.resolve()}")
```
