## Style guide reference

### Headings

Use the HTML classes shown throughout the example above and defined in the CSS cell:

- `usecase-header-bar` for the main title of the usecase, to set the heading apart
- `usecase-title` for the main title of the usecase
- `usecase-details` for the date, duration, level, author, and skill metadata fields
- `usecase-section-heading` for numbered top-level sections (1., 2., 3.)
- `usecase-sub-section-heading` for numbered subsections (1.1., 1.2.)
- `usecase-sub-sub-section-heading` for numbered sub-subsections (1.1.1.) — only use this if a subsection genuinely needs breaking down further
- `usecase-datasets` for highlighting the list of data sets used in the analysis
- `usecase-list` for wrapping standard bulleted or numbered lists to ensure proper spacing and font
- `usecase-unnumbered-heading` for unnumbered headings at the start and end of the use case (e.g., Scenario, Background, Contents, Conclusion)
- `usecase-contents` for the contents page
- `usecase-code-divider` for the block separating template intro from code
- `usecase-body` for any general body of text

### Clickable Table of Contents

To ensure users can click the Contents list and jump to the relevant section, do the following:

1. **Add a unique ID to your section heading tag:** `<div id="my-section" class="usecase-section-heading"><b>Heading Title</b></div>`
2. **Wrap your contents list item in an anchor tag pointing to that ID:** `<a href="#my-section">Heading Title</a>`
