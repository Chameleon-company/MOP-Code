## Scenario

This project focuses on analysing traffic accident data to understand and predict accident severity. The aim is to identify patterns linked with **Fatal accident**, **Serious injury accident**, and **Other injury accident**, and to provide data-driven insights that can support road safety planning and decision-making.

The project combines crash, weather, mobility, temporal, and road-related features to explore accident patterns, train prediction models, interpret key risk factors, and prepare dashboard-ready insights.


## What this use case demonstrates

This use case demonstrates how to:

- prepare and validate an integrated traffic accident dataset
- explore accident severity patterns using visual analysis
- compare baseline and advanced machine learning models
- evaluate models using accuracy, macro F1-score, weighted F1-score, precision, recall, and class-wise metrics
- interpret model behaviour using feature importance
- understand the impact of class imbalance on rare fatal accident prediction
- prepare insights for reporting and dashboard development




## Datasets Used

The project used an integrated traffic accident dataset that combined:

- historical crash records and severity labels
- road and speed zone information
- temporal features such as hour of day, rush hour, and weekend flag
- weather features including rainfall, temperature, humidity, and wind speed
- mobility features including pedestrian density and bicycle density

After validation, the final modelling dataset contained **151,947 records** and three target severity classes:

- Fatal accident
- Serious injury accident
- Other injury accident


## Implementation Summary

The project followed a structured analytics and modelling workflow:

1. **Data validation and preprocessing**  
   The final integrated dataset was checked for missing values, duplicates, feature consistency, and target class suitability.

2. **Exploratory analysis and feature visualisation**  
   Severity patterns were explored across speed zone, time, weather, mobility, and other selected features.

3. **Baseline modelling**  
   Logistic Regression and Decision Tree models were trained to create benchmark results.

4. **Advanced modelling**  
   Random Forest and XGBoost models were trained and compared against the baseline models.

5. **Model evaluation and interpretation**  
   Models were evaluated using overall and class-wise metrics, with special focus on the rare but safety-critical Fatal accident class.

6. **Dashboard and reporting preparation**  
   Model comparison results, class-wise performance, feature importance outputs, and key findings were prepared for final dashboard and report use.



## Main Findings

The advanced models performed better than the baseline models. **Random Forest** was selected as the best final model because it achieved the strongest overall balance across accuracy, macro F1-score, and weighted F1-score.

**XGBoost** achieved higher recall for the Fatal accident class, meaning it detected more actual fatal accident cases. However, it also produced more false positives, so Random Forest was preferred overall.

The most important feature across both advanced models was **speed zone**, confirming that road speed environment was the strongest predictor of accident severity. Weather, mobility, and time-related features also contributed to prediction, but they were less influential than speed zone.

The project also confirmed that Fatal accident prediction remains challenging because fatal crashes are highly under-represented in the dataset. This means accuracy alone is not enough for evaluation, and class-wise metrics are essential.


## Conclusion

This project demonstrates how traffic accident data can be prepared, analysed, modelled, and interpreted to support road safety decision-making. The final model, Random Forest, provided the best overall performance, while fatal accident prediction remains the most important area for future improvement.

The project findings can support local government, transport planners, and road safety teams by helping identify key risk patterns, prioritise further investigation, and communicate accident severity insights through dashboard-based reporting.
