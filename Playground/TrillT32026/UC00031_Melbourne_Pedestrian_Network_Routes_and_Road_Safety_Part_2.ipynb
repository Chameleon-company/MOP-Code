```python
import pandas as pd
import folium
from folium.plugins import FastMarkerCluster, HeatMap

# 1. Load only the columns needed for mapping and diagnosis
cols = [
    "ACCIDENT_NO", "ACCIDENT_DATE", "ACCIDENT_TIME",
    "ACCIDENT_TYPE", "DAY_OF_WEEK", "DCA_CODE_DESCRIPTION", "SEVERITY",
    "LIGHT_CONDITION", "SPEED_ZONE", "ROAD_GEOMETRY",
    "ROAD_NAME", "ROAD_TYPE", "LGA_NAME",
    "DIVIDED", "RMA", "DEG_URBAN_NAME", "RUN_OFFROAD",
    "LATITUDE", "LONGITUDE", "FATALITY", "SERIOUSINJURY",
    "PEDESTRIAN", "BICYCLIST", "MOTORCYCLIST"
]
file = r"C:\Users\route\Downloads\victorian_road_crash_data.csv"

df = pd.read_csv(file, usecols=cols)
df = df.dropna(subset=["LATITUDE", "LONGITUDE"])

# Filter to a specific region or recent years to reduce clutter
# df = df[df["LGA_NAME"] == "GREATER GEELONG"]
df["ACCIDENT_DATE"] = pd.to_datetime(df["ACCIDENT_DATE"])
df = df[df["ACCIDENT_DATE"].dt.year >= 2021]

# 3. Initialise the map centred on your filtered points
m = folium.Map(
    location=[df["LATITUDE"].mean(), df["LONGITUDE"].mean()],
    zoom_start=10,
    tiles="CartoDB positron"  # Clean light background makes crash colours stand out
)

# Overall Density Heatmap (Great for spotting intersections/corridors)
heat_layer = folium.FeatureGroup(name="Crash Density (Heatmap)", show=True)
heat_data = list(zip(df["LATITUDE"], df["LONGITUDE"]))
HeatMap(heat_data, radius=12, blur=15, min_opacity=0.4).add_to(heat_layer)
heat_layer.add_to(m)

# Interactive Clusters with Diagnostic Popups (Serious & Fatal only)
severe_df = df[df["SEVERITY"].isin(["Fatal accident", "Serious injury accident"])].copy()

# Build clean HTML popup text using vectorised string formatting (fast!)
severe_df["popup_html"] = (
    "<b>" + severe_df["SEVERITY"] + "</b> (" + severe_df["ACCIDENT_DATE"].dt.strftime("%Y-%m-%d") + ")<br>"
    "<b>What happened:</b> " + severe_df["DCA_CODE_DESCRIPTION"].fillna("Unknown") + "<br>"
    "<b>Road:</b> " + severe_df["ROAD_NAME"].fillna("") + " " + severe_df["ROAD_TYPE"].fillna("") +
    " (" + severe_df["SPEED_ZONE"].fillna("") + ")<br>"
    "<b>Conditions:</b> " + severe_df["LIGHT_CONDITION"].fillna("") + " | " + severe_df["ROAD_GEOMETRY"].fillna("")
)

# Custom JavaScript callback so FastMarkerCluster displays HTML on click
callback = """
function (row) {
    var marker = L.circleMarker(new L.LatLng(row[0], row[1]), {
        radius: 6,
        color: row[3] > 0 ? '#d90429' : '#f77f00',
        fillOpacity: 0.85
    });
    marker.bindPopup(row[2]);
    return marker;
};
"""

cluster_layer = folium.FeatureGroup(name="Fatal & Serious Crashes (Click for details)", show=True)
cluster_data = list(zip(
    severe_df["LATITUDE"],
    severe_df["LONGITUDE"],
    severe_df["popup_html"],
    severe_df["FATALITY"]
))

FastMarkerCluster(data=cluster_data, callback=callback).add_to(cluster_layer)
cluster_layer.add_to(m)

# add layer toggle control in the top-right corner
folium.LayerControl(collapsed=False).add_to(m)

m.save("victoria_crash_analysis.html")
print(f"Map saved with {len(df)} heatmap points and {len(severe_df)} severe crash markers.")
```

```python
import pandas as pd

# Create a clean binary flag for severe crashes (Fatal or Serious Injury)
df["IS_SEVERE"] = df["SEVERITY"].isin(["Fatal accident", "Serious injury accident"])

# Top 10 Crash Types (DCA Codes) & their Severity Rate
dca_stats = (
    df.groupby("DCA_CODE_DESCRIPTION")
    .agg(
        total_crashes=("ACCIDENT_NO", "count"),
        severe_crashes=("IS_SEVERE", "sum"),
        fatalities=("FATALITY", "sum")
    )
    .assign(severe_pct=lambda x: (x["severe_crashes"] / x["total_crashes"] * 100).round(1))
    .sort_values(by="severe_crashes", ascending=False)
    .head(10)
)
print("\n--- TOP 10 CRASH PATTERNS BY SEVERE COUNT ---")
print(dca_stats)

# light Condition vs. Severity Rate (Does darkness make crashes deadlier?)
light_stats = (
    df.groupby("LIGHT_CONDITION")
    .agg(
        total_crashes=("ACCIDENT_NO", "count"),
        severe_crashes=("IS_SEVERE", "sum"),
        fatalities=("FATALITY", "sum")
    )
    .assign(severe_pct=lambda x: (x["severe_crashes"] / x["total_crashes"] * 100).round(1))
    .sort_values(by="severe_pct", ascending=False)
)
print("\n--- CRASH SEVERITY BY LIGHT CONDITION ---")
print(light_stats)

# Road Geometry & Speed Zone Hotspots
road_stats = (
    df.groupby(["ROAD_GEOMETRY", "SPEED_ZONE"])
    .agg(
        total_crashes=("ACCIDENT_NO", "count"),
        severe_crashes=("IS_SEVERE", "sum")
    )
    .assign(severe_pct=lambda x: (x["severe_crashes"] / x["total_crashes"] * 100).round(1))
    .sort_values(by="severe_crashes", ascending=False)
    .head(10)
)
print("\n--- TOP ROAD GEOMETRY & SPEED COMBINATIONS ---")
print(road_stats)

# top Worst LGAs (Local Government Areas)
lga_stats = (
    df.groupby("LGA_NAME")
    .agg(
        total_crashes=("ACCIDENT_NO", "count"),
        severe_crashes=("IS_SEVERE", "sum"),
        pedestrians=("PEDESTRIAN", "sum"),
        cyclists=("BICYCLIST", "sum"),
        motorcyclists=("MOTORCYCLIST", "sum")
    )
    .sort_values(by="severe_crashes", ascending=False)
    .head(10)
)
print("\n--- TOP 10 LGAs BY SEVERE CRASHES ---")
print(lga_stats)
```

#### Checking in what conditions serious injury accidents happens

```python
import pandas as pd

def find_relationship(df, field1="SEVERITY", field2=None, normalize="index"):
    """
    Creates a percentage matrix showing how `field1` (default: SEVERITY)
    is distributed across each category of `field2`.
    """
    # Compute cross-tabulation (normalize='index' gives row percentages summing to 100%)
    matrix = pd.crosstab(
        index=df[field2].fillna("Unknown"),
        columns=df[field1].fillna("Unknown"),
        normalize=normalize
    )
    
    # Convert proportions (0.452) to clean percentages (45.2%) if normalized
    if normalize:
        matrix = (matrix * 100).round(1)
        
    # Add a total crash count column so you know the sample size for each row
    matrix["Total_Crashes"] = df[field2].fillna("Unknown").value_counts()
    
    return matrix.sort_values(by="Total_Crashes", ascending=False)

# --- key relationship matrices ---
print("\n=== 1. SEVERITY BY SPEED ZONE (%) ===")
print(find_relationship(df, field1="SEVERITY", field2="SPEED_ZONE"))

print("\n=== 2. SEVERITY BY LIGHT CONDITION (%) ===")
print(find_relationship(df, field1="SEVERITY", field2="LIGHT_CONDITION"))

print("\n=== 3. SEVERITY BY ROAD GEOMETRY (%) ===")
print(find_relationship(df, field1="SEVERITY", field2="ROAD_GEOMETRY"))

print("\n=== 4. SEVERITY BY DIVIDED VS UNDIVIDED ROAD (%) ===")
print(find_relationship(df, field1="SEVERITY", field2="DIVIDED"))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Combine the main standard speed zones so tiny sample sizes (like 75 km/hr) don't clutter the chart
main_speeds = ["40 km/hr", "50 km/hr", "60 km/hr", "70 km/hr", "80 km/hr", "100 km/hr", "110 km/hr"]
clean_df = df[df["SPEED_ZONE"].isin(main_speeds)].copy()

# Calculate combined Fatal + Serious Injury rate (%) by Speed Zone
speed_table = find_relationship(clean_df, field1="SEVERITY", field2="SPEED_ZONE")
speed_table = speed_table.reindex(main_speeds)

# Pick the two severity columns from speed_table
import matplotlib.pyplot as plt

top_col = "Serious injury accident"
bottom_col = "Fatal accident"

top_vals = speed_table[top_col]
bottom_vals = speed_table[bottom_col]

# Include sample size (n) on x-axis labels for context
x_labels = [
    f"{zone}\n(n={int(n):,})"
    for zone, n in zip(speed_table.index, speed_table["Total_Crashes"])
]
x_pos = range(len(speed_table.index))

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10.5, 6.8), sharex=True)

# Top Panel: Serious Injury Accident (%)
bars1 = ax1.bar(x_pos, top_vals, color="#f77f00", width=0.55)
ax1.set_ylabel("Serious Injury (%)")
ax1.set_title(
    "Serious Injury vs. Fatal Crash Share by Speed Zone (Victoria)",
    fontsize=13,
    pad=12,
)
ax1.set_ylim(0, max(top_vals.max() * 1.2, 1))
ax1.grid(axis="y", linestyle="--", alpha=0.4)
ax1.spines[["top", "right"]].set_visible(False)

top_pad = max(top_vals.max() * 0.02, 0.4)
for bar in bars1:
  h = bar.get_height()
  ax1.text(
      bar.get_x() + bar.get_width() / 2,
      h + top_pad,
      f"{h:.1f}%",
      ha="center",
      fontsize=9,
      fontweight="bold",
  )

# Bottom Panel: Fatal Accident (%) on its own scale
bars2 = ax2.bar(x_pos, bottom_vals, color="#d90429", width=0.55)
ax2.set_ylabel("Fatal Accident (%)")
ax2.set_xticks(x_pos)
ax2.set_xticklabels(x_labels)
ax2.set_ylim(0, max(bottom_vals.max() * 1.25, 0.5))
ax2.grid(axis="y", linestyle="--", alpha=0.4)
ax2.spines[["top", "right"]].set_visible(False)

bottom_pad = max(bottom_vals.max() * 0.03, 0.08)
for bar in bars2:
  h = bar.get_height()
  ax2.text(
      bar.get_x() + bar.get_width() / 2,
      h + bottom_pad,
      f"{h:.1f}%",
      ha="center",
      fontsize=9,
      fontweight="bold",
  )

plt.tight_layout()
plt.show()
```

Next, we investigate in what times of the day we have the highest frequency of seriously injuried pedestrian.  We can do this by taking the number of pedestrians involved in a crash, taking the accident type from the time of the crash (ACCIDENT_TIME) and converting to a string, converting to a datetime and then for each time of the day creating a plot recording the number of pedestrians having accidents at this time.

```python
# Filter to pedestrian crashes
ped_crashes = df[df["PEDESTRIAN"] > 0].copy()

# Extract hour (0-23) from ACCIDENT_TIME (handles ':' or '.' separators because dont know)
time_str = ped_crashes["ACCIDENT_TIME"].astype(str).str.strip()
parsed_hour = pd.to_datetime(
    time_str.str.replace(".", ":", regex=False), format="%H:%M:%S", errors="coerce"
).dt.hour

# Fallback if ACCIDENT_TIME is already numeric/integer hours
ped_crashes["HOUR"] = parsed_hour.fillna(
    pd.to_numeric(ped_crashes["ACCIDENT_TIME"], errors="coerce")
)
ped_crashes = ped_crashes.dropna(subset=["HOUR"]).copy()
ped_crashes["HOUR"] = ped_crashes["HOUR"].astype(int)

# label hours cleanly as "00:00", "01:00", ..., "23:00"
ped_crashes["HOUR_GROUP"] = ped_crashes["HOUR"].apply(lambda h: f"{h:02d}:00")

frequency_by_daytime = find_relationship(
    ped_crashes, field1="SEVERITY", field2="HOUR_GROUP"
).sort_values("Serious injury accident", ascending=True)

fig, ax = plt.subplots(figsize=(12, 4.5))

bars = ax.bar(
    frequency_by_daytime.index,
    frequency_by_daytime["Serious injury accident"],
    color="#7209b7",
    alpha=0.85,
)

ax.bar_label(bars, fmt="%.1f%%", padding=3, fontsize=8.5)

ax.set_title(
    "Serious Injury Share of Pedestrian Crashes by Hour of Day",
    fontsize=13,
    pad=12,
)
ax.set_xlabel("Time of Day (Hour)")
ax.set_ylabel("Serious Injury Share (%)")
ax.set_ylim(0, frequency_by_daytime["Serious injury accident"].max() * 1.15)
ax.tick_params(axis="x", rotation=45, labelsize=9)
ax.tick_params(axis="y", labelsize=10)
ax.grid(axis="y", linestyle="--", alpha=0.4)
ax.spines[["top", "right"]].set_visible(False)

plt.tight_layout()
plt.show()
```

And we can hone in on particular accidents affecting only cyclists, which we can refer to by saying ["BICYCLISTS"] > 0, and copying that to a new crash dataset for us to run tests on.

In this code we will see how we can take this smaller dataset and plot the relatinship between the highest number of fatal cyclist accidents against the streets on which they occur.  This directly answers one of our questions towards pedestrian and cyclist safety. 

```python
# Filter to cyclist crashes
cyc_crashes = df[df["BICYCLIST"] > 0].copy()

# get ROAD_NAME and ROAD_TYPE into a single label and drop blank names
road_name = cyc_crashes["ROAD_NAME"].fillna("").astype(str).str.strip()
road_type = cyc_crashes["ROAD_TYPE"].fillna("").astype(str).str.strip()

cyc_crashes["STREET"] = (road_name + " " + road_type).str.strip()
cyc_crashes = cyc_crashes[cyc_crashes["STREET"] != ""].copy()

top10_cyc_counts = (
    cyc_crashes.groupby("STREET")
    .agg(
        total_crashes=("SEVERITY", "size"),
        serious_or_fatal=(
            "SEVERITY",
            lambda s: s.isin(["Serious injury accident", "Fatal accident"]).sum(),
        ),
    )
    .sort_values("serious_or_fatal", ascending=False)
    .head(10)
    .sort_values("serious_or_fatal", ascending=True)
)

count_labels = [
    f"{street} (of {int(tot)})"
    for street, tot in zip(
        top10_cyc_counts.index, top10_cyc_counts["total_crashes"]
    )
]

fig, ax = plt.subplots(figsize=(10.5, 5.2))

bars = ax.barh(
    count_labels,
    top10_cyc_counts["serious_or_fatal"],
    color="#d90429",
    alpha=0.85,
)

ax.bar_label(bars, fmt="%d", padding=4, fontsize=9, fontweight="bold")

ax.set_title(
    "Top 10 Streets by Serious & Fatal Crash Count — Cyclist Crashes",
    fontsize=12.5,
    pad=12,
)
ax.set_xlabel("Serious + Fatal Crashes")
ax.set_xlim(0, max(top10_cyc_counts["serious_or_fatal"].max() * 1.15, 2))
ax.grid(axis="x", linestyle="--", alpha=0.4)
ax.spines[["top", "right"]].set_visible(False)

plt.tight_layout()
plt.show()
```

In this section of the workbook, we focus on using Folium to show the parts of the most severe parts of Melbourne to cyclists in terms of fatalaties in the form of a graph. 

```python
import folium
import geopandas as gpd
import numpy as np
import pandas as pd
import shapely.geometry

# Filter to cyclist crashes with coordinates and serious/fatal injury severity
severe_mask = (
    df["BICYCLIST"].gt(0)
    & df["LATITUDE"].notna()
    & df["LONGITUDE"].notna()
    & df["SEVERITY"].isin(["Serious injury accident", "Fatal accident"])
)
severe_cyc = df[severe_mask].copy()

# Create point geometries in WGS84 (EPSG:4326), then project to metres (EPSG:7855)
gdf_severe = gpd.GeoDataFrame(
    severe_cyc,
    geometry=gpd.points_from_xy(
        severe_cyc["LONGITUDE"], severe_cyc["LATITUDE"]
    ),
    crs="EPSG:4326",
).to_crs("EPSG:7855")
```

The below code creates a grid measuring the Postcodes in Melbourne and then creating a severity socring for each of them in terms of the fatalaties to cyclists as well as the use of the severe crashes.

```python
def create_grid(gdf, n_cells=15):
  """Creates a regular square grid over the bounding box of gdf."""
  xmin, ymin, xmax, ymax = gdf.total_bounds
  cell_size = (xmax - xmin) / n_cells
  crs = gdf.crs

  grid_cells = []
  for x0 in np.arange(xmin, xmax + cell_size, cell_size):
    for y0 in np.arange(ymin, ymax + cell_size, cell_size):
      x1 = x0 + cell_size
      y1 = y0 + cell_size
      grid_cells.append(shapely.geometry.box(x0, y0, x1, y1))

  return gpd.GeoDataFrame(grid_cells, columns=["geometry"], crs=crs)


# Create grid and count severe crashes per cell
grid = create_grid(gdf_severe, n_cells=15)
grid["cell_id"] = grid.index.astype(str)

joined = gpd.sjoin(
    gdf_severe, grid[["cell_id", "geometry"]], how="inner", predicate="within"
)
counts = joined.groupby("cell_id").size().rename("severe_crashes")

grid_counts = grid.merge(counts, on="cell_id", how="left")
grid_counts["severe_crashes"] = grid_counts["severe_crashes"].fillna(0).astype(int)

# Keep non-empty cells (or keep all cells to show zero-count squares)
active_grid = grid_counts[grid_counts["severe_crashes"] > 0].copy()
```

This can then be mapped out to an html or viewed in the Jupiter Notebook.

```python
grid_4326 = active_grid.to_crs("EPSG:4326")

# Center map on the centroid of severe cyclist crashes
center_lat = severe_cyc["LATITUDE"].mean()
center_lon = severe_cyc["LONGITUDE"].mean()

m = folium.Map(
    location=[center_lat, center_lon], zoom_start=12, tiles="CartoDB positron"
)

folium.Choropleth(
    geo_data=grid_4326.__geo_interface__,
    data=grid_4326,
    columns=["cell_id", "severe_crashes"],
    key_on="feature.properties.cell_id",
    fill_color="YlOrRd",
    fill_opacity=0.65,
    line_opacity=0.4,
    legend_name="Severe & Fatal Cyclist Crashes per Grid Cell",
).add_to(m)

# Add hover tooltips showing exact counts per cell
folium.GeoJson(
    grid_4326,
    style_function=lambda _: {
        "fillColor": "transparent",
        "color": "#333333",
        "weight": 0.5,
    },
    tooltip=folium.GeoJsonTooltip(
        fields=["cell_id", "severe_crashes"],
        aliases=["Grid Cell:", "Severe/Fatal Crashes:"],
    ),
).add_to(m)

m.save("victoria_cyclist_crash_grid_map.html")
print("Saved grid map of severe/fatal cyclist crashes to 'victoria_cyclist_crash_grid_map.html'.")
```
