```python
import pandas as pd
```

```python
df = pd.read_csv("cleaned_parking.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
def assign_area_by_bay(df):
    df = df.copy()

    areas = [
        "Richmond",
        "South Yarra",
        "Melbourne CBD",
        "Docklands",
        "Carlton",
        "Fitzroy",
        "St Kilda",
        "Brunswick"
    ]

    # Get unique parking bays
    unique_bays = sorted(df["bay_id"].unique())

    print("Number of unique parking bays:", len(unique_bays))

    # Map each bay_id to an area
    bay_to_area = {
        bay: areas[i % len(areas)]
        for i, bay in enumerate(unique_bays)
    }

    # Create new area column
    df["area"] = df["bay_id"].map(bay_to_area)

    return df
```

```python
df = assign_area_by_bay(df)

print("Area counts:")
print(df["area"].value_counts())

df.head()
```

```python
output_file = "cleaned_parking_with_area.csv"
df.to_csv(output_file, index=False)

print(f"Updated dataset saved as: {output_file}")
```
