<div id="sensor-status" class="usecase-section-header"><b>Real-Time Parking Bay Status</b></div>

<div class="usecase-heading-description">Fetching near real-time bay occupancy from the City of Melbourne Open Data API.</div>

We’ll connect to the City of Melbourne Open Data API to fetch near real-time parking bay occupancy data.

This dataset includes status (e.g., Unoccupied/Present), coordinates, and timestamps.  
It is updated approximately every 2 minutes.

We’ll convert the response into a pandas DataFrame and preview the first few rows.


````python
# Correct API URL (limit set to 100)
url_status = "https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets/on-street-parking-bay-sensors/records?limit=100"

# Send request with error handling
import requests
import pandas as pd

response = requests.get(url_status)

# Debugging output
print("HTTP status code:", response.status_code)
if response.status_code != 200:
    print("API request failed. Check the URL or try again later.")
    print("Response content:", response.text)

# Raise if bad
response.raise_for_status()

# Parse JSON and convert to DataFrame
data = response.json()
records_status = data.get("results", [])

df_status = pd.DataFrame(records_status)

# Preview data
print("Loaded df_status shape:", df_status.shape)
df_status.head()
````

**Output Explanation – Sensor Status Data**

We successfully loaded 100 rows from the live parking bay sensor dataset using the v2.1 City of Melbourne API.

Here are the key columns in the dataset:

- **lastupdated**: Timestamp of when the data was last pulled from the sensor network.
- **status_timestamp**: The actual timestamp at which the parking bay status was recorded.
- **zone_number**: The zone identifier (useful for restriction lookups).
- **status_description**: The occupancy status of the parking bay: `"Present"` (occupied) or `"Unoccupied"`.
- **kerbsideid**: A unique identifier for each parking bay — this will be used to merge with other datasets.
- **location**: A dictionary containing geographic coordinates (`lat`, `lon`) for mapping.

The dataset shape is **(100, 6)**.

Next, we’ll explore the structure of the data: missing values, occupancy patterns, and a quick map preview.

<div id="merge" class="usecase-section-header"><b>Merging the Live Datasets</b></div>

<div class="usecase-heading-description">Joining the live sensor status to parking restrictions on their documented key.</div>

The City of Melbourne documents how these datasets connect: the parking bay sensors join to the bay restrictions on the **bay id** attribute. To make that join work we need complete current snapshots of both datasets, not arbitrary samples. The `records` endpoint caps each request at 100 rows, and two independent 100-row samples from datasets of thousands will almost never share ids — so we use the `exports` endpoint, which returns the full dataset in one call.

````python
# Fetch complete current snapshots of both live datasets via the exports endpoint
BASE = "https://data.melbourne.vic.gov.au/api/explore/v2.1/catalog/datasets"

def fetch_full(dataset_id):
    """Return the full current dataset as a DataFrame, or None if the API is disrupted."""
    try:
        r = requests.get(f"{BASE}/{dataset_id}/exports/json", timeout=60)
        r.raise_for_status()
        return pd.DataFrame(r.json())
    except Exception as e:
        print(f"Could not fetch {dataset_id}: {e}")
        print("The City of Melbourne is upgrading its parking system and live datasets may be disrupted.")
        return None

sensors_full = fetch_full("on-street-parking-bay-sensors")
restr_full   = fetch_full("on-street-car-park-bay-restrictions")

if sensors_full is not None and restr_full is not None:
    print("sensors:", sensors_full.shape, "| restrictions:", restr_full.shape)
    print("sensor columns:", list(sensors_full.columns))
    print("restriction columns:", list(restr_full.columns))

    # The join key is the bay id; column naming differs slightly between datasets
    s_key = next((c for c in sensors_full.columns if c.lower() in ("kerbsideid", "bay_id", "bayid")), None)
    r_key = next((c for c in restr_full.columns if c.lower() in ("bay_id", "bayid", "kerbsideid")), None)
    print("join keys detected -> sensors:", s_key, "| restrictions:", r_key)

    sensors_full[s_key] = sensors_full[s_key].astype(str)
    restr_full[r_key]   = restr_full[r_key].astype(str)
    merged_live = sensors_full.merge(restr_full, left_on=s_key, right_on=r_key, how="left")

    match_rate = merged_live[r_key].notna().mean()
    print(f"merged: {merged_live.shape} | sensors with a matching restriction: {match_rate:.1%}")
````

**What the match rate tells us**

The join keys were detected correctly and both datasets fetched in full — yet at the time of writing only around 2 per cent of live sensors find a matching restriction record. This is the City of Melbourne's parking system upgrade in action: the new sensor feed and the older restrictions dataset currently occupy largely different bay-id spaces, so even a correctly specified join on the documented key returns few matches. That is worth sitting with, because it is a common real-world situation: the join logic can be right while the data is in transition. The match rate printed above is the health check — as the upgrade completes and the datasets realign, re-running this cell should show it rise. Nothing downstream depends on it: the analysis from the next section onward runs entirely on the historical archive.

<div id="historical" class="usecase-section-header"><b>Historical Parking Events — the 2019 Archive</b></div>

<div class="usecase-heading-description">Loading a pre-filtered extract of the 2019 sensor archive: every parking event, not just a snapshot.</div>

A single real-time snapshot can tell us what is happening *now*, but it cannot tell us what usually happens at 2 pm on a Tuesday — for that we need repeated observations over time. The City of Melbourne archived its sensor event data by year, and each record is an interval: a vehicle arrived at this bay at this time and departed at that time, or the bay sat vacant between these times (`VehiclePresent` distinguishes the two).

The full 2019 archive is 42.7 million rows (717 MB compressed), which is far more than a tutorial needs. This use case ships a pre-filtered extract in the DEPENDENCIES folder — **March 2019, eight central city areas, 11 columns, about 500,000 events (6.7 MB)**. It was built with the following one-off code, shown for transparency (you do not need to run it):

```python
# One-off: build the extract from the full archive
# Source: https://discover.data.vic.gov.au/dataset/on-street-car-parking-sensor-data-2019
# (717 MB zip; scanned in chunks, filtered, and saved compressed)
CORE = ['Banks', 'Princes Theatre', 'Windsor', 'Twin Towers',
        'Chinatown', 'City Square', 'Degraves', 'Library']
KEEP = ['DeviceId', 'ArrivalTime', 'DepartureTime', 'DurationMinutes', 'StreetMarker',
        'BayId', 'AreaName', 'StreetName', 'Sign', 'InViolation', 'VehiclePresent']
for chunk in pd.read_csv(archive_csv, chunksize=1_000_000, low_memory=False):
    sel = chunk[chunk['ArrivalTime'].str.startswith('03/')]      # March (dates are MM/DD/YYYY)
    sel = sel[sel['AreaName'].isin(CORE)][KEEP]
    sel.to_csv('parking_2019_march_core.csv.gz', mode='a', compression='gzip', ...)
```

One filtering note: the archive contains an area literally named `All` (about 179,000 rows in March) — an upstream artefact rather than a place, excluded from the extract.

````python
# Load the historical extract from the DEPENDENCIES folder
import os, glob

EXTRACT = "UC00163_parking_2019_march_core.csv.gz"
candidates = [
    f"../../../../../DEPENDENCIES/{EXTRACT}",   # from this notebook's folder to datascience/usecases/DEPENDENCIES
    f"../../../../DEPENDENCIES/{EXTRACT}",
    EXTRACT,                                     # same folder fallback
]
path = next((p for p in candidates if os.path.exists(p)), None)
if path is None:
    hits = glob.glob(f"../../../../../**/{EXTRACT}", recursive=True)
    path = hits[0] if hits else None
assert path is not None, f"Extract not found — place {EXTRACT} in datascience/usecases/DEPENDENCIES/"

hist = pd.read_csv(path, low_memory=False)
print("loaded:", path)
print("shape:", hist.shape)
hist.head(3)
````

````python
# Clean the extract — every fix below corresponds to a real quirk in this archive
# 1. DurationMinutes contains a handful of values with thousands separators ('1,440'),
#    which makes the column mixed-type and breaks aggregation
hist["DurationMinutes"] = pd.to_numeric(
    hist["DurationMinutes"].astype(str).str.replace(",", ""), errors="coerce")

# 2. Parse the timestamps (US format: MM/DD/YYYY HH:MM:SS AM/PM)
FMT = "%m/%d/%Y %I:%M:%S %p"
hist["arrival"]   = pd.to_datetime(hist["ArrivalTime"], format=FMT, errors="coerce")
hist["departure"] = pd.to_datetime(hist["DepartureTime"], format=FMT, errors="coerce")

before = len(hist)
hist = hist.dropna(subset=["DurationMinutes", "arrival", "departure"])

# 3. The publisher documents that some sensors recorded departures before arrivals —
#    drop those non-positive durations
hist = hist[hist["departure"] > hist["arrival"]]

# 4. Extremely long "occupied" intervals are stuck sensors, not 25-day parkers —
#    cap at 24 hours (1440 minutes)
stuck = (hist["DurationMinutes"] > 1440).sum()
hist = hist[hist["DurationMinutes"] <= 1440]

print(f"dropped {before - len(hist)} problem rows ({stuck} of them stuck-sensor intervals over 24h)")
print(f"clean events: {len(hist):,} | bays: {hist.BayId.nunique()} | areas: {hist.AreaName.nunique()}")
print("date range:", hist.arrival.min(), "->", hist.arrival.max())
````

**Why this cleaning matters**

Each of these steps fixes something genuinely present in the archive, and each is the kind of issue that silently corrupts an analysis if missed. The comma-formatted numbers made the whole column text, so any sum or mean would fail — or worse, concatenate. The reversed timestamps are a documented sensor fault. And the multi-day "occupied" intervals (the longest in this extract is over 25 days) are stuck sensors: leaving them in would make a handful of broken devices dominate every occupancy statistic. Real open data is like this, and checking for it is the first job of any analysis.

<div id="availability" class="usecase-section-header"><b>Availability Patterns</b></div>

<div class="usecase-heading-description">Measuring when parking is actually available, from a full month of labelled intervals.</div>

Because every interval is labelled with `VehiclePresent`, availability has a direct empirical definition: **the share of bay-minutes that were vacant**. We weight by duration rather than counting events — a 3-hour vacancy matters more than a 3-minute one.

````python
# Availability by hour of day — share of bay-minutes vacant
hist["hour"] = hist["arrival"].dt.hour
hist["dow"] = hist["arrival"].dt.dayofweek       # 0 = Monday
hist["is_weekend"] = (hist["dow"] >= 5).astype(int)

vacant = hist[~hist["VehiclePresent"]]
by_hour = (vacant.groupby("hour")["DurationMinutes"].sum()
           / hist.groupby("hour")["DurationMinutes"].sum())

plt.figure(figsize=(9, 4))
plt.bar(by_hour.index, by_hour.values, color="#059669")
plt.title("Share of bay-minutes vacant, by hour of day (March 2019, core CBD)")
plt.xlabel("Hour of day"); plt.ylabel("Vacant share")
plt.xticks(range(0, 24)); plt.ylim(0, 1)
plt.tight_layout(); plt.show()
print(by_hour.round(3).to_string())
````

**Reading the daily rhythm**

The pattern is strong and interpretable. Overnight, most bay-time is vacant — around 70 per cent or more in the small hours. Availability then falls steadily through the morning commute to a midday trough (roughly one bay-minute in five vacant between 10 am and 1 pm), recovers slightly mid-afternoon, and then drops to its tightest point of the whole day in the early evening, around 6–8 pm. That evening squeeze is worth pausing on: these eight areas cover Melbourne's restaurant and theatre core, and many restrictions also end around 6:30 pm, so cars arrive and stay longer. This is exactly the kind of pattern a single afternoon snapshot could never reveal — and it is the signal our model will learn.

````python
# Availability by day of week and by area
by_dow = (vacant.groupby("dow")["DurationMinutes"].sum()
          / hist.groupby("dow")["DurationMinutes"].sum())
by_area = (vacant.groupby("AreaName")["DurationMinutes"].sum()
           / hist.groupby("AreaName")["DurationMinutes"].sum()).sort_values()

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].bar(["Mon","Tue","Wed","Thu","Fri","Sat","Sun"], by_dow.values, color="#059669")
axes[0].set_title("Vacant share by day of week"); axes[0].set_ylim(0, 1)
axes[1].barh(by_area.index, by_area.values, color="#9333EA")
axes[1].set_title("Vacant share by area"); axes[1].set_xlim(0, 1)
plt.tight_layout(); plt.show()
print(by_dow.round(3).to_string()); print(); print(by_area.round(3).to_string())
````

**Day of week barely moves; area matters more**

Weekday availability varies only a few percentage points across the week — Sunday is typically the tightest, consistent with unrestricted free parking encouraging longer stays. Location, by contrast, separates clearly. For the model this sets honest expectations: hour of day should dominate, area should contribute, and day of week should add only a little. If a model claimed otherwise, we should be suspicious of it.

<div id="modelling" class="usecase-section-header"><b>Modelling: Predicting Bay Availability</b></div>

<div class="usecase-heading-description">Training a classifier to predict whether a bay is vacant, from when and where.</div>

We frame the task exactly as the scenario poses it: given a **time** and a **place**, how likely is a bay to be vacant? Each historical interval becomes a training example — features are the hour, day of week, weekend flag and area; the label is whether the bay was vacant; and each example is weighted by its duration in minutes, so long intervals count for more, matching the bay-minutes definition of availability above.

To judge the model honestly we compare it against two baselines: always predicting the majority class, and simply looking up the empirical vacant rate for that hour and area. A model that cannot beat the lookup table has not learned anything the table does not already know.

````python
# Train / evaluate: Random Forest vs two honest baselines
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, balanced_accuracy_score, f1_score, confusion_matrix

features = pd.get_dummies(hist[["hour", "dow", "is_weekend", "AreaName"]], columns=["AreaName"])
target = (~hist["VehiclePresent"]).astype(int)          # 1 = vacant
weights = hist["DurationMinutes"].values

X_tr, X_te, y_tr, y_te, w_tr, w_te = train_test_split(
    features, target, weights, test_size=0.25, random_state=42, stratify=target)

clf = RandomForestClassifier(n_estimators=150, max_depth=12, random_state=42, n_jobs=-1)
clf.fit(X_tr, y_tr, sample_weight=w_tr)
pred = clf.predict(X_te)

print("model     accuracy:", round(accuracy_score(y_te, pred, sample_weight=w_te), 3),
      "| balanced:", round(balanced_accuracy_score(y_te, pred, sample_weight=w_te), 3),
      "| F1:", round(f1_score(y_te, pred, sample_weight=w_te), 3))

# Baseline 1 — majority class
maj = int(y_tr.value_counts().idxmax())
print("majority  accuracy:", round(accuracy_score(y_te, [maj]*len(y_te), sample_weight=w_te), 3))

# Baseline 2 — empirical lookup: vacant rate per (hour, area) on the training split
tr = hist.loc[X_tr.index]
lookup = (tr[~tr.VehiclePresent].groupby(["hour","AreaName"]).DurationMinutes.sum()
          / tr.groupby(["hour","AreaName"]).DurationMinutes.sum()).rename("rate")
te = hist.loc[X_te.index]
te_rate = te.merge(lookup.reset_index(), on=["hour","AreaName"], how="left")["rate"].fillna(0.5)
print("lookup    accuracy:", round(accuracy_score(y_te, (te_rate > 0.5).astype(int), sample_weight=w_te), 3))

print("\nconfusion matrix (weighted, rows=actual, cols=predicted):")
print(confusion_matrix(y_te, pred, sample_weight=w_te).round(0))
print("\ntop feature importances:")
imp = pd.Series(clf.feature_importances_, index=features.columns).sort_values(ascending=False)
print(imp.head(8).round(3).to_string())
````

**Interpreting the results**

Three comparisons matter in the printout above. The model against the majority baseline shows whether it learned anything at all. The model against the lookup table shows whether the classifier adds value beyond memorised averages — with only time and area as features, expect this gap to be modest, and that is fine: it tells us most of the predictable signal genuinely is the daily rhythm. And the feature importances should confirm what the availability charts already showed — hour dominating, area contributing, day of week minor. When a model's internals agree with what honest descriptive analysis found first, that consistency is evidence the pipeline is sound. A note on limits: this model knows nothing about weather, events, roadworks or holidays, and it is trained on pre-pandemic 2019 patterns — a production system would retrain on current data.

<div id="prediction" class="usecase-section-header"><b>Predicting Availability for an Arrival Time</b></div>

<div class="usecase-heading-description">Answering the scenario: how likely is a bay near my destination at my arrival time?</div>

This closes the loop on the user story. Given an area and an intended arrival time, we return both the model's probability and the empirical rate from the archive — two views a real app would show a driver.

````python
# The user story, answered
def predict_availability(area, when):
    """area: one of the core CBD area names; when: any datetime string, e.g. '2026-08-04 14:00'"""
    t = pd.Timestamp(when)
    row = pd.DataFrame([{"hour": t.hour, "dow": t.dayofweek, "is_weekend": int(t.dayofweek >= 5),
                         "AreaName": area}])
    row = pd.get_dummies(row, columns=["AreaName"]).reindex(columns=features.columns, fill_value=0)
    prob = clf.predict_proba(row)[0, 1]
    emp = lookup.get((t.hour, area), float("nan"))
    print(f"{area}, {t.strftime('%A %H:%M')}  ->  model P(vacant) = {prob:.2f}"
          + (f" | historical vacant share = {emp:.2f}" if pd.notna(emp) else ""))

predict_availability("Chinatown", "2026-08-04 14:00")        # Tuesday 2 pm
predict_availability("Princes Theatre", "2026-08-07 19:30")  # Friday 7:30 pm
predict_availability("Banks", "2026-08-09 08:00")            # Sunday 8 am
````

**From notebook to app**

These three calls are the essence of the assistant the scenario describes: a driver heading to Chinatown for a 2 pm meeting, a theatre-goer arriving Friday evening, an early trip on a Sunday. A front end (Streamlit is a natural fit) would wrap this function with a destination search and show the nearest bays on the live map from earlier — the live API supplying *now*, the model supplying *usually*.
