<div class="usecase-section-header">Scenario</div>


A council inspection team has concrete surface images or video footage from a footpath, kerb, wall, deck, or other inspection area. The team wants a repeatable way to detect likely cracks, estimate crack dimensions, save evidence files, and create a plain-English inspection-support report.

This use case packages that workflow as a concrete crack detection MCP-style server. A user can send an image path, upload an image, process a video, or ask for an MCP-friendly text response. The output is not just a model prediction: it includes CSV rows, JSON summaries, text summaries, annotated visual evidence, and a report generated by Tanvi's reporting workflow.


<div class="usecase-section-header">Step-By-Step Scenario-Based Application</div>


<p class="usecase-subsection-header">Step 1: Collect inspection evidence</p>
A field officer captures still images or video of concrete surfaces. The image or video becomes the input for this system.

<p class="usecase-subsection-header">Step 2: Start the MCP-style API server</p>
The project is started with Docker Compose. Docker creates the same Python 3.13 environment each time, installs dependencies with `uv`, and exposes the FastAPI documentation page at `http://localhost:8000/docs`.

<p class="usecase-subsection-header">Step 3: Submit the inspection input</p>
For a file already on the server, use `/predict`. For a direct image upload, use `/predict/upload`. For inspection footage, use `/predict/video`.

<p class="usecase-subsection-header">Step 4: Run concrete crack detection</p>
Kavita's YOLO pipeline checks the image or selected video frames, finds likely crack areas, estimates crack dimensions, and writes outputs to the `artifacts/` folder.

<p class="usecase-subsection-header">Step 5: Request MCP-friendly output</p>
The `/mcp/predict` endpoint returns readable text and artifact paths. This makes it easier for another assistant, reporting tool, or workflow engine to use the result.

<p class="usecase-subsection-header">Step 6: Generate the inspection-support report</p>
Tanvi's report generator reads the prediction CSV, assigns practical inspection priorities, creates a Plotly chart, and writes a Markdown report, CSV, HTML chart, and JSON summary.

<p class="usecase-subsection-header">Step 7: Human review and decision</p>
A qualified person reviews the evidence. The system supports inspection and maintenance planning; it does not replace engineering judgement.

<p class="usecase-subsection-header">Code Cell Explanation: API And MCP-Style Server</p>

This code cell contains `main.py`, which is the web-server entry point written by Ishika. It uses FastAPI to expose the crack-detection system through HTTP endpoints.

The first imports bring in standard Python tools for JSON, temporary files, paths, and type hints. The FastAPI imports create the API, receive uploaded files, and return meaningful HTTP errors. Pydantic's `BaseModel` defines request shapes so the API knows what fields to expect.

The script then imports Kavita's YOLO image pipeline and Tanvi's report generator. This is the connection point between the web server and the deeper processing code.

`PredictRequest` describes image prediction settings: source path, optional model weights, confidence threshold, and device. `ReportRequest` describes report settings: prediction CSV path, JSON metadata path, output folder, site wording, and report author.

The `/predict` endpoint processes an image path already available on the server. `/predict/upload` handles a user-uploaded image by saving it temporarily and then reusing `/predict`. `/predict/video` processes video frames when the video pipeline is available. `/mcp/predict` returns a compact text-and-artifacts payload for MCP-style usage. `/report/as2870` runs Tanvi's report generator.

The final `uvicorn.run(...)` section starts the server when `main.py` is run directly. In Docker, the compose file starts this same app with `uvicorn main:app`.


````python
from __future__ import annotations

import json
import shutil
import tempfile
from pathlib import Path
from typing import Any

from fastapi import FastAPI, File, HTTPException, UploadFile
from pydantic import BaseModel

from scripts.yolo_crack_bbox_pipeline import (
    PipelineConfig,
    JSON_SUMMARY_NAME,
    TEXT_SUMMARY_NAME,
    CSV_ESTIMATES_NAME,
    estimate_predicted_crack_sizes,
)
from scripts.as2870_report_generator import ReportConfig, write_report

try:
    from scripts.yolo_crack_video_pipeline import process_video
except Exception:
    process_video = None

app = FastAPI(title="Crack Detection API")


class PredictRequest(BaseModel):
    source: str | None = None
    weights: str | None = None
    conf: float = 0.25
    device: str | None = None


class ReportRequest(BaseModel):
    predictions_csv: str | None = None
    usecase_json: str = "UC00236_Concrete_Crack_Detection_mcp_server.json"
    output_root: str = "artifacts/as2870_reports"
    site_classification: str = "Not supplied"
    location_name: str = "City of Melbourne inspection area"
    prepared_for: str = "City of Melbourne"
    prepared_by: str = "Tanvi"


def read_text_safe(path: Path) -> str:
    try:
        return path.read_text()
    except Exception:
        return ""


@app.post("/predict")
async def predict(request: PredictRequest) -> Any:
    if request.source is None:
        raise HTTPException(400, detail="'source' is required and must be a file path on the server.")

    source_path = Path(request.source)
    if not source_path.exists():
        raise HTTPException(404, detail=f"Source not found: {source_path}")

    config = PipelineConfig(output_root=Path("artifacts/api_predictions"), device=request.device)
    config.output_root.mkdir(parents=True, exist_ok=True)

    try:
        prediction_df = estimate_predicted_crack_sizes(
            config,
            source=source_path,
            weights_path=Path(request.weights) if request.weights else None,
            confidence=request.conf,
        )
    except FileNotFoundError as exc:
        raise HTTPException(404, detail=str(exc))
    except Exception as exc:
        raise HTTPException(500, detail=str(exc))

    json_summary = json.loads(read_text_safe(config.predictions_dir / JSON_SUMMARY_NAME) or "[]")
    text_summary = read_text_safe(config.predictions_dir / TEXT_SUMMARY_NAME)
    csv_path = str((config.predictions_dir / CSV_ESTIMATES_NAME).resolve())

    return {
        "predictions": prediction_df.to_dict(orient="records"),
        "json_summary": json_summary,
        "text_summary": text_summary,
        "csv_path": csv_path,
        "predictions_dir": str(config.predictions_dir.resolve()),
    }


@app.post("/predict/upload")
async def predict_upload(file: UploadFile = File(...), conf: float = 0.25, device: str | None = None) -> Any:
    suffix = Path(file.filename).suffix or ".jpg"
    with tempfile.TemporaryDirectory() as td:
        tmp_path = Path(td) / f"upload{suffix}"
        with tmp_path.open("wb") as f:
            shutil.copyfileobj(file.file, f)

        req = PredictRequest(source=str(tmp_path), conf=conf, device=device)
        return await predict(req)


@app.post("/predict/video")
async def predict_video(
    source: str,
    weights: str | None = None,
    conf: float = 0.25,
    frame_stride: int = 30,
    device: str | None = None,
) -> Any:
    if process_video is None:
        raise HTTPException(501, detail="Video processing endpoint unavailable (yolo_crack_video_pipeline not importable).")

    video_path = Path(source)
    if not video_path.exists():
        raise HTTPException(404, detail=f"Video not found: {video_path}")

    output_root = Path("artifacts/api_video_reports")
    args = type("Args", (), {
        "video": video_path,
        "weights": Path(weights) if weights else None,
        "output_root": output_root,
        "imgsz": 640,
        "device": device,
        "conf": conf,
        "frame_stride": frame_stride,
        "max_frames": None,
        "mm_per_pixel": None,
        "save_annotated_video": False,
    })

    try:
        prediction_df = process_video(args)
    except FileNotFoundError as exc:
        raise HTTPException(404, detail=str(exc))
    except Exception as exc:
        raise HTTPException(500, detail=str(exc))

    meta_path = output_root / "video_report_metadata.json"
    meta = json.loads(meta_path.read_text()) if meta_path.exists() else {}
    text_path = Path(meta.get("text_output", ""))
    text = read_text_safe(text_path) if text_path else ""

    return {
        "metadata": meta,
        "text_summary": text,
        "predictions": prediction_df.to_dict(orient="records"),
        "report_dir": str(output_root.resolve()),
    }


@app.post("/mcp/predict")
async def mcp_predict(request: PredictRequest) -> Any:
    resp = await predict(request)
    text = resp.get("text_summary", "")
    if not text:
        texts = [s.get("text") for s in resp.get("json_summary", []) if s.get("text")]
        text = "\n\n".join(texts) if texts else "No detections found."

    mcp_payload = {
        "task": "crack_detection",
        "llm_text": text,
        "artifacts": {
            "csv": resp.get("csv_path"),
            "predictions_dir": resp.get("predictions_dir"),
        },
    }
    return mcp_payload


@app.post("/report/as2870")
async def generate_as2870_report(request: ReportRequest) -> Any:
    try:
        outputs = write_report(
            ReportConfig(
                usecase_json=Path(request.usecase_json),
                predictions_csv=Path(request.predictions_csv) if request.predictions_csv else None,
                output_root=Path(request.output_root),
                site_classification=request.site_classification,
                location_name=request.location_name,
                prepared_for=request.prepared_for,
                prepared_by=request.prepared_by,
            )
        )
    except Exception as exc:
        raise HTTPException(500, detail=str(exc))

    return outputs


def main():
    print("Hello from ishika! API available as `app` variable (FastAPI).")


if __name__ == "__main__":
    import uvicorn

    uvicorn.run("main:app", host="0.0.0.0", port=8000, reload=False)
````

<p class="usecase-subsection-header">Code Cell Explanation: Report Generator</p>

This code cell contains `scripts/as2870_report_generator.py`, authored for report-related work by Tanvi. Its purpose is to convert raw or demonstration crack-detection rows into a report package.

`ReportConfig` centralises report settings. Instead of hard-coding values throughout the file, it stores the use-case JSON path, optional prediction CSV path, output folder, site classification wording, location, client name, author name, and AS 2870-2011 reference.

`load_usecase_metadata()` reads the separate JSON file. `load_predictions()` reads a prediction CSV if one exists. If no CSV is supplied, `build_demo_predictions()` creates a tiny safe demonstration table so the report workflow can still be shown during presentation.

`normalise_prediction_columns()` makes sure required columns exist. This protects the report from failing if a CSV is missing a non-critical column. `calculate_summary_metrics()` creates simple numbers such as total detections, average confidence, maximum crack length, and average crack width.

`assign_inspection_priority()` turns measurements into plain priority labels: high, medium, or low. `add_priority_assessment()` applies that logic to every row. `build_priority_chart()` uses Plotly to create a browser-friendly chart. `build_report_markdown()` writes the human-readable report text, and `write_report()` coordinates the whole workflow.

The command-line section at the bottom lets the report be generated from Terminal as well as from the API or notebook.


````python
from __future__ import annotations

import argparse
import json
from dataclasses import asdict, dataclass
from datetime import datetime
from pathlib import Path
from typing import Any

import pandas as pd
import plotly.express as px


DEFAULT_OUTPUT_ROOT = Path("artifacts/as2870_reports")
DEFAULT_STANDARD_REFERENCE = "AS 2870-2011 Residential slabs and footings"
REPORT_AUTHOR = "Tanvi"


@dataclass
class ReportConfig:
    usecase_json: Path = Path("UC00236_Concrete_Crack_Detection_mcp_server.json")
    predictions_csv: Path | None = None
    output_root: Path = DEFAULT_OUTPUT_ROOT
    site_classification: str = "Not supplied"
    location_name: str = "City of Melbourne inspection area"
    prepared_for: str = "City of Melbourne"
    prepared_by: str = REPORT_AUTHOR
    standard_reference: str = DEFAULT_STANDARD_REFERENCE
    engineer_review_required: bool = True


def load_usecase_metadata(path: Path) -> dict[str, Any]:
    if not path.exists():
        return {
            "title": "UC00236_Concrete_Crack_Detection_mcp_server",
            "description": "Concrete crack detection MCP server inspection report",
            "datasets": [],
            "technology": [],
        }
    return json.loads(path.read_text())


def load_predictions(path: Path | None) -> pd.DataFrame:
    if path is None or not path.exists():
        return build_demo_predictions()
    return pd.read_csv(path)


def build_demo_predictions() -> pd.DataFrame:
    return pd.DataFrame(
        [
            {
                "source": "sample_footpath_image_001.jpg",
                "confidence": 0.86,
                "estimated_crack_length_mm": 2.4,
                "estimated_crack_width_mm": 0.4,
                "length_category": "small",
                "crack_type": "hairline to simple",
            },
            {
                "source": "sample_footpath_image_002.jpg",
                "confidence": 0.79,
                "estimated_crack_length_mm": 8.8,
                "estimated_crack_width_mm": 1.3,
                "length_category": "medium",
                "crack_type": "simple",
            },
            {
                "source": "sample_kerb_image_003.jpg",
                "confidence": 0.91,
                "estimated_crack_length_mm": 14.5,
                "estimated_crack_width_mm": 2.1,
                "length_category": "large",
                "crack_type": "moderate",
            },
        ]
    )


def normalise_prediction_columns(df: pd.DataFrame) -> pd.DataFrame:
    cleaned = df.copy()
    defaults: dict[str, Any] = {
        "source": "unknown source",
        "confidence": 0.0,
        "estimated_crack_length_mm": None,
        "estimated_crack_width_mm": None,
        "length_category": "not measured",
        "crack_type": "not classified",
    }
    for column, value in defaults.items():
        if column not in cleaned.columns:
            cleaned[column] = value
    return cleaned


def calculate_summary_metrics(df: pd.DataFrame) -> dict[str, Any]:
    length_mm = pd.to_numeric(df["estimated_crack_length_mm"], errors="coerce")
    width_mm = pd.to_numeric(df["estimated_crack_width_mm"], errors="coerce")
    confidence = pd.to_numeric(df["confidence"], errors="coerce")

    return {
        "total_detections": int(len(df)),
        "average_confidence": round(float(confidence.mean()), 3) if confidence.notna().any() else None,
        "maximum_crack_length_mm": round(float(length_mm.max()), 2) if length_mm.notna().any() else None,
        "average_crack_length_mm": round(float(length_mm.mean()), 2) if length_mm.notna().any() else None,
        "maximum_crack_width_mm": round(float(width_mm.max()), 2) if width_mm.notna().any() else None,
        "average_crack_width_mm": round(float(width_mm.mean()), 2) if width_mm.notna().any() else None,
    }


def assign_inspection_priority(row: pd.Series) -> str:
    length = pd.to_numeric(row.get("estimated_crack_length_mm"), errors="coerce")
    width = pd.to_numeric(row.get("estimated_crack_width_mm"), errors="coerce")
    confidence = pd.to_numeric(row.get("confidence"), errors="coerce")

    length = 0 if pd.isna(length) else float(length)
    width = 0 if pd.isna(width) else float(width)
    confidence = 0 if pd.isna(confidence) else float(confidence)

    if confidence >= 0.75 and (length >= 10 or width >= 2):
        return "High - inspect soon"
    if confidence >= 0.50 and (length >= 3 or width >= 0.5):
        return "Medium - monitor and schedule review"
    return "Low - record for routine monitoring"


def add_priority_assessment(df: pd.DataFrame) -> pd.DataFrame:
    assessed = normalise_prediction_columns(df)
    assessed["inspection_priority"] = assessed.apply(assign_inspection_priority, axis=1)
    return assessed


def build_priority_chart(df: pd.DataFrame, output_path: Path) -> Path:
    priority_counts = (
        df["inspection_priority"]
        .value_counts()
        .rename_axis("inspection_priority")
        .reset_index(name="count")
    )
    fig = px.bar(
        priority_counts,
        x="inspection_priority",
        y="count",
        color="inspection_priority",
        title="Inspection Priority Summary",
        labels={"inspection_priority": "Inspection priority", "count": "Number of detections"},
    )
    fig.update_layout(showlegend=False)
    fig.write_html(output_path)
    return output_path


def build_report_markdown(
    config: ReportConfig,
    metadata: dict[str, Any],
    predictions: pd.DataFrame,
    chart_path: Path,
) -> str:
    metrics = calculate_summary_metrics(predictions)
    generated_at = datetime.now().strftime("%Y-%m-%d %H:%M")
    top_rows = predictions.head(10).to_markdown(index=False)

    return f"""# AS 2870-2011 Aligned Inspection Support Report

## Plain-English Summary

This is a decision-support report for {config.prepared_for}. It uses the available crack-detection code in this repository to summarise possible surface defects from images or video frames.

The report is aligned to the intent of {config.standard_reference}: structured site review, clear documentation, and professional follow-up where movement or cracking may affect residential-style slabs or footings. It does not replace the Australian Standard, a licensed copy of the Standard, or advice from a qualified structural engineer.

## Use Case

- Title: {metadata.get("title", "UC00236_Concrete_Crack_Detection_mcp_server")}
- Description: {metadata.get("description", "Not supplied")}
- Location: {config.location_name}
- Site classification: {config.site_classification}
- Prepared by: {config.prepared_by}
- Generated: {generated_at}

## Results At A Glance

- Total detections: {metrics["total_detections"]}
- Average confidence: {metrics["average_confidence"]}
- Maximum crack length: {metrics["maximum_crack_length_mm"]} mm
- Average crack length: {metrics["average_crack_length_mm"]} mm
- Maximum crack width: {metrics["maximum_crack_width_mm"]} mm
- Average crack width: {metrics["average_crack_width_mm"]} mm

## Priority Interpretation

- High means the detection has stronger model confidence and larger estimated dimensions, so a human inspection should be scheduled soon.
- Medium means the issue should be monitored and placed into a maintenance review queue.
- Low means the issue should be recorded for routine monitoring.

## Plotly Chart

Open this chart in a browser:

`{chart_path}`

## First Ten Detection Rows

{top_rows}

## AS 2870-2011 Note

AS 2870-2011 is a residential slabs and footings standard. Public descriptions of the standard state that it covers site classification and footing system design for residential-style buildings. This report uses that standard only as a reporting alignment reference. It does not reproduce protected standard clauses or make final compliance decisions.

## Recommended Next Steps

1. Confirm the site classification using the proper engineering process.
2. Review high-priority detections in person.
3. Compare image evidence against asset records and maintenance history.
4. Ask a qualified engineer to make any AS 2870-2011 compliance or structural judgement.
"""


def write_report(config: ReportConfig) -> dict[str, str]:
    config.output_root.mkdir(parents=True, exist_ok=True)
    metadata = load_usecase_metadata(config.usecase_json)
    raw_predictions = load_predictions(config.predictions_csv)
    predictions = add_priority_assessment(raw_predictions)

    assessed_csv = config.output_root / "as2870_assessed_detections.csv"
    chart_path = config.output_root / "inspection_priority_chart.html"
    report_path = config.output_root / "as2870_inspection_support_report.md"
    summary_path = config.output_root / "as2870_report_summary.json"

    predictions.to_csv(assessed_csv, index=False)
    build_priority_chart(predictions, chart_path)
    report_text = build_report_markdown(config, metadata, predictions, chart_path)
    report_path.write_text(report_text)

    summary = {
        "report": str(report_path.resolve()),
        "assessed_csv": str(assessed_csv.resolve()),
        "chart": str(chart_path.resolve()),
        "config": asdict(config),
        "metrics": calculate_summary_metrics(predictions),
    }
    summary_path.write_text(json.dumps(summary, indent=2, default=str))
    return {
        "report": str(report_path.resolve()),
        "assessed_csv": str(assessed_csv.resolve()),
        "chart": str(chart_path.resolve()),
        "summary": str(summary_path.resolve()),
    }


def parse_args() -> argparse.Namespace:
    parser = argparse.ArgumentParser(description="Generate an AS 2870-2011 aligned inspection support report.")
    parser.add_argument("--usecase-json", type=Path, default=Path("UC00236_Concrete_Crack_Detection_mcp_server.json"))
    parser.add_argument("--predictions-csv", type=Path, default=None)
    parser.add_argument("--output-root", type=Path, default=DEFAULT_OUTPUT_ROOT)
    parser.add_argument("--site-classification", default="Not supplied")
    parser.add_argument("--location-name", default="City of Melbourne inspection area")
    parser.add_argument("--prepared-for", default="City of Melbourne")
    parser.add_argument("--prepared-by", default=REPORT_AUTHOR)
    return parser.parse_args()


def main() -> None:
    args = parse_args()
    outputs = write_report(
        ReportConfig(
            usecase_json=args.usecase_json,
            predictions_csv=args.predictions_csv,
            output_root=args.output_root,
            site_classification=args.site_classification,
            location_name=args.location_name,
            prepared_for=args.prepared_for,
            prepared_by=args.prepared_by,
        )
    )
    print(json.dumps(outputs, indent=2))


if __name__ == "__main__":
    main()
````

<p class="usecase-subsection-header">Code Cell Explanation: Image Crack Detection Pipeline</p>

This code cell contains `scripts/yolo_crack_bbox_pipeline.py`, Kavita's main image-processing pipeline. It is the largest technical block because it handles dataset preparation, model inference, measurement estimation, and output writing.

At the top, the imports bring in file handling, JSON, random seeding, SSL/model download support, dataclasses, image processing with OpenCV, numerical processing with NumPy, table processing with pandas, PyTorch device selection, train/validation/test splitting, progress bars, YOLO, and optional MLflow experiment tracking.

The constants define how the SDNET2018 folder names map to readable labels. For example, cracked and uncracked concrete classes are converted into consistent labels. The output filename constants define the CSV, JSON, and text report names. `DETECTION_COLUMNS` defines the exact columns expected in the prediction table.

`PipelineConfig` is the main settings object. The helper functions perform focused jobs: choosing CPU/GPU/MPS, preparing manifests, creating YOLO labels, resolving model weights, running inference, estimating crack dimensions from boxes/masks, converting model results into rows, writing annotated images, and exporting summary files.

In plain language, this pipeline takes SDNET2018-style concrete images, lets YOLO identify likely cracks, estimates useful measurements, and writes evidence files that the API and report generator can use.


````python
from __future__ import annotations

import argparse
import json
import random
import shutil
import ssl
import urllib.request
from contextlib import contextmanager
from dataclasses import asdict, dataclass
from pathlib import Path
from typing import Iterator, Iterable
from collections import Counter

import cv2
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from tqdm.auto import tqdm

try:
    from ultralytics import YOLO
except ImportError:  # pragma: no cover - handled at runtime.
    YOLO = None

try:
    import mlflow
except ImportError:  # pragma: no cover - handled at runtime.
    mlflow = None


SURFACE_NAMES = {"D": "Deck", "P": "Pavement", "W": "Wall"}
SUBCLASS_NAMES = {
    "CD": "Cracked",
    "UD": "Uncracked",
    "CP": "Cracked",
    "UP": "Uncracked",
    "CW": "Cracked",
    "UW": "Uncracked",
}
IMAGE_SUFFIXES = {".jpg", ".jpeg", ".png", ".bmp"}
TEXT_SUMMARY_NAME = "crack_detection_summary.txt"
JSON_SUMMARY_NAME = "crack_detection_summary.json"
CSV_ESTIMATES_NAME = "crack_size_estimates.csv"
DETECTION_COLUMNS = [
    "source",
    "frame_index",
    "timestamp_sec",
    "detection_id",
    "class_id",
    "class_name",
    "confidence",
    "x1",
    "y1",
    "x2",
    "y2",
    "box_width_px",
    "box_height_px",
    "box_area_px",
    "estimated_crack_length_px",
    "estimated_crack_centerline_length_px",
    "estimated_crack_width_px",
    "estimated_crack_mean_width_px",
    "estimated_crack_mask_area_px",
    "estimated_crack_length_mm",
    "estimated_crack_centerline_length_mm",
    "estimated_crack_width_mm",
    "estimated_crack_mean_width_mm",
    "box_area_mm2",
    "image_width_px",
    "image_height_px",
    "annotated_image_path",
    "text_description",
    "length_category",
    "crack_type",
    "measurement_method",
]


@dataclass
class PipelineConfig:
    dataset_root: Path = Path("DEPENDENCIES/SDNET2018")
    fallback_dataset_root: Path = Path("dataset")
    output_root: Path = Path("artifacts/yolo11_sdnet2018_bbox")
    model_name: str = "yolo11n.pt"
    run_name: str = "yolo11n_sdnet2018_crack_bbox"
    image_size: int = 640
    epochs: int = 40
    batch_size: int = 16
    workers: int = 2
    device: str | None = None
    seed: int = 42
    val_fraction: float = 0.15
    test_fraction: float = 0.15
    min_box_area_px: int = 24
    max_boxes_per_image: int = 8
    crack_threshold_percentile: float = 18.0
    mm_per_pixel: float | None = None
    # thresholds in millimetres used to bucket crack length into categories
    hairline_mm: float = 1.0
    small_mm: float = 3.0
    medium_mm: float = 10.0
    # fallback pixel thresholds when mm_per_pixel is not provided
    hairline_px: int = 5
    small_px: int = 15
    medium_px: int = 50
    # width thresholds used to describe crack type from the SDNET-style mask estimate
    hairline_width_mm: float = 0.2
    simple_width_mm: float = 1.0
    moderate_width_mm: float = 3.0
    hairline_width_px: float = 2.0
    simple_width_px: float = 5.0
    moderate_width_px: float = 10.0
    allow_unverified_model_downloads: bool = True
    enable_mlflow: bool = True
    mlflow_experiment_name: str = "crack-yolo-bbox"
    mlflow_tracking_uri: str | None = "sqlite:///mlflow.db"

    @property
    def manifest_path(self) -> Path:
        return self.output_root / "manifests" / "sdnet2018_detection_manifest.csv"

    @property
    def data_yaml_path(self) -> Path:
        return self.output_root / "data.yaml"

    @property
    def yolo_dataset_root(self) -> Path:
        return self.output_root / "dataset"

    @property
    def runs_dir(self) -> Path:
        return self.output_root / "runs"

    @property
    def predictions_dir(self) -> Path:
        return self.output_root / "predictions"


def configure_ssl(config: PipelineConfig) -> None:
    if not config.allow_unverified_model_downloads:
        return
    ssl._create_default_https_context = ssl._create_unverified_context
    urllib.request.install_opener(
        urllib.request.build_opener(
            urllib.request.HTTPSHandler(context=ssl._create_unverified_context())
        )
    )


def select_device(device: str | None = None) -> int | str:
    if device:
        requested = device.lower()
        if requested in {"auto", ""}:
            pass
        elif requested in {"cpu", "mps"}:
            if requested == "mps" and not (
                getattr(torch.backends, "mps", None) and torch.backends.mps.is_available()
            ):
                raise ValueError("MPS was requested, but torch.backends.mps is not available.")
            return requested
        elif requested in {"cuda", "gpu"}:
            if not torch.cuda.is_available():
                raise ValueError("CUDA was requested, but torch.cuda.is_available() is false.")
            return 0
        elif requested.startswith("cuda:"):
            if not torch.cuda.is_available():
                raise ValueError("CUDA was requested, but torch.cuda.is_available() is false.")
            return int(requested.split(":", 1)[1])
        else:
            raise ValueError("device must be one of: auto, cuda, cuda:N, mps, cpu.")

    if torch.cuda.is_available():
        return 0
    if getattr(torch.backends, "mps", None) and torch.backends.mps.is_available():
        return "mps"
    return "cpu"


def config_as_mlflow_params(config: PipelineConfig) -> dict[str, str | int | float | bool]:
    params: dict[str, str | int | float | bool] = {}
    for key, value in asdict(config).items():
        if isinstance(value, Path):
            params[key] = str(value)
        elif value is None:
            params[key] = "None"
        else:
            params[key] = value
    params["device"] = str(select_device(config.device))
    return params


def sanitize_metric_name(name: str) -> str:
    return (
        name.replace("(", "")
        .replace(")", "")
        .replace("/", "_")
        .replace("-", "_")
        .replace(" ", "_")
    )


@contextmanager
def mlflow_run(config: PipelineConfig, stage: str) -> Iterator[bool]:
    if not config.enable_mlflow:
        yield False
        return
    if mlflow is None:
        print("MLflow tracking skipped because mlflow is not installed.")
        yield False
        return

    if config.mlflow_tracking_uri:
        mlflow.set_tracking_uri(config.mlflow_tracking_uri)
    mlflow.set_experiment(config.mlflow_experiment_name)
    with mlflow.start_run(run_name=f"{config.run_name}_{stage}"):
        mlflow.log_params(config_as_mlflow_params(config))
        mlflow.set_tag("pipeline_stage", stage)
        mlflow.set_tag("task", "crack_detection_bbox")
        yield True


def log_artifact_if_exists(path: Path, artifact_path: str | None = None) -> None:
    if mlflow is None or not path.exists():
        return
    if path.is_dir():
        mlflow.log_artifacts(str(path), artifact_path=artifact_path)
    else:
        mlflow.log_artifact(str(path), artifact_path=artifact_path)


def log_preparation_to_mlflow(
    config: PipelineConfig,
    box_df: pd.DataFrame,
    max_images: int | None = None,
) -> None:
    if mlflow is None:
        return

    manifest_df = pd.read_csv(config.manifest_path)
    split_counts = manifest_df.groupby(["split", "has_crack"]).size()
    metrics: dict[str, float] = {
        "images_total": float(len(manifest_df)),
        "pseudo_boxes_total": float(len(box_df)),
        "max_images_requested": float(max_images or 0),
    }
    for (split, has_crack), count in split_counts.items():
        label = "cracked" if bool(has_crack) else "uncracked"
        metrics[f"{split}_{label}_images"] = float(count)
    if not box_df.empty:
        metrics.update(
            {
                "box_length_px_mean": float(box_df["estimated_crack_length_px"].mean()),
                "box_length_px_median": float(box_df["estimated_crack_length_px"].median()),
                "box_width_px_mean": float(box_df["estimated_crack_width_px"].mean()),
                "box_width_px_median": float(box_df["estimated_crack_width_px"].median()),
                "box_area_px_mean": float(box_df["box_area_px"].mean()),
                "box_area_px_median": float(box_df["box_area_px"].median()),
            }
        )
    mlflow.log_metrics(metrics)
    log_artifact_if_exists(config.manifest_path, "manifests")
    log_artifact_if_exists(config.output_root / "manifests" / "generated_bbox_labels.csv", "manifests")
    log_artifact_if_exists(config.data_yaml_path, "dataset")
    log_artifact_if_exists(config.output_root / "pipeline_config.json", "config")


def log_ultralytics_run_to_mlflow(weights_path: Path) -> None:
    if mlflow is None:
        return

    run_dir = weights_path.parents[1]
    results_csv = run_dir / "results.csv"
    if results_csv.exists():
        results_df = pd.read_csv(results_csv)
        if not results_df.empty:
            last_row = results_df.iloc[-1].to_dict()
            metrics = {}
            for key, value in last_row.items():
                if pd.notna(value) and isinstance(value, (int, float, np.integer, np.floating)):
                    metrics[f"yolo_{sanitize_metric_name(str(key))}"] = float(value)
            if metrics:
                mlflow.log_metrics(metrics)
    log_artifact_if_exists(weights_path, "weights")
    log_artifact_if_exists(run_dir, "ultralytics_run")


def log_ultralytics_metrics_to_mlflow(metrics_object: object, prefix: str = "eval") -> None:
    if mlflow is None:
        return
    results_dict = getattr(metrics_object, "results_dict", None)
    if not results_dict:
        return
    metrics = {}
    for key, value in results_dict.items():
        if isinstance(value, (int, float, np.integer, np.floating)):
            metrics[f"{prefix}_{sanitize_metric_name(str(key))}"] = float(value)
    if metrics:
        mlflow.log_metrics(metrics)


def log_predictions_to_mlflow(config: PipelineConfig, prediction_df: pd.DataFrame) -> None:
    if mlflow is None:
        return
    metrics = {"predicted_boxes_total": float(len(prediction_df))}
    if not prediction_df.empty:
        metrics.update(
            {
                "prediction_confidence_mean": float(prediction_df["confidence"].mean()),
                "predicted_crack_length_px_mean": float(prediction_df["estimated_crack_length_px"].mean()),
                "predicted_crack_width_px_mean": float(prediction_df["estimated_crack_width_px"].mean()),
            }
        )
    mlflow.log_metrics(metrics)
    log_artifact_if_exists(config.predictions_dir / CSV_ESTIMATES_NAME, "predictions")
    log_artifact_if_exists(config.predictions_dir / TEXT_SUMMARY_NAME, "predictions")
    log_artifact_if_exists(config.predictions_dir / JSON_SUMMARY_NAME, "predictions")
    log_artifact_if_exists(config.predictions_dir / "annotated_images", "predictions")


def has_sdnet_layout(root: Path) -> bool:
    return all((root / surface / subclass).exists() for surface, subclass in [
        ("D", "CD"),
        ("D", "UD"),
        ("P", "CP"),
        ("P", "UP"),
        ("W", "CW"),
        ("W", "UW"),
    ])


def resolve_dataset_root(config: PipelineConfig) -> Path:
    if has_sdnet_layout(config.dataset_root):
        return config.dataset_root
    if has_sdnet_layout(config.fallback_dataset_root):
        return config.fallback_dataset_root
    raise FileNotFoundError(
        f"Could not find SDNET2018 layout under {config.dataset_root} or {config.fallback_dataset_root}."
    )


def iter_sdnet_images(root: Path) -> Iterable[dict[str, object]]:
    for surface_code in sorted(SURFACE_NAMES):
        surface_dir = root / surface_code
        if not surface_dir.exists():
            continue
        for subclass_dir in sorted(surface_dir.iterdir()):
            if not subclass_dir.is_dir():
                continue
            subclass_code = subclass_dir.name
            crack_status = SUBCLASS_NAMES.get(subclass_code)
            if crack_status is None:
                continue
            for image_path in sorted(subclass_dir.iterdir()):
                if image_path.suffix.lower() not in IMAGE_SUFFIXES:
                    continue
                yield {
                    "path": str(image_path.resolve()),
                    "relative_path": str(image_path.relative_to(root)),
                    "surface_code": surface_code,
                    "surface_name": SURFACE_NAMES[surface_code],
                    "subclass_code": subclass_code,
                    "crack_status": crack_status,
                    "has_crack": crack_status == "Cracked",
                }


def build_manifest(config: PipelineConfig, max_images: int | None = None) -> pd.DataFrame:
    root = resolve_dataset_root(config)
    manifest_df = pd.DataFrame(iter_sdnet_images(root))
    if manifest_df.empty:
        raise RuntimeError(f"No SDNET2018 images found under {root}.")

    if max_images is not None and max_images < len(manifest_df):
        cracked = manifest_df[manifest_df["has_crack"]].sample(
            n=min(max_images // 2, int(manifest_df["has_crack"].sum())),
            random_state=config.seed,
        )
        uncracked = manifest_df[~manifest_df["has_crack"]].sample(
            n=max_images - len(cracked),
            random_state=config.seed,
        )
        manifest_df = pd.concat([cracked, uncracked], ignore_index=True)

    train_df, temp_df = train_test_split(
        manifest_df,
        test_size=config.val_fraction + config.test_fraction,
        random_state=config.seed,
        stratify=manifest_df["has_crack"],
    )
    val_relative = config.test_fraction / (config.val_fraction + config.test_fraction)
    val_df, test_df = train_test_split(
        temp_df,
        test_size=val_relative,
        random_state=config.seed,
        stratify=temp_df["has_crack"],
    )
    split_df = pd.concat(
        [
            train_df.assign(split="train"),
            val_df.assign(split="val"),
            test_df.assign(split="test"),
        ],
        ignore_index=True,
    )
    config.manifest_path.parent.mkdir(parents=True, exist_ok=True)
    split_df.to_csv(config.manifest_path, index=False)
    return split_df


def find_crack_boxes(image_path: Path, config: PipelineConfig) -> tuple[list[tuple[int, int, int, int]], tuple[int, int]]:
    image = cv2.imread(str(image_path), cv2.IMREAD_COLOR)
    if image is None:
        raise ValueError(f"Could not read image: {image_path}")
    height, width = image.shape[:2]
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    gray = cv2.GaussianBlur(gray, (3, 3), 0)

    threshold_value = np.percentile(gray, config.crack_threshold_percentile)
    dark_mask = (gray <= threshold_value).astype(np.uint8) * 255
    edges = cv2.Canny(gray, 40, 120)
    mask = cv2.bitwise_and(dark_mask, cv2.dilate(edges, np.ones((3, 3), np.uint8), iterations=1))
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, np.ones((5, 5), np.uint8), iterations=1)

    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    boxes: list[tuple[int, int, int, int]] = []
    for contour in contours:
        x, y, box_width, box_height = cv2.boundingRect(contour)
        area = box_width * box_height
        if area < config.min_box_area_px:
            continue
        if box_width < 3 or box_height < 3:
            continue
        boxes.append((x, y, x + box_width, y + box_height))

    boxes = sorted(boxes, key=lambda box: (box[2] - box[0]) * (box[3] - box[1]), reverse=True)
    return boxes[: config.max_boxes_per_image], (height, width)


def build_sdnet_crack_mask(image: np.ndarray, config: PipelineConfig) -> np.ndarray:
    """Build a crack-like foreground mask using the same dark-line cue as SDNET pseudo labels."""
    if image.size == 0:
        return np.zeros(image.shape[:2], dtype=np.uint8)

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) if image.ndim == 3 else image.copy()
    gray = cv2.GaussianBlur(gray, (3, 3), 0)
    percentile_threshold = np.percentile(gray, config.crack_threshold_percentile)
    background_threshold = float(gray.mean() - max(gray.std() * 0.5, 5.0))
    threshold_value = min(float(percentile_threshold), background_threshold)
    mask = (gray <= threshold_value).astype(np.uint8) * 255
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, np.ones((3, 3), np.uint8), iterations=1)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, np.ones((2, 2), np.uint8), iterations=1)
    if cv2.countNonZero(mask) > mask.size * 0.6:
        edges = cv2.Canny(gray, 40, 120)
        mask = cv2.bitwise_and(mask, cv2.dilate(edges, np.ones((3, 3), np.uint8), iterations=1))
    return (mask > 0).astype(np.uint8) * 255


def skeletonize_mask(mask: np.ndarray) -> np.ndarray:
    """Return a one-pixel-wide skeleton using OpenCV morphology only."""
    binary = (mask > 0).astype(np.uint8) * 255
    skeleton = np.zeros(binary.shape, dtype=np.uint8)
    element = cv2.getStructuringElement(cv2.MORPH_CROSS, (3, 3))

    max_iterations = max(binary.shape) + 1
    iterations = 0
    while cv2.countNonZero(binary) > 0 and iterations < max_iterations:
        eroded = cv2.erode(binary, element)
        opened = cv2.dilate(eroded, element)
        skeleton = cv2.bitwise_or(skeleton, cv2.subtract(binary, opened))
        binary = eroded
        iterations += 1

    return skeleton


def skeleton_length_px(skeleton: np.ndarray) -> float:
    points = np.argwhere(skeleton > 0)
    if len(points) == 0:
        return 0.0

    skeleton_bool = skeleton > 0
    total = 0.0
    for y, x in points:
        if x + 1 < skeleton.shape[1] and skeleton_bool[y, x + 1]:
            total += 1.0
        if y + 1 < skeleton.shape[0] and skeleton_bool[y + 1, x]:
            total += 1.0
        if y + 1 < skeleton.shape[0] and x + 1 < skeleton.shape[1] and skeleton_bool[y + 1, x + 1]:
            total += float(np.sqrt(2.0))
        if y + 1 < skeleton.shape[0] and x - 1 >= 0 and skeleton_bool[y + 1, x - 1]:
            total += float(np.sqrt(2.0))
    return max(total, float(len(points)))


def estimate_crack_mask_measurements(
    image: np.ndarray | None,
    box: tuple[float, float, float, float],
    config: PipelineConfig,
) -> dict[str, float | str]:
    if image is None or image.size == 0:
        return {}

    height, width = image.shape[:2]
    x1, y1, x2, y2 = box
    ix1 = max(0, min(width - 1, int(np.floor(x1))))
    iy1 = max(0, min(height - 1, int(np.floor(y1))))
    ix2 = max(ix1 + 1, min(width, int(np.ceil(x2))))
    iy2 = max(iy1 + 1, min(height, int(np.ceil(y2))))
    crop = image[iy1:iy2, ix1:ix2]
    if crop.size == 0:
        return {}

    mask = build_sdnet_crack_mask(crop, config)
    mask_area_px = float(cv2.countNonZero(mask))
    if mask_area_px <= 0:
        return {}

    skeleton = skeletonize_mask(mask)
    centerline_length_px = skeleton_length_px(skeleton)
    if centerline_length_px <= 0:
        return {}

    return {
        "estimated_crack_centerline_length_px": centerline_length_px,
        "estimated_crack_mask_area_px": mask_area_px,
        "estimated_crack_mean_width_px": mask_area_px / centerline_length_px,
        "measurement_method": "sdnet_dark_mask_skeleton",
    }


def to_yolo_row(box: tuple[int, int, int, int], image_shape: tuple[int, int]) -> str:
    height, width = image_shape
    x1, y1, x2, y2 = box
    x_center = ((x1 + x2) / 2) / width
    y_center = ((y1 + y2) / 2) / height
    box_width = (x2 - x1) / width
    box_height = (y2 - y1) / height
    return f"0 {x_center:.6f} {y_center:.6f} {box_width:.6f} {box_height:.6f}"


def copy_or_link_image(source: Path, destination: Path) -> None:
    destination.parent.mkdir(parents=True, exist_ok=True)
    if destination.exists():
        return
    try:
        destination.symlink_to(source.resolve())
    except OSError:
        shutil.copy2(source, destination)


def prepare_yolo_detection_dataset(
    config: PipelineConfig,
    rebuild: bool = False,
    max_images: int | None = None,
) -> pd.DataFrame:
    if rebuild and config.yolo_dataset_root.exists():
        shutil.rmtree(config.yolo_dataset_root)

    manifest_df = build_manifest(config, max_images=max_images)
    label_rows: list[dict[str, object]] = []

    for row in tqdm(manifest_df.itertuples(index=False), total=len(manifest_df), desc="Preparing YOLO boxes"):
        source_path = Path(row.path)
        stem = source_path.stem
        split = row.split
        destination_image = config.yolo_dataset_root / "images" / split / f"{stem}{source_path.suffix.lower()}"
        destination_label = config.yolo_dataset_root / "labels" / split / f"{stem}.txt"

        copy_or_link_image(source_path, destination_image)
        destination_label.parent.mkdir(parents=True, exist_ok=True)

        boxes: list[tuple[int, int, int, int]] = []
        image_shape: tuple[int, int] | None = None
        source_image = cv2.imread(str(source_path), cv2.IMREAD_COLOR) if bool(row.has_crack) else None
        if bool(row.has_crack):
            boxes, image_shape = find_crack_boxes(source_path, config)

        if boxes and image_shape:
            destination_label.write_text("\n".join(to_yolo_row(box, image_shape) for box in boxes) + "\n")
        else:
            destination_label.write_text("")

        for box in boxes:
            x1, y1, x2, y2 = box
            width_px = x2 - x1
            height_px = y2 - y1
            mask_measurements = estimate_crack_mask_measurements(source_image, box, config)
            centerline_length_px = float(
                mask_measurements.get("estimated_crack_centerline_length_px", max(width_px, height_px))
            )
            mean_width_px = float(
                mask_measurements.get("estimated_crack_mean_width_px", min(width_px, height_px))
            )
            label_row = {
                "split": split,
                "path": str(source_path),
                "relative_path": row.relative_path,
                "surface_name": row.surface_name,
                "x1": x1,
                "y1": y1,
                "x2": x2,
                "y2": y2,
                "box_width_px": width_px,
                "box_height_px": height_px,
                "box_area_px": width_px * height_px,
                "estimated_crack_length_px": centerline_length_px,
                "estimated_crack_centerline_length_px": centerline_length_px,
                "estimated_crack_width_px": min(width_px, height_px),
                "estimated_crack_mean_width_px": mean_width_px,
                "estimated_crack_mask_area_px": mask_measurements.get("estimated_crack_mask_area_px", ""),
                "measurement_method": mask_measurements.get("measurement_method", "bbox_longest_side"),
            }
            label_row["length_category"] = classify_crack_length(label_row, config)
            label_row["crack_type"] = classify_crack_type(label_row, config)
            label_rows.append(label_row)

    box_df = pd.DataFrame(label_rows)
    if not box_df.empty and config.mm_per_pixel:
        box_df["estimated_crack_length_mm"] = box_df["estimated_crack_length_px"] * config.mm_per_pixel
        box_df["estimated_crack_centerline_length_mm"] = (
            box_df["estimated_crack_centerline_length_px"] * config.mm_per_pixel
        )
        box_df["estimated_crack_width_mm"] = box_df["estimated_crack_width_px"] * config.mm_per_pixel
        box_df["estimated_crack_mean_width_mm"] = box_df["estimated_crack_mean_width_px"] * config.mm_per_pixel
        box_df["box_area_mm2"] = box_df["box_area_px"] * (config.mm_per_pixel ** 2)

    config.output_root.mkdir(parents=True, exist_ok=True)
    box_df.to_csv(config.output_root / "manifests" / "generated_bbox_labels.csv", index=False)
    config.data_yaml_path.write_text(
        "\n".join(
            [
                f"path: {config.yolo_dataset_root.resolve()}",
                "train: images/train",
                "val: images/val",
                "test: images/test",
                "names:",
                "  0: crack",
                "",
            ]
        )
    )
    return box_df


def resolve_best_weights(config: PipelineConfig) -> Path:
    search_roots = [
        config.runs_dir,
        PipelineConfig().runs_dir,
        Path("runs"),
    ]
    candidates: list[Path] = []
    seen: set[Path] = set()
    for root in search_roots:
        for candidate in root.glob(f"**/{config.run_name}/weights/best.pt"):
            resolved = candidate.resolve()
            if resolved not in seen and candidate.exists():
                candidates.append(candidate)
                seen.add(resolved)
    if candidates:
        return max(candidates, key=lambda path: path.stat().st_mtime)

    expected_paths = [
        config.runs_dir / config.run_name / "weights" / "best.pt",
        PipelineConfig().runs_dir / config.run_name / "weights" / "best.pt",
        Path("runs") / config.run_name / "weights" / "best.pt",
    ]
    expected = "\n".join(f"- {path}" for path in expected_paths)
    raise FileNotFoundError(
        "No trained YOLO weights were found. Train the detector first with "
        "`python3 scripts/yolo_crack_bbox_pipeline.py train` or pass a valid "
        "`weights` path in the request. Checked:\n"
        f"{expected}"
    )


def train_yolo(config: PipelineConfig) -> Path:
    if YOLO is None:
        raise ImportError("Install ultralytics before training: `uv add ultralytics` or `pip install ultralytics`.")
    configure_ssl(config)
    model = YOLO(config.model_name)
    train_results = model.train(
        data=str(config.data_yaml_path.resolve()),
        epochs=config.epochs,
        imgsz=config.image_size,
        batch=config.batch_size,
        project=str(config.runs_dir.resolve()),
        name=config.run_name,
        exist_ok=True,
        seed=config.seed,
        device=select_device(config.device),
        workers=config.workers,
        plots=True,
    )
    if train_results is not None and getattr(train_results, "save_dir", None):
        return Path(train_results.save_dir) / "weights" / "best.pt"
    return resolve_best_weights(config)


def evaluate_yolo(config: PipelineConfig, weights_path: Path | None = None) -> object:
    if YOLO is None:
        raise ImportError("Install ultralytics before evaluation.")
    weights = weights_path or resolve_best_weights(config)
    model = YOLO(str(weights))
    return model.val(
        data=str(config.data_yaml_path),
        split="test",
        imgsz=config.image_size,
        device=select_device(config.device),
    )


def add_scaled_measurements(row: dict[str, object], config: PipelineConfig) -> None:
    if not config.mm_per_pixel:
        return
    row["estimated_crack_length_mm"] = row["estimated_crack_length_px"] * config.mm_per_pixel
    row["estimated_crack_centerline_length_mm"] = (
        row.get("estimated_crack_centerline_length_px", row["estimated_crack_length_px"]) * config.mm_per_pixel
    )
    row["estimated_crack_width_mm"] = row["estimated_crack_width_px"] * config.mm_per_pixel
    row["estimated_crack_mean_width_mm"] = (
        row.get("estimated_crack_mean_width_px", row["estimated_crack_width_px"]) * config.mm_per_pixel
    )
    row["box_area_mm2"] = row["box_area_px"] * (config.mm_per_pixel ** 2)


def classify_crack_length(row: dict[str, object], config: PipelineConfig) -> str:
    """Classify a detection into a length category using mm if available, else fall back to px."""
    length_mm = row.get("estimated_crack_length_mm")
    length_px = row.get("estimated_crack_length_px")

    try:
        if length_mm is not None:
            if length_mm < config.hairline_mm:
                return "hairline"
            if length_mm < config.small_mm:
                return "small"
            if length_mm < config.medium_mm:
                return "medium"
            return "large"
    except Exception:
        pass

    # fallback to pixel-based heuristics
    if length_px is None:
        return "unknown"
    if length_px < config.hairline_px:
        return "hairline"
    if length_px < config.small_px:
        return "small"
    if length_px < config.medium_px:
        return "medium"
    return "large"


def classify_crack_type(row: dict[str, object], config: PipelineConfig) -> str:
    """Describe the crack by estimated width first, then by length severity."""
    width_mm = row.get("estimated_crack_mean_width_mm")
    width_px = row.get("estimated_crack_mean_width_px", row.get("estimated_crack_width_px"))

    try:
        if width_mm is not None:
            width = float(width_mm)
            if width <= config.hairline_width_mm:
                return "hairline_crack"
            if width <= config.simple_width_mm:
                return "simple_crack"
            if width <= config.moderate_width_mm:
                return "moderate_crack"
            return "severe_crack"
    except Exception:
        pass

    try:
        width = float(width_px)
    except Exception:
        return "unknown_crack"

    if width <= config.hairline_width_px:
        return "hairline_crack"
    if width <= config.simple_width_px:
        return "simple_crack"
    if width <= config.moderate_width_px:
        return "moderate_crack"
    return "severe_crack"


def detection_description(row: dict[str, object]) -> str:
    crack_type = row.get("crack_type", "unknown_crack")
    mean_width = row.get("estimated_crack_mean_width_px", row.get("estimated_crack_width_px", 0.0))
    return (
        f"Detection {row['detection_id']}: crack confidence {float(row['confidence']):.3f}; "
        f"bbox x1={float(row['x1']):.1f}, y1={float(row['y1']):.1f}, "
        f"x2={float(row['x2']):.1f}, y2={float(row['y2']):.1f}; "
        f"area={float(row['box_area_px']):.1f} px^2; "
        f"type={crack_type}; "
        f"estimated centerline length={float(row['estimated_crack_length_px']):.1f} px; "
        f"estimated mean width={float(mean_width):.1f} px."
    )


def image_summary_text(source: str, rows: list[dict[str, object]]) -> str:
    if not rows:
        return f"{source}: no crack detections were found."
    total_area = sum(float(row["box_area_px"]) for row in rows)
    max_length = max(float(row["estimated_crack_length_px"]) for row in rows)
    mean_confidence = sum(float(row["confidence"]) for row in rows) / len(rows)
    crack_type_counts = Counter(str(row.get("crack_type", "unknown_crack")) for row in rows)
    crack_types = ", ".join(f"{name}: {count}" for name, count in sorted(crack_type_counts.items()))
    details = " ".join(detection_description(row) for row in rows)
    return (
        f"{source}: {len(rows)} crack detection(s). "
        f"Total detected crack bounding-box area is {total_area:.1f} px^2. "
        f"Largest estimated crack length is {max_length:.1f} px. "
        f"Crack types: {crack_types}. "
        f"Mean confidence is {mean_confidence:.3f}. "
        f"{details}"
    )


def write_prediction_reports(
    prediction_df: pd.DataFrame,
    image_summaries: list[dict[str, object]],
    output_dir: Path,
) -> None:
    output_dir.mkdir(parents=True, exist_ok=True)
    if prediction_df.empty:
        prediction_df = pd.DataFrame(columns=DETECTION_COLUMNS)
    prediction_df.to_csv(output_dir / CSV_ESTIMATES_NAME, index=False)
    (output_dir / JSON_SUMMARY_NAME).write_text(json.dumps(image_summaries, indent=2, default=str))
    summary_lines = [str(summary["text"]) for summary in image_summaries]
    (output_dir / TEXT_SUMMARY_NAME).write_text("\n\n".join(summary_lines) + ("\n" if summary_lines else ""))


def result_to_detection_rows(
    result: object,
    config: PipelineConfig,
    annotated_image_path: Path | None = None,
    frame_index: int | None = None,
    timestamp_sec: float | None = None,
) -> list[dict[str, object]]:
    boxes = getattr(result, "boxes", None)
    if boxes is None:
        return []

    rows: list[dict[str, object]] = []
    source_path = str(Path(result.path).resolve()) if getattr(result, "path", None) else ""
    image_shape = getattr(result, "orig_shape", None)
    image_height = int(image_shape[0]) if image_shape else None
    image_width = int(image_shape[1]) if image_shape else None
    source_image = getattr(result, "orig_img", None)

    for detection_index, box in enumerate(boxes, start=1):
        x1, y1, x2, y2 = [float(value) for value in box.xyxy[0].tolist()]
        width_px = x2 - x1
        height_px = y2 - y1
        mask_measurements = estimate_crack_mask_measurements(source_image, (x1, y1, x2, y2), config)
        centerline_length_px = float(
            mask_measurements.get("estimated_crack_centerline_length_px", max(width_px, height_px))
        )
        mean_width_px = float(mask_measurements.get("estimated_crack_mean_width_px", min(width_px, height_px)))
        row: dict[str, object] = {
            "source": source_path,
            "frame_index": frame_index,
            "timestamp_sec": timestamp_sec,
            "detection_id": detection_index,
            "class_id": int(box.cls.item()) if getattr(box, "cls", None) is not None else 0,
            "class_name": "crack",
            "confidence": float(box.conf.item()),
            "x1": x1,
            "y1": y1,
            "x2": x2,
            "y2": y2,
            "box_width_px": width_px,
            "box_height_px": height_px,
            "box_area_px": width_px * height_px,
            "estimated_crack_length_px": centerline_length_px,
            "estimated_crack_centerline_length_px": centerline_length_px,
            "estimated_crack_width_px": min(width_px, height_px),
            "estimated_crack_mean_width_px": mean_width_px,
            "estimated_crack_mask_area_px": mask_measurements.get("estimated_crack_mask_area_px", ""),
            "image_width_px": image_width,
            "image_height_px": image_height,
            "annotated_image_path": str(annotated_image_path) if annotated_image_path else "",
            "measurement_method": mask_measurements.get("measurement_method", "bbox_longest_side"),
        }
        add_scaled_measurements(row, config)
        row["length_category"] = classify_crack_length(row, config)
        row["crack_type"] = classify_crack_type(row, config)
        row["text_description"] = detection_description(row)
        rows.append(row)
    return rows


def save_annotated_result(result: object, output_path: Path) -> Path:
    output_path.parent.mkdir(parents=True, exist_ok=True)
    annotated = result.plot()
    cv2.imwrite(str(output_path), annotated)
    return output_path


def estimate_predicted_crack_sizes(
    config: PipelineConfig,
    source: Path,
    weights_path: Path | None = None,
    confidence: float = 0.25,
) -> pd.DataFrame:
    if YOLO is None:
        raise ImportError("Install ultralytics before prediction.")
    weights = weights_path or resolve_best_weights(config)
    if not weights.exists():
        raise FileNotFoundError(f"YOLO weights not found: {weights}")
    model = YOLO(str(weights))
    results = model.predict(
        source=str(source),
        conf=confidence,
        imgsz=config.image_size,
        device=select_device(config.device),
        save=False,
        verbose=False,
    )
    rows: list[dict[str, object]] = []
    image_summaries: list[dict[str, object]] = []
    annotated_dir = config.predictions_dir / "annotated_images"
    for result in results:
        source_path = Path(result.path)
        annotated_path = annotated_dir / f"{source_path.stem}_bbox.jpg"
        save_annotated_result(result, annotated_path)
        result_rows = result_to_detection_rows(result, config, annotated_image_path=annotated_path)
        rows.extend(result_rows)
        image_summaries.append(
            {
                "source": str(source_path.resolve()),
                "annotated_image_path": str(annotated_path),
                "detections": result_rows,
                "text": image_summary_text(str(source_path.resolve()), result_rows),
            }
        )
    prediction_df = pd.DataFrame(rows, columns=DETECTION_COLUMNS if not rows else None)
    write_prediction_reports(prediction_df, image_summaries, config.predictions_dir)
    return prediction_df


def parse_args() -> argparse.Namespace:
    parser = argparse.ArgumentParser(description="Prepare, train, and use a YOLO crack bounding-box pipeline.")
    parser.add_argument("command", choices=["prepare", "train", "evaluate", "predict", "all"])
    parser.add_argument("--dataset-root", type=Path, default=PipelineConfig.dataset_root)
    parser.add_argument("--output-root", type=Path, default=PipelineConfig.output_root)
    parser.add_argument("--model-name", default=PipelineConfig.model_name)
    parser.add_argument("--epochs", type=int, default=PipelineConfig.epochs)
    parser.add_argument("--batch-size", type=int, default=PipelineConfig.batch_size)
    parser.add_argument("--imgsz", type=int, default=PipelineConfig.image_size)
    parser.add_argument(
        "--device",
        default=None,
        help="Inference/training device: auto, cuda, cuda:N, mps, or cpu. Defaults to CUDA, then MPS, then CPU.",
    )
    parser.add_argument("--max-images", type=int, default=None, help="Optional cap for quick smoke tests.")
    parser.add_argument("--rebuild", action="store_true")
    parser.add_argument("--mm-per-pixel", type=float, default=None)
    parser.add_argument("--source", type=Path, default=None, help="Image or folder for prediction.")
    parser.add_argument("--weights", type=Path, default=None)
    parser.add_argument("--conf", type=float, default=0.25)
    parser.add_argument("--verified-downloads", action="store_true", help="Use normal SSL verification for model downloads.")
    parser.add_argument("--no-mlflow", action="store_true", help="Disable MLflow tracking for this run.")
    parser.add_argument("--mlflow-experiment-name", default=PipelineConfig.mlflow_experiment_name)
    parser.add_argument("--mlflow-tracking-uri", default=PipelineConfig.mlflow_tracking_uri)
    return parser.parse_args()


def main() -> None:
    args = parse_args()
    random.seed(42)
    np.random.seed(42)
    config = PipelineConfig(
        dataset_root=args.dataset_root,
        output_root=args.output_root,
        model_name=args.model_name,
        epochs=args.epochs,
        batch_size=args.batch_size,
        image_size=args.imgsz,
        device=args.device,
        mm_per_pixel=args.mm_per_pixel,
        allow_unverified_model_downloads=not args.verified_downloads,
        enable_mlflow=not args.no_mlflow,
        mlflow_experiment_name=args.mlflow_experiment_name,
        mlflow_tracking_uri=args.mlflow_tracking_uri,
    )
    config.output_root.mkdir(parents=True, exist_ok=True)
    (config.output_root / "pipeline_config.json").write_text(json.dumps(asdict(config), indent=2, default=str))

    with mlflow_run(config, args.command):
        if args.command in {"prepare", "all"}:
            box_df = prepare_yolo_detection_dataset(config, rebuild=args.rebuild, max_images=args.max_images)
            log_preparation_to_mlflow(config, box_df, max_images=args.max_images)
            print(f"Prepared YOLO detection data at {config.yolo_dataset_root}")
            print(f"Generated {len(box_df):,} pseudo bounding boxes.")

        if args.command in {"train", "all"}:
            weights = train_yolo(config)
            log_ultralytics_run_to_mlflow(weights)
            print(f"Best weights: {weights}")

        if args.command == "evaluate":
            metrics = evaluate_yolo(config, weights_path=args.weights)
            log_ultralytics_metrics_to_mlflow(metrics, prefix="test")

        if args.command == "predict":
            if args.source is None:
                raise ValueError("--source is required for predict.")
            prediction_df = estimate_predicted_crack_sizes(
                config,
                source=args.source,
                weights_path=args.weights,
                confidence=args.conf,
            )
            log_predictions_to_mlflow(config, prediction_df)
            print(prediction_df.head())
            print(f"Saved crack-size estimates to {config.predictions_dir / 'crack_size_estimates.csv'}")


if __name__ == "__main__":
    main()
````

<div class="usecase-section-header">Kavita Code: Video Crack Detection Pipeline</div>


<p class="usecase-subsection-header">Scenario For `yolo_crack_video_pipeline.py`: Detecting Cracks In Inspection Footage</p>

Imagine an inspector records a short video while walking along a concrete path. Processing every frame may be slow and unnecessary, so this script samples frames at a chosen interval and applies the image detector to those frames.

A realistic scenario is:

1. The user provides a video path to `/predict/video` or runs the video script directly.
2. The script opens the video with OpenCV.
3. It reads frame information such as FPS, total frame count, width, and height.
4. It processes every Nth frame based on `frame_stride`.
5. Each selected frame is sent through the same YOLO detector used for images.
6. Annotated frame images are saved for visual evidence.
7. Detection rows are collected into a table.
8. Text, JSON, CSV, and metadata files are written for review.

This script expands the use case from still-image inspection to video-based inspection while keeping the output format consistent.


<p class="usecase-subsection-header">Code Cell Explanation: Video Crack Detection Pipeline</p>

This code cell contains `scripts/yolo_crack_video_pipeline.py`, Kavita's video-processing layer. It reuses the image crack detector rather than creating a completely separate video model.

The script imports OpenCV to open videos and read frames, pandas to store detection rows, tqdm for progress bars, and shared functions from the image pipeline. This reuse keeps image and video outputs consistent.

`parse_args()` defines command-line settings: input video path, weights path, output folder, image size, device, confidence threshold, frame stride, optional maximum frames, millimetres-per-pixel, and whether to save an annotated video.

`open_video()` checks whether OpenCV can actually open the video. `build_frame_summary()` creates a readable summary for one processed frame. `process_video()` is the main workflow: it reads the video, processes every Nth frame, runs YOLO on selected frames, saves annotated frame images, converts detections into rows, writes reports, and records metadata such as FPS and processed frame count.

For a layperson, this script means the project can inspect footage by turning selected video frames into still-image crack detection tasks.


````python
from __future__ import annotations

import argparse
import json
from pathlib import Path

import cv2
import pandas as pd
from tqdm.auto import tqdm

try:
    from scripts.yolo_crack_bbox_pipeline import (
        CSV_ESTIMATES_NAME,
        DETECTION_COLUMNS,
        JSON_SUMMARY_NAME,
        TEXT_SUMMARY_NAME,
        PipelineConfig,
        YOLO,
        image_summary_text,
        result_to_detection_rows,
        resolve_best_weights,
        select_device,
        write_prediction_reports,
    )
except ImportError:
    from yolo_crack_bbox_pipeline import (
        CSV_ESTIMATES_NAME,
        DETECTION_COLUMNS,
        JSON_SUMMARY_NAME,
        TEXT_SUMMARY_NAME,
        PipelineConfig,
        YOLO,
        image_summary_text,
        result_to_detection_rows,
        resolve_best_weights,
        select_device,
        write_prediction_reports,
    )


def parse_args() -> argparse.Namespace:
    parser = argparse.ArgumentParser(
        description="Process video frames with a YOLOv11 crack detector and export reporting-ready text."
    )
    parser.add_argument("--video", type=Path, required=True, help="Input video file.")
    parser.add_argument("--weights", type=Path, default=None, help="Trained YOLO detection weights.")
    parser.add_argument("--output-root", type=Path, default=Path("artifacts/yolo11_video_reports"))
    parser.add_argument("--imgsz", type=int, default=PipelineConfig.image_size)
    parser.add_argument(
        "--device",
        default=None,
        help="Inference device: auto, cuda, cuda:N, mps, or cpu. Defaults to CUDA, then MPS, then CPU.",
    )
    parser.add_argument("--conf", type=float, default=0.25)
    parser.add_argument("--frame-stride", type=int, default=30, help="Process every Nth frame.")
    parser.add_argument("--max-frames", type=int, default=None, help="Optional processed-frame cap for testing.")
    parser.add_argument("--mm-per-pixel", type=float, default=None)
    parser.add_argument("--save-annotated-video", action="store_true")
    return parser.parse_args()


def open_video(video_path: Path) -> cv2.VideoCapture:
    capture = cv2.VideoCapture(str(video_path))
    if not capture.isOpened():
        raise FileNotFoundError(f"Could not open video: {video_path}")
    return capture


def build_frame_summary(
    video_path: Path,
    frame_index: int,
    timestamp_sec: float,
    rows: list[dict[str, object]],
    annotated_frame_path: Path,
) -> dict[str, object]:
    frame_label = f"{video_path.resolve()} frame {frame_index} at {timestamp_sec:.2f}s"
    return {
        "video": str(video_path.resolve()),
        "frame_index": frame_index,
        "timestamp_sec": timestamp_sec,
        "annotated_image_path": str(annotated_frame_path),
        "detections": rows,
        "text": image_summary_text(frame_label, rows),
    }


def process_video(args: argparse.Namespace) -> pd.DataFrame:
    if YOLO is None:
        raise ImportError("Install ultralytics before video inference.")

    config = PipelineConfig(
        output_root=args.output_root,
        image_size=args.imgsz,
        device=args.device,
        mm_per_pixel=args.mm_per_pixel,
        enable_mlflow=False,
    )
    weights = args.weights or resolve_best_weights(config)
    model = YOLO(str(weights))

    report_dir = args.output_root
    annotated_frames_dir = report_dir / "annotated_frames"
    annotated_frames_dir.mkdir(parents=True, exist_ok=True)

    capture = open_video(args.video)
    fps = capture.get(cv2.CAP_PROP_FPS) or 30.0
    total_frames = int(capture.get(cv2.CAP_PROP_FRAME_COUNT) or 0)
    frame_width = int(capture.get(cv2.CAP_PROP_FRAME_WIDTH) or 0)
    frame_height = int(capture.get(cv2.CAP_PROP_FRAME_HEIGHT) or 0)

    writer: cv2.VideoWriter | None = None
    if args.save_annotated_video:
        video_output_path = report_dir / f"{args.video.stem}_annotated.mp4"
        video_fps = max(fps / max(args.frame_stride, 1), 1.0)
        writer = cv2.VideoWriter(
            str(video_output_path),
            cv2.VideoWriter_fourcc(*"mp4v"),
            video_fps,
            (frame_width, frame_height),
        )

    rows: list[dict[str, object]] = []
    frame_summaries: list[dict[str, object]] = []
    processed_frames = 0
    progress = tqdm(total=total_frames or None, desc="Processing video frames")

    try:
        frame_index = 0
        while True:
            success, frame = capture.read()
            if not success:
                break

            if frame_index % max(args.frame_stride, 1) != 0:
                frame_index += 1
                progress.update(1)
                continue

            timestamp_sec = frame_index / fps
            result = model.predict(
                source=frame,
                conf=args.conf,
                imgsz=args.imgsz,
                device=select_device(config.device),
                save=False,
                verbose=False,
            )[0]
            annotated = result.plot()
            annotated_frame_path = annotated_frames_dir / f"{args.video.stem}_frame_{frame_index:06d}.jpg"
            cv2.imwrite(str(annotated_frame_path), annotated)
            if writer is not None:
                writer.write(annotated)

            frame_rows = result_to_detection_rows(
                result,
                config,
                annotated_image_path=annotated_frame_path,
                frame_index=frame_index,
                timestamp_sec=timestamp_sec,
            )
            for row in frame_rows:
                row["source"] = str(args.video.resolve())
            rows.extend(frame_rows)
            frame_summaries.append(
                build_frame_summary(args.video, frame_index, timestamp_sec, frame_rows, annotated_frame_path)
            )

            processed_frames += 1
            progress.update(1)
            if args.max_frames is not None and processed_frames >= args.max_frames:
                break

            frame_index += 1
    finally:
        progress.close()
        capture.release()
        if writer is not None:
            writer.release()

    prediction_df = pd.DataFrame(rows, columns=DETECTION_COLUMNS if not rows else None)
    write_prediction_reports(prediction_df, frame_summaries, report_dir)
    metadata = {
        "video": str(args.video.resolve()),
        "weights": str(weights),
        "frame_stride": args.frame_stride,
        "processed_frames": processed_frames,
        "total_video_frames": total_frames,
        "fps": fps,
        "confidence_threshold": args.conf,
        "image_size": args.imgsz,
        "csv_output": str(report_dir / CSV_ESTIMATES_NAME),
        "json_output": str(report_dir / JSON_SUMMARY_NAME),
        "text_output": str(report_dir / TEXT_SUMMARY_NAME),
        "annotated_frames_dir": str(annotated_frames_dir),
    }
    (report_dir / "video_report_metadata.json").write_text(json.dumps(metadata, indent=2))
    return prediction_df


def main() -> None:
    args = parse_args()
    prediction_df = process_video(args)
    print(prediction_df.head())
    print(f"Saved video crack-size estimates to {args.output_root / CSV_ESTIMATES_NAME}")
    print(f"Saved LLM-ready text summary to {args.output_root / TEXT_SUMMARY_NAME}")


if __name__ == "__main__":
    main()
````
